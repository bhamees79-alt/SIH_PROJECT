# Member 1 — HGB Activity Inference Demo

This notebook demonstrates the activity recognition inference pipeline used in the SIH project.

```python
import numpy as np
import joblib
import sys

sys.path.append('/content')

from hgb_inference import (
    LABEL_MAP,
    make_spatial_features,
    predict_activity
)

model_path = '/content/SIH_PROJECT/hgb_spatial_best.joblib'
model = joblib.load(model_path)

print('✅ Member 1 inference setup complete!')
print('Model:', type(model).__name__)
print('Number of classes:', len(LABEL_MAP))
print('Expected sequence shape: (32, 206)')
```

```python
repo_path = '/content/SIH_PROJECT/person2'

X_test = np.load(f'{repo_path}/X_test_scaled_padding_aware.npy')
y_test = np.load(f'{repo_path}/y_test.npy')

print('✅ Test data loaded!')
print('X_test shape:', X_test.shape)
print('y_test shape:', y_test.shape)
print('One sequence shape:', X_test[0].shape)
```

```python
# Predict activity for one 32-frame sequence
sequence = X_test[0]

prediction = predict_activity(sequence, model)

print('✅ Prediction completed!')
print('Predicted activity:', prediction['activity'])
print('Confidence:', round(prediction['confidence'], 4))
print('Predicted label:', prediction['label'])
print('Actual activity:', LABEL_MAP[int(y_test[0])])
```

```python
# Evaluate the HGB model on the test set
X_test_hgb = np.array([
    make_spatial_features(sequence)
    for sequence in X_test
])

y_test_pred = model.predict(X_test_hgb)
accuracy = np.mean(y_test_pred == y_test)

print('✅ Evaluation completed!')
print('Total test sequences:', len(X_test))
print('Development-test accuracy:', f'{accuracy * 100:.2f}%')
```

```python
# Example output for the rest of the SIH system
team_output = {
    'activity': prediction['activity'],
    'confidence': round(prediction['confidence'], 4),
    'timestamp': 12.4,
    'status': 'correct'
}

print('✅ Team output format:')
print(team_output)
```
