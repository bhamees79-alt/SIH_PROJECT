```python
from collections import defaultdict

ids_by_class = defaultdict(set)

results = model.track(
    source='/content/drive/MyDrive/Sih videos/incorrect4.mp4',
    tracker='/content/bytetrack_custom.yaml',
    conf=0.25,
    persist=True,
    stream=True
)

for result in results:

    if result.boxes is None or result.boxes.id is None:
        continue

    classes = result.boxes.cls.cpu().numpy().astype(int)
    ids = result.boxes.id.cpu().numpy().astype(int)

    for cls, track_id in zip(classes, ids):
        ids_by_class[model.names[cls]].add(int(track_id))

for cls, ids in ids_by_class.items():
    print(cls, "→", sorted(ids))
```

```python
import cv2
import mediapipe as mp
from mediapipe.tasks import python
from mediapipe.tasks.python import vision

video_path = "/content/drive/MyDrive/Sih videos/incorrect4.mp4"

base_options = python.BaseOptions(
    model_asset_path="/content/hand_landmarker.task"
)

options = vision.HandLandmarkerOptions(
    base_options=base_options,
    running_mode=vision.RunningMode.IMAGE,
    num_hands=1,
    min_hand_detection_confidence=0.3,
    min_hand_presence_confidence=0.3,
    min_tracking_confidence=0.3
)

detector = vision.HandLandmarker.create_from_options(options)

cap = cv2.VideoCapture(video_path)

total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

# Test 10 frames spread across the video
test_frames = [
    int(i * (total_frames - 1) / 9)
    for i in range(10)
]

print("Total frames:", total_frames)
print("Testing frames:", test_frames)

for frame_no in test_frames:

    cap.set(cv2.CAP_PROP_POS_FRAMES, frame_no)

    ret, frame = cap.read()

    if not ret:
        print(f"Frame {frame_no}: Could not read")
        continue

    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    mp_image = mp.Image(
        image_format=mp.ImageFormat.SRGB,
        data=rgb
    )

    result = detector.detect(mp_image)

    print(
        f"Frame {frame_no}: "
        f"{len(result.hand_landmarks)} hand(s)"
    )

cap.release()
```

```python
from ultralytics import YOLO
import cv2
import os

# Load YOLO model
model = YOLO("/content/best.pt")

# Input video
video_path = "/content/drive/MyDrive/Copy of c10.mp4"

# Output directory
output_dir = "/content/person2_output/c10_tracking"
os.makedirs(output_dir, exist_ok=True)

# Open video
cap = cv2.VideoCapture(video_path)

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Output video
output_video = os.path.join(output_dir, "c10_tracking.mp4")

fourcc = cv2.VideoWriter_fourcc(*"mp4v")
writer = cv2.VideoWriter(
    output_video,
    fourcc,
    fps,
    (width, height)
)

frame_number = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    frame_number += 1

    # YOLO + ByteTrack
    results = model.track(
        frame,
        persist=True,
        tracker="/content/bytetrack_custom.yaml",
        conf=0.5,
        verbose=False
    )

    result = results[0]

    # Draw tracking results
    annotated_frame = result.plot()

    writer.write(annotated_frame)

    # Print progress occasionally
    if frame_number % 50 == 0:
        print(f"Processed {frame_number} frames")

cap.release()
writer.release()

print("\nTracking completed!")
print("Output video:")
print(output_video)
```

```python
from ultralytics import YOLO
import cv2
import pandas as pd

# Load model
model = YOLO("/content/best.pt")

# Input video
video_path = "/content/drive/MyDrive/Copy of c10.mp4"

# Store tracking data
tracking_data = []

# Open video
cap = cv2.VideoCapture(video_path)

frame_number = 0

while True:
    ret, frame = cap.read()

    if not ret:
        break

    frame_number += 1

    # YOLO + ByteTrack
    results = model.track(
        frame,
        persist=True,
        tracker="/content/bytetrack_custom.yaml",
        conf=0.5,
        verbose=False
    )

    result = results[0]

    # Check whether objects were detected
    if result.boxes is not None and len(result.boxes) > 0:

        boxes = result.boxes

        # Bounding boxes
        xyxy = boxes.xyxy.cpu().numpy()

        # Confidence
        confidences = boxes.conf.cpu().numpy()

        # Class IDs
        class_ids = boxes.cls.cpu().numpy().astype(int)

        # Tracking IDs
        if boxes.id is not None:
            track_ids = boxes.id.cpu().numpy().astype(int)
        else:
            track_ids = [-1] * len(xyxy)

        # Save each detected object
        for box, confidence, class_id, track_id in zip(
            xyxy, confidences, class_ids, track_ids
        ):

            x1, y1, x2, y2 = box

            tracking_data.append({
                "frame": frame_number,
                "class_id": class_id,
                "class_name": model.names[class_id],
                "track_id": track_id,
                "x1": x1,
                "y1": y1,
                "x2": x2,
                "y2": y2,
                "confidence": confidence
            })

cap.release()

# Convert to DataFrame
tracking_df = pd.DataFrame(tracking_data)

# Save CSV
csv_path = "/content/person2_output/c10_tracking/c10_tracking_data.csv"

tracking_df.to_csv(csv_path, index=False)

print("Tracking CSV created successfully!")
print("Path:", csv_path)
print()
print("Number of rows:", len(tracking_df))
print()
print(tracking_df.head(10))
```

```python
import pandas as pd
import numpy as np
import cv2

# --------------------------------------------------
# 1. Load tracking CSV
# --------------------------------------------------

csv_path = "/content/person2_output/c10_tracking/c10_tracking_data.csv"

df = pd.read_csv(csv_path)

# --------------------------------------------------
# 2. Get ORIGINAL video dimensions
# --------------------------------------------------

video_path = "/content/drive/MyDrive/Copy of c10.mp4"

cap = cv2.VideoCapture(video_path)

frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

cap.release()

print("Video width :", frame_width)
print("Video height:", frame_height)
print("FPS         :", fps)

# --------------------------------------------------
# 3. Calculate bounding-box spatial features
# --------------------------------------------------

df["center_x"] = (df["x1"] + df["x2"]) / 2
df["center_y"] = (df["y1"] + df["y2"]) / 2

df["width"] = df["x2"] - df["x1"]
df["height"] = df["y2"] - df["y1"]

# --------------------------------------------------
# 4. Normalize spatial coordinates
# --------------------------------------------------

df["norm_center_x"] = df["center_x"] / frame_width
df["norm_center_y"] = df["center_y"] / frame_height

df["norm_width"] = df["width"] / frame_width
df["norm_height"] = df["height"] / frame_height

# --------------------------------------------------
# 5. Calculate movement
# --------------------------------------------------

# Sort properly before calculating movement
df = df.sort_values(
    ["class_name", "track_id", "frame"]
).reset_index(drop=True)

# Previous center position for the SAME tracked object
df["prev_center_x"] = df.groupby(
    ["class_name", "track_id"]
)["center_x"].shift(1)

df["prev_center_y"] = df.groupby(
    ["class_name", "track_id"]
)["center_y"].shift(1)

# Movement in pixels
df["movement_dx"] = (
    df["center_x"] - df["prev_center_x"]
)

df["movement_dy"] = (
    df["center_y"] - df["prev_center_y"]
)

# First observation of a track has no previous position
df["movement_dx"] = df["movement_dx"].fillna(0)
df["movement_dy"] = df["movement_dy"].fillna(0)

# Movement speed in pixels/frame
df["speed"] = np.sqrt(
    df["movement_dx"] ** 2 +
    df["movement_dy"] ** 2
)

# --------------------------------------------------
# 6. Save spatial features
# --------------------------------------------------

features_path = "/content/person2_output/c10_tracking/c10_spatial_features.csv"

df.to_csv(features_path, index=False)

print("\nSpatial features created successfully!")
print("Saved to:", features_path)

print("\nFeature columns:")
print(df.columns.tolist())

print("\nSample:")
display(df.head(10))
```

```python
import cv2
import pandas as pd
import mediapipe as mp
from mediapipe.tasks import python
from mediapipe.tasks.python import vision

# --------------------------------------------------
# Paths
# --------------------------------------------------

video_path = "/content/drive/MyDrive/Copy of c10.mp4"

hand_model_path = "/content/hand_landmarker.task"

output_path = "/content/person2_output/c10_tracking/c10_hand_pose.csv"

# --------------------------------------------------
# Create MediaPipe Hand Landmarker
# --------------------------------------------------

base_options = python.BaseOptions(
    model_asset_path=hand_model_path
)

options = vision.HandLandmarkerOptions(
    base_options=base_options,
    running_mode=vision.RunningMode.VIDEO,
    num_hands=2,
    min_hand_detection_confidence=0.5,
    min_hand_presence_confidence=0.5,
    min_tracking_confidence=0.5
)

landmarker = vision.HandLandmarker.create_from_options(options)

# --------------------------------------------------
# Open video
# --------------------------------------------------

cap = cv2.VideoCapture(video_path)

fps = cap.get(cv2.CAP_PROP_FPS)

print("Video FPS:", fps)

# --------------------------------------------------
# Store hand-pose data
# --------------------------------------------------

hand_data = []

frame_number = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    frame_number += 1

    # OpenCV BGR -> RGB
    rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    # Convert to MediaPipe image
    mp_image = mp.Image(
        image_format=mp.ImageFormat.SRGB,
        data=rgb_frame
    )

    # Timestamp must increase for VIDEO mode
    timestamp_ms = int((frame_number - 1) * 1000 / fps)

    # Detect hands
    result = landmarker.detect_for_video(
        mp_image,
        timestamp_ms
    )

    # --------------------------------------------------
    # Prepare one row per video frame
    # --------------------------------------------------

    row = {
        "frame": frame_number,
        "left_hand_present": 0,
        "right_hand_present": 0
    }

    # Initialize all landmark values with 0
    for hand_side in ["left", "right"]:

        for landmark_id in range(21):

            row[f"{hand_side}_lm{landmark_id}_x"] = 0.0
            row[f"{hand_side}_lm{landmark_id}_y"] = 0.0
            row[f"{hand_side}_lm{landmark_id}_z"] = 0.0

    # --------------------------------------------------
    # Extract detected hands
    # --------------------------------------------------

    for hand_index, landmarks in enumerate(result.hand_landmarks):

        # Determine handedness
        handedness = result.handedness[hand_index][0].category_name

        # MediaPipe may report Left / Right
        hand_side = handedness.lower()

        if hand_side not in ["left", "right"]:
            continue

        row[f"{hand_side}_hand_present"] = 1

        # 21 landmarks
        for landmark_id, landmark in enumerate(landmarks):

            row[f"{hand_side}_lm{landmark_id}_x"] = landmark.x
            row[f"{hand_side}_lm{landmark_id}_y"] = landmark.y
            row[f"{hand_side}_lm{landmark_id}_z"] = landmark.z

    hand_data.append(row)

    if frame_number % 50 == 0:
        print(f"Processed {frame_number} frames")

# --------------------------------------------------
# Close everything
# --------------------------------------------------

cap.release()
landmarker.close()

# --------------------------------------------------
# Create DataFrame
# --------------------------------------------------

hand_df = pd.DataFrame(hand_data)

# --------------------------------------------------
# Save CSV
# --------------------------------------------------

hand_df.to_csv(output_path, index=False)

print("\nHand pose extraction completed!")
print("Saved to:", output_path)

print("\nNumber of frames:", len(hand_df))
print("Number of features:", len(hand_df.columns))

print("\nFirst columns:")
print(hand_df.columns[:15].tolist())
```

```python
import pandas as pd

# --------------------------------------------------
# Load both feature files
# --------------------------------------------------

spatial_path = "/content/person2_output/c10_tracking/c10_spatial_features.csv"
hand_path = "/content/person2_output/c10_tracking/c10_hand_pose.csv"

spatial_df = pd.read_csv(spatial_path)
hand_df = pd.read_csv(hand_path)

print("Spatial data shape:", spatial_df.shape)
print("Hand pose shape:", hand_df.shape)

# --------------------------------------------------
# Merge using frame number
# --------------------------------------------------

combined_df = spatial_df.merge(
    hand_df,
    on="frame",
    how="left"
)

# --------------------------------------------------
# Save combined data
# --------------------------------------------------

combined_path = "/content/person2_output/c10_tracking/c10_combined_features.csv"

combined_df.to_csv(combined_path, index=False)

print("\nCombined features created successfully!")
print("Saved to:", combined_path)

print("\nCombined shape:", combined_df.shape)

print("\nColumns:")
print(combined_df.columns.tolist())
```

```python
import pandas as pd
import numpy as np

# Load combined data
combined_path = "/content/person2_output/c10_tracking/c10_combined_features.csv"

df = pd.read_csv(combined_path)

# --------------------------------------------------
# Video dimensions
# --------------------------------------------------

VIDEO_WIDTH = 1080
VIDEO_HEIGHT = 1920

# --------------------------------------------------
# Object classes
# --------------------------------------------------

object_classes = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# --------------------------------------------------
# Function:
# minimum distance from hand landmarks
# to object bounding box
# --------------------------------------------------

def hand_object_distance(row, hand_side):

    # Check whether hand exists
    if row[f"{hand_side}_hand_present"] == 0:
        return 1.0

    # Object bounding box
    x1 = row["x1"]
    y1 = row["y1"]
    x2 = row["x2"]
    y2 = row["y2"]

    min_distance = np.inf

    # 21 hand landmarks
    for landmark_id in range(21):

        # MediaPipe coordinates are normalized
        lx = row[f"{hand_side}_lm{landmark_id}_x"]
        ly = row[f"{hand_side}_lm{landmark_id}_y"]

        # Convert to pixel coordinates
        px = lx * VIDEO_WIDTH
        py = ly * VIDEO_HEIGHT

        # Distance from point to bounding box
        dx = max(x1 - px, 0, px - x2)
        dy = max(y1 - py, 0, py - y2)

        distance = np.sqrt(dx**2 + dy**2)

        min_distance = min(min_distance, distance)

    # Normalize by image diagonal
    diagonal = np.sqrt(
        VIDEO_WIDTH**2 + VIDEO_HEIGHT**2
    )

    return min_distance / diagonal


# --------------------------------------------------
# Create interaction features
# --------------------------------------------------

for object_class in object_classes:

    for hand_side in ["left", "right"]:

        column_name = (
            f"{hand_side}_hand_to_{object_class}"
        )

        # Only calculate for this object class
        mask = df["class_name"] == object_class

        df.loc[mask, column_name] = df.loc[
            mask
        ].apply(
            lambda row: hand_object_distance(
                row,
                hand_side
            ),
            axis=1
        )

# Fill missing values
interaction_columns = [
    col for col in df.columns
    if "_hand_to_" in col
]

df[interaction_columns] = (
    df[interaction_columns]
    .fillna(1.0)
)

# --------------------------------------------------
# Save
# --------------------------------------------------

output_path = (
    "/content/person2_output/"
    "c10_tracking/c10_interaction_features.csv"
)

df.to_csv(output_path, index=False)

print("Hand-object interaction features created!")
print()
print("Number of interaction features:",
      len(interaction_columns))

print()
print("Interaction columns:")
print(interaction_columns)

print()
print("Saved to:")
print(output_path)
```

```python
import pandas as pd
import numpy as np

# --------------------------------------------------
# Load interaction features
# --------------------------------------------------

input_path = (
    "/content/person2_output/"
    "c10_tracking/c10_interaction_features.csv"
)

df = pd.read_csv(input_path)

# --------------------------------------------------
# Object classes
# --------------------------------------------------

object_classes = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# --------------------------------------------------
# Spatial features we want for each object
# --------------------------------------------------

spatial_features = [
    "norm_center_x",
    "norm_center_y",
    "norm_width",
    "norm_height",
    "confidence",
    "movement_dx",
    "movement_dy",
    "speed"
]

# --------------------------------------------------
# Create frame-level table
# --------------------------------------------------

frames = sorted(df["frame"].unique())

frame_features = []

for frame_number in frames:

    frame_df = df[df["frame"] == frame_number]

    row = {
        "frame": frame_number
    }

    # ------------------------------------------------
    # Object features
    # ------------------------------------------------

    for object_class in object_classes:

        objects = frame_df[
            frame_df["class_name"] == object_class
        ]

        # If object exists
        if len(objects) > 0:

            # Use highest-confidence detection
            obj = objects.loc[
                objects["confidence"].idxmax()
            ]

            row[f"{object_class}_visible"] = 1

            for feature in spatial_features:

                row[f"{object_class}_{feature}"] = (
                    obj[feature]
                )

        # Object not detected
        else:

            row[f"{object_class}_visible"] = 0

            for feature in spatial_features:

                row[f"{object_class}_{feature}"] = 0.0

    # ------------------------------------------------
    # Hand features
    # ------------------------------------------------

    hand_columns = [
        col for col in df.columns
        if col.startswith("left_hand_")
        or col.startswith("right_hand_")
    ]

    # Hand columns are identical for every object
    # within the same frame, so take the first row.
    first_row = frame_df.iloc[0]

    for col in hand_columns:
        row[col] = first_row[col]

    # ------------------------------------------------
    # Hand-object interaction features
    # ------------------------------------------------

    interaction_columns = [
        col for col in df.columns
        if "_hand_to_" in col
    ]

    for col in interaction_columns:

        # Interaction feature only exists on the
        # corresponding object row.
        values = frame_df[col].dropna()

        if len(values) > 0:
            row[col] = values.iloc[0]
        else:
            row[col] = 1.0

    frame_features.append(row)

# --------------------------------------------------
# Create DataFrame
# --------------------------------------------------

frame_df = pd.DataFrame(frame_features)

# --------------------------------------------------
# Sort by frame
# --------------------------------------------------

frame_df = frame_df.sort_values(
    "frame"
).reset_index(drop=True)

# --------------------------------------------------
# Fill missing values
# --------------------------------------------------

frame_df = frame_df.fillna(0)

# --------------------------------------------------
# Save
# --------------------------------------------------

output_path = (
    "/content/person2_output/"
    "c10_tracking/c10_frame_features.csv"
)

frame_df.to_csv(
    output_path,
    index=False
)

# --------------------------------------------------
# Results
# --------------------------------------------------

print("Frame-level feature table created!")
print()
print("Shape:", frame_df.shape)
print()
print("Number of frames:", len(frame_df))
print()
print("Number of features:", len(frame_df.columns))
print()
print("Saved to:")
print(output_path)

print("\nFirst 5 rows:")
display(frame_df.head())
```

```python
import pandas as pd
import numpy as np

# --------------------------------------------------
# Load frame-level features
# --------------------------------------------------

input_path = (
    "/content/person2_output/"
    "c10_tracking/c10_frame_features.csv"
)

df = pd.read_csv(input_path)

# --------------------------------------------------
# Object classes
# --------------------------------------------------

objects = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# --------------------------------------------------
# Calculate object-object distances
# --------------------------------------------------

distance_features = []

for i in range(len(objects)):

    for j in range(i + 1, len(objects)):

        obj1 = objects[i]
        obj2 = objects[j]

        x1 = df[f"{obj1}_norm_center_x"]
        y1 = df[f"{obj1}_norm_center_y"]

        x2 = df[f"{obj2}_norm_center_x"]
        y2 = df[f"{obj2}_norm_center_y"]

        distance = np.sqrt(
            (x1 - x2) ** 2 +
            (y1 - y2) ** 2
        )

        column_name = (
            f"{obj1}_to_{obj2}_distance"
        )

        df[column_name] = distance

        distance_features.append(column_name)

# --------------------------------------------------
# Save
# --------------------------------------------------

output_path = (
    "/content/person2_output/"
    "c10_tracking/c10_final_frame_features.csv"
)

df.to_csv(output_path, index=False)

print("Object-object interaction features created!")
print()
print("Number of new features:", len(distance_features))
print()
print("New features:")

for feature in distance_features:
    print(feature)

print()
print("Final shape:", df.shape)
print()
print("Saved to:")
print(output_path)
```

```python
import os
import cv2
import pandas as pd
from tqdm import tqdm
from ultralytics import YOLO

# Load the trained YOLO model
model = YOLO("/content/best.pt")

# Output directory
test_output_dir = "/content/person2_output/batch_test"
os.makedirs(test_output_dir, exist_ok=True)

# Take only the first 2 project videos for testing
test_videos = project_videos[:2]

print("Testing on:")
for video in test_videos:
    print(" -", video)

print("\nStarting test...\n")

for video_path in test_videos:

    video_name = os.path.splitext(os.path.basename(video_path))[0]
    video_output_dir = os.path.join(test_output_dir, video_name)
    os.makedirs(video_output_dir, exist_ok=True)

    csv_path = os.path.join(
        video_output_dir,
        f"{video_name}_tracking.csv"
    )

    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        print(f"Could not open: {video_path}")
        continue

    frame_number = 0
    tracking_rows = []

    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    print(f"\nProcessing: {video_name}")
    print(f"Total frames: {total_frames}")

    while True:

        ret, frame = cap.read()

        if not ret:
            break

        frame_number += 1

        results = model.track(
            frame,
            persist=True,
            tracker="/content/bytetrack_custom.yaml",
            conf=0.5,
            verbose=False
        )

        result = results[0]

        if result.boxes is not None and len(result.boxes) > 0:

            boxes = result.boxes

            xyxy = boxes.xyxy.cpu().numpy()
            confs = boxes.conf.cpu().numpy()
            classes = boxes.cls.cpu().numpy().astype(int)

            if boxes.id is not None:
                track_ids = boxes.id.cpu().numpy().astype(int)
            else:
                track_ids = [-1] * len(xyxy)

            for box, conf, cls, track_id in zip(
                xyxy, confs, classes, track_ids
            ):

                x1, y1, x2, y2 = box

                tracking_rows.append({
                    "frame": frame_number,
                    "class_id": int(cls),
                    "class_name": model.names[int(cls)],
                    "track_id": int(track_id),
                    "x1": float(x1),
                    "y1": float(y1),
                    "x2": float(x2),
                    "y2": float(y2),
                    "confidence": float(conf)
                })

    cap.release()

    tracking_df = pd.DataFrame(tracking_rows)
    tracking_df.to_csv(csv_path, index=False)

    print("Tracking completed!")
    print("Rows:", len(tracking_df))
    print("Saved:", csv_path)

print("\nTEST BATCH COMPLETED!")
```

```python
import os
import pandas as pd
import numpy as np

for video_name in ["C4", "C5"]:

    tracking_path = (
        f"/content/person2_output/batch_test/"
        f"{video_name}/{video_name}_tracking.csv"
    )

    output_path = (
        f"/content/person2_output/batch_test/"
        f"{video_name}/{video_name}_spatial_features.csv"
    )

    df = pd.read_csv(tracking_path)

    # Get actual video dimensions
    video_path = next(
        p for p in project_videos
        if os.path.splitext(os.path.basename(p))[0] == video_name
    )

    cap = cv2.VideoCapture(video_path)

    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

    cap.release()

    # Bounding-box geometry
    df["center_x"] = (df["x1"] + df["x2"]) / 2
    df["center_y"] = (df["y1"] + df["y2"]) / 2

    df["width"] = df["x2"] - df["x1"]
    df["height"] = df["y2"] - df["y1"]

    # Normalize using actual video dimensions
    df["norm_center_x"] = df["center_x"] / width
    df["norm_center_y"] = df["center_y"] / height

    df["norm_width"] = df["width"] / width
    df["norm_height"] = df["height"] / height

    # Sort before calculating movement
    df = df.sort_values(
        ["class_name", "track_id", "frame"]
    ).reset_index(drop=True)

    # Previous position for each tracked object
    df["prev_center_x"] = (
        df.groupby(["class_name", "track_id"])["center_x"]
        .shift(1)
    )

    df["prev_center_y"] = (
        df.groupby(["class_name", "track_id"])["center_y"]
        .shift(1)
    )

    # Frame-to-frame displacement
    df["movement_dx"] = (
        df["center_x"] - df["prev_center_x"]
    )

    df["movement_dy"] = (
        df["center_y"] - df["prev_center_y"]
    )

    # Speed in pixels/frame
    df["speed"] = np.sqrt(
        df["movement_dx"].fillna(0)**2 +
        df["movement_dy"].fillna(0)**2
    )

    # Replace missing movement values for first observation
    df["movement_dx"] = df["movement_dx"].fillna(0)
    df["movement_dy"] = df["movement_dy"].fillna(0)

    df.to_csv(output_path, index=False)

    print("=" * 60)
    print(video_name)
    print("=" * 60)
    print("Video dimensions:", width, "x", height)
    print("Shape:", df.shape)
    print("Saved:", output_path)
```

```python
import os
import cv2
import pandas as pd
import mediapipe as mp
from mediapipe.tasks import python
from mediapipe.tasks.python import vision

# MediaPipe hand model
base_options = python.BaseOptions(
    model_asset_path="/content/hand_landmarker.task"
)

options = vision.HandLandmarkerOptions(
    base_options=base_options,
    running_mode=vision.RunningMode.VIDEO,
    num_hands=2,
    min_hand_detection_confidence=0.5,
    min_hand_presence_confidence=0.5,
    min_tracking_confidence=0.5
)

hand_landmarker = vision.HandLandmarker.create_from_options(options)

for video_name in ["C4", "C5"]:

    video_path = next(
        p for p in project_videos
        if os.path.splitext(os.path.basename(p))[0] == video_name
    )

    output_path = (
        f"/content/person2_output/batch_test/"
        f"{video_name}/{video_name}_hand_pose.csv"
    )

    cap = cv2.VideoCapture(video_path)

    fps = cap.get(cv2.CAP_PROP_FPS)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    rows = []
    frame_number = 0

    print("=" * 60)
    print(video_name)
    print("FPS:", fps)
    print("Total frames:", total_frames)

    while True:

        ret, frame = cap.read()

        if not ret:
            break

        frame_number += 1

        # Convert BGR -> RGB
        rgb_frame = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2RGB
        )

        mp_image = mp.Image(
            image_format=mp.ImageFormat.SRGB,
            data=rgb_frame
        )

        timestamp_ms = int(
            (frame_number - 1) * 1000 / fps
        )

        result = hand_landmarker.detect_for_video(
            mp_image,
            timestamp_ms
        )

        row = {
            "frame": frame_number,
            "left_hand_present": 0,
            "right_hand_present": 0
        }

        # Initialize all landmark values
        for hand_side in ["left", "right"]:
            for lm_idx in range(21):
                row[f"{hand_side}_lm{lm_idx}_x"] = np.nan
                row[f"{hand_side}_lm{lm_idx}_y"] = np.nan
                row[f"{hand_side}_lm{lm_idx}_z"] = np.nan

        # Store detected hands
        if result.hand_landmarks:

            for hand_idx, landmarks in enumerate(
                result.hand_landmarks
            ):

                if hand_idx >= len(result.handedness):
                    continue

                handedness = result.handedness[hand_idx][0]
                hand_label = handedness.category_name.lower()

                if hand_label not in ["left", "right"]:
                    continue

                row[f"{hand_label}_hand_present"] = 1

                for lm_idx, landmark in enumerate(landmarks):

                    row[f"{hand_label}_lm{lm_idx}_x"] = landmark.x
                    row[f"{hand_label}_lm{lm_idx}_y"] = landmark.y
                    row[f"{hand_label}_lm{lm_idx}_z"] = landmark.z

        rows.append(row)

    cap.release()

    hand_df = pd.DataFrame(rows)

    hand_df.to_csv(
        output_path,
        index=False
    )

    print("Hand pose extraction completed!")
    print("Shape:", hand_df.shape)
    print("Hands detected:")
    print(
        "Left:",
        int(hand_df["left_hand_present"].sum())
    )
    print(
        "Right:",
        int(hand_df["right_hand_present"].sum())
    )
    print("Saved:", output_path)

hand_landmarker.close()

print("\nSTEP 17 COMPLETED!")
```

```python
import cv2
import pandas as pd
import numpy as np
import mediapipe as mp
from mediapipe.tasks import python
from mediapipe.tasks.python import vision

video_name = "C5"

video_path = next(
    p for p in project_videos
    if os.path.splitext(os.path.basename(p))[0] == video_name
)

output_path = (
    f"/content/person2_output/batch_test/"
    f"{video_name}/{video_name}_hand_pose.csv"
)

# Create a FRESH MediaPipe landmarker for C5
base_options = python.BaseOptions(
    model_asset_path="/content/hand_landmarker.task"
)

options = vision.HandLandmarkerOptions(
    base_options=base_options,
    running_mode=vision.RunningMode.VIDEO,
    num_hands=2,
    min_hand_detection_confidence=0.5,
    min_hand_presence_confidence=0.5,
    min_tracking_confidence=0.5
)

hand_landmarker = vision.HandLandmarker.create_from_options(options)

cap = cv2.VideoCapture(video_path)

fps = cap.get(cv2.CAP_PROP_FPS)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

rows = []
frame_number = 0

print("=" * 60)
print(video_name)
print("FPS:", fps)
print("Total frames:", total_frames)

while True:

    ret, frame = cap.read()

    if not ret:
        break

    frame_number += 1

    rgb_frame = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2RGB
    )

    mp_image = mp.Image(
        image_format=mp.ImageFormat.SRGB,
        data=rgb_frame
    )

    # Fresh landmarker => timestamp starts from zero
    timestamp_ms = int(
        (frame_number - 1) * 1000 / fps
    )

    result = hand_landmarker.detect_for_video(
        mp_image,
        timestamp_ms
    )

    row = {
        "frame": frame_number,
        "left_hand_present": 0,
        "right_hand_present": 0
    }

    for hand_side in ["left", "right"]:
        for lm_idx in range(21):
            row[f"{hand_side}_lm{lm_idx}_x"] = np.nan
            row[f"{hand_side}_lm{lm_idx}_y"] = np.nan
            row[f"{hand_side}_lm{lm_idx}_z"] = np.nan

    if result.hand_landmarks:

        for hand_idx, landmarks in enumerate(
            result.hand_landmarks
        ):

            if hand_idx >= len(result.handedness):
                continue

            handedness = result.handedness[hand_idx][0]
            hand_label = handedness.category_name.lower()

            if hand_label not in ["left", "right"]:
                continue

            row[f"{hand_label}_hand_present"] = 1

            for lm_idx, landmark in enumerate(landmarks):

                row[f"{hand_label}_lm{lm_idx}_x"] = landmark.x
                row[f"{hand_label}_lm{lm_idx}_y"] = landmark.y
                row[f"{hand_label}_lm{lm_idx}_z"] = landmark.z

    rows.append(row)

cap.release()
hand_landmarker.close()

hand_df = pd.DataFrame(rows)

hand_df.to_csv(
    output_path,
    index=False
)

print("Hand pose extraction completed!")
print("Shape:", hand_df.shape)
print("Hands detected:")
print("Left:", int(hand_df["left_hand_present"].sum()))
print("Right:", int(hand_df["right_hand_present"].sum()))
print("Saved:", output_path)
```

```python
import os
import numpy as np
import pandas as pd

object_classes = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

for video_name in ["C4", "C5"]:

    base_dir = (
        f"/content/person2_output/batch_test/{video_name}"
    )

    combined_path = os.path.join(
        base_dir,
        f"{video_name}_combined_features.csv"
    )

    output_path = os.path.join(
        base_dir,
        f"{video_name}_interaction_features.csv"
    )

    df = pd.read_csv(combined_path)

    # Get actual video dimensions
    video_path = next(
        p for p in project_videos
        if os.path.splitext(os.path.basename(p))[0] == video_name
    )

    cap = cv2.VideoCapture(video_path)

    img_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    img_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

    cap.release()

    image_diagonal = np.sqrt(
        img_width ** 2 + img_height ** 2
    )

    # Create interaction columns
    for obj in object_classes:

        df[f"left_hand_to_{obj}"] = 1.0
        df[f"right_hand_to_{obj}"] = 1.0

    # Process frame by frame
    for frame_num in df["frame"].unique():

        frame_mask = df["frame"] == frame_num
        frame_rows = df.loc[frame_mask]

        # Hand landmark coordinates
        left_points = []
        right_points = []

        for lm_idx in range(21):

            lx = f"left_lm{lm_idx}_x"
            ly = f"left_lm{lm_idx}_y"

            rx = f"right_lm{lm_idx}_x"
            ry = f"right_lm{lm_idx}_y"

            if (
                lx in df.columns and
                ly in df.columns
            ):
                for _, row in frame_rows.iloc[:1].iterrows():

                    if pd.notna(row[lx]) and pd.notna(row[ly]):
                        left_points.append([
                            row[lx] * img_width,
                            row[ly] * img_height
                        ])

            if (
                rx in df.columns and
                ry in df.columns
            ):
                for _, row in frame_rows.iloc[:1].iterrows():

                    if pd.notna(row[rx]) and pd.notna(row[ry]):
                        right_points.append([
                            row[rx] * img_width,
                            row[ry] * img_height
                        ])

        left_points = np.array(left_points)
        right_points = np.array(right_points)

        # Calculate distance to bounding boxes
        for obj in object_classes:

            obj_rows = frame_rows[
                frame_rows["class_name"] == obj
            ]

            if len(obj_rows) == 0:
                continue

            # Use highest-confidence detection if multiple instances exist
            obj_row = obj_rows.loc[
                obj_rows["confidence"].idxmax()
            ]

            x1 = obj_row["x1"]
            y1 = obj_row["y1"]
            x2 = obj_row["x2"]
            y2 = obj_row["y2"]

            def point_to_box_distance(points):

                if len(points) == 0:
                    return 1.0

                px = points[:, 0]
                py = points[:, 1]

                dx = np.maximum(
                    np.maximum(x1 - px, 0),
                    px - x2
                )

                dy = np.maximum(
                    np.maximum(y1 - py, 0),
                    py - y2
                )

                distances = np.sqrt(
                    dx ** 2 + dy ** 2
                )

                return float(
                    np.min(distances) / image_diagonal
                )

            left_distance = point_to_box_distance(
                left_points
            )

            right_distance = point_to_box_distance(
                right_points
            )

            df.loc[
                frame_mask,
                f"left_hand_to_{obj}"
            ] = left_distance

            df.loc[
                frame_mask,
                f"right_hand_to_{obj}"
            ] = right_distance

    df.to_csv(
        output_path,
        index=False
    )

    print("=" * 60)
    print(video_name)
    print("=" * 60)

    print("Shape:", df.shape)

    print("\nInteraction features:")

    for obj in object_classes:
        print(
            f"left_hand_to_{obj}, "
            f"right_hand_to_{obj}"
        )

    print("\nSaved:", output_path)
```

```python
import os
import numpy as np
import pandas as pd

object_classes = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# All unique object pairs
object_pairs = []

for i in range(len(object_classes)):
    for j in range(i + 1, len(object_classes)):
        object_pairs.append(
            (object_classes[i], object_classes[j])
        )

for video_name in ["C4", "C5"]:

    base_dir = (
        f"/content/person2_output/batch_test/{video_name}"
    )

    interaction_path = os.path.join(
        base_dir,
        f"{video_name}_interaction_features.csv"
    )

    output_path = os.path.join(
        base_dir,
        f"{video_name}_final_object_features.csv"
    )

    df = pd.read_csv(interaction_path)

    # Get actual video dimensions
    video_path = next(
        p for p in project_videos
        if os.path.splitext(os.path.basename(p))[0] == video_name
    )

    cap = cv2.VideoCapture(video_path)

    img_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    img_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

    cap.release()

    image_diagonal = np.sqrt(
        img_width ** 2 + img_height ** 2
    )

    # Create distance columns
    for obj1, obj2 in object_pairs:

        column_name = (
            f"{obj1}_to_{obj2}_distance"
        )

        df[column_name] = 1.0

    # Calculate distances frame-by-frame
    for frame_num in df["frame"].unique():

        frame_mask = df["frame"] == frame_num
        frame_rows = df.loc[frame_mask]

        # Store one representative detection per object
        object_boxes = {}

        for obj in object_classes:

            obj_rows = frame_rows[
                frame_rows["class_name"] == obj
            ]

            if len(obj_rows) == 0:
                continue

            # Highest-confidence detection
            obj_row = obj_rows.loc[
                obj_rows["confidence"].idxmax()
            ]

            object_boxes[obj] = (
                obj_row["center_x"],
                obj_row["center_y"]
            )

        # Calculate pairwise center distances
        for obj1, obj2 in object_pairs:

            column_name = (
                f"{obj1}_to_{obj2}_distance"
            )

            if obj1 not in object_boxes:
                continue

            if obj2 not in object_boxes:
                continue

            x1, y1 = object_boxes[obj1]
            x2, y2 = object_boxes[obj2]

            distance = np.sqrt(
                (x1 - x2) ** 2 +
                (y1 - y2) ** 2
            )

            normalized_distance = (
                distance / image_diagonal
            )

            df.loc[
                frame_mask,
                column_name
            ] = normalized_distance

    df.to_csv(
        output_path,
        index=False
    )

    print("=" * 60)
    print(video_name)
    print("=" * 60)

    print("Shape:", df.shape)

    print("\nObject-object distance features:")

    for obj1, obj2 in object_pairs:
        print(
            f"{obj1}_to_{obj2}_distance"
        )

    print("\nSaved:", output_path)
```

```python
import os
import numpy as np
import pandas as pd

object_classes = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

for video_name in ["C4", "C5"]:

    print("=" * 70)
    print(video_name)
    print("=" * 70)

    base_dir = f"/content/person2_output/batch_test/{video_name}"

    input_path = os.path.join(
        base_dir,
        f"{video_name}_final_object_features.csv"
    )

    output_path = os.path.join(
        base_dir,
        f"{video_name}_frame_features.csv"
    )

    # Load object-level features
    df = pd.read_csv(input_path)

    print("Input shape:", df.shape)

    # Get all unique frames
    frames = sorted(df["frame"].unique())

    frame_rows = []

    for frame_num in frames:

        frame_df = df[df["frame"] == frame_num]

        row = {
            "frame": frame_num
        }

        # --------------------------------------------------
        # Object features
        # --------------------------------------------------

        for obj in object_classes:

            obj_df = frame_df[
                frame_df["class_name"] == obj
            ]

            if len(obj_df) > 0:

                # Keep highest-confidence detection
                best = obj_df.loc[
                    obj_df["confidence"].idxmax()
                ]

                row[f"{obj}_present"] = 1

                row[f"{obj}_confidence"] = best["confidence"]

                row[f"{obj}_center_x"] = best["center_x"]
                row[f"{obj}_center_y"] = best["center_y"]

                row[f"{obj}_width"] = (
                    best["x2"] - best["x1"]
                )

                row[f"{obj}_height"] = (
                    best["y2"] - best["y1"]
                )

                row[f"{obj}_area"] = (
                    (best["x2"] - best["x1"]) *
                    (best["y2"] - best["y1"])
                )

            else:

                row[f"{obj}_present"] = 0

                row[f"{obj}_confidence"] = 0.0

                row[f"{obj}_center_x"] = 0.0
                row[f"{obj}_center_y"] = 0.0

                row[f"{obj}_width"] = 0.0
                row[f"{obj}_height"] = 0.0
                row[f"{obj}_area"] = 0.0

        # --------------------------------------------------
        # Preserve object-object distance features
        # --------------------------------------------------

        distance_columns = [
            col for col in df.columns
            if "_to_" in col and col.endswith("_distance")
        ]

        for col in distance_columns:

            values = frame_df[col].dropna()

            if len(values) > 0:
                row[col] = values.iloc[0]
            else:
                row[col] = 1.0

        # --------------------------------------------------
        # Preserve hand/object interaction features
        # --------------------------------------------------

        interaction_columns = [
            col for col in df.columns
            if (
                col.startswith("left_hand_")
                or col.startswith("right_hand_")
                or col.startswith("left_")
                or col.startswith("right_")
            )
        ]

        for col in interaction_columns:

            if col not in row:

                values = frame_df[col].dropna()

                if len(values) > 0:
                    row[col] = values.iloc[0]
                else:
                    row[col] = np.nan

        frame_rows.append(row)

    # Create frame-level dataframe
    frame_df = pd.DataFrame(frame_rows)

    # Sort by frame
    frame_df = frame_df.sort_values("frame").reset_index(drop=True)

    # Save
    frame_df.to_csv(output_path, index=False)

    print("Output shape:", frame_df.shape)
    print("Number of frames:", len(frame_df))
    print("Saved:", output_path)

    print("\nFirst 5 rows:")
    display(frame_df.head())
```

```python
import os
import pandas as pd
import numpy as np

for video_name in ["C4", "C5"]:

    print("=" * 70)
    print(video_name)
    print("=" * 70)

    base_dir = f"/content/person2_output/batch_test/{video_name}"

    input_path = os.path.join(
        base_dir,
        f"{video_name}_frame_features.csv"
    )

    output_path = os.path.join(
        base_dir,
        f"{video_name}_clean_features.csv"
    )

    df = pd.read_csv(input_path)

    # --------------------------------------------------
    # Identify hand-pose / hand-interaction columns
    # --------------------------------------------------

    hand_columns = [
        col for col in df.columns
        if (
            col.startswith("left_lm")
            or col.startswith("right_lm")
            or col.startswith("left_hand_to_")
            or col.startswith("right_hand_to_")
        )
    ]

    print("Hand-related columns:", len(hand_columns))

    # Missing values before cleaning
    missing_before = df[hand_columns].isna().sum().sum()

    print("Missing hand values before:", missing_before)

    # --------------------------------------------------
    # Interpolate/fill hand features over time
    # --------------------------------------------------

    df[hand_columns] = (
        df[hand_columns]
        .interpolate(method="linear", limit_direction="both")
        .ffill()
        .bfill()
    )

    # Any remaining NaNs are replaced with 0
    df[hand_columns] = df[hand_columns].fillna(0.0)

    # --------------------------------------------------
    # Check result
    # --------------------------------------------------

    missing_after = df[hand_columns].isna().sum().sum()

    print("Missing hand values after:", missing_after)

    # --------------------------------------------------
    # Save
    # --------------------------------------------------

    df.to_csv(output_path, index=False)

    print("Output shape:", df.shape)
    print("Saved:", output_path)

    print()
```

```python
import pandas as pd

video_name = "C4"

path = (
    f"/content/person2_output/batch_test/"
    f"{video_name}/{video_name}_clean_features.csv"
)

df = pd.read_csv(path)

feature_columns = [
    col for col in df.columns
    if col != "frame"
]

groups = {
    "object_presence": [
        c for c in feature_columns
        if c.endswith("_present")
    ],

    "object_confidence": [
        c for c in feature_columns
        if c.endswith("_confidence")
    ],

    "object_geometry": [
        c for c in feature_columns
        if (
            c.endswith("_center_x")
            or c.endswith("_center_y")
            or c.endswith("_width")
            or c.endswith("_height")
            or c.endswith("_area")
        )
    ],

    "object_object_distance": [
        c for c in feature_columns
        if "_to_" in c and c.endswith("_distance")
    ],

    "hand_landmarks": [
        c for c in feature_columns
        if c.startswith("left_lm")
        or c.startswith("right_lm")
    ],

    "hand_object_distance": [
        c for c in feature_columns
        if (
            c.startswith("left_hand_to_")
            or c.startswith("right_hand_to_")
        )
    ]
}

print("=" * 70)
print("FEATURE GROUP SUMMARY")
print("=" * 70)

total = 0

for group_name, columns in groups.items():

    print(f"{group_name:25s}: {len(columns)} features")
    total += len(columns)

print("-" * 70)
print(f"{'TOTAL':25s}: {total} features")
print(f"{'DataFrame features':25s}: {len(feature_columns)} features")

# Check whether every feature belongs to exactly one group
grouped_columns = []

for columns in groups.values():
    grouped_columns.extend(columns)

duplicates = pd.Series(grouped_columns).duplicated().sum()

unassigned = [
    c for c in feature_columns
    if c not in grouped_columns
]

print("\nDuplicate assignments:", duplicates)
print("Unassigned features:", len(unassigned))

if len(unassigned) > 0:
    print("\nUnassigned columns:")
    print(unassigned)
```

```python
import os
import cv2
import pandas as pd
from ultralytics import YOLO
from tqdm import tqdm

# --------------------------------------------------
# SETTINGS
# --------------------------------------------------

video_name = "C6"

video_path = next(
    p for p in project_videos
    if os.path.splitext(os.path.basename(p))[0] == video_name
)

base_dir = f"/content/person2_output/all_videos/{video_name}"
os.makedirs(base_dir, exist_ok=True)

tracking_path = os.path.join(
    base_dir,
    f"{video_name}_tracking.csv"
)

# --------------------------------------------------
# LOAD MODEL
# --------------------------------------------------

model = YOLO("/content/best.pt")

# --------------------------------------------------
# OPEN VIDEO
# --------------------------------------------------

cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    raise RuntimeError(f"Could not open video: {video_path}")

total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
fps = cap.get(cv2.CAP_PROP_FPS)

print("=" * 60)
print("Processing:", video_name)
print("Video:", video_path)
print("Frames:", total_frames)
print("FPS:", fps)
print("=" * 60)

# --------------------------------------------------
# YOLO + BYTETRACK
# --------------------------------------------------

frame_number = 0
tracking_rows = []

while True:

    ret, frame = cap.read()

    if not ret:
        break

    frame_number += 1

    results = model.track(
        frame,
        persist=True,
        tracker="/content/bytetrack_custom.yaml",
        conf=0.5,
        verbose=False
    )

    result = results[0]

    if result.boxes is not None and len(result.boxes) > 0:

        boxes = result.boxes

        xyxy = boxes.xyxy.cpu().numpy()
        confs = boxes.conf.cpu().numpy()
        classes = boxes.cls.cpu().numpy().astype(int)

        if boxes.id is not None:
            track_ids = boxes.id.cpu().numpy().astype(int)
        else:
            track_ids = [-1] * len(xyxy)

        for box, conf, cls, track_id in zip(
            xyxy,
            confs,
            classes,
            track_ids
        ):

            x1, y1, x2, y2 = box

            tracking_rows.append({
                "frame": frame_number,
                "class_id": int(cls),
                "class_name": model.names[int(cls)],
                "track_id": int(track_id),
                "x1": float(x1),
                "y1": float(y1),
                "x2": float(y2),
                "y2": float(y2),
                "confidence": float(conf)
            })

    if frame_number % 50 == 0:
        print(
            f"Processed {frame_number}/{total_frames} frames"
        )

cap.release()

# --------------------------------------------------
# SAVE TRACKING DATA
# --------------------------------------------------

tracking_df = pd.DataFrame(tracking_rows)

tracking_df.to_csv(
    tracking_path,
    index=False
)

print("\nTracking completed!")
print("Rows:", len(tracking_df))
print("Shape:", tracking_df.shape)
print("Saved:", tracking_path)

print("\nDetected classes:")
print(tracking_df["class_name"].value_counts())
```

```python
import os
import pandas as pd
import numpy as np

video_name = "C6"

base_dir = f"/content/person2_output/all_videos/{video_name}"

input_path = os.path.join(
    base_dir,
    f"{video_name}_tracking.csv"
)

output_path = os.path.join(
    base_dir,
    f"{video_name}_spatial_features.csv"
)

df = pd.read_csv(input_path)

print("=" * 70)
print(f"Creating spatial features for {video_name}")
print("=" * 70)

# Object center coordinates
df["center_x"] = (df["x1"] + df["x2"]) / 2
df["center_y"] = (df["y1"] + df["y2"]) / 2

# Bounding-box width and height
df["width"] = df["x2"] - df["x1"]
df["height"] = df["y2"] - df["y1"]

# Bounding-box area
df["area"] = df["width"] * df["height"]

# Normalized center coordinates
df["center_x_norm"] = df["center_x"] / 1920
df["center_y_norm"] = df["center_y"] / 1080

# Normalized width and height
df["width_norm"] = df["width"] / 1920
df["height_norm"] = df["height"] / 1080

# Normalized area
df["area_norm"] = df["area"] / (1920 * 1080)

df.to_csv(output_path, index=False)

print("Input shape:", df.shape)
print("Saved:", output_path)
print()
```

```python
import os
import cv2
import pandas as pd
import numpy as np
import mediapipe as mp

video_name = "C6"

video_path = f"/content/drive/MyDrive/Sih videos/{video_name}.mp4"
output_dir = f"/content/person2_output/all_videos/{video_name}"
output_path = f"{output_dir}/{video_name}_hand_pose.csv"

os.makedirs(output_dir, exist_ok=True)

print("=" * 70)
print(f"Creating hand pose features for {video_name}")
print("=" * 70)

# --------------------------------------------------
# MediaPipe Hand Landmarker
# --------------------------------------------------

BaseOptions = mp.tasks.BaseOptions
HandLandmarker = mp.tasks.vision.HandLandmarker
HandLandmarkerOptions = mp.tasks.vision.HandLandmarkerOptions
VisionRunningMode = mp.tasks.vision.RunningMode

options = HandLandmarkerOptions(
    base_options=BaseOptions(
        model_asset_path="/content/hand_landmarker.task"
    ),
    running_mode=VisionRunningMode.VIDEO,
    num_hands=2
)

# --------------------------------------------------
# Open video
# --------------------------------------------------

cap = cv2.VideoCapture(video_path)

fps = cap.get(cv2.CAP_PROP_FPS)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

print("Frames:", total_frames)
print("FPS:", fps)

rows = []

# IMPORTANT: create a fresh landmarker for this video
with HandLandmarker.create_from_options(options) as landmarker:

    frame_number = 0

    while True:

        ret, frame = cap.read()

        if not ret:
            break

        frame_number += 1

        # Convert BGR → RGB
        rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

        mp_image = mp.Image(
            image_format=mp.ImageFormat.SRGB,
            data=rgb
        )

        # Timestamp must increase for VIDEO mode
        timestamp_ms = int((frame_number - 1) * 1000 / fps)

        result = landmarker.detect_for_video(
            mp_image,
            timestamp_ms
        )

        row = {"frame": frame_number}

        # --------------------------------------------------
        # Initialize 2 hands × 21 landmarks × 3 coordinates
        # --------------------------------------------------

        for hand_side in ["left", "right"]:

            for landmark_id in range(21):

                row[f"{hand_side}_lm{landmark_id}_x"] = np.nan
                row[f"{hand_side}_lm{landmark_id}_y"] = np.nan
                row[f"{hand_side}_lm{landmark_id}_z"] = np.nan

        # --------------------------------------------------
        # Store detected landmarks
        # --------------------------------------------------

        if result.hand_landmarks:

            for hand_index, landmarks in enumerate(
                result.hand_landmarks
            ):

                if hand_index >= len(result.handedness):
                    continue

                handedness = result.handedness[hand_index][0].category_name

                side = handedness.lower()

                if side not in ["left", "right"]:
                    continue

                for landmark_id, landmark in enumerate(landmarks):

                    row[f"{side}_lm{landmark_id}_x"] = landmark.x
                    row[f"{side}_lm{landmark_id}_y"] = landmark.y
                    row[f"{side}_lm{landmark_id}_z"] = landmark.z

        rows.append(row)

        if frame_number % 50 == 0:
            print(
                f"Processed {frame_number}/{total_frames} frames"
            )

cap.release()

# --------------------------------------------------
# Save
# --------------------------------------------------

hand_df = pd.DataFrame(rows)

hand_df.to_csv(output_path, index=False)

print("\n" + "=" * 70)
print("HAND POSE EXTRACTION COMPLETED")
print("=" * 70)

print("Shape:", hand_df.shape)
print("Saved:", output_path)

print("\nLeft-hand detections:",
      hand_df.filter(regex="^left_lm").notna().any(axis=1).sum())

print("Right-hand detections:",
      hand_df.filter(regex="^right_lm").notna().any(axis=1).sum())
```

```python
import os
import pandas as pd

video_name = "C6"
base_dir = f"/content/person2_output/all_videos/{video_name}"

spatial_path = f"{base_dir}/{video_name}_spatial_features.csv"
hand_path = f"{base_dir}/{video_name}_hand_pose.csv"

output_path = f"{base_dir}/{video_name}_frame_features.csv"

print("=" * 70)
print("Creating C6 frame-level features")
print("=" * 70)

spatial_df = pd.read_csv(spatial_path)
hand_df = pd.read_csv(hand_path)

# --------------------------------------------------
# Keep the highest-confidence detection for each
# object class in each frame
# --------------------------------------------------

spatial_df = spatial_df.sort_values(
    ["frame", "confidence"],
    ascending=[True, False]
)

frame_objects = (
    spatial_df
    .drop_duplicates(
        subset=["frame", "class_name"],
        keep="first"
    )
)

# --------------------------------------------------
# Pivot object information so there is ONE row/frame
# --------------------------------------------------

classes = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

frames = pd.DataFrame({
    "frame": range(
        1,
        max(
            spatial_df["frame"].max(),
            hand_df["frame"].max()
        ) + 1
    )
})

result = frames.copy()

# --------------------------------------------------
# Add object features
# --------------------------------------------------

for class_name in classes:

    obj = frame_objects[
        frame_objects["class_name"] == class_name
    ][
        [
            "frame",
            "confidence",
            "center_x",
            "center_y",
            "width",
            "height",
            "area",
            "center_x_norm",
            "center_y_norm",
            "width_norm",
            "height_norm",
            "area_norm"
        ]
    ].copy()

    obj = obj.rename(
        columns={
            "confidence": f"{class_name}_confidence",
            "center_x": f"{class_name}_center_x",
            "center_y": f"{class_name}_center_y",
            "width": f"{class_name}_width",
            "height": f"{class_name}_height",
            "area": f"{class_name}_area",
            "center_x_norm": f"{class_name}_center_x_norm",
            "center_y_norm": f"{class_name}_center_y_norm",
            "width_norm": f"{class_name}_width_norm",
            "height_norm": f"{class_name}_height_norm",
            "area_norm": f"{class_name}_area_norm"
        }
    )

    result = result.merge(
        obj,
        on="frame",
        how="left"
    )

    # Presence flag
    result[f"{class_name}_present"] = (
        result[f"{class_name}_confidence"].notna()
    ).astype(int)

# --------------------------------------------------
# Add hand-pose features
# --------------------------------------------------

result = result.merge(
    hand_df,
    on="frame",
    how="left"
)

# --------------------------------------------------
# Save
# --------------------------------------------------

result.to_csv(output_path, index=False)

print("Spatial rows:", len(spatial_df))
print("Hand-pose rows:", len(hand_df))
print("Final frame rows:", len(result))
print("Final columns:", len(result.columns))
print("Saved:", output_path)

print("\nMissing values:",
      result.drop(columns=["frame"]).isna().sum().sum())
```

```python
import pandas as pd
import numpy as np

video_name = "C6"

path = (
    f"/content/person2_output/all_videos/"
    f"{video_name}/{video_name}_frame_features.csv"
)

output_path = (
    f"/content/person2_output/all_videos/"
    f"{video_name}/{video_name}_clean_features.csv"
)

df = pd.read_csv(path)

print("=" * 70)
print("Cleaning C6 frame-level features")
print("=" * 70)

# --------------------------------------------------
# Hand-pose columns
# --------------------------------------------------

hand_columns = [
    col for col in df.columns
    if (
        col.startswith("left_lm")
        or col.startswith("right_lm")
    )
]

# --------------------------------------------------
# Hand-object distance columns
# --------------------------------------------------

hand_distance_columns = [
    col for col in df.columns
    if (
        col.startswith("left_hand_to_")
        or col.startswith("right_hand_to_")
    )
]

# --------------------------------------------------
# Object columns
# --------------------------------------------------

object_columns = [
    col for col in df.columns
    if any(
        col.startswith(obj + "_")
        for obj in [
            "cotton",
            "dropper",
            "seed_jar",
            "target_plate",
            "water_dish"
        ]
    )
]

# Presence and confidence should be 0 when absent
presence_columns = [
    col for col in df.columns
    if col.endswith("_present")
]

confidence_columns = [
    col for col in df.columns
    if col.endswith("_confidence")
]

df[presence_columns] = df[presence_columns].fillna(0)
df[confidence_columns] = df[confidence_columns].fillna(0)

# --------------------------------------------------
# Interpolate hand features
# --------------------------------------------------

df[hand_columns] = (
    df[hand_columns]
    .interpolate(
        method="linear",
        limit_direction="both"
    )
    .ffill()
    .bfill()
    .fillna(0.0)
)

# --------------------------------------------------
# Hand-object distances
# --------------------------------------------------

df[hand_distance_columns] = (
    df[hand_distance_columns]
    .interpolate(
        method="linear",
        limit_direction="both"
    )
    .ffill()
    .bfill()
    .fillna(0.0)
)

# --------------------------------------------------
# Object geometry
# --------------------------------------------------

geometry_columns = [
    col for col in object_columns
    if (
        col.endswith("_center_x")
        or col.endswith("_center_y")
        or col.endswith("_width")
        or col.endswith("_height")
        or col.endswith("_area")
        or col.endswith("_center_x_norm")
        or col.endswith("_center_y_norm")
        or col.endswith("_width_norm")
        or col.endswith("_height_norm")
        or col.endswith("_area_norm")
    )
]

df[geometry_columns] = (
    df[geometry_columns]
    .fillna(0.0)
)

# --------------------------------------------------
# Final cleanup
# --------------------------------------------------

feature_columns = [
    col for col in df.columns
    if col != "frame"
]

df[feature_columns] = df[feature_columns].replace(
    [np.inf, -np.inf],
    np.nan
)

df[feature_columns] = df[feature_columns].fillna(0.0)

# --------------------------------------------------
# Save
# --------------------------------------------------

df.to_csv(output_path, index=False)

print("Original shape:", df.shape)
print("Remaining missing values:",
      df[feature_columns].isna().sum().sum())

print("Saved:", output_path)
```

```python
import pandas as pd
import numpy as np
import os

video_name = "C6"

base_dir = f"/content/person2_output/all_videos/{video_name}"

input_path = f"{base_dir}/{video_name}_clean_features.csv"
output_path = f"{base_dir}/{video_name}_final_features.csv"

df = pd.read_csv(input_path)

print("=" * 70)
print("Adding object-to-object distances for C6")
print("=" * 70)

objects = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# --------------------------------------------------
# Calculate normalized center-to-center distances
# --------------------------------------------------

for i in range(len(objects)):

    for j in range(i + 1, len(objects)):

        obj1 = objects[i]
        obj2 = objects[j]

        x1 = df[f"{obj1}_center_x_norm"]
        y1 = df[f"{obj1}_center_y_norm"]

        x2 = df[f"{obj2}_center_x_norm"]
        y2 = df[f"{obj2}_center_y_norm"]

        distance = np.sqrt(
            (x1 - x2) ** 2 +
            (y1 - y2) ** 2
        )

        # If either object is absent, distance = 0
        present1 = df[f"{obj1}_present"]
        present2 = df[f"{obj2}_present"]

        distance = distance.where(
            (present1 == 1) & (present2 == 1),
            0.0
        )

        column_name = (
            f"{obj1}_to_{obj2}_distance"
        )

        df[column_name] = distance

        print("Added:", column_name)

# --------------------------------------------------
# Check
# --------------------------------------------------

distance_columns = [
    col for col in df.columns
    if "_to_" in col and col.endswith("_distance")
]

print("\nObject-object distance features:",
      len(distance_columns))

print("Final shape:", df.shape)

# --------------------------------------------------
# Save
# --------------------------------------------------

df.to_csv(output_path, index=False)

print("\nSaved:", output_path)
```

```python
import pandas as pd
import numpy as np

input_path = (
    "/content/person2_output/all_videos/"
    "C6/C6_final_features.csv"
)

output_path = (
    "/content/person2_output/all_videos/"
    "C6/C6_final_features_with_hand_object_distance.csv"
)

df = pd.read_csv(input_path)

objects = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

hands = {
    "left": ("left_lm0_x", "left_lm0_y"),
    "right": ("right_lm0_x", "right_lm0_y")
}

# ------------------------------------------------------------
# Calculate 10 hand-object distances
# ------------------------------------------------------------

for hand_name, (wrist_x, wrist_y) in hands.items():

    for obj in objects:

        distance_col = f"{hand_name}_hand_to_{obj}_distance"

        object_x = f"{obj}_center_x_norm"
        object_y = f"{obj}_center_y_norm"

        # If the hand or object is absent, distance = 0.
        valid = (
            df[wrist_x].notna() &
            df[wrist_y].notna() &
            df[object_x].notna() &
            df[object_y].notna() &
            (df[f"{obj}_present"] == 1)
        )

        df[distance_col] = 0.0

        df.loc[valid, distance_col] = np.sqrt(
            (df.loc[valid, wrist_x] - df.loc[valid, object_x]) ** 2 +
            (df.loc[valid, wrist_y] - df.loc[valid, object_y]) ** 2
        )

# ------------------------------------------------------------
# Save
# ------------------------------------------------------------

df.to_csv(output_path, index=False)

print("=" * 70)
print("C6 HAND-OBJECT DISTANCES ADDED")
print("=" * 70)

print("Rows:", len(df))
print("Total columns:", len(df.columns))
print("Model features:", len(df.columns) - 1)

distance_cols = [
    c for c in df.columns
    if "hand_to_" in c and "distance" in c
]

print("\nHand-object distance features:", len(distance_cols))

for c in distance_cols:
    print(" ", c)

print("\nMissing values:", int(df.isna().sum().sum()))

print("\nSaved:")
print(output_path)
```

```python
import os
import cv2
import numpy as np
import pandas as pd
import mediapipe as mp

from ultralytics import YOLO
from mediapipe.tasks import python
from mediapipe.tasks.python import vision


# ============================================================
# PERSON 2 — STEP 3
# Complete C4 → final 206-feature frame-level dataset
# ============================================================

VIDEO_PATH = "/content/drive/MyDrive/Sih videos/C4.mp4"

OUTPUT_DIR = "/content/person2_output/all_videos/C4"
os.makedirs(OUTPUT_DIR, exist_ok=True)

YOLO_MODEL = "/content/best.pt"
TRACKER_CONFIG = "/content/bytetrack_custom.yaml"
HAND_MODEL = "/content/hand_landmarker.task"


# ------------------------------------------------------------
# Classes
# ------------------------------------------------------------

OBJECTS = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

CLASS_NAMES = {
    0: "cotton",
    1: "dropper",
    2: "seed_jar",
    3: "target_plate",
    4: "water_dish"
}


# ============================================================
# 1. LOAD MODELS
# ============================================================

print("=" * 70)
print("LOADING MODELS")
print("=" * 70)

model = YOLO(YOLO_MODEL)

BaseOptions = python.BaseOptions

options = vision.HandLandmarkerOptions(
    base_options=BaseOptions(model_asset_path=HAND_MODEL),
    running_mode=vision.RunningMode.VIDEO,
    num_hands=2
)

hand_landmarker = vision.HandLandmarker.create_from_options(options)

print("✓ YOLO loaded")
print("✓ MediaPipe HandLandmarker loaded")


# ============================================================
# 2. VIDEO INFORMATION
# ============================================================

cap = cv2.VideoCapture(VIDEO_PATH)

if not cap.isOpened():
    raise RuntimeError(f"Could not open video: {VIDEO_PATH}")

fps = cap.get(cv2.CAP_PROP_FPS)
frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

print("\nVideo:")
print("  Name   :", os.path.basename(VIDEO_PATH))
print("  Frames :", frame_count)
print("  FPS    :", fps)
print("  Size   :", width, "x", height)


# ============================================================
# 3. YOLO + BYTETRACK
# ============================================================

tracking_rows = []
hand_rows = []

frame_number = 0

print("\n" + "=" * 70)
print("PROCESSING C4")
print("=" * 70)

while True:

    ret, frame = cap.read()

    if not ret:
        break

    frame_number += 1

    # --------------------------------------------------------
    # YOLO + ByteTrack
    # --------------------------------------------------------

    results = model.track(
        frame,
        persist=True,
        tracker=TRACKER_CONFIG,
        conf=0.5,
        verbose=False
    )

    result = results[0]

    if result.boxes is not None and len(result.boxes) > 0:

        boxes = result.boxes

        xyxy = boxes.xyxy.cpu().numpy()
        confs = boxes.conf.cpu().numpy()
        classes = boxes.cls.cpu().numpy().astype(int)

        if boxes.id is not None:
            track_ids = boxes.id.cpu().numpy().astype(int)
        else:
            track_ids = [-1] * len(xyxy)

        for box, conf, cls_id, track_id in zip(
            xyxy, confs, classes, track_ids
        ):

            if cls_id not in CLASS_NAMES:
                continue

            x1, y1, x2, y2 = box

            tracking_rows.append({
                "frame": frame_number,
                "class_id": int(cls_id),
                "class_name": CLASS_NAMES[int(cls_id)],
                "track_id": int(track_id),
                "x1": float(x1),
                "y1": float(y1),
                "x2": float(x2),
                "y2": float(y2),
                "confidence": float(conf)
            })

    # --------------------------------------------------------
    # MediaPipe hand landmarks
    # --------------------------------------------------------

    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    mp_image = mp.Image(
        image_format=mp.ImageFormat.SRGB,
        data=rgb
    )

    timestamp_ms = int((frame_number - 1) * 1000 / fps)

    hand_result = hand_landmarker.detect_for_video(
        mp_image,
        timestamp_ms
    )

    hand_data = {
        "frame": frame_number
    }

    # Initialize all landmarks
    for hand_name in ["left", "right"]:
        for lm_id in range(21):
            hand_data[f"{hand_name}_lm{lm_id}_x"] = np.nan
            hand_data[f"{hand_name}_lm{lm_id}_y"] = np.nan
            hand_data[f"{hand_name}_lm{lm_id}_z"] = np.nan

    # --------------------------------------------------------
    # Store landmarks
    # --------------------------------------------------------

    if hand_result.hand_landmarks:

        for hand_idx, landmarks in enumerate(
            hand_result.hand_landmarks
        ):

            if hand_idx >= len(hand_result.handedness):
                continue

            handedness = (
                hand_result.handedness[hand_idx][0].category_name
            )

            # MediaPipe handedness is from image perspective.
            # Keep the label returned by MediaPipe.
            hand_name = handedness.lower()

            if hand_name not in ["left", "right"]:
                continue

            for lm_id, lm in enumerate(landmarks):

                hand_data[
                    f"{hand_name}_lm{lm_id}_x"
                ] = float(lm.x)

                hand_data[
                    f"{hand_name}_lm{lm_id}_y"
                ] = float(lm.y)

                hand_data[
                    f"{hand_name}_lm{lm_id}_z"
                ] = float(lm.z)

    hand_rows.append(hand_data)

    if frame_number % 50 == 0:
        print(
            f"Processed {frame_number}/{frame_count} frames"
        )


cap.release()
hand_landmarker.close()


# ============================================================
# 4. DATAFRAMES
# ============================================================

tracking = pd.DataFrame(tracking_rows)
hands = pd.DataFrame(hand_rows)

print("\nTracking rows:", tracking.shape)
print("Hand rows    :", hands.shape)


# ============================================================
# 5. OBJECT GEOMETRY
# ============================================================

tracking["center_x"] = (
    tracking["x1"] + tracking["x2"]
) / 2

tracking["center_y"] = (
    tracking["y1"] + tracking["y2"]
) / 2

tracking["width"] = tracking["x2"] - tracking["x1"]
tracking["height"] = tracking["y2"] - tracking["y1"]

tracking["area"] = (
    tracking["width"] * tracking["height"]
)

tracking["center_x_norm"] = tracking["center_x"] / width
tracking["center_y_norm"] = tracking["center_y"] / height
tracking["width_norm"] = tracking["width"] / width
tracking["height_norm"] = tracking["height"] / height
tracking["area_norm"] = tracking["area"] / (width * height)


# ============================================================
# 6. ONE ROW PER FRAME
#    Highest-confidence detection per object
# ============================================================

frames = pd.DataFrame({
    "frame": np.arange(1, frame_count + 1)
})

feature_parts = [frames]

for obj in OBJECTS:

    obj_df = tracking[
        tracking["class_name"] == obj
    ].copy()

    if len(obj_df) > 0:

        # Highest-confidence detection if multiple exist
        obj_df = (
            obj_df
            .sort_values(
                ["frame", "confidence"],
                ascending=[True, False]
            )
            .drop_duplicates("frame")
        )

        obj_df = obj_df.set_index("frame")

    else:
        obj_df = pd.DataFrame()

    part = pd.DataFrame(index=frames["frame"])

    # --------------------------------------------------------
    # Confidence
    # --------------------------------------------------------

    part[f"{obj}_confidence"] = np.nan

    # --------------------------------------------------------
    # Geometry
    # --------------------------------------------------------

    for col in [
        "center_x",
        "center_y",
        "width",
        "height",
        "area",
        "center_x_norm",
        "center_y_norm",
        "width_norm",
        "height_norm",
        "area_norm"
    ]:
        part[f"{obj}_{col}"] = np.nan

    # --------------------------------------------------------
    # Presence
    # --------------------------------------------------------

    part[f"{obj}_present"] = 0

    if len(obj_df) > 0:

        common_frames = part.index.intersection(
            obj_df.index
        )

        part.loc[
            common_frames,
            f"{obj}_confidence"
        ] = obj_df.loc[
            common_frames, "confidence"
        ]

        for col in [
            "center_x",
            "center_y",
            "width",
            "height",
            "area",
            "center_x_norm",
            "center_y_norm",
            "width_norm",
            "height_norm",
            "area_norm"
        ]:
            part.loc[
                common_frames,
                f"{obj}_{col}"
            ] = obj_df.loc[
                common_frames, col
            ]

        part.loc[
            common_frames,
            f"{obj}_present"
        ] = 1

    part = part.reset_index()

    feature_parts.append(part)


frame_features = feature_parts[0]

for part in feature_parts[1:]:
    frame_features = frame_features.merge(
        part,
        on="frame",
        how="left"
    )


# ============================================================
# 7. OBJECT-OBJECT DISTANCES
# ============================================================

for i in range(len(OBJECTS)):

    for j in range(i + 1, len(OBJECTS)):

        obj1 = OBJECTS[i]
        obj2 = OBJECTS[j]

        x1 = frame_features[f"{obj1}_center_x_norm"]
        y1 = frame_features[f"{obj1}_center_y_norm"]

        x2 = frame_features[f"{obj2}_center_x_norm"]
        y2 = frame_features[f"{obj2}_center_y_norm"]

        col = f"{obj1}_to_{obj2}_distance"

        valid = (
            (frame_features[f"{obj1}_present"] == 1) &
            (frame_features[f"{obj2}_present"] == 1)
        )

        frame_features[col] = 0.0

        frame_features.loc[valid, col] = np.sqrt(
            (x1[valid] - x2[valid]) ** 2 +
            (y1[valid] - y2[valid]) ** 2
        )


# ============================================================
# 8. MERGE HAND LANDMARKS
# ============================================================

frame_features = frame_features.merge(
    hands,
    on="frame",
    how="left"
)


# ============================================================
# 9. CLEAN HAND LANDMARKS
# ============================================================

hand_columns = [
    c for c in frame_features.columns
    if c.startswith("left_lm") or
       c.startswith("right_lm")
]

frame_features[hand_columns] = (
    frame_features[hand_columns]
    .interpolate(
        method="linear",
        limit_direction="both"
    )
    .ffill()
    .bfill()
    .fillna(0)
)


# ============================================================
# 10. CLEAN OBJECT FEATURES
# ============================================================

presence_columns = [
    f"{obj}_present"
    for obj in OBJECTS
]

confidence_columns = [
    f"{obj}_confidence"
    for obj in OBJECTS
]

geometry_columns = []

for obj in OBJECTS:

    for col in [
        "center_x",
        "center_y",
        "width",
        "height",
        "area",
        "center_x_norm",
        "center_y_norm",
        "width_norm",
        "height_norm",
        "area_norm"
    ]:
        geometry_columns.append(
            f"{obj}_{col}"
        )

frame_features[presence_columns] = (
    frame_features[presence_columns]
    .fillna(0)
)

frame_features[confidence_columns] = (
    frame_features[confidence_columns]
    .fillna(0)
)

frame_features[geometry_columns] = (
    frame_features[geometry_columns]
    .fillna(0)
)


# ============================================================
# 11. HAND → OBJECT DISTANCES
# ============================================================

for hand_name in ["left", "right"]:

    wrist_x = f"{hand_name}_lm0_x"
    wrist_y = f"{hand_name}_lm0_y"

    for obj in OBJECTS:

        object_x = f"{obj}_center_x_norm"
        object_y = f"{obj}_center_y_norm"

        distance_col = (
            f"{hand_name}_hand_to_{obj}_distance"
        )

        valid = (
            frame_features[wrist_x].notna() &
            frame_features[wrist_y].notna() &
            frame_features[object_x].notna() &
            frame_features[object_y].notna() &
            (frame_features[f"{obj}_present"] == 1)
        )

        frame_features[distance_col] = 0.0

        frame_features.loc[valid, distance_col] = np.sqrt(
            (
                frame_features.loc[valid, wrist_x] -
                frame_features.loc[valid, object_x]
            ) ** 2
            +
            (
                frame_features.loc[valid, wrist_y] -
                frame_features.loc[valid, object_y]
            ) ** 2
        )


# ============================================================
# 12. FINAL CLEANUP
# ============================================================

frame_features = frame_features.replace(
    [np.inf, -np.inf],
    np.nan
)

frame_features = frame_features.fillna(0)


# ============================================================
# 13. SAVE FINAL FEATURES
# ============================================================

output_path = os.path.join(
    OUTPUT_DIR,
    "C4_final_features_with_hand_object_distance.csv"
)

frame_features.to_csv(
    output_path,
    index=False
)


# ============================================================
# 14. VALIDATION
# ============================================================

model_features = [
    c for c in frame_features.columns
    if c != "frame"
]

hand_object_features = [
    c for c in model_features
    if "_hand_to_" in c
]

object_object_features = [
    c for c in model_features
    if "_to_" in c and "_hand_to_" not in c
]

hand_landmark_features = [
    c for c in model_features
    if c.startswith("left_lm") or
       c.startswith("right_lm")
]

print("\n" + "=" * 70)
print("C4 FINAL FEATURE VALIDATION")
print("=" * 70)

print("Rows:", len(frame_features))
print("Total columns:", len(frame_features.columns))
print("Model features:", len(model_features))

print("\nHand landmark features:", len(hand_landmark_features))
print("Object-object distances:", len(object_object_features))
print("Hand-object distances:", len(hand_object_features))

print("\nMissing values:",
      int(frame_features.isna().sum().sum()))

print("\nExpected:")
print("  Hand landmarks       = 126")
print("  Object-object        = 10")
print("  Hand-object          = 10")
print("  Total model features = 206")

print("\nSaved:")
print(output_path)

print("=" * 70)
```

```python
# ================================================================
# PERSON 2 — C5 FINAL FEATURE PIPELINE
# YOLO + ByteTrack + MediaPipe Hands + Spatial/Interaction Features
# ================================================================

import os
import cv2
import numpy as np
import pandas as pd
import mediapipe as mp
from ultralytics import YOLO

from mediapipe.tasks import python
from mediapipe.tasks.python import vision


# ================================================================
# PATHS
# ================================================================

VIDEO_PATH = "/content/drive/MyDrive/Sih videos/C5.mp4"

YOLO_PATH = "/content/best.pt"
HAND_MODEL_PATH = "/content/hand_landmarker.task"
TRACKER_PATH = "/content/bytetrack_custom.yaml"

OUT_DIR = "/content/person2_output/all_videos/C5"
os.makedirs(OUT_DIR, exist_ok=True)

FINAL_PATH = os.path.join(
    OUT_DIR,
    "C5_final_features_with_hand_object_distance.csv"
)


# ================================================================
# CONFIG
# ================================================================

CLASS_NAMES = {
    0: "cotton",
    1: "dropper",
    2: "seed_jar",
    3: "target_plate",
    4: "water_dish"
}

OBJECTS = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]


# ================================================================
# LOAD MODELS
# ================================================================

print("=" * 70)
print("LOADING MODELS")
print("=" * 70)

model = YOLO(YOLO_PATH)

print("✓ YOLO loaded")


# Fresh MediaPipe landmarker for this video
# This avoids timestamp/state conflicts between videos.

base_options = python.BaseOptions(
    model_asset_path=HAND_MODEL_PATH
)

options = vision.HandLandmarkerOptions(
    base_options=base_options,
    running_mode=vision.RunningMode.VIDEO,
    num_hands=2,
    min_hand_detection_confidence=0.5,
    min_hand_presence_confidence=0.5,
    min_tracking_confidence=0.5
)

hand_landmarker = vision.HandLandmarker.create_from_options(options)

print("✓ MediaPipe HandLandmarker loaded")


# ================================================================
# VIDEO INFORMATION
# ================================================================

cap = cv2.VideoCapture(VIDEO_PATH)

if not cap.isOpened():
    raise RuntimeError(f"Could not open video: {VIDEO_PATH}")

total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

print()
print("Video:")
print(f"  Name   : C5.mp4")
print(f"  Frames : {total_frames}")
print(f"  FPS    : {fps}")
print(f"  Size   : {width} x {height}")


# ================================================================
# STORAGE
# ================================================================

tracking_rows = []
hand_rows = []


# ================================================================
# HAND LANDMARK COLUMN NAMES
# ================================================================

HAND_COLUMNS = ["frame"]

for hand_name in ["left", "right"]:
    for lm in range(21):
        HAND_COLUMNS += [
            f"{hand_name}_lm{lm}_x",
            f"{hand_name}_lm{lm}_y",
            f"{hand_name}_lm{lm}_z"
        ]


# ================================================================
# PROCESS VIDEO
# ================================================================

print()
print("=" * 70)
print("PROCESSING C5")
print("=" * 70)

frame_idx = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    # ------------------------------------------------------------
    # YOLO + BYTE TRACK
    # ------------------------------------------------------------

    results = model.track(
        frame,
        persist=True,
        tracker=TRACKER_PATH,
        conf=0.5,
        verbose=False
    )

    result = results[0]

    if result.boxes is not None:

        boxes = result.boxes

        xyxy = boxes.xyxy.cpu().numpy()
        confs = boxes.conf.cpu().numpy()
        classes = boxes.cls.cpu().numpy().astype(int)

        if boxes.id is not None:
            track_ids = boxes.id.cpu().numpy().astype(int)
        else:
            track_ids = np.full(len(xyxy), -1)

        for box, conf, cls_id, track_id in zip(
            xyxy,
            confs,
            classes,
            track_ids
        ):

            if cls_id not in CLASS_NAMES:
                continue

            x1, y1, x2, y2 = box

            tracking_rows.append({
                "frame": frame_idx,
                "class_id": cls_id,
                "class_name": CLASS_NAMES[cls_id],
                "track_id": int(track_id),
                "x1": float(x1),
                "y1": float(y1),
                "x2": float(x2),
                "y2": float(y2),
                "confidence": float(conf)
            })


    # ------------------------------------------------------------
    # MEDIAPIPE HAND LANDMARKS
    # ------------------------------------------------------------

    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    mp_image = mp.Image(
        image_format=mp.ImageFormat.SRGB,
        data=rgb
    )

    timestamp_ms = int((frame_idx / fps) * 1000)

    hand_result = hand_landmarker.detect_for_video(
        mp_image,
        timestamp_ms
    )

    hand_data = {
        "frame": frame_idx
    }

    # Initialize all 126 landmark values with NaN
    for hand_name in ["left", "right"]:
        for lm in range(21):
            hand_data[f"{hand_name}_lm{lm}_x"] = np.nan
            hand_data[f"{hand_name}_lm{lm}_y"] = np.nan
            hand_data[f"{hand_name}_lm{lm}_z"] = np.nan


    if hand_result.hand_landmarks:

        for hand_landmarks, handedness in zip(
            hand_result.hand_landmarks,
            hand_result.handedness
        ):

            label = handedness[0].category_name.lower()

            if label not in ["left", "right"]:
                continue

            for lm_idx, landmark in enumerate(hand_landmarks):

                hand_data[f"{label}_lm{lm_idx}_x"] = float(
                    landmark.x
                )

                hand_data[f"{label}_lm{lm_idx}_y"] = float(
                    landmark.y
                )

                hand_data[f"{label}_lm{lm_idx}_z"] = float(
                    landmark.z
                )


    hand_rows.append(hand_data)


    frame_idx += 1

    if frame_idx % 50 == 0:
        print(f"Processed {frame_idx}/{total_frames} frames")


cap.release()
hand_landmarker.close()


# ================================================================
# TRACKING DATAFRAME
# ================================================================

tracking_df = pd.DataFrame(tracking_rows)

print()
print(f"Tracking rows: {tracking_df.shape}")


# ================================================================
# HAND DATAFRAME
# ================================================================

hand_df = pd.DataFrame(
    hand_rows,
    columns=HAND_COLUMNS
)

print(f"Hand rows    : {hand_df.shape}")


# ================================================================
# OBJECT SPATIAL FEATURES
# ================================================================

if len(tracking_df) == 0:
    raise RuntimeError("No YOLO detections found in C5.")

tracking_df["center_x"] = (
    tracking_df["x1"] + tracking_df["x2"]
) / 2

tracking_df["center_y"] = (
    tracking_df["y1"] + tracking_df["y2"]
) / 2

tracking_df["width"] = (
    tracking_df["x2"] - tracking_df["x1"]
)

tracking_df["height"] = (
    tracking_df["y2"] - tracking_df["y1"]
)

tracking_df["area"] = (
    tracking_df["width"] *
    tracking_df["height"]
)

tracking_df["center_x_norm"] = (
    tracking_df["center_x"] / width
)

tracking_df["center_y_norm"] = (
    tracking_df["center_y"] / height
)

tracking_df["width_norm"] = (
    tracking_df["width"] / width
)

tracking_df["height_norm"] = (
    tracking_df["height"] / height
)

tracking_df["area_norm"] = (
    tracking_df["area"] / (width * height)
)


# ================================================================
# ONE ROW PER FRAME
# SELECT HIGHEST-CONFIDENCE DETECTION PER CLASS
# ================================================================

frame_features = pd.DataFrame({
    "frame": np.arange(total_frames)
})

for obj in OBJECTS:

    obj_df = tracking_df[
        tracking_df["class_name"] == obj
    ].copy()

    if len(obj_df) > 0:

        obj_df = (
            obj_df
            .sort_values("confidence", ascending=False)
            .drop_duplicates("frame")
        )

        obj_df = obj_df.set_index("frame")

    else:

        obj_df = pd.DataFrame()

    # ------------------------------------------------------------
    # Presence
    # ------------------------------------------------------------

    presence = np.zeros(total_frames)

    if len(obj_df) > 0:
        valid_frames = obj_df.index[
            obj_df.index < total_frames
        ]

        presence[valid_frames] = 1.0

    frame_features[f"{obj}_presence"] = presence


    # ------------------------------------------------------------
    # Confidence
    # ------------------------------------------------------------

    confidence = np.zeros(total_frames)

    if len(obj_df) > 0:

        for f, value in obj_df["confidence"].items():

            if f < total_frames:
                confidence[f] = value

    frame_features[f"{obj}_confidence"] = confidence


    # ------------------------------------------------------------
    # Raw geometry
    # ------------------------------------------------------------

    raw_columns = [
        "center_x",
        "center_y",
        "width",
        "height",
        "area"
    ]

    for col in raw_columns:

        values = np.zeros(total_frames)

        if len(obj_df) > 0:

            for f, value in obj_df[col].items():

                if f < total_frames:
                    values[f] = value

        frame_features[f"{obj}_{col}"] = values


    # ------------------------------------------------------------
    # Normalized geometry
    # ------------------------------------------------------------

    norm_columns = [
        "center_x_norm",
        "center_y_norm",
        "width_norm",
        "height_norm",
        "area_norm"
    ]

    for col in norm_columns:

        values = np.zeros(total_frames)

        if len(obj_df) > 0:

            for f, value in obj_df[col].items():

                if f < total_frames:
                    values[f] = value

        frame_features[f"{obj}_{col}"] = values


# ================================================================
# OBJECT-OBJECT DISTANCES
# ================================================================

object_pairs = [
    ("cotton", "dropper"),
    ("cotton", "seed_jar"),
    ("cotton", "target_plate"),
    ("cotton", "water_dish"),
    ("dropper", "seed_jar"),
    ("dropper", "target_plate"),
    ("dropper", "water_dish"),
    ("seed_jar", "target_plate"),
    ("seed_jar", "water_dish"),
    ("target_plate", "water_dish")
]


for obj1, obj2 in object_pairs:

    col_name = f"{obj1}_to_{obj2}_distance"

    distance_values = np.zeros(total_frames)

    x1_col = f"{obj1}_center_x_norm"
    y1_col = f"{obj1}_center_y_norm"

    x2_col = f"{obj2}_center_x_norm"
    y2_col = f"{obj2}_center_y_norm"

    valid1 = (
        frame_features[f"{obj1}_presence"].values > 0
    )

    valid2 = (
        frame_features[f"{obj2}_presence"].values > 0
    )

    valid = valid1 & valid2

    distance_values[valid] = np.sqrt(
        (
            frame_features.loc[valid, x1_col].values -
            frame_features.loc[valid, x2_col].values
        ) ** 2
        +
        (
            frame_features.loc[valid, y1_col].values -
            frame_features.loc[valid, y2_col].values
        ) ** 2
    )

    frame_features[col_name] = distance_values


# ================================================================
# MERGE HAND LANDMARKS
# ================================================================

frame_features = frame_features.merge(
    hand_df,
    on="frame",
    how="left"
)


# ================================================================
# CLEAN HAND LANDMARKS
# ================================================================

hand_feature_columns = [
    c for c in frame_features.columns
    if c.startswith("left_lm")
    or c.startswith("right_lm")
]

frame_features[hand_feature_columns] = (
    frame_features[hand_feature_columns]
    .interpolate(
        method="linear",
        limit_direction="both"
    )
)

frame_features[hand_feature_columns] = (
    frame_features[hand_feature_columns]
    .ffill()
    .bfill()
    .fillna(0)
)


# ================================================================
# HAND-OBJECT DISTANCES
# WRIST (LANDMARK 0) → OBJECT CENTER
# ================================================================

for hand_name in ["left", "right"]:

    wrist_x = f"{hand_name}_lm0_x"
    wrist_y = f"{hand_name}_lm0_y"

    for obj in OBJECTS:

        distance_col = (
            f"{hand_name}_hand_to_{obj}_distance"
        )

        values = np.zeros(total_frames)

        hand_valid = (
            frame_features[wrist_x].notna() &
            frame_features[wrist_y].notna()
        )

        object_valid = (
            frame_features[f"{obj}_presence"] > 0
        )

        valid = hand_valid & object_valid

        if valid.any():

            values[valid] = np.sqrt(
                (
                    frame_features.loc[
                        valid, wrist_x
                    ].values
                    -
                    frame_features.loc[
                        valid,
                        f"{obj}_center_x_norm"
                    ].values
                ) ** 2
                +
                (
                    frame_features.loc[
                        valid, wrist_y
                    ].values
                    -
                    frame_features.loc[
                        valid,
                        f"{obj}_center_y_norm"
                    ].values
                ) ** 2
            )

        frame_features[distance_col] = values


# ================================================================
# FINAL CLEANUP
# ================================================================

frame_features = frame_features.replace(
    [np.inf, -np.inf],
    np.nan
)

frame_features = frame_features.fillna(0)


# ================================================================
# SAVE
# ================================================================

frame_features.to_csv(
    FINAL_PATH,
    index=False
)


# ================================================================
# VALIDATION
# ================================================================

feature_columns = [
    c for c in frame_features.columns
    if c != "frame"
]

hand_columns = [
    c for c in feature_columns
    if c.startswith("left_lm")
    or c.startswith("right_lm")
]

object_object_columns = [
    c for c in feature_columns
    if "_to_" in c
    and "_hand_to_" not in c
]

hand_object_columns = [
    c for c in feature_columns
    if "_hand_to_" in c
]

print()
print("=" * 70)
print("C5 FINAL FEATURE VALIDATION")
print("=" * 70)

print(f"Rows: {len(frame_features)}")
print(f"Total columns: {len(frame_features.columns)}")
print(f"Model features: {len(feature_columns)}")
print()
print(f"Hand landmark features: {len(hand_columns)}")
print(f"Object-object distances: {len(object_object_columns)}")
print(f"Hand-object distances: {len(hand_object_columns)}")

print()
print(f"Missing values: {frame_features.isna().sum().sum()}")

print()
print("Expected:")
print("  Hand landmarks       = 126")
print("  Object-object        = 10")
print("  Hand-object          = 10")
print("  Total model features = 206")

print()
print("Saved:")
print(FINAL_PATH)
print("=" * 70)
```

```python
# ================================================================
# PERSON 2 — BATCH 1
# Process next 5 unfinished videos
# Standard output = 206 model features per frame
# ================================================================

import os
import json
import cv2
import numpy as np
import pandas as pd
import mediapipe as mp
from ultralytics import YOLO

from mediapipe.tasks import python
from mediapipe.tasks.python import vision


# ================================================================
# PATHS
# ================================================================

PROJECT_LIST = "/content/person2_output/project_video_list.json"

YOLO_PATH = "/content/best.pt"
HAND_MODEL_PATH = "/content/hand_landmarker.task"
TRACKER_PATH = "/content/bytetrack_custom.yaml"

BASE_OUT = "/content/person2_output/all_videos"

os.makedirs(BASE_OUT, exist_ok=True)


# ================================================================
# LOAD PROJECT VIDEO LIST
# ================================================================

with open(PROJECT_LIST, "r") as f:
    project_videos = json.load(f)

# Handle either a simple list or {"videos": [...]}
if isinstance(project_videos, dict):
    project_videos = project_videos.get("videos", [])

print("=" * 70)
print("PROJECT VIDEO LIST")
print("=" * 70)

print(f"Total project videos: {len(project_videos)}")


# ================================================================
# FIND VIDEO PATH
# ================================================================

SEARCH_DIRS = [
    "/content/drive/MyDrive/BAS_HAR_SIH",
    "/content/drive/MyDrive/Sih videos",
    "/content/drive/MyDrive/sih_project"
]

EXCLUDE_DIRS = [
    "Aarambh 2k22"
]


def find_video(video_name):

    # Search recursively
    for root_dir in SEARCH_DIRS:

        if not os.path.exists(root_dir):
            continue

        for root, dirs, files in os.walk(root_dir):

            # Exclude unrelated directory
            dirs[:] = [
                d for d in dirs
                if d not in EXCLUDE_DIRS
            ]

            if video_name in files:
                return os.path.join(root, video_name)

    return None


# ================================================================
# DETERMINE ALREADY COMPLETED VIDEOS
# ================================================================

def is_completed(video_name):

    stem = os.path.splitext(video_name)[0]

    final_path = os.path.join(
        BASE_OUT,
        stem,
        f"{stem}_final_features_with_hand_object_distance.csv"
    )

    return os.path.exists(final_path)


completed = []
unfinished = []

for video_name in project_videos:

    if is_completed(video_name):
        completed.append(video_name)
    else:
        unfinished.append(video_name)


print(f"Already completed: {len(completed)}")
print(f"Remaining       : {len(unfinished)}")


# ================================================================
# SELECT ONLY NEXT 5
# ================================================================

BATCH = unfinished[:5]

print()
print("This batch:")
for i, video_name in enumerate(BATCH, 1):
    print(f"  {i}. {video_name}")

if len(BATCH) == 0:
    print()
    print("✓ No unfinished videos remain.")
    raise SystemExit


# ================================================================
# LOAD YOLO ONCE
# ================================================================

print()
print("=" * 70)
print("LOADING YOLO")
print("=" * 70)

model = YOLO(YOLO_PATH)

print("✓ YOLO loaded")


# ================================================================
# CONSTANTS
# ================================================================

CLASS_NAMES = {
    0: "cotton",
    1: "dropper",
    2: "seed_jar",
    3: "target_plate",
    4: "water_dish"
}

OBJECTS = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

HAND_COLUMNS = ["frame"]

for hand_name in ["left", "right"]:
    for lm in range(21):
        HAND_COLUMNS += [
            f"{hand_name}_lm{lm}_x",
            f"{hand_name}_lm{lm}_y",
            f"{hand_name}_lm{lm}_z"
        ]

OBJECT_PAIRS = [
    ("cotton", "dropper"),
    ("cotton", "seed_jar"),
    ("cotton", "target_plate"),
    ("cotton", "water_dish"),
    ("dropper", "seed_jar"),
    ("dropper", "target_plate"),
    ("dropper", "water_dish"),
    ("seed_jar", "target_plate"),
    ("seed_jar", "water_dish"),
    ("target_plate", "water_dish")
]


# ================================================================
# PROCESS ONE VIDEO
# ================================================================

def process_video(video_name):

    stem = os.path.splitext(video_name)[0]

    print()
    print()
    print("#" * 70)
    print(f"PROCESSING: {video_name}")
    print("#" * 70)

    video_path = find_video(video_name)

    if video_path is None:
        raise FileNotFoundError(
            f"Could not find video: {video_name}"
        )

    out_dir = os.path.join(BASE_OUT, stem)
    os.makedirs(out_dir, exist_ok=True)

    final_path = os.path.join(
        out_dir,
        f"{stem}_final_features_with_hand_object_distance.csv"
    )

    # ------------------------------------------------------------
    # Skip if already completed
    # ------------------------------------------------------------

    if os.path.exists(final_path):

        print("✓ Already completed — skipping")

        return {
            "video": video_name,
            "status": "already_completed",
            "path": final_path
        }


    # ------------------------------------------------------------
    # Open video
    # ------------------------------------------------------------

    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        raise RuntimeError(
            f"Could not open video: {video_path}"
        )

    total_frames = int(
        cap.get(cv2.CAP_PROP_FRAME_COUNT)
    )

    fps = cap.get(cv2.CAP_PROP_FPS)

    width = int(
        cap.get(cv2.CAP_PROP_FRAME_WIDTH)
    )

    height = int(
        cap.get(cv2.CAP_PROP_FRAME_HEIGHT)
    )

    print()
    print("Video:")
    print(f"  Path   : {video_path}")
    print(f"  Frames : {total_frames}")
    print(f"  FPS    : {fps}")
    print(f"  Size   : {width} x {height}")


    # ------------------------------------------------------------
    # Fresh MediaPipe model for EVERY video
    # ------------------------------------------------------------

    base_options = python.BaseOptions(
        model_asset_path=HAND_MODEL_PATH
    )

    options = vision.HandLandmarkerOptions(
        base_options=base_options,
        running_mode=vision.RunningMode.VIDEO,
        num_hands=2,
        min_hand_detection_confidence=0.5,
        min_hand_presence_confidence=0.5,
        min_tracking_confidence=0.5
    )

    hand_landmarker = (
        vision.HandLandmarker.create_from_options(options)
    )


    # ------------------------------------------------------------
    # Storage
    # ------------------------------------------------------------

    tracking_rows = []
    hand_rows = []

    frame_idx = 0


    # ============================================================
    # FRAME LOOP
    # ============================================================

    while True:

        ret, frame = cap.read()

        if not ret:
            break


        # --------------------------------------------------------
        # YOLO + BYTE TRACK
        # --------------------------------------------------------

        results = model.track(
            frame,
            persist=True,
            tracker=TRACKER_PATH,
            conf=0.5,
            verbose=False
        )

        result = results[0]

        if result.boxes is not None:

            boxes = result.boxes

            xyxy = boxes.xyxy.cpu().numpy()
            confs = boxes.conf.cpu().numpy()
            classes = boxes.cls.cpu().numpy().astype(int)

            if boxes.id is not None:
                track_ids = (
                    boxes.id.cpu().numpy().astype(int)
                )
            else:
                track_ids = np.full(
                    len(xyxy),
                    -1
                )

            for box, conf, cls_id, track_id in zip(
                xyxy,
                confs,
                classes,
                track_ids
            ):

                if cls_id not in CLASS_NAMES:
                    continue

                x1, y1, x2, y2 = box

                tracking_rows.append({
                    "frame": frame_idx,
                    "class_id": int(cls_id),
                    "class_name": CLASS_NAMES[cls_id],
                    "track_id": int(track_id),
                    "x1": float(x1),
                    "y1": float(y1),
                    "x2": float(x2),
                    "y2": float(y2),
                    "confidence": float(conf)
                })


        # --------------------------------------------------------
        # MEDIAPIPE HANDS
        # --------------------------------------------------------

        rgb = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2RGB
        )

        mp_image = mp.Image(
            image_format=mp.ImageFormat.SRGB,
            data=rgb
        )

        timestamp_ms = int(
            (frame_idx / fps) * 1000
        )

        hand_result = (
            hand_landmarker.detect_for_video(
                mp_image,
                timestamp_ms
            )
        )

        hand_data = {
            "frame": frame_idx
        }


        # Initialize 126 landmark values
        for hand_name in ["left", "right"]:

            for lm in range(21):

                hand_data[
                    f"{hand_name}_lm{lm}_x"
                ] = np.nan

                hand_data[
                    f"{hand_name}_lm{lm}_y"
                ] = np.nan

                hand_data[
                    f"{hand_name}_lm{lm}_z"
                ] = np.nan


        # --------------------------------------------------------
        # Store detected hands
        # --------------------------------------------------------

        if hand_result.hand_landmarks:

            for hand_landmarks, handedness in zip(
                hand_result.hand_landmarks,
                hand_result.handedness
            ):

                label = (
                    handedness[0]
                    .category_name
                    .lower()
                )

                if label not in ["left", "right"]:
                    continue

                for lm_idx, landmark in enumerate(
                    hand_landmarks
                ):

                    hand_data[
                        f"{label}_lm{lm_idx}_x"
                    ] = float(landmark.x)

                    hand_data[
                        f"{label}_lm{lm_idx}_y"
                    ] = float(landmark.y)

                    hand_data[
                        f"{label}_lm{lm_idx}_z"
                    ] = float(landmark.z)


        hand_rows.append(hand_data)

        frame_idx += 1

        if frame_idx % 50 == 0:

            print(
                f"Processed "
                f"{frame_idx}/{total_frames} frames"
            )


    # ------------------------------------------------------------
    # Close resources
    # ------------------------------------------------------------

    cap.release()
    hand_landmarker.close()


    # ============================================================
    # DATAFRAMES
    # ============================================================

    tracking_df = pd.DataFrame(
        tracking_rows
    )

    hand_df = pd.DataFrame(
        hand_rows,
        columns=HAND_COLUMNS
    )

    print()
    print(
        f"Tracking rows: {tracking_df.shape}"
    )

    print(
        f"Hand rows    : {hand_df.shape}"
    )


    if len(tracking_df) == 0:

        raise RuntimeError(
            f"No YOLO detections found in {video_name}"
        )


    # ============================================================
    # OBJECT GEOMETRY
    # ============================================================

    tracking_df["center_x"] = (
        tracking_df["x1"] +
        tracking_df["x2"]
    ) / 2

    tracking_df["center_y"] = (
        tracking_df["y1"] +
        tracking_df["y2"]
    ) / 2

    tracking_df["width"] = (
        tracking_df["x2"] -
        tracking_df["x1"]
    )

    tracking_df["height"] = (
        tracking_df["y2"] -
        tracking_df["y1"]
    )

    tracking_df["area"] = (
        tracking_df["width"] *
        tracking_df["height"]
    )

    tracking_df["center_x_norm"] = (
        tracking_df["center_x"] /
        width
    )

    tracking_df["center_y_norm"] = (
        tracking_df["center_y"] /
        height
    )

    tracking_df["width_norm"] = (
        tracking_df["width"] /
        width
    )

    tracking_df["height_norm"] = (
        tracking_df["height"] /
        height
    )

    tracking_df["area_norm"] = (
        tracking_df["area"] /
        (width * height)
    )


    # ============================================================
    # ONE ROW PER FRAME
    # HIGHEST CONFIDENCE DETECTION PER CLASS
    # ============================================================

    frame_features = pd.DataFrame({
        "frame": np.arange(total_frames)
    })


    for obj in OBJECTS:

        obj_df = tracking_df[
            tracking_df["class_name"] == obj
        ].copy()

        if len(obj_df) > 0:

            obj_df = (
                obj_df
                .sort_values(
                    "confidence",
                    ascending=False
                )
                .drop_duplicates("frame")
                .set_index("frame")
            )


        # --------------------------------------------------------
        # Presence
        # --------------------------------------------------------

        presence = np.zeros(total_frames)

        if len(obj_df) > 0:

            valid_frames = obj_df.index[
                obj_df.index < total_frames
            ]

            presence[
                valid_frames
            ] = 1.0

        frame_features[
            f"{obj}_presence"
        ] = presence


        # --------------------------------------------------------
        # Confidence
        # --------------------------------------------------------

        confidence = np.zeros(total_frames)

        if len(obj_df) > 0:

            for f, value in (
                obj_df["confidence"].items()
            ):

                if f < total_frames:
                    confidence[f] = value

        frame_features[
            f"{obj}_confidence"
        ] = confidence


        # --------------------------------------------------------
        # Raw geometry
        # --------------------------------------------------------

        for col in [
            "center_x",
            "center_y",
            "width",
            "height",
            "area"
        ]:

            values = np.zeros(total_frames)

            if len(obj_df) > 0:

                for f, value in obj_df[col].items():

                    if f < total_frames:
                        values[f] = value

            frame_features[
                f"{obj}_{col}"
            ] = values


        # --------------------------------------------------------
        # Normalized geometry
        # --------------------------------------------------------

        for col in [
            "center_x_norm",
            "center_y_norm",
            "width_norm",
            "height_norm",
            "area_norm"
        ]:

            values = np.zeros(total_frames)

            if len(obj_df) > 0:

                for f, value in obj_df[col].items():

                    if f < total_frames:
                        values[f] = value

            frame_features[
                f"{obj}_{col}"
            ] = values


    # ============================================================
    # OBJECT-OBJECT DISTANCES
    # ============================================================

    for obj1, obj2 in OBJECT_PAIRS:

        distance_col = (
            f"{obj1}_to_{obj2}_distance"
        )

        distance_values = np.zeros(
            total_frames
        )

        valid = (
            (
                frame_features[
                    f"{obj1}_presence"
                ].values > 0
            )
            &
            (
                frame_features[
                    f"{obj2}_presence"
                ].values > 0
            )
        )

        if valid.any():

            dx = (
                frame_features.loc[
                    valid,
                    f"{obj1}_center_x_norm"
                ].values
                -
                frame_features.loc[
                    valid,
                    f"{obj2}_center_x_norm"
                ].values
            )

            dy = (
                frame_features.loc[
                    valid,
                    f"{obj1}_center_y_norm"
                ].values
                -
                frame_features.loc[
                    valid,
                    f"{obj2}_center_y_norm"
                ].values
            )

            distance_values[valid] = np.sqrt(
                dx ** 2 + dy ** 2
            )

        frame_features[
            distance_col
        ] = distance_values


    # ============================================================
    # MERGE HAND LANDMARKS
    # ============================================================

    frame_features = frame_features.merge(
        hand_df,
        on="frame",
        how="left"
    )


    # ============================================================
    # CLEAN HAND LANDMARKS
    # ============================================================

    hand_feature_columns = [
        c for c in frame_features.columns
        if c.startswith("left_lm")
        or c.startswith("right_lm")
    ]

    frame_features[
        hand_feature_columns
    ] = (
        frame_features[
            hand_feature_columns
        ]
        .interpolate(
            method="linear",
            limit_direction="both"
        )
        .ffill()
        .bfill()
        .fillna(0)
    )


    # ============================================================
    # HAND-OBJECT DISTANCES
    # WRIST LANDMARK 0 → OBJECT CENTER
    # ============================================================

    for hand_name in ["left", "right"]:

        wrist_x = f"{hand_name}_lm0_x"
        wrist_y = f"{hand_name}_lm0_y"

        for obj in OBJECTS:

            distance_col = (
                f"{hand_name}_hand_to_{obj}_distance"
            )

            values = np.zeros(
                total_frames
            )

            hand_valid = (
                frame_features[wrist_x].notna()
                &
                frame_features[wrist_y].notna()
            )

            object_valid = (
                frame_features[
                    f"{obj}_presence"
                ] > 0
            )

            valid = (
                hand_valid &
                object_valid
            )

            if valid.any():

                dx = (
                    frame_features.loc[
                        valid,
                        wrist_x
                    ].values
                    -
                    frame_features.loc[
                        valid,
                        f"{obj}_center_x_norm"
                    ].values
                )

                dy = (
                    frame_features.loc[
                        valid,
                        wrist_y
                    ].values
                    -
                    frame_features.loc[
                        valid,
                        f"{obj}_center_y_norm"
                    ].values
                )

                values[valid] = np.sqrt(
                    dx ** 2 + dy ** 2
                )

            frame_features[
                distance_col
            ] = values


    # ============================================================
    # FINAL CLEANUP
    # ============================================================

    frame_features = frame_features.replace(
        [np.inf, -np.inf],
        np.nan
    )

    frame_features = frame_features.fillna(0)


    # ============================================================
    # FINAL VALIDATION
    # ============================================================

    feature_columns = [
        c for c in frame_features.columns
        if c != "frame"
    ]

    hand_columns = [
        c for c in feature_columns
        if c.startswith("left_lm")
        or c.startswith("right_lm")
    ]

    object_object_columns = [
        c for c in feature_columns
        if "_to_" in c
        and "_hand_to_" not in c
    ]

    hand_object_columns = [
        c for c in feature_columns
        if "_hand_to_" in c
    ]


    # ------------------------------------------------------------
    # HARD CHECKS
    # ------------------------------------------------------------

    if len(frame_features) != total_frames:
        raise RuntimeError(
            f"Frame count mismatch: "
            f"{len(frame_features)} != {total_frames}"
        )

    if len(hand_columns) != 126:
        raise RuntimeError(
            f"Hand landmark count wrong: "
            f"{len(hand_columns)} != 126"
        )

    if len(object_object_columns) != 10:
        raise RuntimeError(
            f"Object-object distance count wrong: "
            f"{len(object_object_columns)} != 10"
        )

    if len(hand_object_columns) != 10:
        raise RuntimeError(
            f"Hand-object distance count wrong: "
            f"{len(hand_object_columns)} != 10"
        )

    if len(feature_columns) != 206:
        raise RuntimeError(
            f"Feature count wrong: "
            f"{len(feature_columns)} != 206"
        )

    missing = int(
        frame_features.isna().sum().sum()
    )

    if missing != 0:
        raise RuntimeError(
            f"Missing values found: {missing}"
        )


    # ============================================================
    # SAVE
    # ============================================================

    frame_features.to_csv(
        final_path,
        index=False
    )


    # ============================================================
    # REPORT
    # ============================================================

    print()
    print("=" * 70)
    print(f"{stem} FINAL FEATURE VALIDATION")
    print("=" * 70)

    print(
        f"Rows: {len(frame_features)}"
    )

    print(
        f"Total columns: "
        f"{len(frame_features.columns)}"
    )

    print(
        f"Model features: "
        f"{len(feature_columns)}"
    )

    print()
    print(
        f"Hand landmark features: "
        f"{len(hand_columns)}"
    )

    print(
        f"Object-object distances: "
        f"{len(object_object_columns)}"
    )

    print(
        f"Hand-object distances: "
        f"{len(hand_object_columns)}"
    )

    print()
    print(
        f"Missing values: {missing}"
    )

    print()
    print("✓ VALIDATION PASSED")

    print()
    print("Saved:")
    print(final_path)

    print("=" * 70)


    return {
        "video": video_name,
        "status": "completed",
        "path": final_path,
        "frames": total_frames,
        "features": len(feature_columns),
        "missing": missing
    }


# ================================================================
# RUN BATCH
# ================================================================

batch_results = []

for video_name in BATCH:

    try:

        result = process_video(
            video_name
        )

        batch_results.append(result)

    except Exception as e:

        print()
        print("!" * 70)
        print(f"ERROR: {video_name}")
        print(str(e))
        print("!" * 70)

        batch_results.append({
            "video": video_name,
            "status": "ERROR",
            "error": str(e)
        })


# ================================================================
# SAVE BATCH STATUS
# ================================================================

status_path = (
    "/content/person2_output/"
    "batch1_status.json"
)

with open(status_path, "w") as f:

    json.dump(
        batch_results,
        f,
        indent=2
    )


# ================================================================
# FINAL BATCH SUMMARY
# ================================================================

print()
print()
print("=" * 70)
print("BATCH 1 SUMMARY")
print("=" * 70)

for result in batch_results:

    if result["status"] == "completed":

        print(
            f"✓ {result['video']} "
            f"| frames={result['frames']} "
            f"| features={result['features']} "
            f"| missing={result['missing']}"
        )

    elif result["status"] == "already_completed":

        print(
            f"✓ {result['video']} | already completed"
        )

    else:

        print(
            f"✗ {result['video']} "
            f"| ERROR: {result['error']}"
        )

print()
print(f"Status saved: {status_path}")
print("=" * 70)
```

```python
# ================================================================
# PERSON 2 — CORRECTED BATCH 1
# Next 5 unfinished videos
# ================================================================

import os
import json
import cv2
import numpy as np
import pandas as pd
import mediapipe as mp
from ultralytics import YOLO

from mediapipe.tasks import python
from mediapipe.tasks.python import vision


# ================================================================
# PATHS
# ================================================================

PROJECT_LIST = "/content/person2_output/project_video_list.json"

YOLO_PATH = "/content/best.pt"
HAND_MODEL_PATH = "/content/hand_landmarker.task"
TRACKER_PATH = "/content/bytetrack_custom.yaml"

BASE_OUT = "/content/person2_output/all_videos"

os.makedirs(BASE_OUT, exist_ok=True)


# ================================================================
# LOAD VIDEO LIST
# ================================================================

with open(PROJECT_LIST, "r") as f:
    project_videos = json.load(f)

if isinstance(project_videos, dict):
    project_videos = project_videos.get("videos", [])

project_videos = [
    os.path.basename(str(v))
    for v in project_videos
]

project_videos = list(dict.fromkeys(project_videos))


# ================================================================
# SEARCH LOCATIONS
# ================================================================

SEARCH_DIRS = [
    "/content/drive/MyDrive/BAS_HAR_SIH",
    "/content/drive/MyDrive/Sih videos",
    "/content/drive/MyDrive/sih_project"
]

EXCLUDE_DIRS = ["Aarambh 2k22"]


def find_video(video_name):

    for root_dir in SEARCH_DIRS:

        if not os.path.exists(root_dir):
            continue

        for root, dirs, files in os.walk(root_dir):

            dirs[:] = [
                d for d in dirs
                if d not in EXCLUDE_DIRS
            ]

            if video_name in files:
                return os.path.join(root, video_name)

    return None


# ================================================================
# COMPLETED CHECK
# ================================================================

def final_path_for(video_name):

    stem = os.path.splitext(video_name)[0]

    return os.path.join(
        BASE_OUT,
        stem,
        f"{stem}_final_features_with_hand_object_distance.csv"
    )


def is_completed(video_name):

    path = final_path_for(video_name)

    if not os.path.exists(path):
        return False

    try:
        df = pd.read_csv(path, nrows=5)

        return (
            len(df.columns) == 207
            and len([
                c for c in df.columns
                if c != "frame"
            ]) == 206
        )

    except Exception:
        return False


completed = []
unfinished = []

for video_name in project_videos:

    if is_completed(video_name):
        completed.append(video_name)
    else:
        unfinished.append(video_name)


# ================================================================
# BATCH SELECTION
# ================================================================

BATCH = unfinished[:5]


print("=" * 70)
print("PROJECT STATUS")
print("=" * 70)

print(f"Total project videos : {len(project_videos)}")
print(f"Completed            : {len(completed)}")
print(f"Remaining            : {len(unfinished)}")

print()
print("BATCH 1:")

for i, video_name in enumerate(BATCH, 1):
    print(f"  {i}. {video_name}")

print("=" * 70)


if len(BATCH) == 0:

    print("✓ No unfinished videos remain.")

else:

    # ============================================================
    # LOAD YOLO
    # ============================================================

    print()
    print("Loading YOLO...")

    model = YOLO(YOLO_PATH)

    print("✓ YOLO loaded")


    # ============================================================
    # CONSTANTS
    # ============================================================

    CLASS_NAMES = {
        0: "cotton",
        1: "dropper",
        2: "seed_jar",
        3: "target_plate",
        4: "water_dish"
    }

    OBJECTS = [
        "cotton",
        "dropper",
        "seed_jar",
        "target_plate",
        "water_dish"
    ]

    OBJECT_PAIRS = [
        ("cotton", "dropper"),
        ("cotton", "seed_jar"),
        ("cotton", "target_plate"),
        ("cotton", "water_dish"),
        ("dropper", "seed_jar"),
        ("dropper", "target_plate"),
        ("dropper", "water_dish"),
        ("seed_jar", "target_plate"),
        ("seed_jar", "water_dish"),
        ("target_plate", "water_dish")
    ]

    HAND_COLUMNS = ["frame"]

    for hand_name in ["left", "right"]:

        for lm in range(21):

            HAND_COLUMNS += [
                f"{hand_name}_lm{lm}_x",
                f"{hand_name}_lm{lm}_y",
                f"{hand_name}_lm{lm}_z"
            ]


    # ============================================================
    # PROCESS FUNCTION
    # ============================================================

    def process_video(video_name):

        stem = os.path.splitext(video_name)[0]

        print()
        print()
        print("#" * 70)
        print(f"PROCESSING: {video_name}")
        print("#" * 70)

        video_path = find_video(video_name)

        if video_path is None:
            raise FileNotFoundError(
                f"Could not find video: {video_name}"
            )

        print(f"Found: {video_path}")

        out_dir = os.path.join(
            BASE_OUT,
            stem
        )

        os.makedirs(
            out_dir,
            exist_ok=True
        )

        final_path = final_path_for(
            video_name
        )

        # --------------------------------------------------------
        # VIDEO
        # --------------------------------------------------------

        cap = cv2.VideoCapture(
            video_path
        )

        if not cap.isOpened():
            raise RuntimeError(
                f"Could not open: {video_path}"
            )

        total_frames = int(
            cap.get(cv2.CAP_PROP_FRAME_COUNT)
        )

        fps = cap.get(
            cv2.CAP_PROP_FPS
        )

        width = int(
            cap.get(cv2.CAP_PROP_FRAME_WIDTH)
        )

        height = int(
            cap.get(cv2.CAP_PROP_FRAME_HEIGHT)
        )

        print()
        print("Video:")
        print(f"  Frames : {total_frames}")
        print(f"  FPS    : {fps}")
        print(f"  Size   : {width} x {height}")


        # --------------------------------------------------------
        # FRESH MEDIAPIPE MODEL
        # --------------------------------------------------------

        base_options = python.BaseOptions(
            model_asset_path=HAND_MODEL_PATH
        )

        options = vision.HandLandmarkerOptions(
            base_options=base_options,
            running_mode=vision.RunningMode.VIDEO,
            num_hands=2,
            min_hand_detection_confidence=0.5,
            min_hand_presence_confidence=0.5,
            min_tracking_confidence=0.5
        )

        hand_landmarker = (
            vision.HandLandmarker.create_from_options(
                options
            )
        )


        tracking_rows = []
        hand_rows = []

        frame_idx = 0


        # ========================================================
        # FRAME PROCESSING
        # ========================================================

        while True:

            ret, frame = cap.read()

            if not ret:
                break


            # ----------------------------------------------------
            # YOLO TRACKING
            # ----------------------------------------------------

            results = model.track(
                frame,
                persist=True,
                tracker=TRACKER_PATH,
                conf=0.5,
                verbose=False
            )

            result = results[0]

            if result.boxes is not None:

                boxes = result.boxes

                xyxy = boxes.xyxy.cpu().numpy()
                confs = boxes.conf.cpu().numpy()
                classes = boxes.cls.cpu().numpy().astype(int)

                if boxes.id is not None:

                    track_ids = (
                        boxes.id
                        .cpu()
                        .numpy()
                        .astype(int)
                    )

                else:

                    track_ids = np.full(
                        len(xyxy),
                        -1
                    )


                for box, conf, cls_id, track_id in zip(
                    xyxy,
                    confs,
                    classes,
                    track_ids
                ):

                    if cls_id not in CLASS_NAMES:
                        continue

                    x1, y1, x2, y2 = box

                    tracking_rows.append({
                        "frame": frame_idx,
                        "class_id": int(cls_id),
                        "class_name": CLASS_NAMES[cls_id],
                        "track_id": int(track_id),
                        "x1": float(x1),
                        "y1": float(y1),
                        "x2": float(x2),
                        "y2": float(y2),
                        "confidence": float(conf)
                    })


            # ----------------------------------------------------
            # MEDIAPIPE HANDS
            # ----------------------------------------------------

            rgb = cv2.cvtColor(
                frame,
                cv2.COLOR_BGR2RGB
            )

            mp_image = mp.Image(
                image_format=mp.ImageFormat.SRGB,
                data=rgb
            )

            timestamp_ms = int(
                (frame_idx / fps) * 1000
            )

            hand_result = (
                hand_landmarker.detect_for_video(
                    mp_image,
                    timestamp_ms
                )
            )

            hand_data = {
                "frame": frame_idx
            }


            for hand_name in ["left", "right"]:

                for lm in range(21):

                    hand_data[
                        f"{hand_name}_lm{lm}_x"
                    ] = np.nan

                    hand_data[
                        f"{hand_name}_lm{lm}_y"
                    ] = np.nan

                    hand_data[
                        f"{hand_name}_lm{lm}_z"
                    ] = np.nan


            if hand_result.hand_landmarks:

                for hand_landmarks, handedness in zip(
                    hand_result.hand_landmarks,
                    hand_result.handedness
                ):

                    label = (
                        handedness[0]
                        .category_name
                        .lower()
                    )

                    if label not in [
                        "left",
                        "right"
                    ]:
                        continue

                    for lm_idx, landmark in enumerate(
                        hand_landmarks
                    ):

                        hand_data[
                            f"{label}_lm{lm_idx}_x"
                        ] = float(
                            landmark.x
                        )

                        hand_data[
                            f"{label}_lm{lm_idx}_y"
                        ] = float(
                            landmark.y
                        )

                        hand_data[
                            f"{label}_lm{lm_idx}_z"
                        ] = float(
                            landmark.z
                        )


            hand_rows.append(
                hand_data
            )

            frame_idx += 1

            if frame_idx % 50 == 0:

                print(
                    f"Processed "
                    f"{frame_idx}/{total_frames} frames"
                )


        cap.release()
        hand_landmarker.close()


        # ========================================================
        # DATAFRAMES
        # ========================================================

        tracking_df = pd.DataFrame(
            tracking_rows
        )

        hand_df = pd.DataFrame(
            hand_rows,
            columns=HAND_COLUMNS
        )

        print()
        print(
            f"Tracking rows: "
            f"{tracking_df.shape}"
        )

        print(
            f"Hand rows    : "
            f"{hand_df.shape}"
        )


        if len(tracking_df) == 0:

            raise RuntimeError(
                "No YOLO detections found."
            )


        # ========================================================
        # OBJECT GEOMETRY
        # ========================================================

        tracking_df["center_x"] = (
            tracking_df["x1"] +
            tracking_df["x2"]
        ) / 2

        tracking_df["center_y"] = (
            tracking_df["y1"] +
            tracking_df["y2"]
        ) / 2

        tracking_df["width"] = (
            tracking_df["x2"] -
            tracking_df["x1"]
        )

        tracking_df["height"] = (
            tracking_df["y2"] -
            tracking_df["y1"]
        )

        tracking_df["area"] = (
            tracking_df["width"] *
            tracking_df["height"]
        )

        tracking_df["center_x_norm"] = (
            tracking_df["center_x"] /
            width
        )

        tracking_df["center_y_norm"] = (
            tracking_df["center_y"] /
            height
        )

        tracking_df["width_norm"] = (
            tracking_df["width"] /
            width
        )

        tracking_df["height_norm"] = (
            tracking_df["height"] /
            height
        )

        tracking_df["area_norm"] = (
            tracking_df["area"] /
            (width * height)
        )


        # ========================================================
        # ONE ROW PER FRAME
        # ========================================================

        frame_features = pd.DataFrame({
            "frame": np.arange(
                total_frames
            )
        })


        for obj in OBJECTS:

            obj_df = tracking_df[
                tracking_df["class_name"] == obj
            ].copy()

            if len(obj_df) > 0:

                obj_df = (
                    obj_df
                    .sort_values(
                        "confidence",
                        ascending=False
                    )
                    .drop_duplicates(
                        "frame"
                    )
                    .set_index("frame")
                )


            # ----------------------------------------------------
            # Presence
            # ----------------------------------------------------

            values = np.zeros(
                total_frames
            )

            if len(obj_df) > 0:

                idx = obj_df.index[
                    obj_df.index < total_frames
                ]

                values[idx] = 1.0

            frame_features[
                f"{obj}_presence"
            ] = values


            # ----------------------------------------------------
            # Confidence
            # ----------------------------------------------------

            values = np.zeros(
                total_frames
            )

            if len(obj_df) > 0:

                for f, value in (
                    obj_df["confidence"].items()
                ):

                    if f < total_frames:
                        values[f] = value

            frame_features[
                f"{obj}_confidence"
            ] = values


            # ----------------------------------------------------
            # Raw geometry
            # ----------------------------------------------------

            for col in [
                "center_x",
                "center_y",
                "width",
                "height",
                "area"
            ]:

                values = np.zeros(
                    total_frames
                )

                if len(obj_df) > 0:

                    for f, value in (
                        obj_df[col].items()
                    ):

                        if f < total_frames:
                            values[f] = value

                frame_features[
                    f"{obj}_{col}"
                ] = values


            # ----------------------------------------------------
            # Normalized geometry
            # ----------------------------------------------------

            for col in [
                "center_x_norm",
                "center_y_norm",
                "width_norm",
                "height_norm",
                "area_norm"
            ]:

                values = np.zeros(
                    total_frames
                )

                if len(obj_df) > 0:

                    for f, value in (
                        obj_df[col].items()
                    ):

                        if f < total_frames:
                            values[f] = value

                frame_features[
                    f"{obj}_{col}"
                ] = values


        # ========================================================
        # OBJECT-OBJECT DISTANCES
        # ========================================================

        for obj1, obj2 in OBJECT_PAIRS:

            distance_col = (
                f"{obj1}_to_{obj2}_distance"
            )

            values = np.zeros(
                total_frames
            )

            valid = (
                (
                    frame_features[
                        f"{obj1}_presence"
                    ].values > 0
                )
                &
                (
                    frame_features[
                        f"{obj2}_presence"
                    ].values > 0
                )
            )

            if valid.any():

                dx = (
                    frame_features.loc[
                        valid,
                        f"{obj1}_center_x_norm"
                    ].values
                    -
                    frame_features.loc[
                        valid,
                        f"{obj2}_center_x_norm"
                    ].values
                )

                dy = (
                    frame_features.loc[
                        valid,
                        f"{obj1}_center_y_norm"
                    ].values
                    -
                    frame_features.loc[
                        valid,
                        f"{obj2}_center_y_norm"
                    ].values
                )

                values[valid] = np.sqrt(
                    dx ** 2 + dy ** 2
                )

            frame_features[
                distance_col
            ] = values


        # ========================================================
        # HAND LANDMARKS
        # ========================================================

        frame_features = frame_features.merge(
            hand_df,
            on="frame",
            how="left"
        )

        hand_feature_columns = [
            c for c in frame_features.columns
            if c.startswith("left_lm")
            or c.startswith("right_lm")
        ]

        frame_features[
            hand_feature_columns
        ] = (
            frame_features[
                hand_feature_columns
            ]
            .interpolate(
                method="linear",
                limit_direction="both"
            )
            .ffill()
            .bfill()
            .fillna(0)
        )


        # ========================================================
        # HAND-OBJECT DISTANCES
        # ========================================================

        for hand_name in [
            "left",
            "right"
        ]:

            wrist_x = (
                f"{hand_name}_lm0_x"
            )

            wrist_y = (
                f"{hand_name}_lm0_y"
            )

            for obj in OBJECTS:

                distance_col = (
                    f"{hand_name}_hand_to_{obj}_distance"
                )

                values = np.zeros(
                    total_frames
                )

                hand_valid = (
                    frame_features[wrist_x].notna()
                    &
                    frame_features[wrist_y].notna()
                )

                object_valid = (
                    frame_features[
                        f"{obj}_presence"
                    ] > 0
                )

                valid = (
                    hand_valid &
                    object_valid
                )

                if valid.any():

                    dx = (
                        frame_features.loc[
                            valid,
                            wrist_x
                        ].values
                        -
                        frame_features.loc[
                            valid,
                            f"{obj}_center_x_norm"
                        ].values
                    )

                    dy = (
                        frame_features.loc[
                            valid,
                            wrist_y
                        ].values
                        -
                        frame_features.loc[
                            valid,
                            f"{obj}_center_y_norm"
                        ].values
                    )

                    values[valid] = np.sqrt(
                        dx ** 2 + dy ** 2
                    )

                frame_features[
                    distance_col
                ] = values


        # ========================================================
        # CLEAN
        # ========================================================

        frame_features = (
            frame_features
            .replace(
                [np.inf, -np.inf],
                np.nan
            )
            .fillna(0)
        )


        # ========================================================
        # VALIDATION
        # ========================================================

        feature_columns = [
            c for c in frame_features.columns
            if c != "frame"
        ]

        hand_columns = [
            c for c in feature_columns
            if c.startswith("left_lm")
            or c.startswith("right_lm")
        ]

        object_object_columns = [
            c for c in feature_columns
            if "_to_" in c
            and "_hand_to_" not in c
        ]

        hand_object_columns = [
            c for c in feature_columns
            if "_hand_to_" in c
        ]

        missing = int(
            frame_features.isna()
            .sum()
            .sum()
        )


        # HARD CHECKS

        if len(frame_features) != total_frames:
            raise RuntimeError(
                f"Rows {len(frame_features)} "
                f"!= video frames {total_frames}"
            )

        if len(hand_columns) != 126:
            raise RuntimeError(
                f"Hand features = "
                f"{len(hand_columns)}, expected 126"
            )

        if len(object_object_columns) != 10:
            raise RuntimeError(
                f"Object-object distances = "
                f"{len(object_object_columns)}, expected 10"
            )

        if len(hand_object_columns) != 10:
            raise RuntimeError(
                f"Hand-object distances = "
                f"{len(hand_object_columns)}, expected 10"
            )

        if len(feature_columns) != 206:
            raise RuntimeError(
                f"Model features = "
                f"{len(feature_columns)}, expected 206"
            )

        if missing != 0:
            raise RuntimeError(
                f"Missing values = {missing}"
            )


        # ========================================================
        # SAVE
        # ========================================================

        frame_features.to_csv(
            final_path,
            index=False
        )


        # ========================================================
        # REPORT
        # ========================================================

        print()
        print("=" * 70)
        print(f"{stem} FINAL FEATURE VALIDATION")
        print("=" * 70)

        print(
            f"Rows: {len(frame_features)}"
        )

        print(
            f"Total columns: "
            f"{len(frame_features.columns)}"
        )

        print(
            f"Model features: "
            f"{len(feature_columns)}"
        )

        print()
        print(
            f"Hand landmark features: "
            f"{len(hand_columns)}"
        )

        print(
            f"Object-object distances: "
            f"{len(object_object_columns)}"
        )

        print(
            f"Hand-object distances: "
            f"{len(hand_object_columns)}"
        )

        print()
        print(
            f"Missing values: {missing}"
        )

        print()
        print("✓ VALIDATION PASSED")

        print()
        print("Saved:")
        print(final_path)

        print("=" * 70)


        return {
            "video": video_name,
            "status": "completed",
            "path": final_path,
            "frames": total_frames,
            "features": len(feature_columns),
            "missing": missing
        }


    # ============================================================
    # RUN BATCH
    # ============================================================

    batch_results = []

    for video_name in BATCH:

        try:

            result = process_video(
                video_name
            )

            batch_results.append(
                result
            )

        except Exception as e:

            print()
            print("!" * 70)
            print(
                f"ERROR: {video_name}"
            )
            print(str(e))
            print("!" * 70)

            batch_results.append({
                "video": video_name,
                "status": "ERROR",
                "error": str(e)
            })


    # ============================================================
    # SAVE STATUS
    # ============================================================

    status_path = (
        "/content/person2_output/"
        "batch1_status.json"
    )

    with open(status_path, "w") as f:

        json.dump(
            batch_results,
            f,
            indent=2
        )


    # ============================================================
    # SUMMARY
    # ============================================================

    print()
    print()
    print("=" * 70)
    print("BATCH 1 SUMMARY")
    print("=" * 70)

    for result in batch_results:

        if result["status"] == "completed":

            print(
                f"✓ {result['video']} "
                f"| frames={result['frames']} "
                f"| features={result['features']} "
                f"| missing={result['missing']}"
            )

        else:

            print(
                f"✗ {result['video']} "
                f"| ERROR: {result['error']}"
            )

    print()
    print(f"Status saved: {status_path}")
    print("=" * 70)
```

```python
# ============================================================
# PERSON 2 — BATCH 2
# Process next 5 unfinished project videos
# ============================================================

import os
import json
import glob
import cv2
import numpy as np
import pandas as pd
from ultralytics import YOLO
import mediapipe as mp
from mediapipe.tasks import python
from mediapipe.tasks.python import vision

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

PROJECT_LIST = "/content/person2_output/project_video_list.json"
OUTPUT_ROOT = "/content/person2_output/all_videos"
STATUS_PATH = "/content/person2_output/batch2_status.json"

YOLO_WEIGHTS = "/content/best.pt"
TRACKER_CONFIG = "/content/bytetrack_custom.yaml"
HAND_MODEL = "/content/hand_landmarker.task"

VIDEO_ROOTS = [
    "/content/drive/MyDrive/BAS_HAR_SIH",
    "/content/drive/MyDrive/Sih videos",
    "/content/drive/MyDrive/sih_project"
]

# ------------------------------------------------------------
# SETTINGS
# ------------------------------------------------------------

BATCH_SIZE = 5

CLASS_NAMES = {
    0: "cotton",
    1: "dropper",
    2: "seed_jar",
    3: "target_plate",
    4: "water_dish"
}

OBJECTS = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# ------------------------------------------------------------
# LOAD PROJECT VIDEO LIST
# ------------------------------------------------------------

with open(PROJECT_LIST, "r") as f:
    project_videos = json.load(f)

# Handle either strings or dictionary entries
video_names = []

for item in project_videos:
    if isinstance(item, str):
        video_names.append(os.path.basename(item))
    elif isinstance(item, dict):
        name = item.get("video_name") or item.get("video_file") or item.get("name")
        if name:
            video_names.append(os.path.basename(name))

video_names = sorted(set(video_names))

print("=" * 70)
print("PROJECT VIDEO LIST")
print("=" * 70)
print("Total project videos:", len(video_names))

# ------------------------------------------------------------
# FIND VIDEO FILES RECURSIVELY
# ------------------------------------------------------------

video_lookup = {}

for root in VIDEO_ROOTS:
    if not os.path.exists(root):
        continue

    for path in glob.glob(os.path.join(root, "**", "*"), recursive=True):

        if not os.path.isfile(path):
            continue

        # Exclude unrelated Aarambh folder
        if "Aarambh 2k22" in path:
            continue

        base = os.path.basename(path)

        if base.lower().endswith((
            ".mp4", ".avi", ".mov", ".mkv", ".webm", ".mpeg", ".mpg"
        )):
            video_lookup[base] = path

print("Videos found on Drive:", len(video_lookup))

# ------------------------------------------------------------
# DETERMINE COMPLETED VIDEOS
# ------------------------------------------------------------

def is_complete(video_name):

    stem = os.path.splitext(video_name)[0]

    output_file = os.path.join(
        OUTPUT_ROOT,
        stem,
        f"{stem}_final_features_with_hand_object_distance.csv"
    )

    if not os.path.exists(output_file):
        return False

    try:
        df = pd.read_csv(output_file)

        # Must contain frame + exactly 206 model features
        if df.shape[1] != 207:
            return False

        if "frame" not in df.columns:
            return False

        if df.drop(columns=["frame"]).shape[1] != 206:
            return False

        if df.isna().sum().sum() != 0:
            return False

        return True

    except Exception:
        return False


completed = [
    v for v in video_names
    if is_complete(v)
]

unfinished = [
    v for v in video_names
    if not is_complete(v)
]

batch_videos = unfinished[:BATCH_SIZE]

print()
print("Completed videos :", len(completed))
print("Unfinished videos:", len(unfinished))

print()
print("BATCH 2 VIDEOS:")
for i, v in enumerate(batch_videos, 1):
    print(f"{i}. {v}")

if len(batch_videos) == 0:
    print("\n🎉 ALL PROJECT VIDEOS ARE COMPLETE!")
    raise SystemExit

# ------------------------------------------------------------
# LOAD YOLO ONCE
# ------------------------------------------------------------

print()
print("=" * 70)
print("LOADING MODELS")
print("=" * 70)

model = YOLO(YOLO_WEIGHTS)
print("✓ YOLO loaded")

# ------------------------------------------------------------
# HELPER: OBJECT FEATURES FROM TRACKING
# ------------------------------------------------------------

def build_object_features(tracking_df, frame_width, frame_height):

    rows = []

    # frame range
    all_frames = sorted(tracking_df["frame"].unique())

    for frame_num in all_frames:

        frame_data = tracking_df[
            tracking_df["frame"] == frame_num
        ]

        row = {
            "frame": int(frame_num)
        }

        # ----------------------------------------------------
        # Presence + confidence
        # ----------------------------------------------------

        centers = {}

        for obj in OBJECTS:

            obj_data = frame_data[
                frame_data["class_name"] == obj
            ]

            if len(obj_data) == 0:

                row[f"{obj}_presence"] = 0
                row[f"{obj}_confidence"] = 0.0

                row[f"{obj}_x1"] = 0.0
                row[f"{obj}_y1"] = 0.0
                row[f"{obj}_x2"] = 0.0
                row[f"{obj}_y2"] = 0.0
                row[f"{obj}_cx"] = 0.0
                row[f"{obj}_cy"] = 0.0
                row[f"{obj}_width"] = 0.0
                row[f"{obj}_height"] = 0.0

                row[f"{obj}_nx1"] = 0.0
                row[f"{obj}_ny1"] = 0.0
                row[f"{obj}_nx2"] = 0.0
                row[f"{obj}_ny2"] = 0.0
                row[f"{obj}_ncx"] = 0.0
                row[f"{obj}_ncy"] = 0.0
                row[f"{obj}_nwidth"] = 0.0
                row[f"{obj}_nheight"] = 0.0

                centers[obj] = None

            else:

                # Use highest-confidence detection for this class
                det = obj_data.loc[
                    obj_data["confidence"].idxmax()
                ]

                x1 = float(det["x1"])
                y1 = float(det["y1"])
                x2 = float(det["x2"])
                y2 = float(det["y2"])
                conf = float(det["confidence"])

                cx = (x1 + x2) / 2
                cy = (y1 + y2) / 2

                width = x2 - x1
                height = y2 - y1

                row[f"{obj}_presence"] = 1
                row[f"{obj}_confidence"] = conf

                # Raw geometry
                row[f"{obj}_x1"] = x1
                row[f"{obj}_y1"] = y1
                row[f"{obj}_x2"] = x2
                row[f"{obj}_y2"] = y2
                row[f"{obj}_cx"] = cx
                row[f"{obj}_cy"] = cy
                row[f"{obj}_width"] = width
                row[f"{obj}_height"] = height

                # Normalized geometry
                row[f"{obj}_nx1"] = x1 / frame_width
                row[f"{obj}_ny1"] = y1 / frame_height
                row[f"{obj}_nx2"] = x2 / frame_width
                row[f"{obj}_ny2"] = y2 / frame_height
                row[f"{obj}_ncx"] = cx / frame_width
                row[f"{obj}_ncy"] = cy / frame_height
                row[f"{obj}_nwidth"] = width / frame_width
                row[f"{obj}_nheight"] = height / frame_height

                centers[obj] = (
                    cx / frame_width,
                    cy / frame_height
                )

        # ----------------------------------------------------
        # Object-object normalized center distances
        # ----------------------------------------------------

        for i in range(len(OBJECTS)):

            for j in range(i + 1, len(OBJECTS)):

                a = OBJECTS[i]
                b = OBJECTS[j]

                key = f"{a}_to_{b}_distance"

                if centers[a] is None or centers[b] is None:
                    row[key] = 0.0
                else:
                    dx = centers[a][0] - centers[b][0]
                    dy = centers[a][1] - centers[b][1]

                    row[key] = float(
                        np.sqrt(dx * dx + dy * dy)
                    )

        rows.append(row)

    return pd.DataFrame(rows)

# ------------------------------------------------------------
# PROCESS ONE VIDEO
# ------------------------------------------------------------

def process_video(video_name):

    video_path = video_lookup.get(video_name)

    if video_path is None:
        raise FileNotFoundError(
            f"Could not find video: {video_name}"
        )

    stem = os.path.splitext(video_name)[0]

    output_dir = os.path.join(
        OUTPUT_ROOT,
        stem
    )

    os.makedirs(output_dir, exist_ok=True)

    print()
    print("=" * 70)
    print(f"PROCESSING: {video_name}")
    print("=" * 70)

    # --------------------------------------------------------
    # VIDEO INFO
    # --------------------------------------------------------

    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        raise RuntimeError(
            f"Could not open video: {video_path}"
        )

    frame_count = int(
        cap.get(cv2.CAP_PROP_FRAME_COUNT)
    )

    fps = float(
        cap.get(cv2.CAP_PROP_FPS)
    )

    frame_width = int(
        cap.get(cv2.CAP_PROP_FRAME_WIDTH)
    )

    frame_height = int(
        cap.get(cv2.CAP_PROP_FRAME_HEIGHT)
    )

    print("Path   :", video_path)
    print("Frames :", frame_count)
    print("FPS    :", fps)
    print("Size   :", frame_height, "x", frame_width)

    # --------------------------------------------------------
    # FRESH MEDIAPIPE LANDMARKER FOR THIS VIDEO
    # --------------------------------------------------------

    base_options = python.BaseOptions(
        model_asset_path=HAND_MODEL
    )

    options = vision.HandLandmarkerOptions(
        base_options=base_options,
        running_mode=vision.RunningMode.VIDEO,
        num_hands=2,
        min_hand_detection_confidence=0.3,
        min_hand_presence_confidence=0.3,
        min_tracking_confidence=0.3
    )

    hand_landmarker = vision.HandLandmarker.create_from_options(
        options
    )

    # --------------------------------------------------------
    # TRACKING STORAGE
    # --------------------------------------------------------

    tracking_rows = []
    hand_rows = []

    frame_number = 0

    while True:

        ret, frame = cap.read()

        if not ret:
            break

        # ----------------------------------------------------
        # YOLO + BYTETRACK
        # ----------------------------------------------------

        results = model.track(
            frame,
            persist=True,
            tracker=TRACKER_CONFIG,
            conf=0.5,
            verbose=False
        )

        result = results[0]

        if (
            result.boxes is not None
            and len(result.boxes) > 0
        ):

            boxes = result.boxes

            xyxy = boxes.xyxy.cpu().numpy()
            confs = boxes.conf.cpu().numpy()
            classes = boxes.cls.cpu().numpy().astype(int)

            if boxes.id is not None:
                track_ids = (
                    boxes.id.cpu().numpy().astype(int)
                )
            else:
                track_ids = [-1] * len(xyxy)

            for box, conf, cls_id, track_id in zip(
                xyxy,
                confs,
                classes,
                track_ids
            ):

                if cls_id not in CLASS_NAMES:
                    continue

                x1, y1, x2, y2 = box

                tracking_rows.append({
                    "frame": frame_number,
                    "class_id": int(cls_id),
                    "class_name": CLASS_NAMES[cls_id],
                    "track_id": int(track_id),
                    "x1": float(x1),
                    "y1": float(y1),
                    "x2": float(x2),
                    "y2": float(y2),
                    "confidence": float(conf)
                })

        # ----------------------------------------------------
        # MEDIAPIPE HAND LANDMARKS
        # ----------------------------------------------------

        rgb = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2RGB
        )

        mp_image = mp.Image(
            image_format=mp.ImageFormat.SRGB,
            data=rgb
        )

        timestamp_ms = int(
            frame_number * 1000 / fps
        )

        hand_result = hand_landmarker.detect_for_video(
            mp_image,
            timestamp_ms
        )

        hand_row = {
            "frame": frame_number
        }

        # Initialize all 126 values
        # 2 hands × 21 landmarks × 3 coordinates

        for hand_name in ["left", "right"]:

            for landmark_idx in range(21):

                hand_row[
                    f"{hand_name}_hand_{landmark_idx}_x"
                ] = np.nan

                hand_row[
                    f"{hand_name}_hand_{landmark_idx}_y"
                ] = np.nan

                hand_row[
                    f"{hand_name}_hand_{landmark_idx}_z"
                ] = np.nan

        # MediaPipe handedness
        for hand_idx, landmarks in enumerate(
            hand_result.hand_landmarks
        ):

            if hand_idx >= len(
                hand_result.handedness
            ):
                continue

            handedness = (
                hand_result.handedness[hand_idx][0]
            )

            hand_name = (
                handedness.category_name.lower()
            )

            if hand_name not in ["left", "right"]:
                continue

            for landmark_idx, landmark in enumerate(
                landmarks
            ):

                hand_row[
                    f"{hand_name}_hand_{landmark_idx}_x"
                ] = float(landmark.x)

                hand_row[
                    f"{hand_name}_hand_{landmark_idx}_y"
                ] = float(landmark.y)

                hand_row[
                    f"{hand_name}_hand_{landmark_idx}_z"
                ] = float(landmark.z)

        hand_rows.append(hand_row)

        frame_number += 1

        if frame_number % 50 == 0:
            print(
                f"Processed {frame_number}/{frame_count} frames"
            )

    cap.release()
    hand_landmarker.close()

    # --------------------------------------------------------
    # DATAFRAMES
    # --------------------------------------------------------

    tracking_df = pd.DataFrame(
        tracking_rows
    )

    hand_df = pd.DataFrame(
        hand_rows
    )

    print()
    print("Tracking rows:", tracking_df.shape)
    print("Hand rows    :", hand_df.shape)

    # --------------------------------------------------------
    # OBJECT FEATURES
    # --------------------------------------------------------

    object_df = build_object_features(
        tracking_df,
        frame_width,
        frame_height
    )

    # --------------------------------------------------------
    # HAND FEATURES
    # --------------------------------------------------------

    # Ensure one row per frame
    hand_df = hand_df.set_index(
        "frame"
    ).reindex(
        range(frame_count)
    ).reset_index()

    hand_feature_cols = [
        c for c in hand_df.columns
        if c != "frame"
    ]

    # Interpolate short gaps and fill remaining gaps
    hand_df[hand_feature_cols] = (
        hand_df[hand_feature_cols]
        .interpolate(limit_direction="both")
        .ffill()
        .bfill()
        .fillna(0.0)
    )

    # --------------------------------------------------------
    # COMBINE
    # --------------------------------------------------------

    frame_features = object_df.merge(
        hand_df,
        on="frame",
        how="left"
    )

    # --------------------------------------------------------
    # HAND-OBJECT DISTANCES
    # --------------------------------------------------------

    for hand_name in ["left", "right"]:

        wrist_x = f"{hand_name}_hand_0_x"
        wrist_y = f"{hand_name}_hand_0_y"

        for obj in OBJECTS:

            distance_col = (
                f"{hand_name}_hand_to_{obj}_distance"
            )

            object_x = f"{obj}_ncx"
            object_y = f"{obj}_ncy"

            dx = (
                frame_features[wrist_x]
                - frame_features[object_x]
            )

            dy = (
                frame_features[wrist_y]
                - frame_features[object_y]
            )

            distance = np.sqrt(
                dx * dx + dy * dy
            )

            # Distance is meaningful only when
            # the hand and object are present.
            hand_valid = (
                frame_features[wrist_x].notna()
                & frame_features[wrist_y].notna()
            )

            object_present = (
                frame_features[f"{obj}_presence"] == 1
            )

            frame_features[distance_col] = np.where(
                hand_valid & object_present,
                distance,
                0.0
            )

    # --------------------------------------------------------
    # FINAL CLEANUP
    # --------------------------------------------------------

    frame_features = (
        frame_features
        .replace([np.inf, -np.inf], np.nan)
        .fillna(0.0)
    )

    # --------------------------------------------------------
    # VALIDATION
    # --------------------------------------------------------

    model_feature_cols = [
        c for c in frame_features.columns
        if c != "frame"
    ]

    hand_cols = [
        c for c in model_feature_cols
        if "_hand_" in c
        and "_to_" not in c
    ]

    object_distance_cols = [
        c for c in model_feature_cols
        if "_to_" in c
        and "_hand_" not in c
    ]

    hand_object_cols = [
        c for c in model_feature_cols
        if "_hand_to_" in c
    ]

    missing_values = int(
        frame_features.isna().sum().sum()
    )

    print()
    print("=" * 70)
    print(f"{stem} FINAL FEATURE VALIDATION")
    print("=" * 70)

    print("Rows:", len(frame_features))
    print("Total columns:", len(frame_features.columns))
    print("Model features:", len(model_feature_cols))
    print()
    print("Hand landmark features:", len(hand_cols))
    print("Object-object distances:", len(object_distance_cols))
    print("Hand-object distances:", len(hand_object_cols))
    print()
    print("Missing values:", missing_values)

    # --------------------------------------------------------
    # STRICT VALIDATION
    # --------------------------------------------------------

    assert len(frame_features) == frame_count
    assert len(hand_cols) == 126
    assert len(object_distance_cols) == 10
    assert len(hand_object_cols) == 10
    assert len(model_feature_cols) == 206
    assert missing_values == 0

    # --------------------------------------------------------
    # SAVE
    # --------------------------------------------------------

    output_file = os.path.join(
        output_dir,
        f"{stem}_final_features_with_hand_object_distance.csv"
    )

    frame_features.to_csv(
        output_file,
        index=False
    )

    print()
    print("✓ VALIDATION PASSED")
    print("Saved:")
    print(output_file)

    return {
        "video": video_name,
        "frames": int(frame_count),
        "features": int(len(model_feature_cols)),
        "missing": int(missing_values),
        "output": output_file,
        "status": "success"
    }

# ------------------------------------------------------------
# RUN BATCH 2
# ------------------------------------------------------------

batch_status = []

for video_name in batch_videos:

    try:

        result = process_video(video_name)
        batch_status.append(result)

    except Exception as e:

        print()
        print("✗ ERROR:", video_name)
        print(str(e))

        batch_status.append({
            "video": video_name,
            "status": "error",
            "error": str(e)
        })

# ------------------------------------------------------------
# SAVE STATUS
# ------------------------------------------------------------

with open(STATUS_PATH, "w") as f:
    json.dump(
        batch_status,
        f,
        indent=2
    )

# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

print()
print("=" * 70)
print("BATCH 2 SUMMARY")
print("=" * 70)

for item in batch_status:

    if item["status"] == "success":

        print(
            f"✓ {item['video']} | "
            f"frames={item['frames']} | "
            f"features={item['features']} | "
            f"missing={item['missing']}"
        )

    else:

        print(
            f"✗ {item['video']} | "
            f"ERROR: {item['error']}"
        )

print()
print("Status saved:", STATUS_PATH)
print("=" * 70)
```

```python
# ============================================================
# I2.mp4 — FULL PIPELINE DIAGNOSTIC
# ============================================================

import os
import cv2
import numpy as np
import pandas as pd
import traceback
import mediapipe as mp

from mediapipe.tasks import python
from mediapipe.tasks.python import vision

VIDEO_PATH = "/content/drive/MyDrive/Sih videos/I2.mp4"
YOLO_WEIGHTS = "/content/best.pt"
TRACKER_CONFIG = "/content/bytetrack_custom.yaml"
HAND_MODEL = "/content/hand_landmarker.task"

CLASS_NAMES = {
    0: "cotton",
    1: "dropper",
    2: "seed_jar",
    3: "target_plate",
    4: "water_dish"
}

print("=" * 70)
print("I2.mp4 — FULL PIPELINE DIAGNOSTIC")
print("=" * 70)

stage = "initialization"

try:

    # --------------------------------------------------------
    # 1. LOAD YOLO
    # --------------------------------------------------------

    stage = "YOLO loading"

    print("\n[1] Loading YOLO...")

    from ultralytics import YOLO

    diagnostic_model = YOLO(YOLO_WEIGHTS)

    print("✓ YOLO loaded")

    # --------------------------------------------------------
    # 2. OPEN VIDEO
    # --------------------------------------------------------

    stage = "video opening"

    print("\n[2] Opening video...")

    cap = cv2.VideoCapture(VIDEO_PATH)

    if not cap.isOpened():
        raise RuntimeError("Could not open video")

    frame_count = int(
        cap.get(cv2.CAP_PROP_FRAME_COUNT)
    )

    fps = float(
        cap.get(cv2.CAP_PROP_FPS)
    )

    width = int(
        cap.get(cv2.CAP_PROP_FRAME_WIDTH)
    )

    height = int(
        cap.get(cv2.CAP_PROP_FRAME_HEIGHT)
    )

    print("Frames:", frame_count)
    print("FPS:", fps)
    print("Size:", width, "x", height)

    # --------------------------------------------------------
    # 3. CREATE MEDIAPIPE
    # --------------------------------------------------------

    stage = "MediaPipe initialization"

    print("\n[3] Creating MediaPipe...")

    base_options = python.BaseOptions(
        model_asset_path=HAND_MODEL
    )

    options = vision.HandLandmarkerOptions(
        base_options=base_options,
        running_mode=vision.RunningMode.VIDEO,
        num_hands=2,
        min_hand_detection_confidence=0.3,
        min_hand_presence_confidence=0.3,
        min_tracking_confidence=0.3
    )

    hand_landmarker = vision.HandLandmarker.create_from_options(
        options
    )

    print("✓ MediaPipe created")

    # --------------------------------------------------------
    # 4. PROCESS VIDEO
    # --------------------------------------------------------

    tracking_rows = []
    hand_rows = []

    frame_number = 0

    print("\n[4] Processing frames...")

    while True:

        ret, frame = cap.read()

        if not ret:
            break

        # ----------------------------------------------------
        # YOLO
        # ----------------------------------------------------

        stage = f"YOLO frame {frame_number}"

        results = diagnostic_model.track(
            frame,
            persist=True,
            tracker=TRACKER_CONFIG,
            conf=0.5,
            verbose=False
        )

        result = results[0]

        if (
            result.boxes is not None
            and len(result.boxes) > 0
        ):

            boxes = result.boxes

            xyxy = boxes.xyxy.cpu().numpy()
            confs = boxes.conf.cpu().numpy()
            classes = boxes.cls.cpu().numpy().astype(int)

            if boxes.id is not None:
                track_ids = (
                    boxes.id.cpu().numpy().astype(int)
                )
            else:
                track_ids = [-1] * len(xyxy)

            for box, conf, cls_id, track_id in zip(
                xyxy,
                confs,
                classes,
                track_ids
            ):

                if cls_id not in CLASS_NAMES:
                    continue

                x1, y1, x2, y2 = box

                tracking_rows.append({
                    "frame": frame_number,
                    "class_id": int(cls_id),
                    "class_name": CLASS_NAMES[cls_id],
                    "track_id": int(track_id),
                    "x1": float(x1),
                    "y1": float(y1),
                    "x2": float(x2),
                    "y2": float(y2),
                    "confidence": float(conf)
                })

        # ----------------------------------------------------
        # MEDIAPIPE
        # ----------------------------------------------------

        stage = f"MediaPipe frame {frame_number}"

        rgb = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2RGB
        )

        mp_image = mp.Image(
            image_format=mp.ImageFormat.SRGB,
            data=rgb
        )

        timestamp_ms = int(
            frame_number * 1000 / fps
        )

        hand_result = hand_landmarker.detect_for_video(
            mp_image,
            timestamp_ms
        )

        hand_row = {
            "frame": frame_number
        }

        for hand_name in ["left", "right"]:

            for landmark_idx in range(21):

                hand_row[
                    f"{hand_name}_hand_{landmark_idx}_x"
                ] = np.nan

                hand_row[
                    f"{hand_name}_hand_{landmark_idx}_y"
                ] = np.nan

                hand_row[
                    f"{hand_name}_hand_{landmark_idx}_z"
                ] = np.nan

        for hand_idx, landmarks in enumerate(
            hand_result.hand_landmarks
        ):

            if hand_idx >= len(
                hand_result.handedness
            ):
                continue

            handedness = (
                hand_result.handedness[hand_idx][0]
            )

            hand_name = (
                handedness.category_name.lower()
            )

            if hand_name not in ["left", "right"]:
                continue

            for landmark_idx, landmark in enumerate(
                landmarks
            ):

                hand_row[
                    f"{hand_name}_hand_{landmark_idx}_x"
                ] = float(landmark.x)

                hand_row[
                    f"{hand_name}_hand_{landmark_idx}_y"
                ] = float(landmark.y)

                hand_row[
                    f"{hand_name}_hand_{landmark_idx}_z"
                ] = float(landmark.z)

        hand_rows.append(hand_row)

        frame_number += 1

        if frame_number % 50 == 0:
            print(
                f"  Processed {frame_number}/{frame_count}"
            )

    cap.release()
    hand_landmarker.close()

    print("✓ Frame processing completed")

    # --------------------------------------------------------
    # 5. CREATE DATAFRAMES
    # --------------------------------------------------------

    stage = "DataFrame construction"

    print("\n[5] Creating DataFrames...")

    tracking_df = pd.DataFrame(
        tracking_rows
    )

    hand_df = pd.DataFrame(
        hand_rows
    )

    print("Tracking:", tracking_df.shape)
    print("Hands   :", hand_df.shape)

    # --------------------------------------------------------
    # 6. OBJECT FEATURES
    # --------------------------------------------------------

    stage = "Object feature construction"

    print("\n[6] Building object features...")

    objects = [
        "cotton",
        "dropper",
        "seed_jar",
        "target_plate",
        "water_dish"
    ]

    rows = []

    for frame_num in range(frame_count):

        frame_data = tracking_df[
            tracking_df["frame"] == frame_num
        ]

        row = {
            "frame": frame_num
        }

        centers = {}

        for obj in objects:

            obj_data = frame_data[
                frame_data["class_name"] == obj
            ]

            if len(obj_data) == 0:

                row[f"{obj}_presence"] = 0
                row[f"{obj}_confidence"] = 0.0

                for suffix in [
                    "x1", "y1", "x2", "y2",
                    "cx", "cy", "width", "height",
                    "nx1", "ny1", "nx2", "ny2",
                    "ncx", "ncy", "nwidth", "nheight"
                ]:
                    row[f"{obj}_{suffix}"] = 0.0

                centers[obj] = None

            else:

                det = obj_data.loc[
                    obj_data["confidence"].idxmax()
                ]

                x1 = float(det["x1"])
                y1 = float(det["y1"])
                x2 = float(det["x2"])
                y2 = float(det["y2"])

                conf = float(
                    det["confidence"]
                )

                cx = (x1 + x2) / 2
                cy = (y1 + y2) / 2

                bw = x2 - x1
                bh = y2 - y1

                row[f"{obj}_presence"] = 1
                row[f"{obj}_confidence"] = conf

                row[f"{obj}_x1"] = x1
                row[f"{obj}_y1"] = y1
                row[f"{obj}_x2"] = x2
                row[f"{obj}_y2"] = y2
                row[f"{obj}_cx"] = cx
                row[f"{obj}_cy"] = cy
                row[f"{obj}_width"] = bw
                row[f"{obj}_height"] = bh

                row[f"{obj}_nx1"] = x1 / width
                row[f"{obj}_ny1"] = y1 / height
                row[f"{obj}_nx2"] = x2 / width
                row[f"{obj}_ny2"] = y2 / height
                row[f"{obj}_ncx"] = cx / width
                row[f"{obj}_ncy"] = cy / height
                row[f"{obj}_nwidth"] = bw / width
                row[f"{obj}_nheight"] = bh / height

                centers[obj] = (
                    cx / width,
                    cy / height
                )

        # Object-object distances

        for i in range(len(objects)):

            for j in range(i + 1, len(objects)):

                a = objects[i]
                b = objects[j]

                if (
                    centers[a] is None
                    or centers[b] is None
                ):
                    distance = 0.0

                else:

                    dx = (
                        centers[a][0]
                        - centers[b][0]
                    )

                    dy = (
                        centers[a][1]
                        - centers[b][1]
                    )

                    distance = np.sqrt(
                        dx * dx + dy * dy
                    )

                row[
                    f"{a}_to_{b}_distance"
                ] = float(distance)

        rows.append(row)

    object_df = pd.DataFrame(rows)

    print("Object features:", object_df.shape)

    # --------------------------------------------------------
    # 7. HAND FEATURES
    # --------------------------------------------------------

    stage = "Hand feature processing"

    print("\n[7] Processing hand features...")

    hand_df = (
        hand_df
        .set_index("frame")
        .reindex(range(frame_count))
        .reset_index()
    )

    hand_feature_cols = [
        c for c in hand_df.columns
        if c != "frame"
    ]

    print(
        "Hand feature columns:",
        len(hand_feature_cols)
    )

    hand_df[hand_feature_cols] = (
        hand_df[hand_feature_cols]
        .interpolate(limit_direction="both")
        .ffill()
        .bfill()
        .fillna(0.0)
    )

    # --------------------------------------------------------
    # 8. MERGE
    # --------------------------------------------------------

    stage = "Feature merging"

    print("\n[8] Merging features...")

    frame_features = object_df.merge(
        hand_df,
        on="frame",
        how="left"
    )

    print(
        "After merge:",
        frame_features.shape
    )

    # --------------------------------------------------------
    # 9. HAND-OBJECT DISTANCES
    # --------------------------------------------------------

    stage = "Hand-object distance construction"

    print("\n[9] Building hand-object distances...")

    for hand_name in ["left", "right"]:

        wrist_x = f"{hand_name}_hand_0_x"
        wrist_y = f"{hand_name}_hand_0_y"

        for obj in objects:

            distance_col = (
                f"{hand_name}_hand_to_{obj}_distance"
            )

            dx = (
                frame_features[wrist_x]
                - frame_features[f"{obj}_ncx"]
            )

            dy = (
                frame_features[wrist_y]
                - frame_features[f"{obj}_ncy"]
            )

            distance = np.sqrt(
                dx * dx + dy * dy
            )

            valid = (
                frame_features[wrist_x].notna()
                & frame_features[wrist_y].notna()
                & (
                    frame_features[
                        f"{obj}_presence"
                    ] == 1
                )
            )

            frame_features[distance_col] = np.where(
                valid,
                distance,
                0.0
            )

    print(
        "After hand-object distances:",
        frame_features.shape
    )

    # --------------------------------------------------------
    # 10. FINAL CLEANUP
    # --------------------------------------------------------

    stage = "Final cleanup and validation"

    print("\n[10] Final cleanup...")

    frame_features = (
        frame_features
        .replace([np.inf, -np.inf], np.nan)
        .fillna(0.0)
    )

    model_feature_cols = [
        c for c in frame_features.columns
        if c != "frame"
    ]

    print()
    print("=" * 70)
    print("FINAL DIAGNOSTIC RESULT")
    print("=" * 70)

    print("Rows:", len(frame_features))
    print("Total columns:", len(frame_features.columns))
    print("Model features:", len(model_feature_cols))
    print(
        "Missing values:",
        frame_features.isna().sum().sum()
    )

    print()
    print("✓ FULL PIPELINE PASSED")

except Exception as e:

    print()
    print("=" * 70)
    print("✗ PIPELINE FAILED")
    print("=" * 70)

    print("FAILED STAGE:")
    print(stage)

    print()
    print("EXCEPTION TYPE:")
    print(type(e).__name__)

    print()
    print("EXCEPTION MESSAGE:")
    print(repr(e))

    print()
    print("FULL TRACEBACK:")
    traceback.print_exc()

finally:

    try:
        cap.release()
    except:
        pass

    try:
        hand_landmarker.close()
    except:
        pass
```

```python
import os
import json
import traceback
import cv2
import numpy as np
import pandas as pd
from ultralytics import YOLO
import mediapipe as mp

# ============================================================
# BATCH 2 — CORRECTED STANDARDIZED PROCESSOR
# ============================================================

VIDEO_ROOTS = [
    "/content/drive/MyDrive/BAS_HAR_SIH",
    "/content/drive/MyDrive/Sih videos",
    "/content/drive/MyDrive/sih_project",
]

LIST_PATH = "/content/person2_output/project_video_list.json"
BASE_OUTPUT = "/content/person2_output/all_videos"

TRACKER_PATH = "/content/bytetrack_custom.yaml"
YOLO_PATH = "/content/best.pt"
MP_PATH = "/content/hand_landmarker.task"

BATCH_SIZE = 5

OBJECTS = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

DISTANCE_PAIRS = [
    ("cotton", "dropper"),
    ("cotton", "seed_jar"),
    ("cotton", "target_plate"),
    ("cotton", "water_dish"),
    ("dropper", "seed_jar"),
    ("dropper", "target_plate"),
    ("dropper", "water_dish"),
    ("seed_jar", "target_plate"),
    ("seed_jar", "water_dish"),
    ("target_plate", "water_dish"),
]

# ------------------------------------------------------------
# Find videos
# ------------------------------------------------------------

with open(LIST_PATH, "r") as f:
    project_videos = json.load(f)

project_videos = [
    os.path.basename(x) for x in project_videos
]

video_lookup = {}

for root in VIDEO_ROOTS:
    if not os.path.exists(root):
        continue

    for dirpath, _, filenames in os.walk(root):

        if "Aarambh 2k22" in dirpath:
            continue

        for filename in filenames:

            if filename in project_videos:
                video_lookup[filename] = os.path.join(
                    dirpath, filename
                )

# ------------------------------------------------------------
# Determine completed videos
# ------------------------------------------------------------

completed = []

for name in project_videos:

    stem = os.path.splitext(name)[0]

    output_path = os.path.join(
        BASE_OUTPUT,
        stem,
        f"{stem}_final_features_with_hand_object_distance.csv"
    )

    if os.path.exists(output_path):

        try:
            check = pd.read_csv(output_path, nrows=2)

            if (
                len(check.columns) == 207
                and "frame" in check.columns
            ):
                completed.append(name)

        except Exception:
            pass

unfinished = [
    x for x in project_videos
    if x not in completed
]

batch = unfinished[:BATCH_SIZE]

print("=" * 70)
print("BATCH 2 — CORRECTED STANDARDIZED PROCESSOR")
print("=" * 70)

print("\nCompleted:", len(completed))
print("Remaining:", len(unfinished))

print("\nThis batch:")
for i, name in enumerate(batch, 1):
    print(f"{i}. {name}")

if not batch:
    print("\n✓ NO UNFINISHED VIDEOS")
    raise SystemExit

# ------------------------------------------------------------
# Load YOLO once
# ------------------------------------------------------------

print("\nLoading YOLO...")
model = YOLO(YOLO_PATH)
print("✓ YOLO loaded")

batch_results = []

# ============================================================
# PROCESS EACH VIDEO
# ============================================================

for video_name in batch:

    print("\n" + "=" * 70)
    print(f"PROCESSING: {video_name}")
    print("=" * 70)

    try:

        video_path = video_lookup.get(video_name)

        if video_path is None:
            raise FileNotFoundError(
                f"Could not find video: {video_name}"
            )

        print("Path:", video_path)

        # ----------------------------------------------------
        # Video
        # ----------------------------------------------------

        cap = cv2.VideoCapture(video_path)

        if not cap.isOpened():
            raise RuntimeError(
                f"Could not open video: {video_path}"
            )

        total_frames = int(
            cap.get(cv2.CAP_PROP_FRAME_COUNT)
        )

        fps = cap.get(cv2.CAP_PROP_FPS)
        width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
        height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

        print(
            f"Frames: {total_frames} | "
            f"FPS: {fps:.3f} | "
            f"Size: {width}x{height}"
        )

        # ----------------------------------------------------
        # Fresh MediaPipe instance PER VIDEO
        # ----------------------------------------------------

        BaseOptions = mp.tasks.BaseOptions
        VisionRunningMode = mp.tasks.vision.RunningMode
        HandLandmarker = mp.tasks.vision.HandLandmarker
        HandLandmarkerOptions = mp.tasks.vision.HandLandmarkerOptions

        mp_options = HandLandmarkerOptions(
            base_options=BaseOptions(
                model_asset_path=MP_PATH
            ),
            running_mode=VisionRunningMode.IMAGE,
            num_hands=2,
            min_hand_detection_confidence=0.3,
            min_hand_presence_confidence=0.3,
            min_tracking_confidence=0.3
        )

        hand_landmarker = (
            HandLandmarker.create_from_options(mp_options)
        )

        # ----------------------------------------------------
        # Process frames
        # ----------------------------------------------------

        tracking_rows = []
        hand_rows = []

        frame_idx = 0

        while True:

            ret, frame = cap.read()

            if not ret:
                break

            # -------------------------
            # YOLO + ByteTrack
            # -------------------------

            results = model.track(
                frame,
                persist=True,
                tracker=TRACKER_PATH,
                conf=0.5,
                verbose=False
            )

            result = results[0]

            if (
                result.boxes is not None
                and len(result.boxes) > 0
            ):

                boxes = result.boxes

                xyxy = boxes.xyxy.cpu().numpy()
                confs = boxes.conf.cpu().numpy()
                classes = boxes.cls.cpu().numpy().astype(int)

                if boxes.id is not None:
                    track_ids = (
                        boxes.id.cpu().numpy().astype(int)
                    )
                else:
                    track_ids = [-1] * len(xyxy)

                for box, conf, cls_id, track_id in zip(
                    xyxy,
                    confs,
                    classes,
                    track_ids
                ):

                    cls_name = model.names[int(cls_id)]

                    x1, y1, x2, y2 = box

                    tracking_rows.append([
                        frame_idx,
                        int(cls_id),
                        cls_name,
                        int(track_id),
                        float(x1),
                        float(y1),
                        float(x2),
                        float(y2),
                        float(conf)
                    ])

            # -------------------------
            # MediaPipe
            # -------------------------

            rgb = cv2.cvtColor(
                frame,
                cv2.COLOR_BGR2RGB
            )

            mp_image = mp.Image(
                image_format=mp.ImageFormat.SRGB,
                data=rgb
            )

            hand_result = hand_landmarker.detect(
                mp_image
            )

            hand_data = {
                "frame": frame_idx
            }

            for side in ["left", "right"]:
                for landmark_id in range(21):

                    hand_data[
                        f"{side}_hand_{landmark_id}_x"
                    ] = 0.0

                    hand_data[
                        f"{side}_hand_{landmark_id}_y"
                    ] = 0.0

                    hand_data[
                        f"{side}_hand_{landmark_id}_z"
                    ] = 0.0

            if hand_result.hand_landmarks:

                for hand_landmarks, handedness in zip(
                    hand_result.hand_landmarks,
                    hand_result.handedness
                ):

                    label = (
                        handedness[0]
                        .category_name
                        .lower()
                    )

                    if "left" in label:
                        side = "left"
                    elif "right" in label:
                        side = "right"
                    else:
                        continue

                    for landmark_id, lm in enumerate(
                        hand_landmarks
                    ):

                        hand_data[
                            f"{side}_hand_{landmark_id}_x"
                        ] = float(lm.x)

                        hand_data[
                            f"{side}_hand_{landmark_id}_y"
                        ] = float(lm.y)

                        hand_data[
                            f"{side}_hand_{landmark_id}_z"
                        ] = float(lm.z)

            hand_rows.append(hand_data)

            frame_idx += 1

            if frame_idx % 50 == 0:
                print(
                    f"  Processed "
                    f"{frame_idx}/{total_frames}"
                )

        cap.release()
        hand_landmarker.close()

        print("✓ Frame processing completed")

        # ----------------------------------------------------
        # Tracking DataFrame
        # ----------------------------------------------------

        tracking_df = pd.DataFrame(
            tracking_rows,
            columns=[
                "frame",
                "class_id",
                "class_name",
                "track_id",
                "x1",
                "y1",
                "x2",
                "y2",
                "confidence"
            ]
        )

        hand_df = pd.DataFrame(hand_rows)

        print("Tracking:", tracking_df.shape)
        print("Hands   :", hand_df.shape)

        # ----------------------------------------------------
        # Object features
        # ----------------------------------------------------

        rows = []

        for frame in range(total_frames):

            row = {"frame": frame}

            frame_objects = tracking_df[
                tracking_df["frame"] == frame
            ]

            centers = {}

            for obj in OBJECTS:

                obj_rows = frame_objects[
                    frame_objects["class_name"] == obj
                ]

                if len(obj_rows) > 0:

                    best = obj_rows.loc[
                        obj_rows["confidence"].idxmax()
                    ]

                    presence = 1.0
                    confidence = float(
                        best["confidence"]
                    )

                    x1 = float(best["x1"])
                    y1 = float(best["y1"])
                    x2 = float(best["x2"])
                    y2 = float(best["y2"])

                else:

                    presence = 0.0
                    confidence = 0.0

                    x1 = y1 = x2 = y2 = 0.0

                cx = (x1 + x2) / 2.0
                cy = (y1 + y2) / 2.0

                row[f"{obj}_presence"] = presence
                row[f"{obj}_confidence"] = confidence

                # EXACTLY 5 raw geometry
                row[f"{obj}_x1"] = x1
                row[f"{obj}_y1"] = y1
                row[f"{obj}_x2"] = x2
                row[f"{obj}_y2"] = y2
                row[f"{obj}_center_x"] = cx

                # EXACTLY 5 normalized geometry
                row[f"{obj}_x1_norm"] = x1 / width
                row[f"{obj}_y1_norm"] = y1 / height
                row[f"{obj}_x2_norm"] = x2 / width
                row[f"{obj}_y2_norm"] = y2 / height
                row[f"{obj}_center_x_norm"] = cx / width

                centers[obj] = (
                    cx / width,
                    cy / height
                )

            rows.append(row)

        object_df = pd.DataFrame(rows)

        print("Object features:", object_df.shape)

        # ----------------------------------------------------
        # Object-object distances — exactly 10
        # ----------------------------------------------------

        for a, b in DISTANCE_PAIRS:

            ax = object_df[
                f"{a}_center_x_norm"
            ]

            ay = (
                object_df[f"{a}_y1"] +
                object_df[f"{a}_y2"]
            ) / (2.0 * height)

            bx = object_df[
                f"{b}_center_x_norm"
            ]

            by = (
                object_df[f"{b}_y1"] +
                object_df[f"{b}_y2"]
            ) / (2.0 * height)

            distance = np.sqrt(
                (ax - bx) ** 2 +
                (ay - by) ** 2
            )

            valid = (
                (object_df[f"{a}_presence"] > 0) &
                (object_df[f"{b}_presence"] > 0)
            )

            object_df[
                f"{a}_to_{b}_distance"
            ] = np.where(
                valid,
                distance,
                0.0
            )

        # ----------------------------------------------------
        # Hand features — exactly 126
        # ----------------------------------------------------

        hand_feature_cols = [
            c for c in hand_df.columns
            if c != "frame"
        ]

        if len(hand_feature_cols) != 126:
            raise ValueError(
                f"Expected 126 hand features, "
                f"got {len(hand_feature_cols)}"
            )

        hand_df[hand_feature_cols] = (
            hand_df[hand_feature_cols]
            .replace([np.inf, -np.inf], np.nan)
            .interpolate(limit_direction="both")
            .ffill()
            .bfill()
            .fillna(0.0)
        )

        # ----------------------------------------------------
        # Merge
        # ----------------------------------------------------

        final_df = object_df.merge(
            hand_df,
            on="frame",
            how="left"
        )

        # ----------------------------------------------------
        # Hand-object distances — exactly 10
        # ----------------------------------------------------

        for side in ["left", "right"]:

            wrist_x = final_df[
                f"{side}_hand_0_x"
            ]

            wrist_y = final_df[
                f"{side}_hand_0_y"
            ]

            for obj in OBJECTS:

                obj_x = final_df[
                    f"{obj}_center_x_norm"
                ]

                obj_y = (
                    final_df[f"{obj}_y1"] +
                    final_df[f"{obj}_y2"]
                ) / (2.0 * height)

                distance = np.sqrt(
                    (wrist_x - obj_x) ** 2 +
                    (wrist_y - obj_y) ** 2
                )

                hand_exists = (
                    (wrist_x != 0) |
                    (wrist_y != 0)
                )

                object_exists = (
                    final_df[
                        f"{obj}_presence"
                    ] > 0
                )

                valid = (
                    hand_exists &
                    object_exists
                )

                final_df[
                    f"{side}_hand_to_{obj}_distance"
                ] = np.where(
                    valid,
                    distance,
                    0.0
                )

        # ----------------------------------------------------
        # Cleanup
        # ----------------------------------------------------

        final_df = final_df.replace(
            [np.inf, -np.inf],
            np.nan
        )

        final_df = final_df.fillna(0.0)

        final_df = (
            final_df
            .sort_values("frame")
            .reset_index(drop=True)
        )

        # ----------------------------------------------------
        # Validation
        # ----------------------------------------------------

        model_features = [
            c for c in final_df.columns
            if c != "frame"
        ]

        hand_cols = [
            c for c in final_df.columns
            if "_hand_" in c
            and c.endswith(("_x", "_y", "_z"))
        ]

        object_distance_cols = [
            c for c in final_df.columns
            if "_to_" in c
            and "distance" in c
            and "hand" not in c
        ]

        hand_object_cols = [
            c for c in final_df.columns
            if "_hand_to_" in c
            and "distance" in c
        ]

        print("\nVALIDATION")
        print("Rows:", len(final_df))
        print("Total columns:", len(final_df.columns))
        print("Model features:", len(model_features))
        print("Hand landmarks:", len(hand_cols))
        print("Object-object distances:", len(object_distance_cols))
        print("Hand-object distances:", len(hand_object_cols))
        print(
            "Missing values:",
            int(final_df.isna().sum().sum())
        )

        assert len(final_df) == total_frames
        assert len(hand_cols) == 126
        assert len(object_distance_cols) == 10
        assert len(hand_object_cols) == 10
        assert len(model_features) == 206
        assert len(final_df.columns) == 207
        assert final_df.isna().sum().sum() == 0

        # ----------------------------------------------------
        # Save
        # ----------------------------------------------------

        stem = os.path.splitext(video_name)[0]

        output_dir = os.path.join(
            BASE_OUTPUT,
            stem
        )

        os.makedirs(output_dir, exist_ok=True)

        output_path = os.path.join(
            output_dir,
            f"{stem}_final_features_with_hand_object_distance.csv"
        )

        final_df.to_csv(
            output_path,
            index=False
        )

        print("\n✓ SUCCESS:", video_name)
        print("Saved:", output_path)

        batch_results.append({
            "video": video_name,
            "status": "SUCCESS",
            "frames": len(final_df),
            "features": len(model_features),
            "missing": int(
                final_df.isna().sum().sum()
            )
        })

    except Exception as e:

        print("\n✗ FAILED:", video_name)
        print("ERROR TYPE:", type(e).__name__)
        print("ERROR:", repr(e))

        print("\nFULL TRACEBACK:")
        traceback.print_exc()

        batch_results.append({
            "video": video_name,
            "status": "FAILED",
            "error_type": type(e).__name__,
            "error": repr(e)
        })

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

status_path = "/content/person2_output/batch2_status.json"

with open(status_path, "w") as f:
    json.dump(
        batch_results,
        f,
        indent=2
    )

print("\n" + "=" * 70)
print("BATCH 2 SUMMARY")
print("=" * 70)

for r in batch_results:

    if r["status"] == "SUCCESS":

        print(
            f"✓ {r['video']} | "
            f"frames={r['frames']} | "
            f"features={r['features']} | "
            f"missing={r['missing']}"
        )

    else:

        print(
            f"✗ {r['video']} | "
            f"{r['error_type']}: {r['error']}"
        )

print("\nStatus saved:", status_path)
```

```python
import os
import json
import traceback
import cv2
import numpy as np
import pandas as pd
from ultralytics import YOLO
import mediapipe as mp

# ============================================================
# BATCH 2 — CORRECTED STANDARDIZED PROCESSOR
# ============================================================

VIDEO_ROOTS = [
    "/content/drive/MyDrive/BAS_HAR_SIH",
    "/content/drive/MyDrive/Sih videos",
    "/content/drive/MyDrive/sih_project",
]

LIST_PATH = "/content/person2_output/project_video_list.json"
BASE_OUTPUT = "/content/person2_output/all_videos"

TRACKER_PATH = "/content/bytetrack_custom.yaml"
YOLO_PATH = "/content/best.pt"
MP_PATH = "/content/hand_landmarker.task"

BATCH_SIZE = 5

OBJECTS = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

DISTANCE_PAIRS = [
    ("cotton", "dropper"),
    ("cotton", "seed_jar"),
    ("cotton", "target_plate"),
    ("cotton", "water_dish"),
    ("dropper", "seed_jar"),
    ("dropper", "target_plate"),
    ("dropper", "water_dish"),
    ("seed_jar", "target_plate"),
    ("seed_jar", "water_dish"),
    ("target_plate", "water_dish"),
]

# ------------------------------------------------------------
# Find videos
# ------------------------------------------------------------

with open(LIST_PATH, "r") as f:
    project_videos = json.load(f)

project_videos = [
    os.path.basename(x) for x in project_videos
]

video_lookup = {}

for root in VIDEO_ROOTS:
    if not os.path.exists(root):
        continue

    for dirpath, _, filenames in os.walk(root):

        if "Aarambh 2k22" in dirpath:
            continue

        for filename in filenames:

            if filename in project_videos:
                video_lookup[filename] = os.path.join(
                    dirpath, filename
                )

# ------------------------------------------------------------
# Determine completed videos
# ------------------------------------------------------------

completed = []

for name in project_videos:

    stem = os.path.splitext(name)[0]

    output_path = os.path.join(
        BASE_OUTPUT,
        stem,
        f"{stem}_final_features_with_hand_object_distance.csv"
    )

    if os.path.exists(output_path):

        try:
            check = pd.read_csv(output_path, nrows=2)

            if (
                len(check.columns) == 207
                and "frame" in check.columns
            ):
                completed.append(name)

        except Exception:
            pass

unfinished = [
    x for x in project_videos
    if x not in completed
]

batch = unfinished[:BATCH_SIZE]

print("=" * 70)
print("BATCH 2 — CORRECTED STANDARDIZED PROCESSOR")
print("=" * 70)

print("\nCompleted:", len(completed))
print("Remaining:", len(unfinished))

print("\nThis batch:")
for i, name in enumerate(batch, 1):
    print(f"{i}. {name}")

if not batch:
    print("\n✓ NO UNFINISHED VIDEOS")
    raise SystemExit

# ------------------------------------------------------------
# Load YOLO once
# ------------------------------------------------------------

print("\nLoading YOLO...")
model = YOLO(YOLO_PATH)
print("✓ YOLO loaded")

batch_results = []

# ============================================================
# PROCESS EACH VIDEO
# ============================================================

for video_name in batch:

    print("\n" + "=" * 70)
    print(f"PROCESSING: {video_name}")
    print("=" * 70)

    try:

        video_path = video_lookup.get(video_name)

        if video_path is None:
            raise FileNotFoundError(
                f"Could not find video: {video_name}"
            )

        print("Path:", video_path)

        # ----------------------------------------------------
        # Video
        # ----------------------------------------------------

        cap = cv2.VideoCapture(video_path)

        if not cap.isOpened():
            raise RuntimeError(
                f"Could not open video: {video_path}"
            )

        total_frames = int(
            cap.get(cv2.CAP_PROP_FRAME_COUNT)
        )

        fps = cap.get(cv2.CAP_PROP_FPS)
        width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
        height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

        print(
            f"Frames: {total_frames} | "
            f"FPS: {fps:.3f} | "
            f"Size: {width}x{height}"
        )

        # ----------------------------------------------------
        # Fresh MediaPipe instance PER VIDEO
        # ----------------------------------------------------

        BaseOptions = mp.tasks.BaseOptions
        VisionRunningMode = mp.tasks.vision.RunningMode
        HandLandmarker = mp.tasks.vision.HandLandmarker
        HandLandmarkerOptions = mp.tasks.vision.HandLandmarkerOptions

        mp_options = HandLandmarkerOptions(
            base_options=BaseOptions(
                model_asset_path=MP_PATH
            ),
            running_mode=VisionRunningMode.IMAGE,
            num_hands=2,
            min_hand_detection_confidence=0.3,
            min_hand_presence_confidence=0.3,
            min_tracking_confidence=0.3
        )

        hand_landmarker = (
            HandLandmarker.create_from_options(mp_options)
        )

        # ----------------------------------------------------
        # Process frames
        # ----------------------------------------------------

        tracking_rows = []
        hand_rows = []

        frame_idx = 0

        while True:

            ret, frame = cap.read()

            if not ret:
                break

            # -------------------------
            # YOLO + ByteTrack
            # -------------------------

            results = model.track(
                frame,
                persist=True,
                tracker=TRACKER_PATH,
                conf=0.5,
                verbose=False
            )

            result = results[0]

            if (
                result.boxes is not None
                and len(result.boxes) > 0
            ):

                boxes = result.boxes

                xyxy = boxes.xyxy.cpu().numpy()
                confs = boxes.conf.cpu().numpy()
                classes = boxes.cls.cpu().numpy().astype(int)

                if boxes.id is not None:
                    track_ids = (
                        boxes.id.cpu().numpy().astype(int)
                    )
                else:
                    track_ids = [-1] * len(xyxy)

                for box, conf, cls_id, track_id in zip(
                    xyxy,
                    confs,
                    classes,
                    track_ids
                ):

                    cls_name = model.names[int(cls_id)]

                    x1, y1, x2, y2 = box

                    tracking_rows.append([
                        frame_idx,
                        int(cls_id),
                        cls_name,
                        int(track_id),
                        float(x1),
                        float(y1),
                        float(x2),
                        float(y2),
                        float(conf)
                    ])

            # -------------------------
            # MediaPipe
            # -------------------------

            rgb = cv2.cvtColor(
                frame,
                cv2.COLOR_BGR2RGB
            )

            mp_image = mp.Image(
                image_format=mp.ImageFormat.SRGB,
                data=rgb
            )

            hand_result = hand_landmarker.detect(
                mp_image
            )

            hand_data = {
                "frame": frame_idx
            }

            for side in ["left", "right"]:
                for landmark_id in range(21):

                    hand_data[
                        f"{side}_hand_{landmark_id}_x"
                    ] = 0.0

                    hand_data[
                        f"{side}_hand_{landmark_id}_y"
                    ] = 0.0

                    hand_data[
                        f"{side}_hand_{landmark_id}_z"
                    ] = 0.0

            if hand_result.hand_landmarks:

                for hand_landmarks, handedness in zip(
                    hand_result.hand_landmarks,
                    hand_result.handedness
                ):

                    label = (
                        handedness[0]
                        .category_name
                        .lower()
                    )

                    if "left" in label:
                        side = "left"
                    elif "right" in label:
                        side = "right"
                    else:
                        continue

                    for landmark_id, lm in enumerate(
                        hand_landmarks
                    ):

                        hand_data[
                            f"{side}_hand_{landmark_id}_x"
                        ] = float(lm.x)

                        hand_data[
                            f"{side}_hand_{landmark_id}_y"
                        ] = float(lm.y)

                        hand_data[
                            f"{side}_hand_{landmark_id}_z"
                        ] = float(lm.z)

            hand_rows.append(hand_data)

            frame_idx += 1

            if frame_idx % 50 == 0:
                print(
                    f"  Processed "
                    f"{frame_idx}/{total_frames}"
                )

        cap.release()
        hand_landmarker.close()

        print("✓ Frame processing completed")

        # ----------------------------------------------------
        # Tracking DataFrame
        # ----------------------------------------------------

        tracking_df = pd.DataFrame(
            tracking_rows,
            columns=[
                "frame",
                "class_id",
                "class_name",
                "track_id",
                "x1",
                "y1",
                "x2",
                "y2",
                "confidence"
            ]
        )

        hand_df = pd.DataFrame(hand_rows)

        print("Tracking:", tracking_df.shape)
        print("Hands   :", hand_df.shape)

        # ----------------------------------------------------
        # Object features
        # ----------------------------------------------------

        rows = []

        for frame in range(total_frames):

            row = {"frame": frame}

            frame_objects = tracking_df[
                tracking_df["frame"] == frame
            ]

            centers = {}

            for obj in OBJECTS:

                obj_rows = frame_objects[
                    frame_objects["class_name"] == obj
                ]

                if len(obj_rows) > 0:

                    best = obj_rows.loc[
                        obj_rows["confidence"].idxmax()
                    ]

                    presence = 1.0
                    confidence = float(
                        best["confidence"]
                    )

                    x1 = float(best["x1"])
                    y1 = float(best["y1"])
                    x2 = float(best["x2"])
                    y2 = float(best["y2"])

                else:

                    presence = 0.0
                    confidence = 0.0

                    x1 = y1 = x2 = y2 = 0.0

                cx = (x1 + x2) / 2.0
                cy = (y1 + y2) / 2.0

                row[f"{obj}_presence"] = presence
                row[f"{obj}_confidence"] = confidence

                # EXACTLY 5 raw geometry
                row[f"{obj}_x1"] = x1
                row[f"{obj}_y1"] = y1
                row[f"{obj}_x2"] = x2
                row[f"{obj}_y2"] = y2
                row[f"{obj}_center_x"] = cx

                # EXACTLY 5 normalized geometry
                row[f"{obj}_x1_norm"] = x1 / width
                row[f"{obj}_y1_norm"] = y1 / height
                row[f"{obj}_x2_norm"] = x2 / width
                row[f"{obj}_y2_norm"] = y2 / height
                row[f"{obj}_center_x_norm"] = cx / width

                centers[obj] = (
                    cx / width,
                    cy / height
                )

            rows.append(row)

        object_df = pd.DataFrame(rows)

        print("Object features:", object_df.shape)

        # ----------------------------------------------------
        # Object-object distances — exactly 10
        # ----------------------------------------------------

        for a, b in DISTANCE_PAIRS:

            ax = object_df[
                f"{a}_center_x_norm"
            ]

            ay = (
                object_df[f"{a}_y1"] +
                object_df[f"{a}_y2"]
            ) / (2.0 * height)

            bx = object_df[
                f"{b}_center_x_norm"
            ]

            by = (
                object_df[f"{b}_y1"] +
                object_df[f"{b}_y2"]
            ) / (2.0 * height)

            distance = np.sqrt(
                (ax - bx) ** 2 +
                (ay - by) ** 2
            )

            valid = (
                (object_df[f"{a}_presence"] > 0) &
                (object_df[f"{b}_presence"] > 0)
            )

            object_df[
                f"{a}_to_{b}_distance"
            ] = np.where(
                valid,
                distance,
                0.0
            )

        # ----------------------------------------------------
        # Hand features — exactly 126
        # ----------------------------------------------------

        hand_feature_cols = [
            c for c in hand_df.columns
            if c != "frame"
        ]

        if len(hand_feature_cols) != 126:
            raise ValueError(
                f"Expected 126 hand features, "
                f"got {len(hand_feature_cols)}"
            )

        hand_df[hand_feature_cols] = (
            hand_df[hand_feature_cols]
            .replace([np.inf, -np.inf], np.nan)
            .interpolate(limit_direction="both")
            .ffill()
            .bfill()
            .fillna(0.0)
        )

        # ----------------------------------------------------
        # Merge
        # ----------------------------------------------------

        final_df = object_df.merge(
            hand_df,
            on="frame",
            how="left"
        )

        # ----------------------------------------------------
        # Hand-object distances — exactly 10
        # ----------------------------------------------------

        for side in ["left", "right"]:

            wrist_x = final_df[
                f"{side}_hand_0_x"
            ]

            wrist_y = final_df[
                f"{side}_hand_0_y"
            ]

            for obj in OBJECTS:

                obj_x = final_df[
                    f"{obj}_center_x_norm"
                ]

                obj_y = (
                    final_df[f"{obj}_y1"] +
                    final_df[f"{obj}_y2"]
                ) / (2.0 * height)

                distance = np.sqrt(
                    (wrist_x - obj_x) ** 2 +
                    (wrist_y - obj_y) ** 2
                )

                hand_exists = (
                    (wrist_x != 0) |
                    (wrist_y != 0)
                )

                object_exists = (
                    final_df[
                        f"{obj}_presence"
                    ] > 0
                )

                valid = (
                    hand_exists &
                    object_exists
                )

                final_df[
                    f"{side}_hand_to_{obj}_distance"
                ] = np.where(
                    valid,
                    distance,
                    0.0
                )

        # ----------------------------------------------------
        # Cleanup
        # ----------------------------------------------------

        final_df = final_df.replace(
            [np.inf, -np.inf],
            np.nan
        )

        final_df = final_df.fillna(0.0)

        final_df = (
            final_df
            .sort_values("frame")
            .reset_index(drop=True)
        )

        # ----------------------------------------------------
        # Validation
        # ----------------------------------------------------

        model_features = [
            c for c in final_df.columns
            if c != "frame"
        ]

        hand_cols = [
            c for c in final_df.columns
            if "_hand_" in c
            and c.endswith(("_x", "_y", "_z"))
        ]

        object_distance_cols = [
            c for c in final_df.columns
            if "_to_" in c
            and "distance" in c
            and "hand" not in c
        ]

        hand_object_cols = [
            c for c in final_df.columns
            if "_hand_to_" in c
            and "distance" in c
        ]

        print("\nVALIDATION")
        print("Rows:", len(final_df))
        print("Total columns:", len(final_df.columns))
        print("Model features:", len(model_features))
        print("Hand landmarks:", len(hand_cols))
        print("Object-object distances:", len(object_distance_cols))
        print("Hand-object distances:", len(hand_object_cols))
        print(
            "Missing values:",
            int(final_df.isna().sum().sum())
        )

        assert len(final_df) == total_frames
        assert len(hand_cols) == 126
        assert len(object_distance_cols) == 10
        assert len(hand_object_cols) == 10
        assert len(model_features) == 206
        assert len(final_df.columns) == 207
        assert final_df.isna().sum().sum() == 0

        # ----------------------------------------------------
        # Save
        # ----------------------------------------------------

        stem = os.path.splitext(video_name)[0]

        output_dir = os.path.join(
            BASE_OUTPUT,
            stem
        )

        os.makedirs(output_dir, exist_ok=True)

        output_path = os.path.join(
            output_dir,
            f"{stem}_final_features_with_hand_object_distance.csv"
        )

        final_df.to_csv(
            output_path,
            index=False
        )

        print("\n✓ SUCCESS:", video_name)
        print("Saved:", output_path)

        batch_results.append({
            "video": video_name,
            "status": "SUCCESS",
            "frames": len(final_df),
            "features": len(model_features),
            "missing": int(
                final_df.isna().sum().sum()
            )
        })

    except Exception as e:

        print("\n✗ FAILED:", video_name)
        print("ERROR TYPE:", type(e).__name__)
        print("ERROR:", repr(e))

        print("\nFULL TRACEBACK:")
        traceback.print_exc()

        batch_results.append({
            "video": video_name,
            "status": "FAILED",
            "error_type": type(e).__name__,
            "error": repr(e)
        })

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

status_path = "/content/person2_output/batch2_status.json"

with open(status_path, "w") as f:
    json.dump(
        batch_results,
        f,
        indent=2
    )

print("\n" + "=" * 70)
print("BATCH 2 SUMMARY")
print("=" * 70)

for r in batch_results:

    if r["status"] == "SUCCESS":

        print(
            f"✓ {r['video']} | "
            f"frames={r['frames']} | "
            f"features={r['features']} | "
            f"missing={r['missing']}"
        )

    else:

        print(
            f"✗ {r['video']} | "
            f"{r['error_type']}: {r['error']}"
        )

print("\nStatus saved:", status_path)
```

```python
import os
import json
import traceback
import cv2
import numpy as np
import pandas as pd
from ultralytics import YOLO
import mediapipe as mp

# ============================================================
# BATCH 2 — CORRECTED STANDARDIZED PROCESSOR
# ============================================================

VIDEO_ROOTS = [
    "/content/drive/MyDrive/BAS_HAR_SIH",
    "/content/drive/MyDrive/Sih videos",
    "/content/drive/MyDrive/sih_project",
]

LIST_PATH = "/content/person2_output/project_video_list.json"
BASE_OUTPUT = "/content/person2_output/all_videos"

TRACKER_PATH = "/content/bytetrack_custom.yaml"
YOLO_PATH = "/content/best.pt"
MP_PATH = "/content/hand_landmarker.task"

BATCH_SIZE = 5

OBJECTS = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

DISTANCE_PAIRS = [
    ("cotton", "dropper"),
    ("cotton", "seed_jar"),
    ("cotton", "target_plate"),
    ("cotton", "water_dish"),
    ("dropper", "seed_jar"),
    ("dropper", "target_plate"),
    ("dropper", "water_dish"),
    ("seed_jar", "target_plate"),
    ("seed_jar", "water_dish"),
    ("target_plate", "water_dish"),
]

# ------------------------------------------------------------
# Find videos
# ------------------------------------------------------------

with open(LIST_PATH, "r") as f:
    project_videos = json.load(f)

project_videos = [
    os.path.basename(x) for x in project_videos
]

video_lookup = {}

for root in VIDEO_ROOTS:
    if not os.path.exists(root):
        continue

    for dirpath, _, filenames in os.walk(root):

        if "Aarambh 2k22" in dirpath:
            continue

        for filename in filenames:

            if filename in project_videos:
                video_lookup[filename] = os.path.join(
                    dirpath, filename
                )

# ------------------------------------------------------------
# Determine completed videos
# ------------------------------------------------------------

completed = []

for name in project_videos:

    stem = os.path.splitext(name)[0]

    output_path = os.path.join(
        BASE_OUTPUT,
        stem,
        f"{stem}_final_features_with_hand_object_distance.csv"
    )

    if os.path.exists(output_path):

        try:
            check = pd.read_csv(output_path, nrows=2)

            if (
                len(check.columns) == 207
                and "frame" in check.columns
            ):
                completed.append(name)

        except Exception:
            pass

unfinished = [
    x for x in project_videos
    if x not in completed
]

batch = unfinished[:BATCH_SIZE]

print("=" * 70)
print("BATCH 2 — CORRECTED STANDARDIZED PROCESSOR")
print("=" * 70)

print("\nCompleted:", len(completed))
print("Remaining:", len(unfinished))

print("\nThis batch:")
for i, name in enumerate(batch, 1):
    print(f"{i}. {name}")

if not batch:
    print("\n✓ NO UNFINISHED VIDEOS")
    raise SystemExit

# ------------------------------------------------------------
# Load YOLO once
# ------------------------------------------------------------

print("\nLoading YOLO...")
model = YOLO(YOLO_PATH)
print("✓ YOLO loaded")

batch_results = []

# ============================================================
# PROCESS EACH VIDEO
# ============================================================

for video_name in batch:

    print("\n" + "=" * 70)
    print(f"PROCESSING: {video_name}")
    print("=" * 70)

    try:

        video_path = video_lookup.get(video_name)

        if video_path is None:
            raise FileNotFoundError(
                f"Could not find video: {video_name}"
            )

        print("Path:", video_path)

        # ----------------------------------------------------
        # Video
        # ----------------------------------------------------

        cap = cv2.VideoCapture(video_path)

        if not cap.isOpened():
            raise RuntimeError(
                f"Could not open video: {video_path}"
            )

        total_frames = int(
            cap.get(cv2.CAP_PROP_FRAME_COUNT)
        )

        fps = cap.get(cv2.CAP_PROP_FPS)
        width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
        height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

        print(
            f"Frames: {total_frames} | "
            f"FPS: {fps:.3f} | "
            f"Size: {width}x{height}"
        )

        # ----------------------------------------------------
        # Fresh MediaPipe instance PER VIDEO
        # ----------------------------------------------------

        BaseOptions = mp.tasks.BaseOptions
        VisionRunningMode = mp.tasks.vision.RunningMode
        HandLandmarker = mp.tasks.vision.HandLandmarker
        HandLandmarkerOptions = mp.tasks.vision.HandLandmarkerOptions

        mp_options = HandLandmarkerOptions(
            base_options=BaseOptions(
                model_asset_path=MP_PATH
            ),
            running_mode=VisionRunningMode.IMAGE,
            num_hands=2,
            min_hand_detection_confidence=0.3,
            min_hand_presence_confidence=0.3,
            min_tracking_confidence=0.3
        )

        hand_landmarker = (
            HandLandmarker.create_from_options(mp_options)
        )

        # ----------------------------------------------------
        # Process frames
        # ----------------------------------------------------

        tracking_rows = []
        hand_rows = []

        frame_idx = 0

        while True:

            ret, frame = cap.read()

            if not ret:
                break

            # -------------------------
            # YOLO + ByteTrack
            # -------------------------

            results = model.track(
                frame,
                persist=True,
                tracker=TRACKER_PATH,
                conf=0.5,
                verbose=False
            )

            result = results[0]

            if (
                result.boxes is not None
                and len(result.boxes) > 0
            ):

                boxes = result.boxes

                xyxy = boxes.xyxy.cpu().numpy()
                confs = boxes.conf.cpu().numpy()
                classes = boxes.cls.cpu().numpy().astype(int)

                if boxes.id is not None:
                    track_ids = (
                        boxes.id.cpu().numpy().astype(int)
                    )
                else:
                    track_ids = [-1] * len(xyxy)

                for box, conf, cls_id, track_id in zip(
                    xyxy,
                    confs,
                    classes,
                    track_ids
                ):

                    cls_name = model.names[int(cls_id)]

                    x1, y1, x2, y2 = box

                    tracking_rows.append([
                        frame_idx,
                        int(cls_id),
                        cls_name,
                        int(track_id),
                        float(x1),
                        float(y1),
                        float(x2),
                        float(y2),
                        float(conf)
                    ])

            # -------------------------
            # MediaPipe
            # -------------------------

            rgb = cv2.cvtColor(
                frame,
                cv2.COLOR_BGR2RGB
            )

            mp_image = mp.Image(
                image_format=mp.ImageFormat.SRGB,
                data=rgb
            )

            hand_result = hand_landmarker.detect(
                mp_image
            )

            hand_data = {
                "frame": frame_idx
            }

            for side in ["left", "right"]:
                for landmark_id in range(21):

                    hand_data[
                        f"{side}_hand_{landmark_id}_x"
                    ] = 0.0

                    hand_data[
                        f"{side}_hand_{landmark_id}_y"
                    ] = 0.0

                    hand_data[
                        f"{side}_hand_{landmark_id}_z"
                    ] = 0.0

            if hand_result.hand_landmarks:

                for hand_landmarks, handedness in zip(
                    hand_result.hand_landmarks,
                    hand_result.handedness
                ):

                    label = (
                        handedness[0]
                        .category_name
                        .lower()
                    )

                    if "left" in label:
                        side = "left"
                    elif "right" in label:
                        side = "right"
                    else:
                        continue

                    for landmark_id, lm in enumerate(
                        hand_landmarks
                    ):

                        hand_data[
                            f"{side}_hand_{landmark_id}_x"
                        ] = float(lm.x)

                        hand_data[
                            f"{side}_hand_{landmark_id}_y"
                        ] = float(lm.y)

                        hand_data[
                            f"{side}_hand_{landmark_id}_z"
                        ] = float(lm.z)

            hand_rows.append(hand_data)

            frame_idx += 1

            if frame_idx % 50 == 0:
                print(
                    f"  Processed "
                    f"{frame_idx}/{total_frames}"
                )

        cap.release()
        hand_landmarker.close()

        print("✓ Frame processing completed")

        # ----------------------------------------------------
        # Tracking DataFrame
        # ----------------------------------------------------

        tracking_df = pd.DataFrame(
            tracking_rows,
            columns=[
                "frame",
                "class_id",
                "class_name",
                "track_id",
                "x1",
                "y1",
                "x2",
                "y2",
                "confidence"
            ]
        )

        hand_df = pd.DataFrame(hand_rows)

        print("Tracking:", tracking_df.shape)
        print("Hands   :", hand_df.shape)

        # ----------------------------------------------------
        # Object features
        # ----------------------------------------------------

        rows = []

        for frame in range(total_frames):

            row = {"frame": frame}

            frame_objects = tracking_df[
                tracking_df["frame"] == frame
            ]

            centers = {}

            for obj in OBJECTS:

                obj_rows = frame_objects[
                    frame_objects["class_name"] == obj
                ]

                if len(obj_rows) > 0:

                    best = obj_rows.loc[
                        obj_rows["confidence"].idxmax()
                    ]

                    presence = 1.0
                    confidence = float(
                        best["confidence"]
                    )

                    x1 = float(best["x1"])
                    y1 = float(best["y1"])
                    x2 = float(best["x2"])
                    y2 = float(best["y2"])

                else:

                    presence = 0.0
                    confidence = 0.0

                    x1 = y1 = x2 = y2 = 0.0

                cx = (x1 + x2) / 2.0
                cy = (y1 + y2) / 2.0

                row[f"{obj}_presence"] = presence
                row[f"{obj}_confidence"] = confidence

                # EXACTLY 5 raw geometry
                row[f"{obj}_x1"] = x1
                row[f"{obj}_y1"] = y1
                row[f"{obj}_x2"] = x2
                row[f"{obj}_y2"] = y2
                row[f"{obj}_center_x"] = cx

                # EXACTLY 5 normalized geometry
                row[f"{obj}_x1_norm"] = x1 / width
                row[f"{obj}_y1_norm"] = y1 / height
                row[f"{obj}_x2_norm"] = x2 / width
                row[f"{obj}_y2_norm"] = y2 / height
                row[f"{obj}_center_x_norm"] = cx / width

                centers[obj] = (
                    cx / width,
                    cy / height
                )

            rows.append(row)

        object_df = pd.DataFrame(rows)

        print("Object features:", object_df.shape)

        # ----------------------------------------------------
        # Object-object distances — exactly 10
        # ----------------------------------------------------

        for a, b in DISTANCE_PAIRS:

            ax = object_df[
                f"{a}_center_x_norm"
            ]

            ay = (
                object_df[f"{a}_y1"] +
                object_df[f"{a}_y2"]
            ) / (2.0 * height)

            bx = object_df[
                f"{b}_center_x_norm"
            ]

            by = (
                object_df[f"{b}_y1"] +
                object_df[f"{b}_y2"]
            ) / (2.0 * height)

            distance = np.sqrt(
                (ax - bx) ** 2 +
                (ay - by) ** 2
            )

            valid = (
                (object_df[f"{a}_presence"] > 0) &
                (object_df[f"{b}_presence"] > 0)
            )

            object_df[
                f"{a}_to_{b}_distance"
            ] = np.where(
                valid,
                distance,
                0.0
            )

        # ----------------------------------------------------
        # Hand features — exactly 126
        # ----------------------------------------------------

        hand_feature_cols = [
            c for c in hand_df.columns
            if c != "frame"
        ]

        if len(hand_feature_cols) != 126:
            raise ValueError(
                f"Expected 126 hand features, "
                f"got {len(hand_feature_cols)}"
            )

        hand_df[hand_feature_cols] = (
            hand_df[hand_feature_cols]
            .replace([np.inf, -np.inf], np.nan)
            .interpolate(limit_direction="both")
            .ffill()
            .bfill()
            .fillna(0.0)
        )

        # ----------------------------------------------------
        # Merge
        # ----------------------------------------------------

        final_df = object_df.merge(
            hand_df,
            on="frame",
            how="left"
        )

        # ----------------------------------------------------
        # Hand-object distances — exactly 10
        # ----------------------------------------------------

        for side in ["left", "right"]:

            wrist_x = final_df[
                f"{side}_hand_0_x"
            ]

            wrist_y = final_df[
                f"{side}_hand_0_y"
            ]

            for obj in OBJECTS:

                obj_x = final_df[
                    f"{obj}_center_x_norm"
                ]

                obj_y = (
                    final_df[f"{obj}_y1"] +
                    final_df[f"{obj}_y2"]
                ) / (2.0 * height)

                distance = np.sqrt(
                    (wrist_x - obj_x) ** 2 +
                    (wrist_y - obj_y) ** 2
                )

                hand_exists = (
                    (wrist_x != 0) |
                    (wrist_y != 0)
                )

                object_exists = (
                    final_df[
                        f"{obj}_presence"
                    ] > 0
                )

                valid = (
                    hand_exists &
                    object_exists
                )

                final_df[
                    f"{side}_hand_to_{obj}_distance"
                ] = np.where(
                    valid,
                    distance,
                    0.0
                )

        # ----------------------------------------------------
        # Cleanup
        # ----------------------------------------------------

        final_df = final_df.replace(
            [np.inf, -np.inf],
            np.nan
        )

        final_df = final_df.fillna(0.0)

        final_df = (
            final_df
            .sort_values("frame")
            .reset_index(drop=True)
        )

        # ----------------------------------------------------
        # Validation
        # ----------------------------------------------------

        model_features = [
            c for c in final_df.columns
            if c != "frame"
        ]

        hand_cols = [
            c for c in final_df.columns
            if "_hand_" in c
            and c.endswith(("_x", "_y", "_z"))
        ]

        object_distance_cols = [
            c for c in final_df.columns
            if "_to_" in c
            and "distance" in c
            and "hand" not in c
        ]

        hand_object_cols = [
            c for c in final_df.columns
            if "_hand_to_" in c
            and "distance" in c
        ]

        print("\nVALIDATION")
        print("Rows:", len(final_df))
        print("Total columns:", len(final_df.columns))
        print("Model features:", len(model_features))
        print("Hand landmarks:", len(hand_cols))
        print("Object-object distances:", len(object_distance_cols))
        print("Hand-object distances:", len(hand_object_cols))
        print(
            "Missing values:",
            int(final_df.isna().sum().sum())
        )

        assert len(final_df) == total_frames
        assert len(hand_cols) == 126
        assert len(object_distance_cols) == 10
        assert len(hand_object_cols) == 10
        assert len(model_features) == 206
        assert len(final_df.columns) == 207
        assert final_df.isna().sum().sum() == 0

        # ----------------------------------------------------
        # Save
        # ----------------------------------------------------

        stem = os.path.splitext(video_name)[0]

        output_dir = os.path.join(
            BASE_OUTPUT,
            stem
        )

        os.makedirs(output_dir, exist_ok=True)

        output_path = os.path.join(
            output_dir,
            f"{stem}_final_features_with_hand_object_distance.csv"
        )

        final_df.to_csv(
            output_path,
            index=False
        )

        print("\n✓ SUCCESS:", video_name)
        print("Saved:", output_path)

        batch_results.append({
            "video": video_name,
            "status": "SUCCESS",
            "frames": len(final_df),
            "features": len(model_features),
            "missing": int(
                final_df.isna().sum().sum()
            )
        })

    except Exception as e:

        print("\n✗ FAILED:", video_name)
        print("ERROR TYPE:", type(e).__name__)
        print("ERROR:", repr(e))

        print("\nFULL TRACEBACK:")
        traceback.print_exc()

        batch_results.append({
            "video": video_name,
            "status": "FAILED",
            "error_type": type(e).__name__,
            "error": repr(e)
        })

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

status_path = "/content/person2_output/batch2_status.json"

with open(status_path, "w") as f:
    json.dump(
        batch_results,
        f,
        indent=2
    )

print("\n" + "=" * 70)
print("BATCH 2 SUMMARY")
print("=" * 70)

for r in batch_results:

    if r["status"] == "SUCCESS":

        print(
            f"✓ {r['video']} | "
            f"frames={r['frames']} | "
            f"features={r['features']} | "
            f"missing={r['missing']}"
        )

    else:

        print(
            f"✗ {r['video']} | "
            f"{r['error_type']}: {r['error']}"
        )

print("\nStatus saved:", status_path)
```

```python
import os
import json
import traceback
import cv2
import numpy as np
import pandas as pd
from ultralytics import YOLO
import mediapipe as mp

# ============================================================
# BATCH 2 — CORRECTED STANDARDIZED PROCESSOR
# ============================================================

VIDEO_ROOTS = [
    "/content/drive/MyDrive/BAS_HAR_SIH",
    "/content/drive/MyDrive/Sih videos",
    "/content/drive/MyDrive/sih_project",
]

LIST_PATH = "/content/person2_output/project_video_list.json"
BASE_OUTPUT = "/content/person2_output/all_videos"

TRACKER_PATH = "/content/bytetrack_custom.yaml"
YOLO_PATH = "/content/best.pt"
MP_PATH = "/content/hand_landmarker.task"

BATCH_SIZE = 5

OBJECTS = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

DISTANCE_PAIRS = [
    ("cotton", "dropper"),
    ("cotton", "seed_jar"),
    ("cotton", "target_plate"),
    ("cotton", "water_dish"),
    ("dropper", "seed_jar"),
    ("dropper", "target_plate"),
    ("dropper", "water_dish"),
    ("seed_jar", "target_plate"),
    ("seed_jar", "water_dish"),
    ("target_plate", "water_dish"),
]

# ------------------------------------------------------------
# Find videos
# ------------------------------------------------------------

with open(LIST_PATH, "r") as f:
    project_videos = json.load(f)

project_videos = [
    os.path.basename(x) for x in project_videos
]

video_lookup = {}

for root in VIDEO_ROOTS:
    if not os.path.exists(root):
        continue

    for dirpath, _, filenames in os.walk(root):

        if "Aarambh 2k22" in dirpath:
            continue

        for filename in filenames:

            if filename in project_videos:
                video_lookup[filename] = os.path.join(
                    dirpath, filename
                )

# ------------------------------------------------------------
# Determine completed videos
# ------------------------------------------------------------

completed = []

for name in project_videos:

    stem = os.path.splitext(name)[0]

    output_path = os.path.join(
        BASE_OUTPUT,
        stem,
        f"{stem}_final_features_with_hand_object_distance.csv"
    )

    if os.path.exists(output_path):

        try:
            check = pd.read_csv(output_path, nrows=2)

            if (
                len(check.columns) == 207
                and "frame" in check.columns
            ):
                completed.append(name)

        except Exception:
            pass

unfinished = [
    x for x in project_videos
    if x not in completed
]

batch = unfinished[:BATCH_SIZE]

print("=" * 70)
print("BATCH 2 — CORRECTED STANDARDIZED PROCESSOR")
print("=" * 70)

print("\nCompleted:", len(completed))
print("Remaining:", len(unfinished))

print("\nThis batch:")
for i, name in enumerate(batch, 1):
    print(f"{i}. {name}")

if not batch:
    print("\n✓ NO UNFINISHED VIDEOS")
    raise SystemExit

# ------------------------------------------------------------
# Load YOLO once
# ------------------------------------------------------------

print("\nLoading YOLO...")
model = YOLO(YOLO_PATH)
print("✓ YOLO loaded")

batch_results = []

# ============================================================
# PROCESS EACH VIDEO
# ============================================================

for video_name in batch:

    print("\n" + "=" * 70)
    print(f"PROCESSING: {video_name}")
    print("=" * 70)

    try:

        video_path = video_lookup.get(video_name)

        if video_path is None:
            raise FileNotFoundError(
                f"Could not find video: {video_name}"
            )

        print("Path:", video_path)

        # ----------------------------------------------------
        # Video
        # ----------------------------------------------------

        cap = cv2.VideoCapture(video_path)

        if not cap.isOpened():
            raise RuntimeError(
                f"Could not open video: {video_path}"
            )

        total_frames = int(
            cap.get(cv2.CAP_PROP_FRAME_COUNT)
        )

        fps = cap.get(cv2.CAP_PROP_FPS)
        width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
        height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

        print(
            f"Frames: {total_frames} | "
            f"FPS: {fps:.3f} | "
            f"Size: {width}x{height}"
        )

        # ----------------------------------------------------
        # Fresh MediaPipe instance PER VIDEO
        # ----------------------------------------------------

        BaseOptions = mp.tasks.BaseOptions
        VisionRunningMode = mp.tasks.vision.RunningMode
        HandLandmarker = mp.tasks.vision.HandLandmarker
        HandLandmarkerOptions = mp.tasks.vision.HandLandmarkerOptions

        mp_options = HandLandmarkerOptions(
            base_options=BaseOptions(
                model_asset_path=MP_PATH
            ),
            running_mode=VisionRunningMode.IMAGE,
            num_hands=2,
            min_hand_detection_confidence=0.3,
            min_hand_presence_confidence=0.3,
            min_tracking_confidence=0.3
        )

        hand_landmarker = (
            HandLandmarker.create_from_options(mp_options)
        )

        # ----------------------------------------------------
        # Process frames
        # ----------------------------------------------------

        tracking_rows = []
        hand_rows = []

        frame_idx = 0

        while True:

            ret, frame = cap.read()

            if not ret:
                break

            # -------------------------
            # YOLO + ByteTrack
            # -------------------------

            results = model.track(
                frame,
                persist=True,
                tracker=TRACKER_PATH,
                conf=0.5,
                verbose=False
            )

            result = results[0]

            if (
                result.boxes is not None
                and len(result.boxes) > 0
            ):

                boxes = result.boxes

                xyxy = boxes.xyxy.cpu().numpy()
                confs = boxes.conf.cpu().numpy()
                classes = boxes.cls.cpu().numpy().astype(int)

                if boxes.id is not None:
                    track_ids = (
                        boxes.id.cpu().numpy().astype(int)
                    )
                else:
                    track_ids = [-1] * len(xyxy)

                for box, conf, cls_id, track_id in zip(
                    xyxy,
                    confs,
                    classes,
                    track_ids
                ):

                    cls_name = model.names[int(cls_id)]

                    x1, y1, x2, y2 = box

                    tracking_rows.append([
                        frame_idx,
                        int(cls_id),
                        cls_name,
                        int(track_id),
                        float(x1),
                        float(y1),
                        float(x2),
                        float(y2),
                        float(conf)
                    ])

            # -------------------------
            # MediaPipe
            # -------------------------

            rgb = cv2.cvtColor(
                frame,
                cv2.COLOR_BGR2RGB
            )

            mp_image = mp.Image(
                image_format=mp.ImageFormat.SRGB,
                data=rgb
            )

            hand_result = hand_landmarker.detect(
                mp_image
            )

            hand_data = {
                "frame": frame_idx
            }

            for side in ["left", "right"]:
                for landmark_id in range(21):

                    hand_data[
                        f"{side}_hand_{landmark_id}_x"
                    ] = 0.0

                    hand_data[
                        f"{side}_hand_{landmark_id}_y"
                    ] = 0.0

                    hand_data[
                        f"{side}_hand_{landmark_id}_z"
                    ] = 0.0

            if hand_result.hand_landmarks:

                for hand_landmarks, handedness in zip(
                    hand_result.hand_landmarks,
                    hand_result.handedness
                ):

                    label = (
                        handedness[0]
                        .category_name
                        .lower()
                    )

                    if "left" in label:
                        side = "left"
                    elif "right" in label:
                        side = "right"
                    else:
                        continue

                    for landmark_id, lm in enumerate(
                        hand_landmarks
                    ):

                        hand_data[
                            f"{side}_hand_{landmark_id}_x"
                        ] = float(lm.x)

                        hand_data[
                            f"{side}_hand_{landmark_id}_y"
                        ] = float(lm.y)

                        hand_data[
                            f"{side}_hand_{landmark_id}_z"
                        ] = float(lm.z)

            hand_rows.append(hand_data)

            frame_idx += 1

            if frame_idx % 50 == 0:
                print(
                    f"  Processed "
                    f"{frame_idx}/{total_frames}"
                )

        cap.release()
        hand_landmarker.close()

        print("✓ Frame processing completed")

        # ----------------------------------------------------
        # Tracking DataFrame
        # ----------------------------------------------------

        tracking_df = pd.DataFrame(
            tracking_rows,
            columns=[
                "frame",
                "class_id",
                "class_name",
                "track_id",
                "x1",
                "y1",
                "x2",
                "y2",
                "confidence"
            ]
        )

        hand_df = pd.DataFrame(hand_rows)

        print("Tracking:", tracking_df.shape)
        print("Hands   :", hand_df.shape)

        # ----------------------------------------------------
        # Object features
        # ----------------------------------------------------

        rows = []

        for frame in range(total_frames):

            row = {"frame": frame}

            frame_objects = tracking_df[
                tracking_df["frame"] == frame
            ]

            centers = {}

            for obj in OBJECTS:

                obj_rows = frame_objects[
                    frame_objects["class_name"] == obj
                ]

                if len(obj_rows) > 0:

                    best = obj_rows.loc[
                        obj_rows["confidence"].idxmax()
                    ]

                    presence = 1.0
                    confidence = float(
                        best["confidence"]
                    )

                    x1 = float(best["x1"])
                    y1 = float(best["y1"])
                    x2 = float(best["x2"])
                    y2 = float(best["y2"])

                else:

                    presence = 0.0
                    confidence = 0.0

                    x1 = y1 = x2 = y2 = 0.0

                cx = (x1 + x2) / 2.0
                cy = (y1 + y2) / 2.0

                row[f"{obj}_presence"] = presence
                row[f"{obj}_confidence"] = confidence

                # EXACTLY 5 raw geometry
                row[f"{obj}_x1"] = x1
                row[f"{obj}_y1"] = y1
                row[f"{obj}_x2"] = x2
                row[f"{obj}_y2"] = y2
                row[f"{obj}_center_x"] = cx

                # EXACTLY 5 normalized geometry
                row[f"{obj}_x1_norm"] = x1 / width
                row[f"{obj}_y1_norm"] = y1 / height
                row[f"{obj}_x2_norm"] = x2 / width
                row[f"{obj}_y2_norm"] = y2 / height
                row[f"{obj}_center_x_norm"] = cx / width

                centers[obj] = (
                    cx / width,
                    cy / height
                )

            rows.append(row)

        object_df = pd.DataFrame(rows)

        print("Object features:", object_df.shape)

        # ----------------------------------------------------
        # Object-object distances — exactly 10
        # ----------------------------------------------------

        for a, b in DISTANCE_PAIRS:

            ax = object_df[
                f"{a}_center_x_norm"
            ]

            ay = (
                object_df[f"{a}_y1"] +
                object_df[f"{a}_y2"]
            ) / (2.0 * height)

            bx = object_df[
                f"{b}_center_x_norm"
            ]

            by = (
                object_df[f"{b}_y1"] +
                object_df[f"{b}_y2"]
            ) / (2.0 * height)

            distance = np.sqrt(
                (ax - bx) ** 2 +
                (ay - by) ** 2
            )

            valid = (
                (object_df[f"{a}_presence"] > 0) &
                (object_df[f"{b}_presence"] > 0)
            )

            object_df[
                f"{a}_to_{b}_distance"
            ] = np.where(
                valid,
                distance,
                0.0
            )

        # ----------------------------------------------------
        # Hand features — exactly 126
        # ----------------------------------------------------

        hand_feature_cols = [
            c for c in hand_df.columns
            if c != "frame"
        ]

        if len(hand_feature_cols) != 126:
            raise ValueError(
                f"Expected 126 hand features, "
                f"got {len(hand_feature_cols)}"
            )

        hand_df[hand_feature_cols] = (
            hand_df[hand_feature_cols]
            .replace([np.inf, -np.inf], np.nan)
            .interpolate(limit_direction="both")
            .ffill()
            .bfill()
            .fillna(0.0)
        )

        # ----------------------------------------------------
        # Merge
        # ----------------------------------------------------

        final_df = object_df.merge(
            hand_df,
            on="frame",
            how="left"
        )

        # ----------------------------------------------------
        # Hand-object distances — exactly 10
        # ----------------------------------------------------

        for side in ["left", "right"]:

            wrist_x = final_df[
                f"{side}_hand_0_x"
            ]

            wrist_y = final_df[
                f"{side}_hand_0_y"
            ]

            for obj in OBJECTS:

                obj_x = final_df[
                    f"{obj}_center_x_norm"
                ]

                obj_y = (
                    final_df[f"{obj}_y1"] +
                    final_df[f"{obj}_y2"]
                ) / (2.0 * height)

                distance = np.sqrt(
                    (wrist_x - obj_x) ** 2 +
                    (wrist_y - obj_y) ** 2
                )

                hand_exists = (
                    (wrist_x != 0) |
                    (wrist_y != 0)
                )

                object_exists = (
                    final_df[
                        f"{obj}_presence"
                    ] > 0
                )

                valid = (
                    hand_exists &
                    object_exists
                )

                final_df[
                    f"{side}_hand_to_{obj}_distance"
                ] = np.where(
                    valid,
                    distance,
                    0.0
                )

        # ----------------------------------------------------
        # Cleanup
        # ----------------------------------------------------

        final_df = final_df.replace(
            [np.inf, -np.inf],
            np.nan
        )

        final_df = final_df.fillna(0.0)

        final_df = (
            final_df
            .sort_values("frame")
            .reset_index(drop=True)
        )

        # ----------------------------------------------------
        # Validation
        # ----------------------------------------------------

        model_features = [
            c for c in final_df.columns
            if c != "frame"
        ]

        hand_cols = [
            c for c in final_df.columns
            if "_hand_" in c
            and c.endswith(("_x", "_y", "_z"))
        ]

        object_distance_cols = [
            c for c in final_df.columns
            if "_to_" in c
            and "distance" in c
            and "hand" not in c
        ]

        hand_object_cols = [
            c for c in final_df.columns
            if "_hand_to_" in c
            and "distance" in c
        ]

        print("\nVALIDATION")
        print("Rows:", len(final_df))
        print("Total columns:", len(final_df.columns))
        print("Model features:", len(model_features))
        print("Hand landmarks:", len(hand_cols))
        print("Object-object distances:", len(object_distance_cols))
        print("Hand-object distances:", len(hand_object_cols))
        print(
            "Missing values:",
            int(final_df.isna().sum().sum())
        )

        assert len(final_df) == total_frames
        assert len(hand_cols) == 126
        assert len(object_distance_cols) == 10
        assert len(hand_object_cols) == 10
        assert len(model_features) == 206
        assert len(final_df.columns) == 207
        assert final_df.isna().sum().sum() == 0

        # ----------------------------------------------------
        # Save
        # ----------------------------------------------------

        stem = os.path.splitext(video_name)[0]

        output_dir = os.path.join(
            BASE_OUTPUT,
            stem
        )

        os.makedirs(output_dir, exist_ok=True)

        output_path = os.path.join(
            output_dir,
            f"{stem}_final_features_with_hand_object_distance.csv"
        )

        final_df.to_csv(
            output_path,
            index=False
        )

        print("\n✓ SUCCESS:", video_name)
        print("Saved:", output_path)

        batch_results.append({
            "video": video_name,
            "status": "SUCCESS",
            "frames": len(final_df),
            "features": len(model_features),
            "missing": int(
                final_df.isna().sum().sum()
            )
        })

    except Exception as e:

        print("\n✗ FAILED:", video_name)
        print("ERROR TYPE:", type(e).__name__)
        print("ERROR:", repr(e))

        print("\nFULL TRACEBACK:")
        traceback.print_exc()

        batch_results.append({
            "video": video_name,
            "status": "FAILED",
            "error_type": type(e).__name__,
            "error": repr(e)
        })

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

status_path = "/content/person2_output/batch2_status.json"

with open(status_path, "w") as f:
    json.dump(
        batch_results,
        f,
        indent=2
    )

print("\n" + "=" * 70)
print("BATCH 2 SUMMARY")
print("=" * 70)

for r in batch_results:

    if r["status"] == "SUCCESS":

        print(
            f"✓ {r['video']} | "
            f"frames={r['frames']} | "
            f"features={r['features']} | "
            f"missing={r['missing']}"
        )

    else:

        print(
            f"✗ {r['video']} | "
            f"{r['error_type']}: {r['error']}"
        )

print("\nStatus saved:", status_path)
```

```python
import os
import glob
import pandas as pd

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------
FEATURE_ROOT = "/content/person2_output/all_videos"

TRAIN_CSV = "/content/train (1).csv"
VAL_CSV   = "/content/val.csv"
TEST_CSV  = "/content/test.csv"

OUTPUT_ROOT = "/content/person2_output/aligned"

os.makedirs(OUTPUT_ROOT, exist_ok=True)

# ------------------------------------------------------------
# LOAD PERSON 1 METADATA
# ------------------------------------------------------------
splits = {
    "train": pd.read_csv(TRAIN_CSV),
    "val":   pd.read_csv(VAL_CSV),
    "test":  pd.read_csv(TEST_CSV)
}

# ------------------------------------------------------------
# BUILD VIDEO -> FEATURE FILE MAP
# ------------------------------------------------------------
feature_files = glob.glob(
    os.path.join(
        FEATURE_ROOT,
        "*",
        "*_final_features_with_hand_object_distance.csv"
    )
)

feature_map = {}

for path in feature_files:
    filename = os.path.basename(path)
    stem = filename.replace(
        "_final_features_with_hand_object_distance.csv",
        ""
    )

    feature_map[stem.lower()] = path

print("=" * 70)
print("PERSON 2 — FRAME-LEVEL ALIGNMENT")
print("=" * 70)

print(f"Feature files found: {len(feature_files)}")

# ------------------------------------------------------------
# ALIGN EACH SPLIT
# ------------------------------------------------------------
for split_name, metadata in splits.items():

    aligned_parts = []
    missing_videos = []
    missing_frames = []

    print(f"\nProcessing {split_name.upper()}...")

    for _, clip in metadata.iterrows():

        video_file = str(clip["video_file"])
        video_stem = os.path.splitext(video_file)[0]

        start_frame = int(clip["start_frame"])
        end_frame   = int(clip["end_frame"])

        feature_path = feature_map.get(video_stem.lower())

        if feature_path is None:
            missing_videos.append(video_file)
            continue

        features = pd.read_csv(feature_path)

        # Select the exact temporal range
        clip_features = features[
            (features["frame"] >= start_frame) &
            (features["frame"] <= end_frame)
        ].copy()

        expected_frames = end_frame - start_frame + 1

        if len(clip_features) != expected_frames:
            missing_frames.append({
                "clip_id": clip["clip_id"],
                "video_file": video_file,
                "expected": expected_frames,
                "found": len(clip_features),
                "start_frame": start_frame,
                "end_frame": end_frame
            })
            continue

        # Add Person 1 temporal metadata
        clip_features.insert(0, "clip_id", clip["clip_id"])
        clip_features.insert(1, "video_id", clip["video_id"])
        clip_features.insert(2, "activity", clip["activity"])
        clip_features.insert(3, "label", int(clip["label"]))
        clip_features.insert(4, "step_id", clip["step_id"])
        clip_features.insert(5, "behavior", clip["behavior"])
        clip_features.insert(6, "variation", clip["variation"])

        aligned_parts.append(clip_features)

    # --------------------------------------------------------
    # COMBINE SPLIT
    # --------------------------------------------------------
    if aligned_parts:
        aligned_df = pd.concat(
            aligned_parts,
            ignore_index=True
        )
    else:
        aligned_df = pd.DataFrame()

    output_path = os.path.join(
        OUTPUT_ROOT,
        f"{split_name}_aligned.csv"
    )

    aligned_df.to_csv(output_path, index=False)

    # --------------------------------------------------------
    # REPORT
    # --------------------------------------------------------
    print(f"\n{split_name.upper()} RESULTS")
    print("-" * 50)
    print(f"Metadata clips       : {len(metadata)}")
    print(f"Aligned clips        : {aligned_df['clip_id'].nunique() if not aligned_df.empty else 0}")
    print(f"Aligned frames       : {len(aligned_df)}")
    print(f"Missing videos       : {len(missing_videos)}")
    print(f"Bad frame ranges     : {len(missing_frames)}")
    print(f"Saved                : {output_path}")

    if missing_videos:
        print("\nMissing videos:")
        for v in sorted(set(missing_videos)):
            print("  -", v)

    if missing_frames:
        print("\nBad frame ranges:")
        for item in missing_frames[:10]:
            print("  -", item)

print("\n" + "=" * 70)
print("ALIGNMENT COMPLETE")
print("=" * 70)
```

```python
import pandas as pd
import os

# ------------------------------------------------------------
# LOAD METADATA + ALIGNED OUTPUTS
# ------------------------------------------------------------
metadata_files = {
    "train": "/content/train (1).csv",
    "val":   "/content/val.csv",
    "test":  "/content/test.csv"
}

aligned_files = {
    "train": "/content/person2_output/aligned/train_aligned.csv",
    "val":   "/content/person2_output/aligned/val_aligned.csv",
    "test":  "/content/person2_output/aligned/test_aligned.csv"
}

print("=" * 80)
print("PERSON 2 — ALIGNMENT DIAGNOSTICS")
print("=" * 80)

for split in ["train", "val", "test"]:

    metadata = pd.read_csv(metadata_files[split])
    aligned = pd.read_csv(aligned_files[split])

    aligned_clip_ids = set(aligned["clip_id"].astype(str))

    missing = metadata[
        ~metadata["clip_id"].astype(str).isin(aligned_clip_ids)
    ].copy()

    print(f"\n{split.upper()}")
    print("-" * 80)
    print(f"Metadata clips : {len(metadata)}")
    print(f"Aligned clips  : {len(aligned_clip_ids)}")
    print(f"Missing clips  : {len(missing)}")

    if len(missing) == 0:
        print("✓ No missing clips")
        continue

    # --------------------------------------------------------
    # CHECK EACH MISSING CLIP
    # --------------------------------------------------------
    for _, row in missing.iterrows():

        video_file = str(row["video_file"])
        start = int(row["start_frame"])
        end = int(row["end_frame"])

        feature_path = os.path.join(
            "/content/person2_output/all_videos",
            os.path.splitext(video_file)[0],
            f"{os.path.splitext(video_file)[0]}_final_features_with_hand_object_distance.csv"
        )

        print(
            f"\nclip_id={row['clip_id']} | "
            f"video={video_file} | "
            f"range={start}-{end} | "
            f"expected={end-start+1}"
        )

        if not os.path.exists(feature_path):
            print("  ERROR: feature file not found")
            continue

        features = pd.read_csv(feature_path)

        min_frame = int(features["frame"].min())
        max_frame = int(features["frame"].max())

        available = features[
            (features["frame"] >= start) &
            (features["frame"] <= end)
        ]

        print(
            f"  Feature frames available: "
            f"{min_frame}-{max_frame}"
        )

        print(
            f"  Matching frames found: "
            f"{len(available)}"
        )

        if start > end:
            print("  ⚠ INVALID RANGE: start_frame > end_frame")

        elif start < min_frame or end > max_frame:
            print("  ⚠ RANGE OUTSIDE FEATURE FRAME RANGE")

        else:
            print("  ⚠ Unexpected alignment failure")

print("\n" + "=" * 80)
print("DIAGNOSTICS COMPLETE")
print("=" * 80)
```

```python
import pandas as pd
import os

# ------------------------------------------------------------
# INPUTS
# ------------------------------------------------------------
metadata_files = {
    "train": "/content/train (1).csv",
    "val":   "/content/val.csv",
    "test":  "/content/test.csv"
}

aligned_files = {
    "train": "/content/person2_output/aligned/train_aligned.csv",
    "val":   "/content/person2_output/aligned/val_aligned.csv",
    "test":  "/content/person2_output/aligned/test_aligned.csv"
}

OUTPUT_ROOT = "/content/person2_output/aligned"
os.makedirs(OUTPUT_ROOT, exist_ok=True)

# ------------------------------------------------------------
# KNOWN INVALID TEMPORAL CLIPS
# ------------------------------------------------------------
invalid_clips = {
    "V098_4_0",
    "V098_5_0",
    "V098_6_0"
}

# ------------------------------------------------------------
# PROCESS EACH SPLIT
# ------------------------------------------------------------
exclusion_rows = []

for split in ["train", "val", "test"]:

    metadata = pd.read_csv(metadata_files[split])
    aligned = pd.read_csv(aligned_files[split])

    # Remove any invalid clips if present
    cleaned = aligned[
        ~aligned["clip_id"].astype(str).isin(invalid_clips)
    ].copy()

    # Save cleaned aligned file
    output_path = os.path.join(
        OUTPUT_ROOT,
        f"{split}_aligned_clean.csv"
    )

    cleaned.to_csv(output_path, index=False)

    # Record invalid clips belonging to this split
    split_invalid = metadata[
        metadata["clip_id"].astype(str).isin(invalid_clips)
    ].copy()

    for _, row in split_invalid.iterrows():
        exclusion_rows.append({
            "split": split,
            "clip_id": row["clip_id"],
            "video_file": row["video_file"],
            "activity": row["activity"],
            "start_frame": row["start_frame"],
            "end_frame": row["end_frame"],
            "reason": "Invalid temporal range: start_frame > end_frame"
        })

    print(f"\n{split.upper()}")
    print("-" * 60)
    print("Clean aligned clips :", cleaned["clip_id"].nunique())
    print("Clean aligned rows  :", len(cleaned))
    print("Saved               :", output_path)

# ------------------------------------------------------------
# SAVE EXCLUSION REPORT
# ------------------------------------------------------------
exclusion_df = pd.DataFrame(exclusion_rows)

exclusion_path = os.path.join(
    OUTPUT_ROOT,
    "excluded_temporal_clips.csv"
)

exclusion_df.to_csv(exclusion_path, index=False)

# ------------------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("PERSON 2 — FINAL ALIGNMENT SUMMARY")
print("=" * 70)

print("Excluded clips:", len(exclusion_df))

if len(exclusion_df) > 0:
    print("\nExcluded clips:")
    print(
        exclusion_df[
            [
                "split",
                "clip_id",
                "video_file",
                "activity",
                "start_frame",
                "end_frame",
                "reason"
            ]
        ].to_string(index=False)
    )

print("\nUsable clips:")
print(
    "Train:",
    pd.read_csv(
        "/content/person2_output/aligned/train_aligned_clean.csv"
    )["clip_id"].nunique()
)

print(
    "Val  :",
    pd.read_csv(
        "/content/person2_output/aligned/val_aligned_clean.csv"
    )["clip_id"].nunique()
)

print(
    "Test :",
    pd.read_csv(
        "/content/person2_output/aligned/test_aligned_clean.csv"
    )["clip_id"].nunique()
)

print("\nExclusion report:")
print(exclusion_path)

print("=" * 70)
print("✓ CLEAN ALIGNMENT DATASET READY")
print("=" * 70)
```

```python
import pandas as pd
import os

print("=" * 80)
print("PERSON 2 — CLEAN ALIGNMENT INTEGRITY CHECK")
print("=" * 80)

for split in ["train (1)", "val", "test"]:

    metadata_path = f"/content/{split}.csv"
    aligned_path = (
        f"/content/person2_output/aligned/"
        f"{split}_aligned_clean.csv"
    )

    metadata = pd.read_csv(metadata_path)
    aligned = pd.read_csv(aligned_path)

    # Remove the known invalid metadata rows
    metadata = metadata[
        metadata["clip_id"].astype(str).isin(
            set(aligned["clip_id"].astype(str))
        )
    ].copy()

    # Expected frame count from metadata
    expected = (
        metadata
        .groupby("clip_id")
        .apply(
            lambda x: (
                x["end_frame"].iloc[0]
                - x["start_frame"].iloc[0]
                + 1
            )
        )
        .rename("expected_frames")
        .reset_index()
    )

    # Actual frame count in aligned data
    actual = (
        aligned
        .groupby("clip_id")
        .size()
        .rename("actual_frames")
        .reset_index()
    )

    check = expected.merge(
        actual,
        on="clip_id",
        how="outer"
    )

    check["match"] = (
        check["expected_frames"]
        == check["actual_frames"]
    )

    bad = check[~check["match"]].copy()

    print(f"\n{split.upper()}")
    print("-" * 80)
    print("Unique clips :", aligned["clip_id"].nunique())
    print("Total frames :", len(aligned))
    print("Frame-count mismatches :", len(bad))

    if len(bad) > 0:
        print("\nMISMATCHES:")
        print(bad.to_string(index=False))
    else:
        print("✓ Every clean clip has the expected number of frames.")

    # Check duplicate clip IDs in metadata
    duplicate_ids = (
        metadata["clip_id"]
        .value_counts()
    )

    duplicate_ids = duplicate_ids[
        duplicate_ids > 1
    ]

    print("Duplicate clip IDs in metadata :", len(duplicate_ids))

    if len(duplicate_ids) > 0:
        print(duplicate_ids.to_string())

print("\n" + "=" * 80)
print("INTEGRITY CHECK COMPLETE")
print("=" * 80)
```

```python
import os
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------
ALIGNED_ROOT = "/content/person2_output/aligned"
SEQUENCE_ROOT = "/content/person2_output/sequences"

os.makedirs(SEQUENCE_ROOT, exist_ok=True)

# ------------------------------------------------------------
# SETTINGS
# ------------------------------------------------------------
SEQ_LEN = 32

split_files = {
    "train": os.path.join(
        ALIGNED_ROOT,
        "train_aligned_clean.csv"
    ),
    "val": os.path.join(
        ALIGNED_ROOT,
        "val_aligned_clean.csv"
    ),
    "test": os.path.join(
        ALIGNED_ROOT,
        "test_aligned_clean.csv"
    )
}

# ------------------------------------------------------------
# IDENTIFY MODEL FEATURES
# ------------------------------------------------------------
# We do NOT use these temporal/metadata columns as features.
META_COLUMNS = [
    "clip_id",
    "video_id",
    "activity",
    "label",
    "step_id",
    "behavior",
    "variation"
]

# frame is also not a model feature
NON_FEATURE_COLUMNS = META_COLUMNS + ["frame"]

# ------------------------------------------------------------
# PROCESS EACH SPLIT
# ------------------------------------------------------------
for split, path in split_files.items():

    print("\n" + "=" * 80)
    print(f"CREATING {split.upper()} SEQUENCES")
    print("=" * 80)

    df = pd.read_csv(path)

    # --------------------------------------------------------
    # FEATURE COLUMNS
    # --------------------------------------------------------
    feature_columns = [
        c for c in df.columns
        if c not in NON_FEATURE_COLUMNS
    ]

    print("Input rows        :", len(df))
    print("Unique clips      :", df["clip_id"].nunique())
    print("Feature columns   :", len(feature_columns))

    # --------------------------------------------------------
    # CHECK EXPECTED FEATURE COUNT
    # --------------------------------------------------------
    if len(feature_columns) != 206:
        raise ValueError(
            f"{split}: Expected 206 model features, "
            f"but found {len(feature_columns)}"
        )

    X_sequences = []
    y_sequences = []
    clip_ids = []

    padded_clips = 0
    exact_clips = 0
    skipped_clips = 0

    # --------------------------------------------------------
    # PROCESS ONE CLIP AT A TIME
    # --------------------------------------------------------
    for clip_id, clip_df in df.groupby(
        "clip_id",
        sort=False
    ):

        # Keep frames in temporal order
        clip_df = clip_df.sort_values("frame")

        X = clip_df[feature_columns].to_numpy(
            dtype=np.float32
        )

        # ----------------------------------------------------
        # LABEL
        # ----------------------------------------------------
        labels = clip_df["label"].unique()

        if len(labels) != 1:
            print(
                f"⚠ Skipping {clip_id}: "
                f"multiple labels found: {labels}"
            )
            skipped_clips += 1
            continue

        label = int(labels[0])

        # ----------------------------------------------------
        # EXACTLY 32 FRAMES
        # ----------------------------------------------------
        if len(X) > SEQ_LEN:
            # This should not normally happen.
            # Keep the first 32 frames only.
            X = X[:SEQ_LEN]

        elif len(X) < SEQ_LEN:
            # Zero-pad shorter clips at the END.
            padding = np.zeros(
                (
                    SEQ_LEN - len(X),
                    len(feature_columns)
                ),
                dtype=np.float32
            )

            X = np.vstack([X, padding])

            padded_clips += 1

        else:
            exact_clips += 1

        X_sequences.append(X)
        y_sequences.append(label)
        clip_ids.append(clip_id)

    # --------------------------------------------------------
    # CONVERT TO NUMPY
    # --------------------------------------------------------
    X_sequences = np.asarray(
        X_sequences,
        dtype=np.float32
    )

    y_sequences = np.asarray(
        y_sequences,
        dtype=np.int64
    )

    clip_ids = np.asarray(
        clip_ids
    )

    # --------------------------------------------------------
    # VALIDATION
    # --------------------------------------------------------
    print("\nSequence results:")
    print("Sequences created :", len(X_sequences))
    print("X shape           :", X_sequences.shape)
    print("y shape           :", y_sequences.shape)
    print("Exact 32-frame    :", exact_clips)
    print("Padded clips      :", padded_clips)
    print("Skipped clips     :", skipped_clips)

    print("\nLabels:")
    print(
        pd.Series(y_sequences)
        .value_counts()
        .sort_index()
        .to_string()
    )

    # --------------------------------------------------------
    # SAVE
    # --------------------------------------------------------
    np.save(
        os.path.join(
            SEQUENCE_ROOT,
            f"X_{split}.npy"
        ),
        X_sequences
    )

    np.save(
        os.path.join(
            SEQUENCE_ROOT,
            f"y_{split}.npy"
        ),
        y_sequences
    )

    np.save(
        os.path.join(
            SEQUENCE_ROOT,
            f"clip_ids_{split}.npy"
        ),
        clip_ids
    )

    print("\nSaved:")
    print(
        os.path.join(
            SEQUENCE_ROOT,
            f"X_{split}.npy"
        )
    )
    print(
        os.path.join(
            SEQUENCE_ROOT,
            f"y_{split}.npy"
        )
    )
    print(
        os.path.join(
            SEQUENCE_ROOT,
            f"clip_ids_{split}.npy"
        )
    )

print("\n" + "=" * 80)
print("✓ FIXED-LENGTH SEQUENCE CREATION COMPLETE")
print("=" * 80)
```

```python
import pandas as pd

PATH = "/content/person2_output/aligned/train_aligned_clean.csv"

df = pd.read_csv(PATH)

print("=" * 80)
print("PERSON 2 — INSPECT ALIGNED FEATURE COLUMNS")
print("=" * 80)

print("Rows    :", len(df))
print("Columns :", len(df.columns))

print("\nALL COLUMNS:")
for i, col in enumerate(df.columns):
    print(f"{i:3d}: {col}")

print("\n" + "=" * 80)
print("COLUMN GROUP SUMMARY")
print("=" * 80)

prefix_groups = {
    "presence": [],
    "confidence": [],
    "raw_geometry": [],
    "normalized_geometry": [],
    "object_object_distance": [],
    "hand_landmark": [],
    "hand_object_distance": [],
    "metadata": [],
    "other": []
}

for col in df.columns:

    c = col.lower()

    if col in [
        "clip_id",
        "video_id",
        "activity",
        "label",
        "step_id",
        "behavior",
        "variation",
        "frame"
    ]:
        prefix_groups["metadata"].append(col)

    elif "presence" in c:
        prefix_groups["presence"].append(col)

    elif "confidence" in c:
        prefix_groups["confidence"].append(col)

    elif "normalized" in c:
        prefix_groups["normalized_geometry"].append(col)

    elif "hand_object" in c:
        prefix_groups["hand_object_distance"].append(col)

    elif "hand_" in c or "left_" in c or "right_" in c:
        prefix_groups["hand_landmark"].append(col)

    elif "_to_" in c and "distance" in c:
        prefix_groups["object_object_distance"].append(col)

    elif any(x in c for x in [
        "_x1", "_y1", "_x2", "_y2",
        "_cx", "_cy", "_width", "_height"
    ]):
        prefix_groups["raw_geometry"].append(col)

    else:
        prefix_groups["other"].append(col)

print()

for group, cols in prefix_groups.items():
    print(f"{group:25s}: {len(cols)}")

print("\n" + "=" * 80)
print("INSPECTION COMPLETE")
print("=" * 80)
```

```python
import pandas as pd

PATH = "/content/person2_output/aligned/train_aligned_clean.csv"

df = pd.read_csv(PATH)

OBJECTS = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# ---------------------------------------------------------
# 1. Presence + confidence
# ---------------------------------------------------------
presence_cols = [
    f"{obj}_presence"
    for obj in OBJECTS
]

confidence_cols = [
    f"{obj}_confidence"
    for obj in OBJECTS
]

# ---------------------------------------------------------
# 2. Raw geometry
# Exactly 5 features per object
# ---------------------------------------------------------
raw_geometry_cols = []

for obj in OBJECTS:
    raw_geometry_cols.extend([
        f"{obj}_x1",
        f"{obj}_y1",
        f"{obj}_x2",
        f"{obj}_y2",
        f"{obj}_center_x",
    ])

# ---------------------------------------------------------
# 3. Normalized geometry
# Exactly 5 features per object
# ---------------------------------------------------------
normalized_geometry_cols = []

for obj in OBJECTS:
    normalized_geometry_cols.extend([
        f"{obj}_x1_norm",
        f"{obj}_y1_norm",
        f"{obj}_x2_norm",
        f"{obj}_y2_norm",
        f"{obj}_center_x_norm",
    ])

# ---------------------------------------------------------
# 4. Object-object distances
# ---------------------------------------------------------
object_object_cols = [
    "cotton_to_dropper_distance",
    "cotton_to_seed_jar_distance",
    "cotton_to_target_plate_distance",
    "cotton_to_water_dish_distance",
    "dropper_to_seed_jar_distance",
    "dropper_to_target_plate_distance",
    "dropper_to_water_dish_distance",
    "seed_jar_to_target_plate_distance",
    "seed_jar_to_water_dish_distance",
    "target_plate_to_water_dish_distance",
]

# ---------------------------------------------------------
# 5. Hand landmarks
# 21 landmarks × 3 coordinates × 2 hands = 126
# ---------------------------------------------------------
hand_landmark_cols = []

for hand in ["left", "right"]:
    for landmark in range(21):
        for coord in ["x", "y", "z"]:
            hand_landmark_cols.append(
                f"{hand}_hand_{landmark}_{coord}"
            )

# ---------------------------------------------------------
# 6. Hand-object distances
# 2 hands × 5 objects = 10
# ---------------------------------------------------------
hand_object_cols = []

for hand in ["left", "right"]:
    for obj in OBJECTS:
        hand_object_cols.append(
            f"{hand}_hand_to_{obj}_distance"
        )

# ---------------------------------------------------------
# Combine
# ---------------------------------------------------------
MODEL_FEATURES = (
    presence_cols
    + confidence_cols
    + raw_geometry_cols
    + normalized_geometry_cols
    + object_object_cols
    + hand_landmark_cols
    + hand_object_cols
)

print("=" * 80)
print("PERSON 2 — STANDARDIZED 206-FEATURE SCHEMA CHECK")
print("=" * 80)

print("Presence features          :", len(presence_cols))
print("Confidence features        :", len(confidence_cols))
print("Raw geometry features      :", len(raw_geometry_cols))
print("Normalized geometry        :", len(normalized_geometry_cols))
print("Object-object distances    :", len(object_object_cols))
print("Hand landmark features     :", len(hand_landmark_cols))
print("Hand-object distances      :", len(hand_object_cols))
print("-" * 80)
print("TOTAL EXPECTED FEATURES    :", len(MODEL_FEATURES))

missing = [c for c in MODEL_FEATURES if c not in df.columns]

print("\nMissing standardized columns:", len(missing))

if missing:
    print("\nMISSING COLUMNS:")
    for col in missing:
        print(col)

extra = [
    c for c in df.columns
    if c not in MODEL_FEATURES
    and c not in [
        "clip_id",
        "video_id",
        "activity",
        "label",
        "step_id",
        "behavior",
        "variation",
        "frame"
    ]
]

print("\nExtra/non-standardized feature columns:", len(extra))

print("\n" + "=" * 80)

if len(MODEL_FEATURES) == 206 and len(missing) == 0:
    print("✓ EXACTLY 206 STANDARDIZED FEATURES FOUND")
    print("✓ All required columns exist")
else:
    print("✗ SCHEMA CHECK FAILED")

print("=" * 80)
```

```python
import os
import numpy as np
import pandas as pd

# ============================================================
# PERSON 2 — STEP 2
# CREATE FIXED-LENGTH SEQUENCES
# ============================================================

ALIGNED_ROOT = "/content/person2_output/aligned"
SEQUENCE_ROOT = "/content/person2_output/sequences"

os.makedirs(SEQUENCE_ROOT, exist_ok=True)

SEQ_LEN = 32

# ------------------------------------------------------------
# Metadata columns
# ------------------------------------------------------------
META_COLUMNS = [
    "clip_id",
    "video_id",
    "activity",
    "label",
    "step_id",
    "behavior",
    "variation"
]

# ------------------------------------------------------------
# Objects
# ------------------------------------------------------------
OBJECTS = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# ------------------------------------------------------------
# 1. Presence
# ------------------------------------------------------------
presence_cols = [
    f"{obj}_presence"
    for obj in OBJECTS
]

# ------------------------------------------------------------
# 2. Confidence
# ------------------------------------------------------------
confidence_cols = [
    f"{obj}_confidence"
    for obj in OBJECTS
]

# ------------------------------------------------------------
# 3. Raw geometry
# ------------------------------------------------------------
raw_geometry_cols = []

for obj in OBJECTS:
    raw_geometry_cols.extend([
        f"{obj}_x1",
        f"{obj}_y1",
        f"{obj}_x2",
        f"{obj}_y2",
        f"{obj}_center_x",
    ])

# ------------------------------------------------------------
# 4. Normalized geometry
# ------------------------------------------------------------
normalized_geometry_cols = []

for obj in OBJECTS:
    normalized_geometry_cols.extend([
        f"{obj}_x1_norm",
        f"{obj}_y1_norm",
        f"{obj}_x2_norm",
        f"{obj}_y2_norm",
        f"{obj}_center_x_norm",
    ])

# ------------------------------------------------------------
# 5. Object-object distances
# ------------------------------------------------------------
object_object_cols = [
    "cotton_to_dropper_distance",
    "cotton_to_seed_jar_distance",
    "cotton_to_target_plate_distance",
    "cotton_to_water_dish_distance",
    "dropper_to_seed_jar_distance",
    "dropper_to_target_plate_distance",
    "dropper_to_water_dish_distance",
    "seed_jar_to_target_plate_distance",
    "seed_jar_to_water_dish_distance",
    "target_plate_to_water_dish_distance",
]

# ------------------------------------------------------------
# 6. Hand landmarks
# 21 landmarks × 3 coordinates × 2 hands = 126
# ------------------------------------------------------------
hand_landmark_cols = []

for hand in ["left", "right"]:
    for landmark in range(21):
        for coord in ["x", "y", "z"]:
            hand_landmark_cols.append(
                f"{hand}_hand_{landmark}_{coord}"
            )

# ------------------------------------------------------------
# 7. Hand-object distances
# 2 hands × 5 objects = 10
# ------------------------------------------------------------
hand_object_cols = []

for hand in ["left", "right"]:
    for obj in OBJECTS:
        hand_object_cols.append(
            f"{hand}_hand_to_{obj}_distance"
        )

# ------------------------------------------------------------
# Complete standardized feature list
# ------------------------------------------------------------
MODEL_FEATURES = (
    presence_cols
    + confidence_cols
    + raw_geometry_cols
    + normalized_geometry_cols
    + object_object_cols
    + hand_landmark_cols
    + hand_object_cols
)

assert len(MODEL_FEATURES) == 206

# ------------------------------------------------------------
# Input files
# ------------------------------------------------------------
split_files = {
    "train": os.path.join(
        ALIGNED_ROOT,
        "train_aligned_clean.csv"
    ),
    "val": os.path.join(
        ALIGNED_ROOT,
        "val_aligned_clean.csv"
    ),
    "test": os.path.join(
        ALIGNED_ROOT,
        "test_aligned_clean.csv"
    )
}

# ------------------------------------------------------------
# Containers
# ------------------------------------------------------------
sequence_data = {}
sequence_labels = {}
sequence_metadata = {}

print("=" * 80)
print("PERSON 2 — CREATE FIXED-LENGTH SEQUENCES")
print("=" * 80)
print("Sequence length :", SEQ_LEN)
print("Feature count   :", len(MODEL_FEATURES))
print()

# ============================================================
# PROCESS EACH SPLIT
# ============================================================

for split, path in split_files.items():

    print("-" * 80)
    print(f"PROCESSING {split.upper()}")
    print("-" * 80)

    df = pd.read_csv(path)

    print("Input rows       :", len(df))
    print("Input clips      :", df["clip_id"].nunique())

    # --------------------------------------------------------
    # Verify all required columns
    # --------------------------------------------------------
    missing = [
        c for c in MODEL_FEATURES
        if c not in df.columns
    ]

    if missing:
        raise ValueError(
            f"{split}: Missing {len(missing)} required features"
        )

    # --------------------------------------------------------
    # Ensure frame ordering
    # --------------------------------------------------------
    df = df.sort_values(
        ["clip_id", "frame"]
    ).reset_index(drop=True)

    X_list = []
    y_list = []
    meta_list = []

    skipped = 0
    padded = 0
    exact = 0

    # --------------------------------------------------------
    # Process each clip
    # --------------------------------------------------------
    for clip_id, clip_df in df.groupby(
        "clip_id",
        sort=False
    ):

        # Sort by frame
        clip_df = clip_df.sort_values(
            "frame"
        )

        # ----------------------------------------------------
        # Feature matrix
        # ----------------------------------------------------
        features = clip_df[
            MODEL_FEATURES
        ].to_numpy(
            dtype=np.float32
        )

        # ----------------------------------------------------
        # Label
        # ----------------------------------------------------
        labels = clip_df["label"].unique()

        if len(labels) != 1:
            raise ValueError(
                f"{split}: Clip {clip_id} has "
                f"multiple labels: {labels}"
            )

        label = int(labels[0])

        # ----------------------------------------------------
        # Metadata
        # ----------------------------------------------------
        first_row = clip_df.iloc[0]

        metadata = {
            "clip_id": clip_id,
            "video_id": first_row["video_id"],
            "activity": first_row["activity"],
            "label": label,
            "step_id": first_row["step_id"],
            "behavior": first_row["behavior"],
            "variation": first_row["variation"],
            "original_length": len(features)
        }

        # ----------------------------------------------------
        # Skip clips longer than SEQ_LEN
        # ----------------------------------------------------
        if len(features) > SEQ_LEN:
            skipped += 1

            print(
                f"WARNING: {clip_id} has "
                f"{len(features)} frames — skipped"
            )

            continue

        # ----------------------------------------------------
        # Create fixed-length sequence
        # ----------------------------------------------------
        sequence = np.zeros(
            (SEQ_LEN, len(MODEL_FEATURES)),
            dtype=np.float32
        )

        sequence[:len(features)] = features

        # ----------------------------------------------------
        # Statistics
        # ----------------------------------------------------
        if len(features) < SEQ_LEN:
            padded += 1
        else:
            exact += 1

        X_list.append(sequence)
        y_list.append(label)
        meta_list.append(metadata)

    # --------------------------------------------------------
    # Convert to numpy arrays
    # --------------------------------------------------------
    X = np.stack(X_list).astype(
        np.float32
    )

    y = np.array(
        y_list,
        dtype=np.int64
    )

    metadata_df = pd.DataFrame(
        meta_list
    )

    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------
    np.save(
        os.path.join(
            SEQUENCE_ROOT,
            f"X_{split}.npy"
        ),
        X
    )

    np.save(
        os.path.join(
            SEQUENCE_ROOT,
            f"y_{split}.npy"
        ),
        y
    )

    metadata_df.to_csv(
        os.path.join(
            SEQUENCE_ROOT,
            f"{split}_sequence_metadata.csv"
        ),
        index=False
    )

    # --------------------------------------------------------
    # Store
    # --------------------------------------------------------
    sequence_data[split] = X
    sequence_labels[split] = y
    sequence_metadata[split] = metadata_df

    # --------------------------------------------------------
    # Report
    # --------------------------------------------------------
    print()
    print("Created sequences :", len(X))
    print("Sequence shape    :", X.shape)
    print("Labels shape      :", y.shape)
    print("Exact 32 frames   :", exact)
    print("Zero padded       :", padded)
    print("Skipped (>32)     :", skipped)
    print("Label distribution:")

    print(
        pd.Series(y)
        .value_counts()
        .sort_index()
        .to_string()
    )

print()
print("=" * 80)
print("SEQUENCE CREATION COMPLETE")
print("=" * 80)

print("\nSaved files:")

for split in split_files:
    print(
        f"  {split}: "
        f"{os.path.join(SEQUENCE_ROOT, f'X_{split}.npy')}"
    )
    print(
        f"  {split}: "
        f"{os.path.join(SEQUENCE_ROOT, f'y_{split}.npy')}"
    )
    print(
        f"  {split}: "
        f"{os.path.join(SEQUENCE_ROOT, f'{split}_sequence_metadata.csv')}"
    )

print("\nExpected tensor format:")
print("  X_train: (N_train, 32, 206)")
print("  X_val  : (N_val,   32, 206)")
print("  X_test : (N_test,  32, 206)")

print("=" * 80)
```

```python
import pandas as pd
import os

ALIGNED_ROOT = "/content/person2_output/aligned"

split_files = {
    "train": os.path.join(
        ALIGNED_ROOT,
        "train_aligned_clean.csv"
    ),
    "val": os.path.join(
        ALIGNED_ROOT,
        "val_aligned_clean.csv"
    ),
    "test": os.path.join(
        ALIGNED_ROOT,
        "test_aligned_clean.csv"
    )
}

print("=" * 80)
print("PERSON 2 — INVESTIGATE CLIPS LONGER THAN 32 FRAMES")
print("=" * 80)

all_long = []

for split, path in split_files.items():

    df = pd.read_csv(path)

    clip_lengths = (
        df.groupby("clip_id")
          .size()
          .reset_index(name="actual_aligned_frames")
    )

    long_clips = clip_lengths[
        clip_lengths["actual_aligned_frames"] > 32
    ].copy()

    long_clips["split"] = split

    # Add label/activity/video information
    info = (
        df[
            [
                "clip_id",
                "video_id",
                "activity",
                "label",
                "step_id",
                "behavior",
                "variation"
            ]
        ]
        .drop_duplicates("clip_id")
    )

    long_clips = long_clips.merge(
        info,
        on="clip_id",
        how="left"
    )

    all_long.append(long_clips)

    print()
    print(f"{split.upper()}")
    print("-" * 80)
    print("Long clips:", len(long_clips))

    if len(long_clips):
        print(
            long_clips[
                [
                    "clip_id",
                    "video_id",
                    "activity",
                    "label",
                    "actual_aligned_frames"
                ]
            ].to_string(index=False)
        )

combined = pd.concat(
    all_long,
    ignore_index=True
)

print()
print("=" * 80)
print("SUMMARY")
print("=" * 80)

print("Total clips > 32:", len(combined))

print("\nBy split:")
print(
    combined["split"]
    .value_counts()
    .sort_index()
    .to_string()
)

print("\nBy label:")
print(
    combined["label"]
    .value_counts()
    .sort_index()
    .to_string()
)

print("\nFrame-count distribution:")
print(
    combined["actual_aligned_frames"]
    .value_counts()
    .sort_index()
    .to_string()
)

print("=" * 80)
```

```python
import os
import pandas as pd

ALIGNED_ROOT = "/content/person2_output/aligned"

split_files = {
    "train": os.path.join(
        ALIGNED_ROOT,
        "train_aligned_clean.csv"
    ),
    "val": os.path.join(
        ALIGNED_ROOT,
        "val_aligned_clean.csv"
    ),
    "test": os.path.join(
        ALIGNED_ROOT,
        "test_aligned_clean.csv"
    )
}

# Original Person 1 metadata
META_FILES = {
    "train": "/content/train (1).csv",
    "val": "/content/val.csv",
    "test": "/content/test.csv"
}

print("=" * 100)
print("PERSON 2 — LONG CLIP METADATA DIAGNOSTIC")
print("=" * 100)

for split in ["train", "val", "test"]:

    aligned = pd.read_csv(split_files[split])
    metadata = pd.read_csv(META_FILES[split])

    # Find clips whose aligned frame count > 32
    lengths = (
        aligned.groupby("clip_id")
        .size()
        .reset_index(name="aligned_rows")
    )

    long_ids = lengths.loc[
        lengths["aligned_rows"] > 32,
        "clip_id"
    ].tolist()

    if not long_ids:
        continue

    print()
    print("=" * 100)
    print(split.upper())
    print("=" * 100)

    for clip_id in long_ids:

        print()
        print("-" * 100)
        print("CLIP:", clip_id)

        # ----------------------------------------------------
        # Original metadata rows for this clip_id
        # ----------------------------------------------------
        meta_rows = metadata[
            metadata["clip_id"] == clip_id
        ].copy()

        print("\nORIGINAL PERSON 1 METADATA ROW(S):")

        if len(meta_rows) == 0:
            print("  *** NO METADATA ROW FOUND ***")
        else:
            print(
                meta_rows[
                    [
                        "clip_id",
                        "video_id",
                        "video_file",
                        "step_id",
                        "activity",
                        "start_frame",
                        "end_frame",
                        "sequence_length",
                        "label"
                    ]
                ].to_string(index=False)
            )

        # ----------------------------------------------------
        # Aligned rows
        # ----------------------------------------------------
        clip = aligned[
            aligned["clip_id"] == clip_id
        ].sort_values("frame")

        frames = clip["frame"].astype(int).tolist()

        print("\nALIGNED DATA:")
        print("  Rows          :", len(clip))
        print("  Unique frames :", clip["frame"].nunique())
        print("  First frame   :", frames[0] if frames else None)
        print("  Last frame    :", frames[-1] if frames else None)

        # ----------------------------------------------------
        # Check continuity
        # ----------------------------------------------------
        if frames:
            expected = list(
                range(frames[0], frames[-1] + 1)
            )

            missing_frames = sorted(
                set(expected) - set(frames)
            )

            duplicate_frames = (
                clip["frame"]
                .value_counts()
            )

            duplicate_frames = duplicate_frames[
                duplicate_frames > 1
            ]

            print(
                "  Continuous     :",
                len(missing_frames) == 0
            )

            print(
                "  Missing frames :",
                len(missing_frames)
            )

            if missing_frames:
                print(
                    "  Missing list   :",
                    missing_frames[:30]
                )

            print(
                "  Duplicate frames:",
                len(duplicate_frames)
            )

            if len(duplicate_frames):
                print(
                    "  Duplicate list :",
                    duplicate_frames.to_dict()
                )

        # ----------------------------------------------------
        # Compare metadata ranges with aligned frames
        # ----------------------------------------------------
        print("\nRANGE COMPARISON:")

        for idx, row in meta_rows.iterrows():

            start = int(row["start_frame"])
            end = int(row["end_frame"])

            expected_len = end - start + 1

            actual_in_range = clip[
                (clip["frame"] >= start) &
                (clip["frame"] <= end)
            ]["frame"].nunique()

            print(
                f"  Metadata row {idx}: "
                f"{start}–{end} | "
                f"metadata length={expected_len} | "
                f"aligned frames in range={actual_in_range}"
            )

print()
print("=" * 100)
print("LONG CLIP DIAGNOSTIC COMPLETE")
print("=" * 100)
```

```python
import os
import glob
import pandas as pd

# ============================================================
# PERSON 2 — CORRECTED ROW-LEVEL ALIGNMENT
# ============================================================

FEATURE_ROOT = "/content/person2_output/all_videos"
OUTPUT_ROOT = "/content/person2_output/aligned_rowwise"
os.makedirs(OUTPUT_ROOT, exist_ok=True)

META_FILES = {
    "train": "/content/train (1).csv",
    "val": "/content/val.csv",
    "test": "/content/test.csv"
}

# ------------------------------------------------------------
# Build feature-file lookup by video filename
# ------------------------------------------------------------

feature_files = glob.glob(
    os.path.join(
        FEATURE_ROOT,
        "*",
        "*_final_features_with_hand_object_distance.csv"
    )
)

feature_map = {}

for path in feature_files:
    filename = os.path.basename(path)

    # Example:
    # C4_final_features_with_hand_object_distance.csv
    stem = filename.replace(
        "_final_features_with_hand_object_distance.csv",
        ""
    )

    feature_map[stem.lower()] = path

print("Feature files found:", len(feature_files))


# ------------------------------------------------------------
# Process each split
# ------------------------------------------------------------

for split, meta_path in META_FILES.items():

    print("\n" + "=" * 80)
    print(split.upper())
    print("=" * 80)

    meta = pd.read_csv(meta_path)

    aligned_parts = []
    excluded_rows = []

    for row_idx, row in meta.iterrows():

        video_file = str(row["video_file"])
        video_stem = os.path.splitext(
            os.path.basename(video_file)
        )[0].lower()

        clip_id = str(row["clip_id"])

        start_frame = int(row["start_frame"])
        end_frame = int(row["end_frame"])

        # ----------------------------------------------------
        # Unique segment identifier
        # IMPORTANT:
        # Do NOT use clip_id alone.
        # ----------------------------------------------------

        segment_uid = f"{split}_{row_idx:04d}_{clip_id}"

        # ----------------------------------------------------
        # Validate temporal range
        # ----------------------------------------------------

        if start_frame > end_frame:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "reason": "Invalid temporal range"
            })

            continue

        # ----------------------------------------------------
        # Find Person 2 feature file
        # ----------------------------------------------------

        feature_path = feature_map.get(video_stem)

        if feature_path is None:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "reason": "Person 2 feature file not found"
            })

            continue

        # ----------------------------------------------------
        # Load features
        # ----------------------------------------------------

        features = pd.read_csv(feature_path)

        # ----------------------------------------------------
        # Select ONLY this metadata row's frame range
        # ----------------------------------------------------

        segment = features[
            (features["frame"] >= start_frame) &
            (features["frame"] <= end_frame)
        ].copy()

        expected_frames = end_frame - start_frame + 1

        actual_frames = len(segment)

        # ----------------------------------------------------
        # Validate frame coverage
        # ----------------------------------------------------

        if actual_frames != expected_frames:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "expected_frames": expected_frames,
                "actual_frames": actual_frames,
                "reason": "Frame count mismatch"
            })

            continue

        # ----------------------------------------------------
        # Add metadata
        # ----------------------------------------------------

        segment["segment_uid"] = segment_uid
        segment["clip_id"] = clip_id
        segment["video_id"] = row["video_id"]
        segment["activity"] = row["activity"]
        segment["label"] = row["label"]
        segment["step_id"] = row["step_id"]
        segment["behavior"] = row["behavior"]
        segment["variation"] = row["variation"]
        segment["metadata_row"] = row_idx

        aligned_parts.append(segment)

    # --------------------------------------------------------
    # Combine
    # --------------------------------------------------------

    if aligned_parts:

        aligned = pd.concat(
            aligned_parts,
            ignore_index=True
        )

    else:
        aligned = pd.DataFrame()

    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------

    output_path = os.path.join(
        OUTPUT_ROOT,
        f"{split}_aligned_rowwise.csv"
    )

    aligned.to_csv(
        output_path,
        index=False
    )

    # Save exclusions
    excluded_df = pd.DataFrame(excluded_rows)

    excluded_path = os.path.join(
        OUTPUT_ROOT,
        f"{split}_excluded.csv"
    )

    excluded_df.to_csv(
        excluded_path,
        index=False
    )

    # --------------------------------------------------------
    # Summary
    # --------------------------------------------------------

    print("Original metadata rows :", len(meta))
    print("Aligned segment rows   :", len(aligned_parts))
    print("Aligned frame rows     :", len(aligned))

    if len(aligned_parts) > 0:
        print(
            "Unique segment UIDs    :",
            aligned["segment_uid"].nunique()
        )

    print("Excluded metadata rows :", len(excluded_df))

    print("Saved:")
    print(output_path)
    print(excluded_path)

print("\n" + "=" * 80)
print("ROW-LEVEL ALIGNMENT COMPLETE")
print("=" * 80)
```

```python
import os
import pandas as pd

ALIGN_ROOT = "/content/person2_output/aligned_rowwise"

SPLITS = {
    "train": {
        "aligned": os.path.join(ALIGN_ROOT, "train_aligned_rowwise.csv"),
        "excluded": os.path.join(ALIGN_ROOT, "train_excluded.csv"),
        "expected_segments": 1334
    },
    "val": {
        "aligned": os.path.join(ALIGN_ROOT, "val_aligned_rowwise.csv"),
        "excluded": os.path.join(ALIGN_ROOT, "val_excluded.csv"),
        "expected_segments": 313
    },
    "test": {
        "aligned": os.path.join(ALIGN_ROOT, "test_aligned_rowwise.csv"),
        "excluded": os.path.join(ALIGN_ROOT, "test_excluded.csv"),
        "expected_segments": 294
    }
}

# ---------------------------------------------------------
# STANDARDIZED 206-FEATURE SCHEMA
# ---------------------------------------------------------

presence_features = [
    "cotton_present",
    "dropper_present",
    "seed_jar_present",
    "target_plate_present",
    "water_dish_present"
]

confidence_features = [
    "cotton_confidence",
    "dropper_confidence",
    "seed_jar_confidence",
    "target_plate_confidence",
    "water_dish_confidence"
]

object_classes = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

raw_geometry_features = []
normalized_geometry_features = []

for cls in object_classes:
    raw_geometry_features += [
        f"{cls}_x1",
        f"{cls}_y1",
        f"{cls}_x2",
        f"{cls}_y2",
        f"{cls}_center_x"
    ]

    normalized_geometry_features += [
        f"{cls}_x1_norm",
        f"{cls}_y1_norm",
        f"{cls}_x2_norm",
        f"{cls}_y2_norm",
        f"{cls}_center_x_norm"
    ]

# Object-object distances
object_object_features = []

for i in range(len(object_classes)):
    for j in range(i + 1, len(object_classes)):
        object_object_features.append(
            f"{object_classes[i]}_to_{object_classes[j]}_distance"
        )

# Hand landmarks
hand_landmark_features = []

for hand in ["left_hand", "right_hand"]:
    for landmark in range(21):
        for coord in ["x", "y", "z"]:
            hand_landmark_features.append(
                f"{hand}_{landmark}_{coord}"
            )

# Hand-object distances
hand_object_features = [
    f"{hand}_to_{obj}_distance"
    for hand in ["left_hand", "right_hand"]
    for obj in object_classes
]

FEATURE_COLUMNS = (
    presence_features
    + confidence_features
    + raw_geometry_features
    + normalized_geometry_features
    + object_object_features
    + hand_landmark_features
    + hand_object_features
)

print("Expected standardized features:", len(FEATURE_COLUMNS))

assert len(FEATURE_COLUMNS) == 206


# ---------------------------------------------------------
# VERIFY EACH SPLIT
# ---------------------------------------------------------

all_passed = True

for split, info in SPLITS.items():

    print("\n" + "=" * 80)
    print(split.upper())
    print("=" * 80)

    df = pd.read_csv(info["aligned"])

    # ---------------------------------------------
    # Basic structure
    # ---------------------------------------------

    segment_count = df["segment_uid"].nunique()

    print("Rows                :", len(df))
    print("Unique segments     :", segment_count)
    print("Expected segments   :", info["expected_segments"])

    if segment_count != info["expected_segments"]:
        print("❌ Segment count mismatch")
        all_passed = False
    else:
        print("✓ Segment count correct")

    # ---------------------------------------------
    # Required columns
    # ---------------------------------------------

    missing_columns = [
        col for col in FEATURE_COLUMNS
        if col not in df.columns
    ]

    print("Missing feature columns:", len(missing_columns))

    if missing_columns:
        print("❌ Missing standardized features")
        print(missing_columns[:20])
        all_passed = False
    else:
        print("✓ All 206 standardized features present")

    # ---------------------------------------------
    # Segment frame counts
    # ---------------------------------------------

    segment_lengths = (
        df.groupby("segment_uid")
          .size()
    )

    print(
        "Segment length range:",
        segment_lengths.min(),
        "to",
        segment_lengths.max()
    )

    over_32 = segment_lengths[segment_lengths > 32]

    print("Segments > 32 frames:", len(over_32))

    if len(over_32) > 0:
        print("❌ Segments longer than 32 found")
        print(over_32.head(20))
        all_passed = False
    else:
        print("✓ No segment exceeds 32 frames")

    # ---------------------------------------------
    # Check frame continuity
    # ---------------------------------------------

    continuity_errors = 0

    for segment_uid, group in df.groupby("segment_uid"):

        frames = group["frame"].sort_values().to_numpy()

        if len(frames) > 1:
            expected = list(range(frames[0], frames[-1] + 1))

            if frames.tolist() != expected:
                continuity_errors += 1

    print("Frame continuity errors:", continuity_errors)

    if continuity_errors > 0:
        print("❌ Frame continuity problem")
        all_passed = False
    else:
        print("✓ Frame ranges continuous")

    # ---------------------------------------------
    # Check duplicate frames inside segments
    # ---------------------------------------------

    duplicate_frames = (
        df.groupby("segment_uid")["frame"]
          .apply(lambda x: x.duplicated().sum())
          .sum()
    )

    print("Duplicate frames:", duplicate_frames)

    if duplicate_frames > 0:
        print("❌ Duplicate frames found")
        all_passed = False
    else:
        print("✓ No duplicate frames")

    # ---------------------------------------------
    # Missing values
    # ---------------------------------------------

    feature_missing = df[FEATURE_COLUMNS].isna().sum().sum()

    print("Missing feature values:", feature_missing)

    if feature_missing > 0:
        print("❌ Missing feature values found")
        all_passed = False
    else:
        print("✓ No missing feature values")

    # ---------------------------------------------
    # Label check
    # ---------------------------------------------

    labels = sorted(df["label"].dropna().unique().tolist())

    print("Labels found:", labels)

    expected_labels = [0, 1, 2, 3, 4, 5]

    if labels != expected_labels:
        print("⚠️ Label set differs from expected:", expected_labels)
    else:
        print("✓ All 6 labels present")


# ---------------------------------------------------------
# EXCLUDED ROWS
# ---------------------------------------------------------

print("\n" + "=" * 80)
print("EXCLUDED METADATA ROWS")
print("=" * 80)

for split, info in SPLITS.items():

    excluded = pd.read_csv(info["excluded"])

    print(f"\n{split.upper()}: {len(excluded)} excluded")

    if len(excluded) > 0:
        print(
            excluded[
                [
                    "metadata_row",
                    "segment_uid",
                    "clip_id",
                    "start_frame",
                    "end_frame",
                    "reason"
                ]
            ].to_string(index=False)
        )


# ---------------------------------------------------------
# FINAL RESULT
# ---------------------------------------------------------

print("\n" + "=" * 80)
print("ALIGNMENT VERIFICATION RESULT")
print("=" * 80)

if all_passed:
    print("🎉 ALL ALIGNMENT CHECKS PASSED!")
    print("✓ Train segments are correct")
    print("✓ Validation segments are correct")
    print("✓ Test segments are correct")
    print("✓ 206 standardized features present")
    print("✓ No missing values")
    print("✓ No segments > 32 frames")
    print("✓ Frame ranges continuous")
    print("✓ No duplicate frames")
    print("\nREADY FOR SEQUENCE GENERATION.")
else:
    print("❌ SOME ALIGNMENT CHECKS FAILED")
    print("Do NOT generate the final sequences yet.")
```

```python
import os
import pandas as pd

ALIGN_ROOT = "/content/person2_output/aligned_rowwise"

object_classes = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# 5 presence
presence_features = [
    f"{obj}_presence"
    for obj in object_classes
]

# 5 confidence
confidence_features = [
    f"{obj}_confidence"
    for obj in object_classes
]

# 25 raw geometry
raw_geometry_features = []

for obj in object_classes:
    raw_geometry_features += [
        f"{obj}_x1",
        f"{obj}_y1",
        f"{obj}_x2",
        f"{obj}_y2",
        f"{obj}_center_x"
    ]

# 25 normalized geometry
normalized_geometry_features = []

for obj in object_classes:
    normalized_geometry_features += [
        f"{obj}_x1_norm",
        f"{obj}_y1_norm",
        f"{obj}_x2_norm",
        f"{obj}_y2_norm",
        f"{obj}_center_x_norm"
    ]

# 10 object-object distances
object_object_features = []

for i in range(len(object_classes)):
    for j in range(i + 1, len(object_classes)):
        object_object_features.append(
            f"{object_classes[i]}_to_{object_classes[j]}_distance"
        )

# 126 hand landmarks
hand_landmark_features = []

for hand in ["left_hand", "right_hand"]:
    for landmark in range(21):
        for coord in ["x", "y", "z"]:
            hand_landmark_features.append(
                f"{hand}_{landmark}_{coord}"
            )

# 10 hand-object distances
hand_object_features = [
    f"{hand}_to_{obj}_distance"
    for hand in ["left_hand", "right_hand"]
    for obj in object_classes
]

FEATURE_COLUMNS = (
    presence_features
    + confidence_features
    + raw_geometry_features
    + normalized_geometry_features
    + object_object_features
    + hand_landmark_features
    + hand_object_features
)

assert len(FEATURE_COLUMNS) == 206

print("206-feature schema confirmed.\n")


for split in ["train", "val", "test"]:

    path = os.path.join(
        ALIGN_ROOT,
        f"{split}_aligned_rowwise.csv"
    )

    df = pd.read_csv(path)

    model_df = df[FEATURE_COLUMNS]

    missing = model_df.isna().sum()
    missing = missing[missing > 0].sort_values(
        ascending=False
    )

    print("\n" + "=" * 90)
    print(split.upper())
    print("=" * 90)

    print("Rows:", len(df))
    print("Model features:", model_df.shape[1])

    print(
        "Total missing cells in 206 features:",
        int(missing.sum())
    )

    print(
        "Features containing NaN:",
        len(missing),
        "/ 206"
    )

    if len(missing) > 0:
        print("\nMissing values by feature:")

        print(missing.to_string())

    # -----------------------------------------------------
    # Missing by feature group
    # -----------------------------------------------------

    groups = {
        "Presence": presence_features,
        "Confidence": confidence_features,
        "Raw geometry": raw_geometry_features,
        "Normalized geometry": normalized_geometry_features,
        "Object-object distances": object_object_features,
        "Hand landmarks": hand_landmark_features,
        "Hand-object distances": hand_object_features
    }

    print("\nMissing values by feature group:")

    for group_name, columns in groups.items():

        count = int(
            model_df[columns]
            .isna()
            .sum()
            .sum()
        )

        total = len(df) * len(columns)

        percentage = (
            100 * count / total
            if total > 0 else 0
        )

        print(
            f"{group_name:30s}: "
            f"{count:8d} / {total:8d} "
            f"({percentage:6.2f}%)"
        )

    # -----------------------------------------------------
    # Rows containing NaN
    # -----------------------------------------------------

    rows_with_nan = model_df.isna().any(axis=1).sum()

    print(
        "\nRows containing at least one NaN:",
        int(rows_with_nan),
        f"/ {len(df)}"
    )
```

```python
import os
import pandas as pd

ALIGN_ROOT = "/content/person2_output/aligned_rowwise"

for split in ["train", "val", "test"]:

    path = os.path.join(
        ALIGN_ROOT,
        f"{split}_aligned_rowwise.csv"
    )

    df = pd.read_csv(path)

    print("\n" + "=" * 90)
    print(split.upper())
    print("=" * 90)

    # Standardized feature columns
    presence_cols = [
        "cotton_presence",
        "dropper_presence",
        "seed_jar_presence",
        "target_plate_presence",
        "water_dish_presence"
    ]

    confidence_cols = [
        "cotton_confidence",
        "dropper_confidence",
        "seed_jar_confidence",
        "target_plate_confidence",
        "water_dish_confidence"
    ]

    raw_geom_cols = [
        f"{obj}_{coord}"
        for obj in [
            "cotton",
            "dropper",
            "seed_jar",
            "target_plate",
            "water_dish"
        ]
        for coord in ["x1", "y1", "x2", "y2"]
    ]

    norm_geom_cols = [
        f"{obj}_{coord}_norm"
        for obj in [
            "cotton",
            "dropper",
            "seed_jar",
            "target_plate",
            "water_dish"
        ]
        for coord in ["x1", "y1", "x2", "y2"]
    ]

    object_distance_cols = [
        "cotton_to_dropper_distance",
        "cotton_to_seed_jar_distance",
        "cotton_to_target_plate_distance",
        "cotton_to_water_dish_distance",
        "dropper_to_seed_jar_distance",
        "dropper_to_target_plate_distance",
        "dropper_to_water_dish_distance",
        "seed_jar_to_target_plate_distance",
        "seed_jar_to_water_dish_distance",
        "target_plate_to_water_dish_distance"
    ]

    hand_cols = [
        f"{hand}_hand_{i}_{coord}"
        for hand in ["left", "right"]
        for i in range(21)
        for coord in ["x", "y", "z"]
    ]

    hand_object_cols = [
        "left_hand_to_cotton_distance",
        "left_hand_to_dropper_distance",
        "left_hand_to_seed_jar_distance",
        "left_hand_to_target_plate_distance",
        "left_hand_to_water_dish_distance",
        "right_hand_to_cotton_distance",
        "right_hand_to_dropper_distance",
        "right_hand_to_seed_jar_distance",
        "right_hand_to_target_plate_distance",
        "right_hand_to_water_dish_distance"
    ]

    feature_cols = (
        presence_cols +
        confidence_cols +
        raw_geom_cols +
        norm_geom_cols +
        object_distance_cols +
        hand_cols +
        hand_object_cols
    )

    # Rows with any NaN
    nan_mask = df[feature_cols].isna().any(axis=1)
    nan_rows = df[nan_mask].copy()

    print("Total rows:", len(df))
    print("Rows containing NaN:", len(nan_rows))
    print("Percentage:", round(100 * len(nan_rows) / len(df), 2), "%")

    # Frames with NaNs
    print("\nFirst 20 NaN frames:")
    print(
        nan_rows[
            ["segment_uid", "clip_id", "video_id", "frame", "label"]
        ].head(20).to_string(index=False)
    )

    # How many NaN frames per segment?
    segment_nan_counts = (
        nan_rows
        .groupby("segment_uid")
        .size()
        .sort_values(ascending=False)
    )

    print("\nSegments containing NaNs:", len(segment_nan_counts))

    print("\nTop 20 segments by NaN-frame count:")
    print(segment_nan_counts.head(20).to_string())

    # Maximum consecutive NaN frames within each segment
    max_gap = 0
    worst_segment = None

    for segment_uid, group in df.groupby("segment_uid", sort=False):

        group = group.sort_values("frame")

        row_nan = group[feature_cols].isna().any(axis=1).tolist()

        current = 0
        local_max = 0

        for value in row_nan:
            if value:
                current += 1
                local_max = max(local_max, current)
            else:
                current = 0

        if local_max > max_gap:
            max_gap = local_max
            worst_segment = segment_uid

    print("\nMaximum consecutive NaN frames:", max_gap)
    print("Worst segment:", worst_segment)
```

```python
import os
import glob
import pandas as pd

# Person 1 metadata
META_FILES = {
    "train": "/content/train (1).csv",
    "val": "/content/val.csv",
    "test": "/content/test.csv"
}

FEATURE_ROOT = "/content/person2_output/all_videos"

# ---------------------------------------------------------
# 1. Get every video expected by Person 1
# ---------------------------------------------------------

expected_videos = set()

for split, path in META_FILES.items():

    df = pd.read_csv(path)

    for video_file in df["video_file"].dropna():

        stem = os.path.splitext(
            os.path.basename(str(video_file))
        )[0].lower()

        expected_videos.add(stem)


# ---------------------------------------------------------
# 2. Get every video actually processed by Person 2
# ---------------------------------------------------------

feature_files = glob.glob(
    os.path.join(
        FEATURE_ROOT,
        "*",
        "*_final_features_with_hand_object_distance.csv"
    )
)

processed_videos = set()

for path in feature_files:

    filename = os.path.basename(path)

    stem = filename.replace(
        "_final_features_with_hand_object_distance.csv",
        ""
    ).lower()

    processed_videos.add(stem)


# ---------------------------------------------------------
# 3. Compare
# ---------------------------------------------------------

missing_videos = sorted(
    expected_videos - processed_videos
)

extra_videos = sorted(
    processed_videos - expected_videos
)


print("=" * 90)
print("PERSON 1 ↔ PERSON 2 VIDEO COVERAGE CHECK")
print("=" * 90)

print("\nExpected videos from Person 1 metadata :", len(expected_videos))
print("Processed feature videos              :", len(processed_videos))

print("\nMissing Person 2 feature videos:",
      len(missing_videos))

for i, video in enumerate(missing_videos, 1):
    print(f"{i:3d}. {video}.mp4")

print("\nExtra Person 2 feature videos:",
      len(extra_videos))

for i, video in enumerate(extra_videos, 1):
    print(f"{i:3d}. {video}.mp4")
```

```python
import os
import pandas as pd
import numpy as np

FEATURE_PATH = (
    "/content/person2_output/all_videos/C5/"
    "C5_final_features_with_hand_object_distance.csv"
)

features = pd.read_csv(FEATURE_PATH)

# ---------------------------------------------------------
# Canonical object list
# ---------------------------------------------------------

objects = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# ---------------------------------------------------------
# Build canonical 206-feature row representation
# ---------------------------------------------------------

canonical = pd.DataFrame()

canonical["frame"] = features["frame"]

# Presence + confidence
for obj in objects:
    canonical[f"{obj}_presence"] = features[f"{obj}_presence"]
    canonical[f"{obj}_confidence"] = features[f"{obj}_confidence"]

# Raw geometry
#
# Original feature files contain:
# center_x, center_y, width, height, area
#
# Reconstruct:
# x1 = center_x - width/2
# y1 = center_y - height/2
# x2 = center_x + width/2
# y2 = center_y + height/2
# center_x = center_x

for obj in objects:

    cx = features[f"{obj}_center_x"]
    cy = features[f"{obj}_center_y"]
    w = features[f"{obj}_width"]
    h = features[f"{obj}_height"]

    canonical[f"{obj}_x1"] = cx - w / 2
    canonical[f"{obj}_y1"] = cy - h / 2
    canonical[f"{obj}_x2"] = cx + w / 2
    canonical[f"{obj}_y2"] = cy + h / 2
    canonical[f"{obj}_center_x"] = cx

# Normalized geometry
for obj in objects:

    cx = features[f"{obj}_center_x_norm"]
    cy = features[f"{obj}_center_y_norm"]
    w = features[f"{obj}_width_norm"]
    h = features[f"{obj}_height_norm"]

    canonical[f"{obj}_x1_norm"] = cx - w / 2
    canonical[f"{obj}_y1_norm"] = cy - h / 2
    canonical[f"{obj}_x2_norm"] = cx + w / 2
    canonical[f"{obj}_y2_norm"] = cy + h / 2
    canonical[f"{obj}_center_x_norm"] = cx

# Object-object distances
distance_cols = [
    "cotton_to_dropper_distance",
    "cotton_to_seed_jar_distance",
    "cotton_to_target_plate_distance",
    "cotton_to_water_dish_distance",
    "dropper_to_seed_jar_distance",
    "dropper_to_target_plate_distance",
    "dropper_to_water_dish_distance",
    "seed_jar_to_target_plate_distance",
    "seed_jar_to_water_dish_distance",
    "target_plate_to_water_dish_distance"
]

for col in distance_cols:
    canonical[col] = features[col]

# Hand landmarks
for hand in ["left", "right"]:
    for landmark in range(21):
        for coord in ["x", "y", "z"]:

            source_col = f"{hand}_lm{landmark}_{coord}"
            target_col = f"{hand}_hand_{landmark}_{coord}"

            canonical[target_col] = features[source_col]

# Hand-object distances
for hand in ["left", "right"]:
    for obj in objects:

        col = f"{hand}_hand_to_{obj}_distance"

        # Actual source column is:
        # left_hand_to_object_distance
        canonical[col] = features[col]

# ---------------------------------------------------------
# Select V024_1_0 frames 18–24
# ---------------------------------------------------------

segment = canonical[
    canonical["frame"].between(18, 24)
].copy()

print("=" * 90)
print("CANONICAL C5 TEST — V024_1_0")
print("=" * 90)

print("Rows:", len(segment))
print("Total columns including frame:", len(canonical))

model_columns = [
    col for col in canonical.columns
    if col != "frame"
]

print("Model features:", len(model_columns))

print(
    "NaN cells in complete canonical C5:",
    int(canonical[model_columns].isna().sum().sum())
)

print(
    "NaN cells in frames 18–24:",
    int(segment[model_columns].isna().sum().sum())
)

print("\nFrames:")
print(segment["frame"].tolist())

print("\nRepresentative values:")
print(
    segment[
        [
            "frame",
            "cotton_presence",
            "cotton_confidence",
            "cotton_x1",
            "cotton_y1",
            "cotton_x2",
            "cotton_y2",
            "cotton_center_x",
            "left_hand_hand_0_x",
            "left_hand_hand_0_y",
            "left_hand_hand_0_z",
            "right_hand_hand_0_x",
            "right_hand_hand_0_y",
            "right_hand_hand_0_z",
            "cotton_to_dropper_distance",
            "left_hand_to_cotton_distance"
        ]
    ].to_string(index=False)
)
```

```python
import os
import pandas as pd

FEATURE_PATH = (
    "/content/person2_output/all_videos/C5/"
    "C5_final_features_with_hand_object_distance.csv"
)

features = pd.read_csv(FEATURE_PATH)

objects = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# ---------------------------------------------------------
# Build canonical 206-feature dataframe
# ---------------------------------------------------------

data = {}

# Presence + confidence
for obj in objects:
    data[f"{obj}_presence"] = features[f"{obj}_presence"]
    data[f"{obj}_confidence"] = features[f"{obj}_confidence"]

# Raw geometry
for obj in objects:

    cx = features[f"{obj}_center_x"]
    cy = features[f"{obj}_center_y"]
    w = features[f"{obj}_width"]
    h = features[f"{obj}_height"]

    data[f"{obj}_x1"] = cx - w / 2
    data[f"{obj}_y1"] = cy - h / 2
    data[f"{obj}_x2"] = cx + w / 2
    data[f"{obj}_y2"] = cy + h / 2
    data[f"{obj}_center_x"] = cx

# Normalized geometry
for obj in objects:

    cx = features[f"{obj}_center_x_norm"]
    cy = features[f"{obj}_center_y_norm"]
    w = features[f"{obj}_width_norm"]
    h = features[f"{obj}_height_norm"]

    data[f"{obj}_x1_norm"] = cx - w / 2
    data[f"{obj}_y1_norm"] = cy - h / 2
    data[f"{obj}_x2_norm"] = cx + w / 2
    data[f"{obj}_y2_norm"] = cy + h / 2
    data[f"{obj}_center_x_norm"] = cx

# Object-object distances
distance_cols = [
    "cotton_to_dropper_distance",
    "cotton_to_seed_jar_distance",
    "cotton_to_target_plate_distance",
    "cotton_to_water_dish_distance",
    "dropper_to_seed_jar_distance",
    "dropper_to_target_plate_distance",
    "dropper_to_water_dish_distance",
    "seed_jar_to_target_plate_distance",
    "seed_jar_to_water_dish_distance",
    "target_plate_to_water_dish_distance"
]

for col in distance_cols:
    data[col] = features[col]

# Hand landmarks
for hand in ["left", "right"]:
    for landmark in range(21):
        for coord in ["x", "y", "z"]:

            source_col = f"{hand}_lm{landmark}_{coord}"
            target_col = f"{hand}_hand_{landmark}_{coord}"

            data[target_col] = features[source_col]

# Hand-object distances
for hand in ["left", "right"]:
    for obj in objects:

        col = f"{hand}_hand_to_{obj}_distance"
        data[col] = features[col]

# Create dataframe in one operation
canonical = pd.DataFrame(data)

canonical.insert(0, "frame", features["frame"])

model_columns = [
    col for col in canonical.columns
    if col != "frame"
]

# ---------------------------------------------------------
# Checks
# ---------------------------------------------------------

print("=" * 90)
print("CANONICAL C5 TEST — V024_1_0")
print("=" * 90)

print("Original C5 shape:", features.shape)
print("Canonical shape:", canonical.shape)

print("Model features:", len(model_columns))

print(
    "Original C5 total NaNs:",
    int(features.isna().sum().sum())
)

print(
    "Canonical total NaNs:",
    int(canonical[model_columns].isna().sum().sum())
)

# V024_1_0 = frames 18–24
segment = canonical[
    canonical["frame"].between(18, 24)
].copy()

print("\nFrames:", segment["frame"].tolist())

print(
    "Canonical NaNs in frames 18–24:",
    int(segment[model_columns].isna().sum().sum())
)

print("\nRepresentative values:")
print(
    segment[
        [
            "frame",
            "cotton_presence",
            "cotton_confidence",
            "cotton_x1",
            "cotton_y1",
            "cotton_x2",
            "cotton_y2",
            "cotton_center_x",
            "left_hand_0_x",
            "left_hand_0_y",
            "left_hand_0_z",
            "right_hand_0_x",
            "right_hand_0_y",
            "right_hand_0_z",
            "cotton_to_dropper_distance",
            "left_hand_to_cotton_distance"
        ]
    ].to_string(index=False)
)
```

```python
import os
import glob
import pandas as pd

FEATURE_ROOT = "/content/person2_output/all_videos"

OUTPUT_ROOT = "/content/person2_output/aligned_corrected"
os.makedirs(OUTPUT_ROOT, exist_ok=True)

META_FILES = {
    "train": "/content/train (1).csv",
    "val": "/content/val.csv",
    "test": "/content/test.csv"
}

OBJECTS = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# ---------------------------------------------------------
# Build feature-file map
# ---------------------------------------------------------

feature_files = glob.glob(
    os.path.join(
        FEATURE_ROOT,
        "*",
        "*_final_features_with_hand_object_distance.csv"
    )
)

feature_map = {}

for path in feature_files:

    filename = os.path.basename(path)

    stem = filename.replace(
        "_final_features_with_hand_object_distance.csv",
        ""
    ).lower()

    feature_map[stem] = path

print("Feature files found:", len(feature_map))


# ---------------------------------------------------------
# Convert one Person 2 feature file to canonical schema
# ---------------------------------------------------------

def make_canonical_features(features):

    data = {}

    # Presence + confidence
    for obj in OBJECTS:

        data[f"{obj}_presence"] = features[
            f"{obj}_presence"
        ]

        data[f"{obj}_confidence"] = features[
            f"{obj}_confidence"
        ]

    # Raw geometry
    #
    # Source:
    # center_x, center_y, width, height
    #
    # Canonical:
    # x1, y1, x2, y2, center_x

    for obj in OBJECTS:

        cx = features[f"{obj}_center_x"]
        cy = features[f"{obj}_center_y"]
        w = features[f"{obj}_width"]
        h = features[f"{obj}_height"]

        data[f"{obj}_x1"] = cx - w / 2
        data[f"{obj}_y1"] = cy - h / 2
        data[f"{obj}_x2"] = cx + w / 2
        data[f"{obj}_y2"] = cy + h / 2
        data[f"{obj}_center_x"] = cx

    # Normalized geometry
    for obj in OBJECTS:

        cx = features[f"{obj}_center_x_norm"]
        cy = features[f"{obj}_center_y_norm"]
        w = features[f"{obj}_width_norm"]
        h = features[f"{obj}_height_norm"]

        data[f"{obj}_x1_norm"] = cx - w / 2
        data[f"{obj}_y1_norm"] = cy - h / 2
        data[f"{obj}_x2_norm"] = cx + w / 2
        data[f"{obj}_y2_norm"] = cy + h / 2
        data[f"{obj}_center_x_norm"] = cx

    # Object-object distances
    distance_cols = [
        "cotton_to_dropper_distance",
        "cotton_to_seed_jar_distance",
        "cotton_to_target_plate_distance",
        "cotton_to_water_dish_distance",
        "dropper_to_seed_jar_distance",
        "dropper_to_target_plate_distance",
        "dropper_to_water_dish_distance",
        "seed_jar_to_target_plate_distance",
        "seed_jar_to_water_dish_distance",
        "target_plate_to_water_dish_distance"
    ]

    for col in distance_cols:
        data[col] = features[col]

    # Hand landmarks
    for hand in ["left", "right"]:

        for landmark in range(21):

            for coord in ["x", "y", "z"]:

                source_col = (
                    f"{hand}_lm{landmark}_{coord}"
                )

                target_col = (
                    f"{hand}_hand_{landmark}_{coord}"
                )

                data[target_col] = features[source_col]

    # Hand-object distances
    for hand in ["left", "right"]:

        for obj in OBJECTS:

            col = f"{hand}_hand_to_{obj}_distance"

            data[col] = features[col]

    canonical = pd.DataFrame(data)

    canonical.insert(
        0,
        "frame",
        features["frame"]
    )

    return canonical


# ---------------------------------------------------------
# Process each split
# ---------------------------------------------------------

for split, meta_path in META_FILES.items():

    print("\n" + "=" * 90)
    print("REBUILDING", split.upper())
    print("=" * 90)

    meta = pd.read_csv(meta_path)

    aligned_parts = []
    excluded_rows = []

    # Cache canonical video features so a video is converted
    # only once even when it has many temporal segments.
    canonical_cache = {}

    for row_idx, row in meta.iterrows():

        video_file = str(row["video_file"])

        video_stem = os.path.splitext(
            os.path.basename(video_file)
        )[0].lower()

        clip_id = str(row["clip_id"])

        start_frame = int(row["start_frame"])
        end_frame = int(row["end_frame"])

        segment_uid = (
            f"{split}_{row_idx:04d}_{clip_id}"
        )

        # ---------------------------------------------
        # Invalid temporal range
        # ---------------------------------------------

        if start_frame > end_frame:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "reason": "Invalid temporal range"
            })

            continue

        # ---------------------------------------------
        # Find Person 2 feature file
        # ---------------------------------------------

        feature_path = feature_map.get(video_stem)

        if feature_path is None:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "reason": "Person 2 feature file not found"
            })

            continue

        # ---------------------------------------------
        # Load / canonicalize video once
        # ---------------------------------------------

        if video_stem not in canonical_cache:

            source = pd.read_csv(feature_path)

            canonical_cache[video_stem] = (
                make_canonical_features(source)
            )

        features = canonical_cache[video_stem]

        # ---------------------------------------------
        # Extract temporal segment
        # ---------------------------------------------

        segment = features[
            (features["frame"] >= start_frame) &
            (features["frame"] <= end_frame)
        ].copy()

        expected_frames = (
            end_frame - start_frame + 1
        )

        actual_frames = len(segment)

        if actual_frames != expected_frames:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "expected_frames": expected_frames,
                "actual_frames": actual_frames,
                "reason": "Frame count mismatch"
            })

            continue

        # ---------------------------------------------
        # Add metadata
        # ---------------------------------------------

        segment["segment_uid"] = segment_uid
        segment["clip_id"] = clip_id
        segment["video_id"] = row["video_id"]
        segment["activity"] = row["activity"]
        segment["label"] = row["label"]
        segment["step_id"] = row["step_id"]
        segment["behavior"] = row["behavior"]
        segment["variation"] = row["variation"]
        segment["metadata_row"] = row_idx

        aligned_parts.append(segment)

    # -------------------------------------------------
    # Combine
    # -------------------------------------------------

    aligned = pd.concat(
        aligned_parts,
        ignore_index=True
    )

    excluded_df = pd.DataFrame(excluded_rows)

    aligned_path = os.path.join(
        OUTPUT_ROOT,
        f"{split}_aligned_corrected.csv"
    )

    excluded_path = os.path.join(
        OUTPUT_ROOT,
        f"{split}_excluded.csv"
    )

    aligned.to_csv(
        aligned_path,
        index=False
    )

    # Even if empty, create a useful excluded CSV
    if len(excluded_df) == 0:
        excluded_df = pd.DataFrame(columns=[
            "split",
            "metadata_row",
            "segment_uid",
            "clip_id",
            "video_id",
            "video_file",
            "start_frame",
            "end_frame",
            "reason"
        ])

    excluded_df.to_csv(
        excluded_path,
        index=False
    )

    # -------------------------------------------------
    # Validation
    # -------------------------------------------------

    feature_columns = [
        c for c in aligned.columns
        if c not in [
            "frame",
            "segment_uid",
            "clip_id",
            "video_id",
            "activity",
            "label",
            "step_id",
            "behavior",
            "variation",
            "metadata_row"
        ]
    ]

    print("Original metadata rows :", len(meta))
    print("Aligned frame rows     :", len(aligned))
    print("Unique segments        :", aligned["segment_uid"].nunique())
    print("Excluded rows           :", len(excluded_df))
    print("Total columns           :", len(aligned.columns))
    print("Model features          :", len(feature_columns))
    print(
        "Missing feature cells   :",
        int(aligned[feature_columns].isna().sum().sum())
    )

    print("\nSaved:")
    print(aligned_path)
    print(excluded_path)

print("\n" + "=" * 90)
print("CORRECTED ALIGNMENT COMPLETE")
print("=" * 90)
```

```python
import os
import glob
import pandas as pd

FEATURE_ROOT = "/content/person2_output/all_videos"
META_FILES = {
    "train": "/content/train (1).csv",
    "val": "/content/val.csv",
    "test": "/content/test.csv"
}

OUTPUT_ROOT = "/content/person2_output/aligned_canonical"
os.makedirs(OUTPUT_ROOT, exist_ok=True)

# ------------------------------------------------------------
# 1. Define the EXACT 206-feature schema
# ------------------------------------------------------------

objects = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

presence_features = [
    f"{obj}_presence" for obj in objects
]

confidence_features = [
    f"{obj}_confidence" for obj in objects
]

raw_geometry_features = []

for obj in objects:
    raw_geometry_features += [
        f"{obj}_x1",
        f"{obj}_y1",
        f"{obj}_x2",
        f"{obj}_y2",
        f"{obj}_center_x"
    ]

normalized_geometry_features = []

for obj in objects:
    normalized_geometry_features += [
        f"{obj}_x1_norm",
        f"{obj}_y1_norm",
        f"{obj}_x2_norm",
        f"{obj}_y2_norm",
        f"{obj}_center_x_norm"
    ]

object_object_features = []

for i in range(len(objects)):
    for j in range(i + 1, len(objects)):
        object_object_features.append(
            f"{objects[i]}_to_{objects[j]}_distance"
        )

hand_landmark_features = []

for hand in ["left_hand", "right_hand"]:
    for landmark in range(21):
        for coord in ["x", "y", "z"]:
            hand_landmark_features.append(
                f"{hand}_{landmark}_{coord}"
            )

hand_object_features = []

for hand in ["left_hand", "right_hand"]:
    for obj in objects:
        hand_object_features.append(
            f"{hand}_to_{obj}_distance"
        )

FEATURE_COLUMNS = (
    presence_features
    + confidence_features
    + raw_geometry_features
    + normalized_geometry_features
    + object_object_features
    + hand_landmark_features
    + hand_object_features
)

assert len(FEATURE_COLUMNS) == 206

print("206-feature schema confirmed.")


# ------------------------------------------------------------
# 2. Build feature-file map
# ------------------------------------------------------------

feature_files = glob.glob(
    os.path.join(
        FEATURE_ROOT,
        "*",
        "*_final_features_with_hand_object_distance.csv"
    )
)

feature_map = {}

for path in feature_files:

    filename = os.path.basename(path)

    stem = filename.replace(
        "_final_features_with_hand_object_distance.csv",
        ""
    )

    feature_map[stem.lower()] = path

print("Feature files found:", len(feature_map))


# ------------------------------------------------------------
# 3. Process each split
# ------------------------------------------------------------

for split, meta_path in META_FILES.items():

    print("\n" + "=" * 90)
    print(f"REBUILDING {split.upper()}")
    print("=" * 90)

    meta = pd.read_csv(meta_path)

    aligned_parts = []
    excluded_rows = []

    # Cache each video after canonical extraction
    canonical_cache = {}

    for row_idx, row in meta.iterrows():

        video_file = str(row["video_file"])

        video_stem = os.path.splitext(
            os.path.basename(video_file)
        )[0].lower()

        clip_id = str(row["clip_id"])

        start_frame = int(row["start_frame"])
        end_frame = int(row["end_frame"])

        segment_uid = (
            f"{split}_{row_idx:04d}_{clip_id}"
        )

        # ----------------------------------------------------
        # Invalid temporal range
        # ----------------------------------------------------

        if start_frame > end_frame:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "reason": "Invalid temporal range"
            })

            continue

        # ----------------------------------------------------
        # Locate feature file
        # ----------------------------------------------------

        feature_path = feature_map.get(video_stem)

        if feature_path is None:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "reason": "Feature file not found"
            })

            continue

        # ----------------------------------------------------
        # Load canonical 206 features ONCE per video
        # ----------------------------------------------------

        if video_stem not in canonical_cache:

            source = pd.read_csv(feature_path)

            missing_columns = [
                col for col in FEATURE_COLUMNS
                if col not in source.columns
            ]

            if missing_columns:

                raise ValueError(
                    f"{video_stem} is missing canonical "
                    f"features: {missing_columns}"
                )

            canonical = source[
                ["frame"] + FEATURE_COLUMNS
            ].copy()

            # Verify source itself is clean
            nan_count = int(
                canonical[FEATURE_COLUMNS]
                .isna()
                .sum()
                .sum()
            )

            if nan_count != 0:

                raise ValueError(
                    f"{video_stem} contains "
                    f"{nan_count} NaN cells in the "
                    f"206 canonical features."
                )

            canonical_cache[video_stem] = canonical

        else:

            canonical = canonical_cache[video_stem]

        # ----------------------------------------------------
        # Extract temporal segment
        # ----------------------------------------------------

        segment = canonical[
            (canonical["frame"] >= start_frame) &
            (canonical["frame"] <= end_frame)
        ].copy()

        expected_frames = (
            end_frame - start_frame + 1
        )

        actual_frames = len(segment)

        # ----------------------------------------------------
        # Verify exact frame coverage
        # ----------------------------------------------------

        if actual_frames != expected_frames:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "expected_frames": expected_frames,
                "actual_frames": actual_frames,
                "reason": "Frame count mismatch"
            })

            continue

        # ----------------------------------------------------
        # Add metadata
        # ----------------------------------------------------

        segment["segment_uid"] = segment_uid
        segment["clip_id"] = clip_id
        segment["video_id"] = row["video_id"]
        segment["activity"] = row["activity"]
        segment["label"] = row["label"]
        segment["step_id"] = row["step_id"]
        segment["behavior"] = row["behavior"]
        segment["variation"] = row["variation"]
        segment["metadata_row"] = row_idx

        aligned_parts.append(segment)

    # --------------------------------------------------------
    # Combine
    # --------------------------------------------------------

    aligned = pd.concat(
        aligned_parts,
        ignore_index=True
    )

    # --------------------------------------------------------
    # Final validation
    # --------------------------------------------------------

    model_df = aligned[FEATURE_COLUMNS]

    total_nan = int(
        model_df.isna().sum().sum()
    )

    duplicate_frames = int(
        aligned.duplicated(
            subset=["segment_uid", "frame"]
        ).sum()
    )

    unique_segments = aligned[
        "segment_uid"
    ].nunique()

    print("\nRESULT")
    print("-" * 90)
    print("Original metadata rows :", len(meta))
    print("Aligned segment rows   :", unique_segments)
    print("Aligned frame rows     :", len(aligned))
    print("Excluded metadata rows :", len(excluded_rows))
    print("Model features         :", len(FEATURE_COLUMNS))
    print("NaN cells in features  :", total_nan)
    print("Duplicate segment/frame:", duplicate_frames)

    if total_nan != 0:
        raise ValueError(
            f"{split}: canonical alignment "
            f"still contains NaNs."
        )

    if duplicate_frames != 0:
        raise ValueError(
            f"{split}: duplicate segment/frame rows found."
        )

    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------

    output_path = os.path.join(
        OUTPUT_ROOT,
        f"{split}_aligned_canonical.csv"
    )

    aligned.to_csv(
        output_path,
        index=False
    )

    excluded_path = os.path.join(
        OUTPUT_ROOT,
        f"{split}_excluded.csv"
    )

    pd.DataFrame(excluded_rows).to_csv(
        excluded_path,
        index=False
    )

    print("Saved:")
    print(output_path)
    print(excluded_path)

print("\n" + "=" * 90)
print("CANONICAL ALIGNMENT COMPLETE")
print("=" * 90)
```

```python
import pandas as pd

FEATURE_PATH = (
    "/content/person2_output/all_videos/C5/"
    "C5_final_features_with_hand_object_distance.csv"
)

features = pd.read_csv(FEATURE_PATH)

objects = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

data = {}

# ---------------------------------------------------------
# Presence + confidence
# ---------------------------------------------------------

for obj in objects:
    data[f"{obj}_presence"] = features[f"{obj}_presence"]
    data[f"{obj}_confidence"] = features[f"{obj}_confidence"]


# ---------------------------------------------------------
# Raw geometry
# ---------------------------------------------------------

for obj in objects:

    cx = features[f"{obj}_center_x"]
    cy = features[f"{obj}_center_y"]
    w = features[f"{obj}_width"]
    h = features[f"{obj}_height"]

    data[f"{obj}_x1"] = cx - w / 2
    data[f"{obj}_y1"] = cy - h / 2
    data[f"{obj}_x2"] = cx + w / 2
    data[f"{obj}_y2"] = cy + h / 2
    data[f"{obj}_center_x"] = cx


# ---------------------------------------------------------
# Normalized geometry
# ---------------------------------------------------------

for obj in objects:

    cx = features[f"{obj}_center_x_norm"]
    cy = features[f"{obj}_center_y_norm"]
    w = features[f"{obj}_width_norm"]
    h = features[f"{obj}_height_norm"]

    data[f"{obj}_x1_norm"] = cx - w / 2
    data[f"{obj}_y1_norm"] = cy - h / 2
    data[f"{obj}_x2_norm"] = cx + w / 2
    data[f"{obj}_y2_norm"] = cy + h / 2
    data[f"{obj}_center_x_norm"] = cx


# ---------------------------------------------------------
# Object-object distances
# ---------------------------------------------------------

distance_cols = [
    "cotton_to_dropper_distance",
    "cotton_to_seed_jar_distance",
    "cotton_to_target_plate_distance",
    "cotton_to_water_dish_distance",
    "dropper_to_seed_jar_distance",
    "dropper_to_target_plate_distance",
    "dropper_to_water_dish_distance",
    "seed_jar_to_target_plate_distance",
    "seed_jar_to_water_dish_distance",
    "target_plate_to_water_dish_distance"
]

for col in distance_cols:
    data[col] = features[col]


# ---------------------------------------------------------
# Hand landmarks
# ---------------------------------------------------------

for hand in ["left", "right"]:

    for landmark in range(21):

        for coord in ["x", "y", "z"]:

            source_col = (
                f"{hand}_lm{landmark}_{coord}"
            )

            target_col = (
                f"{hand}_hand_{landmark}_{coord}"
            )

            data[target_col] = features[source_col]


# ---------------------------------------------------------
# Hand-object distances
# ---------------------------------------------------------

for hand in ["left", "right"]:

    for obj in objects:

        source_col = (
            f"{hand}_hand_to_{obj}_distance"
        )

        target_col = (
            f"{hand}_hand_to_{obj}_distance"
        )

        data[target_col] = features[source_col]


# ---------------------------------------------------------
# Build canonical dataframe
# ---------------------------------------------------------

canonical = pd.DataFrame(data)

canonical.insert(
    0,
    "frame",
    features["frame"]
)

model_columns = [
    c for c in canonical.columns
    if c != "frame"
]


# ---------------------------------------------------------
# Validation
# ---------------------------------------------------------

print("=" * 90)
print("C5 CANONICAL CONVERSION — FINAL TEST")
print("=" * 90)

print("Source shape       :", features.shape)
print("Canonical shape    :", canonical.shape)
print("Model features     :", len(model_columns))
print("Source NaNs        :", int(features.isna().sum().sum()))
print("Canonical NaNs     :", int(
    canonical[model_columns].isna().sum().sum()
))

print("\nExpected:")
print("Total columns      : 207")
print("Model features     : 206")
print("Canonical NaNs     : 0")

# V024_1_0
segment = canonical[
    canonical["frame"].between(18, 24)
].copy()

print("\nV024_1_0")
print("Frames             :", segment["frame"].tolist())
print("Frames found       :", len(segment))
print("Expected frames    : 7")
print("Segment NaNs       :", int(
    segment[model_columns].isna().sum().sum()
))

# Check duplicate frames
print(
    "Duplicate frames   :",
    int(canonical["frame"].duplicated().sum())
)

print("\nFirst 15 canonical columns:")
print(canonical.columns[:15].tolist())

print("\nLast 10 canonical columns:")
print(canonical.columns[-10:].tolist())
```

```python
import os
import glob
import pandas as pd

FEATURE_ROOT = "/content/person2_output/all_videos"
OUTPUT_ROOT = "/content/person2_output/aligned_canonical"

os.makedirs(OUTPUT_ROOT, exist_ok=True)

META_FILES = {
    "train": "/content/train (1).csv",
    "val": "/content/val.csv",
    "test": "/content/test.csv"
}

OBJECTS = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# ============================================================
# 1. EXACT CANONICAL 206-FEATURE SCHEMA
# ============================================================

presence_features = [
    f"{obj}_presence"
    for obj in OBJECTS
]

confidence_features = [
    f"{obj}_confidence"
    for obj in OBJECTS
]

raw_geometry_features = []

for obj in OBJECTS:
    raw_geometry_features += [
        f"{obj}_x1",
        f"{obj}_y1",
        f"{obj}_x2",
        f"{obj}_y2",
        f"{obj}_center_x"
    ]

normalized_geometry_features = []

for obj in OBJECTS:
    normalized_geometry_features += [
        f"{obj}_x1_norm",
        f"{obj}_y1_norm",
        f"{obj}_x2_norm",
        f"{obj}_y2_norm",
        f"{obj}_center_x_norm"
    ]

object_object_features = []

for i in range(len(OBJECTS)):
    for j in range(i + 1, len(OBJECTS)):
        object_object_features.append(
            f"{OBJECTS[i]}_to_{OBJECTS[j]}_distance"
        )

hand_landmark_features = []

for hand in ["left_hand", "right_hand"]:
    for landmark in range(21):
        for coord in ["x", "y", "z"]:
            hand_landmark_features.append(
                f"{hand}_{landmark}_{coord}"
            )

hand_object_features = []

for hand in ["left_hand", "right_hand"]:
    for obj in OBJECTS:
        hand_object_features.append(
            f"{hand}_to_{obj}_distance"
        )

FEATURE_COLUMNS = (
    presence_features
    + confidence_features
    + raw_geometry_features
    + normalized_geometry_features
    + object_object_features
    + hand_landmark_features
    + hand_object_features
)

assert len(FEATURE_COLUMNS) == 206

print("Canonical model features:", len(FEATURE_COLUMNS))


# ============================================================
# 2. BUILD PERSON 2 FEATURE FILE MAP
# ============================================================

feature_files = glob.glob(
    os.path.join(
        FEATURE_ROOT,
        "*",
        "*_final_features_with_hand_object_distance.csv"
    )
)

feature_map = {}

for path in feature_files:

    filename = os.path.basename(path)

    stem = filename.replace(
        "_final_features_with_hand_object_distance.csv",
        ""
    ).lower()

    feature_map[stem] = path

print("Feature files found:", len(feature_map))

if len(feature_map) != 117:
    raise ValueError(
        f"Expected 117 feature files, found {len(feature_map)}"
    )


# ============================================================
# 3. FUNCTION: SOURCE → CANONICAL
# ============================================================

def convert_to_canonical(source):

    data = {}

    # --------------------------------------------------------
    # Presence + confidence
    # --------------------------------------------------------

    for obj in OBJECTS:

        data[f"{obj}_presence"] = source[
            f"{obj}_presence"
        ]

        data[f"{obj}_confidence"] = source[
            f"{obj}_confidence"
        ]

    # --------------------------------------------------------
    # Raw geometry
    #
    # Source:
    # center_x, center_y, width, height
    #
    # Canonical:
    # x1, y1, x2, y2, center_x
    # --------------------------------------------------------

    for obj in OBJECTS:

        cx = source[f"{obj}_center_x"]
        cy = source[f"{obj}_center_y"]
        w = source[f"{obj}_width"]
        h = source[f"{obj}_height"]

        data[f"{obj}_x1"] = cx - w / 2
        data[f"{obj}_y1"] = cy - h / 2
        data[f"{obj}_x2"] = cx + w / 2
        data[f"{obj}_y2"] = cy + h / 2
        data[f"{obj}_center_x"] = cx

    # --------------------------------------------------------
    # Normalized geometry
    # --------------------------------------------------------

    for obj in OBJECTS:

        cx = source[f"{obj}_center_x_norm"]
        cy = source[f"{obj}_center_y_norm"]
        w = source[f"{obj}_width_norm"]
        h = source[f"{obj}_height_norm"]

        data[f"{obj}_x1_norm"] = cx - w / 2
        data[f"{obj}_y1_norm"] = cy - h / 2
        data[f"{obj}_x2_norm"] = cx + w / 2
        data[f"{obj}_y2_norm"] = cy + h / 2
        data[f"{obj}_center_x_norm"] = cx

    # --------------------------------------------------------
    # Object-object distances
    # --------------------------------------------------------

    for col in object_object_features:
        data[col] = source[col]

    # --------------------------------------------------------
    # Hand landmarks
    #
    # Source:
    # left_lm0_x
    #
    # Canonical:
    # left_hand_0_x
    # --------------------------------------------------------

    for hand in ["left", "right"]:

        for landmark in range(21):

            for coord in ["x", "y", "z"]:

                source_col = (
                    f"{hand}_lm{landmark}_{coord}"
                )

                target_col = (
                    f"{hand}_hand_{landmark}_{coord}"
                )

                data[target_col] = source[source_col]

    # --------------------------------------------------------
    # Hand-object distances
    # --------------------------------------------------------

    for hand in ["left_hand", "right_hand"]:

        for obj in OBJECTS:

            col = (
                f"{hand}_to_{obj}_distance"
            )

            data[col] = source[col]

    # --------------------------------------------------------
    # Final dataframe
    # --------------------------------------------------------

    canonical = pd.DataFrame(data)

    canonical.insert(
        0,
        "frame",
        source["frame"]
    )

    return canonical


# ============================================================
# 4. PROCESS TRAIN / VAL / TEST
# ============================================================

for split, meta_path in META_FILES.items():

    print("\n" + "=" * 90)
    print(f"REBUILDING {split.upper()}")
    print("=" * 90)

    meta = pd.read_csv(meta_path)

    aligned_parts = []
    excluded_rows = []

    # Cache canonical data for each video
    canonical_cache = {}

    for row_idx, row in meta.iterrows():

        video_file = str(row["video_file"])

        video_stem = os.path.splitext(
            os.path.basename(video_file)
        )[0].lower()

        clip_id = str(row["clip_id"])

        start_frame = int(row["start_frame"])
        end_frame = int(row["end_frame"])

        segment_uid = (
            f"{split}_{row_idx:04d}_{clip_id}"
        )

        # ----------------------------------------------------
        # Invalid temporal range
        # ----------------------------------------------------

        if start_frame > end_frame:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "reason": "Invalid temporal range"
            })

            continue

        # ----------------------------------------------------
        # Locate Person 2 file
        # ----------------------------------------------------

        feature_path = feature_map.get(video_stem)

        if feature_path is None:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "reason": "Person 2 feature file not found"
            })

            continue

        # ----------------------------------------------------
        # Canonicalize video once
        # ----------------------------------------------------

        if video_stem not in canonical_cache:

            source = pd.read_csv(feature_path)

            canonical = convert_to_canonical(source)

            # Verify exact structure
            expected_columns = [
                "frame"
            ] + FEATURE_COLUMNS

            if list(canonical.columns) != expected_columns:

                raise ValueError(
                    f"{video_stem}: canonical column "
                    f"structure mismatch."
                )

            # Verify source/canonical are clean
            source_nan = int(
                source.isna().sum().sum()
            )

            canonical_nan = int(
                canonical[FEATURE_COLUMNS]
                .isna()
                .sum()
                .sum()
            )

            if source_nan != 0:

                raise ValueError(
                    f"{video_stem}: source has "
                    f"{source_nan} NaN cells."
                )

            if canonical_nan != 0:

                raise ValueError(
                    f"{video_stem}: canonical has "
                    f"{canonical_nan} NaN cells."
                )

            canonical_cache[video_stem] = canonical

        canonical = canonical_cache[video_stem]

        # ----------------------------------------------------
        # Extract exact temporal range
        # ----------------------------------------------------

        segment = canonical[
            (canonical["frame"] >= start_frame) &
            (canonical["frame"] <= end_frame)
        ].copy()

        expected_frames = (
            end_frame - start_frame + 1
        )

        actual_frames = len(segment)

        # ----------------------------------------------------
        # Exact frame-count validation
        # ----------------------------------------------------

        if actual_frames != expected_frames:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "expected_frames": expected_frames,
                "actual_frames": actual_frames,
                "reason": "Frame count mismatch"
            })

            continue

        # ----------------------------------------------------
        # Verify no duplicate frames
        # ----------------------------------------------------

        if segment["frame"].duplicated().any():

            raise ValueError(
                f"{segment_uid}: duplicate frames detected."
            )

        # ----------------------------------------------------
        # Add metadata
        # ----------------------------------------------------

        segment["segment_uid"] = segment_uid
        segment["clip_id"] = clip_id
        segment["video_id"] = row["video_id"]
        segment["activity"] = row["activity"]
        segment["label"] = row["label"]
        segment["step_id"] = row["step_id"]
        segment["behavior"] = row["behavior"]
        segment["variation"] = row["variation"]
        segment["metadata_row"] = row_idx

        aligned_parts.append(segment)

    # ========================================================
    # COMBINE
    # ========================================================

    aligned = pd.concat(
        aligned_parts,
        ignore_index=True
    )

    excluded_df = pd.DataFrame(excluded_rows)

    # ========================================================
    # FINAL VALIDATION
    # ========================================================

    total_nan = int(
        aligned[FEATURE_COLUMNS]
        .isna()
        .sum()
        .sum()
    )

    duplicate_segment_frames = int(
        aligned.duplicated(
            subset=["segment_uid", "frame"]
        ).sum()
    )

    unique_segments = aligned[
        "segment_uid"
    ].nunique()

    # Segment lengths
    segment_lengths = (
        aligned
        .groupby("segment_uid")["frame"]
        .count()
    )

    print("\nRESULT")
    print("-" * 90)

    print(
        "Original metadata rows :",
        len(meta)
    )

    print(
        "Aligned segments       :",
        unique_segments
    )

    print(
        "Aligned frame rows     :",
        len(aligned)
    )

    print(
        "Excluded metadata rows :",
        len(excluded_rows)
    )

    print(
        "Total columns          :",
        len(aligned.columns)
    )

    print(
        "Model features         :",
        len(FEATURE_COLUMNS)
    )

    print(
        "Missing feature cells  :",
        total_nan
    )

    print(
        "Duplicate segment/frame:",
        duplicate_segment_frames
    )

    print(
        "Segment length min/max :",
        segment_lengths.min(),
        "/",
        segment_lengths.max()
    )

    print(
        "Segments exactly 32    :",
        int((segment_lengths == 32).sum())
    )

    print(
        "Segments shorter than32:",
        int((segment_lengths < 32).sum())
    )

    print(
        "Segments longer than32 :",
        int((segment_lengths > 32).sum())
    )

    # --------------------------------------------------------
    # Hard validation
    # --------------------------------------------------------

    if total_nan != 0:
        raise ValueError(
            f"{split}: NaNs detected."
        )

    if duplicate_segment_frames != 0:
        raise ValueError(
            f"{split}: duplicate segment/frame rows."
        )

    if len(FEATURE_COLUMNS) != 206:
        raise ValueError(
            "Canonical feature count is not 206."
        )

    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------

    aligned_path = os.path.join(
        OUTPUT_ROOT,
        f"{split}_aligned_canonical.csv"
    )

    excluded_path = os.path.join(
        OUTPUT_ROOT,
        f"{split}_excluded.csv"
    )

    aligned.to_csv(
        aligned_path,
        index=False
    )

    excluded_df.to_csv(
        excluded_path,
        index=False
    )

    print("\nSaved:")
    print(aligned_path)
    print(excluded_path)


print("\n" + "=" * 90)
print("ALL CANONICAL ALIGNMENT COMPLETE")
print("=" * 90)
```

```python
import pandas as pd

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

FEATURE_PATH = (
    "/content/person2_output/all_videos/C5/"
    "C5_final_features_with_hand_object_distance.csv"
)

META_PATH = "/content/train (1).csv"


# ------------------------------------------------------------
# Load
# ------------------------------------------------------------

features = pd.read_csv(FEATURE_PATH)
meta = pd.read_csv(META_PATH)

print("C5 source shape:", features.shape)


# ------------------------------------------------------------
# Build canonical features EXACTLY as validated
# ------------------------------------------------------------

objects = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

data = {}

# Presence + confidence
for obj in objects:
    data[f"{obj}_presence"] = features[f"{obj}_presence"]
    data[f"{obj}_confidence"] = features[f"{obj}_confidence"]

# Raw geometry
for obj in objects:

    cx = features[f"{obj}_center_x"]
    cy = features[f"{obj}_center_y"]
    w = features[f"{obj}_width"]
    h = features[f"{obj}_height"]

    data[f"{obj}_x1"] = cx - w / 2
    data[f"{obj}_y1"] = cy - h / 2
    data[f"{obj}_x2"] = cx + w / 2
    data[f"{obj}_y2"] = cy + h / 2
    data[f"{obj}_center_x"] = cx

# Normalized geometry
for obj in objects:

    cx = features[f"{obj}_center_x_norm"]
    cy = features[f"{obj}_center_y_norm"]
    w = features[f"{obj}_width_norm"]
    h = features[f"{obj}_height_norm"]

    data[f"{obj}_x1_norm"] = cx - w / 2
    data[f"{obj}_y1_norm"] = cy - h / 2
    data[f"{obj}_x2_norm"] = cx + w / 2
    data[f"{obj}_y2_norm"] = cy + h / 2
    data[f"{obj}_center_x_norm"] = cx

# Object-object distances
distance_cols = [
    "cotton_to_dropper_distance",
    "cotton_to_seed_jar_distance",
    "cotton_to_target_plate_distance",
    "cotton_to_water_dish_distance",
    "dropper_to_seed_jar_distance",
    "dropper_to_target_plate_distance",
    "dropper_to_water_dish_distance",
    "seed_jar_to_target_plate_distance",
    "seed_jar_to_water_dish_distance",
    "target_plate_to_water_dish_distance"
]

for col in distance_cols:
    data[col] = features[col]

# Hand landmarks
for hand in ["left", "right"]:

    for landmark in range(21):

        for coord in ["x", "y", "z"]:

            source_col = f"{hand}_lm{landmark}_{coord}"
            target_col = f"{hand}_hand_{landmark}_{coord}"

            data[target_col] = features[source_col]

# Hand-object distances
for hand in ["left", "right"]:

    for obj in objects:

        col = f"{hand}_hand_to_{obj}_distance"

        data[col] = features[col]


canonical = pd.DataFrame(data)

canonical.insert(
    0,
    "frame",
    features["frame"]
)

model_columns = [
    c for c in canonical.columns
    if c != "frame"
]

print("Canonical shape:", canonical.shape)
print("Model features:", len(model_columns))
print("Canonical NaNs:", int(
    canonical[model_columns].isna().sum().sum()
))


# ------------------------------------------------------------
# Find ALL TRAIN metadata rows belonging to C5
# ------------------------------------------------------------

c5_meta = meta[
    meta["video_file"].astype(str)
       .apply(lambda x: x.lower().endswith("c5.mp4"))
].copy()

print("\nC5 metadata rows:", len(c5_meta))

print(
    c5_meta[
        [
            "clip_id",
            "video_id",
            "video_file",
            "start_frame",
            "end_frame",
            "label"
        ]
    ].to_string(index=False)
)


# ------------------------------------------------------------
# Align C5 metadata rows
# ------------------------------------------------------------

aligned_parts = []

for row_idx, row in c5_meta.iterrows():

    start = int(row["start_frame"])
    end = int(row["end_frame"])

    if start > end:
        print(
            "\nEXCLUDED:",
            row["clip_id"],
            start,
            end
        )
        continue

    segment = canonical[
        canonical["frame"].between(start, end)
    ].copy()

    expected = end - start + 1

    print(
        f"\n{row['clip_id']}: "
        f"{start}-{end} | "
        f"expected={expected}, "
        f"actual={len(segment)}"
    )

    if len(segment) != expected:
        print("  ❌ FRAME COUNT MISMATCH")
        continue

    segment["segment_uid"] = (
        f"train_{row_idx:04d}_{row['clip_id']}"
    )

    segment["clip_id"] = row["clip_id"]
    segment["video_id"] = row["video_id"]
    segment["activity"] = row["activity"]
    segment["label"] = row["label"]
    segment["step_id"] = row["step_id"]
    segment["behavior"] = row["behavior"]
    segment["variation"] = row["variation"]
    segment["metadata_row"] = row_idx

    aligned_parts.append(segment)


# ------------------------------------------------------------
# Combine C5
# ------------------------------------------------------------

c5_aligned = pd.concat(
    aligned_parts,
    ignore_index=True
)

print("\n" + "=" * 90)
print("C5 ALIGNMENT TEST RESULT")
print("=" * 90)

print("Aligned rows       :", len(c5_aligned))
print("Aligned segments   :", c5_aligned["segment_uid"].nunique())
print("Total columns      :", len(c5_aligned.columns))
print("Model features     :", len(model_columns))
print("Feature NaNs       :", int(
    c5_aligned[model_columns].isna().sum().sum()
))
print("Duplicate seg/frame:",
      int(
          c5_aligned.duplicated(
              subset=["segment_uid", "frame"]
          ).sum()
      )
)

print("\nExpected feature NaNs: 0")
print("Expected duplicate seg/frame: 0")
```

```python
import os
import glob
import pandas as pd

FEATURE_ROOT = "/content/person2_output/all_videos"
OUTPUT_ROOT = "/content/person2_output/aligned_canonical"

os.makedirs(OUTPUT_ROOT, exist_ok=True)

META_FILES = {
    "train": "/content/train (1).csv",
    "val": "/content/val.csv",
    "test": "/content/test.csv"
}

OBJECTS = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# ============================================================
# BUILD FEATURE FILE MAP
# ============================================================

feature_files = glob.glob(
    os.path.join(
        FEATURE_ROOT,
        "*",
        "*_final_features_with_hand_object_distance.csv"
    )
)

feature_map = {}

for path in feature_files:

    filename = os.path.basename(path)

    stem = filename.replace(
        "_final_features_with_hand_object_distance.csv",
        ""
    ).lower()

    feature_map[stem] = path

print("Feature files found:", len(feature_map))


# ============================================================
# EXACT CONVERSION THAT PASSED C5 TEST
# ============================================================

def convert_to_canonical(features):

    data = {}

    # --------------------------------------------------------
    # Presence + confidence
    # --------------------------------------------------------

    for obj in OBJECTS:

        data[f"{obj}_presence"] = features[
            f"{obj}_presence"
        ]

        data[f"{obj}_confidence"] = features[
            f"{obj}_confidence"
        ]

    # --------------------------------------------------------
    # Raw geometry
    # --------------------------------------------------------

    for obj in OBJECTS:

        cx = features[f"{obj}_center_x"]
        cy = features[f"{obj}_center_y"]
        w = features[f"{obj}_width"]
        h = features[f"{obj}_height"]

        data[f"{obj}_x1"] = cx - w / 2
        data[f"{obj}_y1"] = cy - h / 2
        data[f"{obj}_x2"] = cx + w / 2
        data[f"{obj}_y2"] = cy + h / 2
        data[f"{obj}_center_x"] = cx

    # --------------------------------------------------------
    # Normalized geometry
    # --------------------------------------------------------

    for obj in OBJECTS:

        cx = features[f"{obj}_center_x_norm"]
        cy = features[f"{obj}_center_y_norm"]
        w = features[f"{obj}_width_norm"]
        h = features[f"{obj}_height_norm"]

        data[f"{obj}_x1_norm"] = cx - w / 2
        data[f"{obj}_y1_norm"] = cy - h / 2
        data[f"{obj}_x2_norm"] = cx + w / 2
        data[f"{obj}_y2_norm"] = cy + h / 2
        data[f"{obj}_center_x_norm"] = cx

    # --------------------------------------------------------
    # Object-object distances
    # --------------------------------------------------------

    distance_cols = [
        "cotton_to_dropper_distance",
        "cotton_to_seed_jar_distance",
        "cotton_to_target_plate_distance",
        "cotton_to_water_dish_distance",
        "dropper_to_seed_jar_distance",
        "dropper_to_target_plate_distance",
        "dropper_to_water_dish_distance",
        "seed_jar_to_target_plate_distance",
        "seed_jar_to_water_dish_distance",
        "target_plate_to_water_dish_distance"
    ]

    for col in distance_cols:
        data[col] = features[col]

    # --------------------------------------------------------
    # Hand landmarks
    # --------------------------------------------------------

    for hand in ["left", "right"]:

        for landmark in range(21):

            for coord in ["x", "y", "z"]:

                source_col = (
                    f"{hand}_lm{landmark}_{coord}"
                )

                target_col = (
                    f"{hand}_hand_{landmark}_{coord}"
                )

                data[target_col] = features[source_col]

    # --------------------------------------------------------
    # Hand-object distances
    # --------------------------------------------------------

    for hand in ["left", "right"]:

        for obj in OBJECTS:

            col = (
                f"{hand}_hand_to_{obj}_distance"
            )

            data[col] = features[col]

    # --------------------------------------------------------
    # Final canonical dataframe
    # --------------------------------------------------------

    canonical = pd.DataFrame(data)

    canonical.insert(
        0,
        "frame",
        features["frame"]
    )

    return canonical


# ============================================================
# BUILD EACH SPLIT
# ============================================================

for split, meta_path in META_FILES.items():

    print("\n" + "=" * 90)
    print(f"REBUILDING {split.upper()}")
    print("=" * 90)

    meta = pd.read_csv(meta_path)

    aligned_parts = []
    excluded_rows = []

    # Cache canonical video data
    canonical_cache = {}

    for row_idx, row in meta.iterrows():

        video_file = str(row["video_file"])

        video_stem = os.path.splitext(
            os.path.basename(video_file)
        )[0].lower()

        clip_id = str(row["clip_id"])

        start_frame = int(row["start_frame"])
        end_frame = int(row["end_frame"])

        segment_uid = (
            f"{split}_{row_idx:04d}_{clip_id}"
        )

        # ----------------------------------------------------
        # Invalid temporal range
        # ----------------------------------------------------

        if start_frame > end_frame:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "reason": "Invalid temporal range"
            })

            continue

        # ----------------------------------------------------
        # Find Person 2 feature file
        # ----------------------------------------------------

        feature_path = feature_map.get(video_stem)

        if feature_path is None:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "reason": "Feature file not found"
            })

            continue

        # ----------------------------------------------------
        # Canonicalize video once
        # ----------------------------------------------------

        if video_stem not in canonical_cache:

            source = pd.read_csv(feature_path)

            canonical = convert_to_canonical(source)

            # Expected:
            # frame + 206 features = 207 columns

            if canonical.shape[1] != 207:

                raise ValueError(
                    f"{video_stem}: expected 207 canonical "
                    f"columns, got {canonical.shape[1]}"
                )

            # Verify no NaNs in source
            source_nan = int(
                source.isna().sum().sum()
            )

            if source_nan != 0:

                raise ValueError(
                    f"{video_stem}: source contains "
                    f"{source_nan} NaN cells"
                )

            # Verify no NaNs after conversion
            canonical_nan = int(
                canonical.iloc[:, 1:]
                .isna()
                .sum()
                .sum()
            )

            if canonical_nan != 0:

                raise ValueError(
                    f"{video_stem}: canonical contains "
                    f"{canonical_nan} NaN cells"
                )

            canonical_cache[video_stem] = canonical

        canonical = canonical_cache[video_stem]

        # ----------------------------------------------------
        # Extract exact segment
        # ----------------------------------------------------

        segment = canonical[
            canonical["frame"].between(
                start_frame,
                end_frame
            )
        ].copy()

        expected_frames = (
            end_frame - start_frame + 1
        )

        actual_frames = len(segment)

        # ----------------------------------------------------
        # Exact frame validation
        # ----------------------------------------------------

        if actual_frames != expected_frames:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "expected_frames": expected_frames,
                "actual_frames": actual_frames,
                "reason": "Frame count mismatch"
            })

            continue

        # ----------------------------------------------------
        # No duplicate frames inside segment
        # ----------------------------------------------------

        if segment["frame"].duplicated().any():

            raise ValueError(
                f"{segment_uid}: duplicate frames detected"
            )

        # ----------------------------------------------------
        # Add Person 1 metadata
        # ----------------------------------------------------

        segment["segment_uid"] = segment_uid
        segment["clip_id"] = clip_id
        segment["video_id"] = row["video_id"]
        segment["activity"] = row["activity"]
        segment["label"] = row["label"]
        segment["step_id"] = row["step_id"]
        segment["behavior"] = row["behavior"]
        segment["variation"] = row["variation"]
        segment["metadata_row"] = row_idx

        aligned_parts.append(segment)

    # ========================================================
    # COMBINE
    # ========================================================

    aligned = pd.concat(
        aligned_parts,
        ignore_index=True
    )

    excluded_df = pd.DataFrame(excluded_rows)

    # ========================================================
    # VALIDATION
    # ========================================================

    model_columns = [
        c for c in aligned.columns
        if c not in [
            "frame",
            "segment_uid",
            "clip_id",
            "video_id",
            "activity",
            "label",
            "step_id",
            "behavior",
            "variation",
            "metadata_row"
        ]
    ]

    # There should be exactly 206 model columns
    assert len(model_columns) == 206

    total_nan = int(
        aligned[model_columns]
        .isna()
        .sum()
        .sum()
    )

    duplicate_segment_frames = int(
        aligned.duplicated(
            subset=["segment_uid", "frame"]
        ).sum()
    )

    segment_lengths = (
        aligned
        .groupby("segment_uid")["frame"]
        .count()
    )

    print("\nRESULT")
    print("-" * 90)

    print(
        "Original metadata rows :",
        len(meta)
    )

    print(
        "Aligned segments       :",
        aligned["segment_uid"].nunique()
    )

    print(
        "Aligned frame rows     :",
        len(aligned)
    )

    print(
        "Excluded metadata rows :",
        len(excluded_rows)
    )

    print(
        "Total columns          :",
        len(aligned.columns)
    )

    print(
        "Model features         :",
        len(model_columns)
    )

    print(
        "Feature NaNs           :",
        total_nan
    )

    print(
        "Duplicate seg/frame    :",
        duplicate_segment_frames
    )

    print(
        "Segment length min/max :",
        segment_lengths.min(),
        "/",
        segment_lengths.max()
    )

    print(
        "Exactly 32 frames      :",
        int((segment_lengths == 32).sum())
    )

    print(
        "Shorter than 32        :",
        int((segment_lengths < 32).sum())
    )

    print(
        "Longer than 32         :",
        int((segment_lengths > 32).sum())
    )

    # ========================================================
    # HARD VALIDATION
    # ========================================================

    if total_nan != 0:
        raise ValueError(
            f"{split}: feature NaNs detected."
        )

    if duplicate_segment_frames != 0:
        raise ValueError(
            f"{split}: duplicate segment/frame rows detected."
        )

    # ========================================================
    # SAVE
    # ========================================================

    aligned_path = os.path.join(
        OUTPUT_ROOT,
        f"{split}_aligned_canonical.csv"
    )

    excluded_path = os.path.join(
        OUTPUT_ROOT,
        f"{split}_excluded.csv"
    )

    aligned.to_csv(
        aligned_path,
        index=False
    )

    excluded_df.to_csv(
        excluded_path,
        index=False
    )

    print("\nSaved:")
    print(aligned_path)
    print(excluded_path)


print("\n" + "=" * 90)
print("CANONICAL ALIGNMENT FINISHED")
print("=" * 90)
```

```python
import os
import glob
import pandas as pd

FEATURE_ROOT = "/content/person2_output/all_videos"

OBJECTS = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# ------------------------------------------------------------
# Required source columns for the canonical conversion
# ------------------------------------------------------------

required_columns = []

for obj in OBJECTS:

    required_columns += [
        f"{obj}_presence",
        f"{obj}_confidence",
        f"{obj}_center_x",
        f"{obj}_center_y",
        f"{obj}_width",
        f"{obj}_height",
        f"{obj}_center_x_norm",
        f"{obj}_center_y_norm",
        f"{obj}_width_norm",
        f"{obj}_height_norm",
    ]

distance_cols = [
    "cotton_to_dropper_distance",
    "cotton_to_seed_jar_distance",
    "cotton_to_target_plate_distance",
    "cotton_to_water_dish_distance",
    "dropper_to_seed_jar_distance",
    "dropper_to_target_plate_distance",
    "dropper_to_water_dish_distance",
    "seed_jar_to_target_plate_distance",
    "seed_jar_to_water_dish_distance",
    "target_plate_to_water_dish_distance"
]

required_columns += distance_cols

for hand in ["left", "right"]:
    for landmark in range(21):
        for coord in ["x", "y", "z"]:
            required_columns.append(
                f"{hand}_lm{landmark}_{coord}"
            )

for hand in ["left", "right"]:
    for obj in OBJECTS:
        required_columns.append(
            f"{hand}_hand_to_{obj}_distance"
        )

# ------------------------------------------------------------
# Scan every feature file
# ------------------------------------------------------------

feature_files = glob.glob(
    os.path.join(
        FEATURE_ROOT,
        "*",
        "*_final_features_with_hand_object_distance.csv"
    )
)

print("=" * 100)
print("SOURCE SCHEMA DIAGNOSTIC — ALL 117 VIDEOS")
print("=" * 100)

problems = []

for path in sorted(feature_files):

    stem = os.path.basename(path).replace(
        "_final_features_with_hand_object_distance.csv",
        ""
    )

    df = pd.read_csv(path)

    missing = [
        col for col in required_columns
        if col not in df.columns
    ]

    nan_count = int(
        df.isna().sum().sum()
    )

    if missing or nan_count != 0:

        problems.append({
            "video": stem,
            "shape": df.shape,
            "missing_columns": missing,
            "missing_count": len(missing),
            "nan_count": nan_count
        })

        print("\n❌ PROBLEM:", stem)
        print("Shape:", df.shape)
        print("Missing required columns:", len(missing))

        if missing:
            print("Missing:")
            print(missing[:20])

        print("NaN count:", nan_count)

    else:

        print(
            f"✓ {stem:<25} "
            f"shape={str(df.shape):<14} "
            f"schema=OK"
        )

print("\n" + "=" * 100)
print("SUMMARY")
print("=" * 100)

print("Feature files checked :", len(feature_files))
print("Problematic files     :", len(problems))

if problems:

    print("\nProblematic videos:")

    for p in problems:
        print(
            f"  {p['video']}: "
            f"missing={p['missing_count']}, "
            f"NaNs={p['nan_count']}"
        )

else:

    print("🎉 ALL 117 SOURCE FILES HAVE THE REQUIRED SCHEMA")
```

```python
import os
import pandas as pd

FEATURE_ROOT = "/content/person2_output/all_videos"

OBJECTS = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# ------------------------------------------------------------
# Canonical 206-feature column order
# ------------------------------------------------------------

CANONICAL_FEATURES = []

# 1. Presence + confidence
for obj in OBJECTS:
    CANONICAL_FEATURES.append(f"{obj}_presence")

for obj in OBJECTS:
    CANONICAL_FEATURES.append(f"{obj}_confidence")

# 2. Raw geometry
for obj in OBJECTS:
    CANONICAL_FEATURES += [
        f"{obj}_x1",
        f"{obj}_y1",
        f"{obj}_x2",
        f"{obj}_y2",
        f"{obj}_center_x"
    ]

# 3. Normalized geometry
for obj in OBJECTS:
    CANONICAL_FEATURES += [
        f"{obj}_x1_norm",
        f"{obj}_y1_norm",
        f"{obj}_x2_norm",
        f"{obj}_y2_norm",
        f"{obj}_center_x_norm"
    ]

# 4. Object-object distances
CANONICAL_FEATURES += [
    "cotton_to_dropper_distance",
    "cotton_to_seed_jar_distance",
    "cotton_to_target_plate_distance",
    "cotton_to_water_dish_distance",
    "dropper_to_seed_jar_distance",
    "dropper_to_target_plate_distance",
    "dropper_to_water_dish_distance",
    "seed_jar_to_target_plate_distance",
    "seed_jar_to_water_dish_distance",
    "target_plate_to_water_dish_distance"
]

# 5. Hand landmarks
for hand in ["left", "right"]:
    for landmark in range(21):
        for coord in ["x", "y", "z"]:
            CANONICAL_FEATURES.append(
                f"{hand}_hand_{landmark}_{coord}"
            )

# 6. Hand-object distances
for hand in ["left", "right"]:
    for obj in OBJECTS:
        CANONICAL_FEATURES.append(
            f"{hand}_hand_to_{obj}_distance"
        )

print("Canonical feature count:", len(CANONICAL_FEATURES))

# ------------------------------------------------------------
# Universal converter
# ------------------------------------------------------------

def convert_to_canonical(source):

    df = source.copy()

    out = pd.DataFrame(index=df.index)

    # --------------------------------------------------------
    # Presence
    # Supports:
    #   *_presence
    #   *_present
    # --------------------------------------------------------

    for obj in OBJECTS:

        presence_col = f"{obj}_presence"
        present_col = f"{obj}_present"

        if presence_col in df.columns:
            out[presence_col] = df[presence_col]

        elif present_col in df.columns:
            out[presence_col] = df[present_col]

        else:
            raise KeyError(
                f"Missing presence column for {obj}. "
                f"Expected {presence_col} or {present_col}"
            )

    # --------------------------------------------------------
    # Confidence
    # --------------------------------------------------------

    for obj in OBJECTS:

        col = f"{obj}_confidence"

        if col not in df.columns:
            raise KeyError(f"Missing confidence column: {col}")

        out[col] = df[col]

    # --------------------------------------------------------
    # Geometry
    #
    # Supports BOTH:
    #
    # A) canonical:
    #    x1,y1,x2,y2,center_x
    #
    # B) center/size:
    #    center_x,center_y,width,height
    # --------------------------------------------------------

    for obj in OBJECTS:

        x1 = f"{obj}_x1"
        y1 = f"{obj}_y1"
        x2 = f"{obj}_x2"
        y2 = f"{obj}_y2"

        cx = f"{obj}_center_x"
        cy = f"{obj}_center_y"
        w  = f"{obj}_width"
        h  = f"{obj}_height"

        # Canonical raw geometry
        if all(c in df.columns for c in [x1, y1, x2, y2]):

            out[x1] = df[x1]
            out[y1] = df[y1]
            out[x2] = df[x2]
            out[y2] = df[y2]

        # Center/width/height raw geometry
        elif all(c in df.columns for c in [cx, cy, w, h]):

            out[x1] = df[cx] - df[w] / 2
            out[y1] = df[cy] - df[h] / 2
            out[x2] = df[cx] + df[w] / 2
            out[y2] = df[cy] + df[h] / 2

        else:
            raise KeyError(
                f"Cannot reconstruct raw geometry for {obj}"
            )

        # Center X
        if cx in df.columns:
            out[cx] = df[cx]

        else:
            out[cx] = (
                out[x1] + out[x2]
            ) / 2

        # ----------------------------------------------------
        # Normalized geometry
        # ----------------------------------------------------

        x1n = f"{obj}_x1_norm"
        y1n = f"{obj}_y1_norm"
        x2n = f"{obj}_x2_norm"
        y2n = f"{obj}_y2_norm"

        cxn = f"{obj}_center_x_norm"
        cyn = f"{obj}_center_y_norm"
        wn  = f"{obj}_width_norm"
        hn  = f"{obj}_height_norm"

        # Canonical normalized geometry
        if all(
            c in df.columns
            for c in [x1n, y1n, x2n, y2n]
        ):

            out[x1n] = df[x1n]
            out[y1n] = df[y1n]
            out[x2n] = df[x2n]
            out[y2n] = df[y2n]

        # Center/size normalized geometry
        elif all(
            c in df.columns
            for c in [cxn, cyn, wn, hn]
        ):

            out[x1n] = df[cxn] - df[wn] / 2
            out[y1n] = df[cyn] - df[hn] / 2
            out[x2n] = df[cxn] + df[wn] / 2
            out[y2n] = df[cyn] + df[hn] / 2

        else:
            raise KeyError(
                f"Cannot reconstruct normalized geometry "
                f"for {obj}"
            )

        # Normalized center X
        if cxn in df.columns:
            out[cxn] = df[cxn]

        else:
            out[cxn] = (
                out[x1n] + out[x2n]
            ) / 2

    # --------------------------------------------------------
    # Object-object distances
    # --------------------------------------------------------

    distance_cols = [
        "cotton_to_dropper_distance",
        "cotton_to_seed_jar_distance",
        "cotton_to_target_plate_distance",
        "cotton_to_water_dish_distance",
        "dropper_to_seed_jar_distance",
        "dropper_to_target_plate_distance",
        "dropper_to_water_dish_distance",
        "seed_jar_to_target_plate_distance",
        "seed_jar_to_water_dish_distance",
        "target_plate_to_water_dish_distance"
    ]

    for col in distance_cols:

        if col not in df.columns:
            raise KeyError(
                f"Missing object-object distance: {col}"
            )

        out[col] = df[col]

    # --------------------------------------------------------
    # Hand landmarks
    #
    # Supports:
    #   left_lm0_x
    #   left_hand_0_x
    #
    # Same for right hand.
    # --------------------------------------------------------

    for hand in ["left", "right"]:

        for landmark in range(21):

            for coord in ["x", "y", "z"]:

                canonical = (
                    f"{hand}_hand_{landmark}_{coord}"
                )

                old_style = (
                    f"{hand}_lm{landmark}_{coord}"
                )

                if canonical in df.columns:
                    out[canonical] = df[canonical]

                elif old_style in df.columns:
                    out[canonical] = df[old_style]

                else:
                    raise KeyError(
                        f"Missing hand landmark: "
                        f"{canonical} / {old_style}"
                    )

    # --------------------------------------------------------
    # Hand-object distances
    # --------------------------------------------------------

    for hand in ["left", "right"]:

        for obj in OBJECTS:

            col = f"{hand}_hand_to_{obj}_distance"

            if col not in df.columns:
                raise KeyError(
                    f"Missing hand-object distance: {col}"
                )

            out[col] = df[col]

    # --------------------------------------------------------
    # Force exact canonical order
    # --------------------------------------------------------

    out = out[CANONICAL_FEATURES]

    return out


# ------------------------------------------------------------
# Test C4, C5, I12
# ------------------------------------------------------------

TEST_FILES = {
    "C4": os.path.join(
        FEATURE_ROOT,
        "C4",
        "C4_final_features_with_hand_object_distance.csv"
    ),
    "C5": os.path.join(
        FEATURE_ROOT,
        "C5",
        "C5_final_features_with_hand_object_distance.csv"
    ),
    "I12": os.path.join(
        FEATURE_ROOT,
        "I12",
        "I12_final_features_with_hand_object_distance.csv"
    )
}

for name, path in TEST_FILES.items():

    print("\n" + "=" * 100)
    print(f"TESTING {name}")
    print("=" * 100)

    source = pd.read_csv(path)

    print("Source shape   :", source.shape)

    try:

        canonical = convert_to_canonical(source)

        print("Canonical shape:", canonical.shape)
        print(
            "Model features:",
            canonical.shape[1]
        )

        print(
            "NaN values    :",
            int(canonical.isna().sum().sum())
        )

        print(
            "Columns exact :",
            list(canonical.columns) == CANONICAL_FEATURES
        )

        print(
            "First columns :",
            list(canonical.columns[:15])
        )

        print(
            "Last columns  :",
            list(canonical.columns[-10:])
        )

        if (
            canonical.shape[1] == 206
            and canonical.isna().sum().sum() == 0
            and list(canonical.columns) == CANONICAL_FEATURES
        ):
            print(f"✅ {name} PASSED")

        else:
            print(f"❌ {name} FAILED VALIDATION")

    except Exception as e:

        print(f"❌ {name} FAILED")
        print("Error:", repr(e))

print("\n" + "=" * 100)
print("3-FILE CONVERTER TEST COMPLETE")
print("=" * 100)
```

```python
# ============================================================
# PERSON 2 — STEP 1
# CONVERT ALL 117 FEATURE FILES TO CANONICAL FORMAT
# ============================================================

import os
import glob
import traceback
import pandas as pd

FEATURE_ROOT = "/content/person2_output/all_videos"
CANONICAL_ROOT = "/content/person2_output/canonical_features"

os.makedirs(CANONICAL_ROOT, exist_ok=True)

OBJECTS = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# ------------------------------------------------------------
# CANONICAL FEATURE ORDER
# ------------------------------------------------------------

CANONICAL_FEATURES = []

# 5 presence
for obj in OBJECTS:
    CANONICAL_FEATURES.append(
        f"{obj}_presence"
    )

# 5 confidence
for obj in OBJECTS:
    CANONICAL_FEATURES.append(
        f"{obj}_confidence"
    )

# 25 raw geometry
for obj in OBJECTS:
    CANONICAL_FEATURES += [
        f"{obj}_x1",
        f"{obj}_y1",
        f"{obj}_x2",
        f"{obj}_y2",
        f"{obj}_center_x"
    ]

# 25 normalized geometry
for obj in OBJECTS:
    CANONICAL_FEATURES += [
        f"{obj}_x1_norm",
        f"{obj}_y1_norm",
        f"{obj}_x2_norm",
        f"{obj}_y2_norm",
        f"{obj}_center_x_norm"
    ]

# 10 object-object distances
DISTANCE_COLS = [
    "cotton_to_dropper_distance",
    "cotton_to_seed_jar_distance",
    "cotton_to_target_plate_distance",
    "cotton_to_water_dish_distance",
    "dropper_to_seed_jar_distance",
    "dropper_to_target_plate_distance",
    "dropper_to_water_dish_distance",
    "seed_jar_to_target_plate_distance",
    "seed_jar_to_water_dish_distance",
    "target_plate_to_water_dish_distance"
]

CANONICAL_FEATURES += DISTANCE_COLS

# 126 hand landmarks
for hand in ["left", "right"]:
    for landmark in range(21):
        for coord in ["x", "y", "z"]:
            CANONICAL_FEATURES.append(
                f"{hand}_hand_{landmark}_{coord}"
            )

# 10 hand-object distances
for hand in ["left", "right"]:
    for obj in OBJECTS:
        CANONICAL_FEATURES.append(
            f"{hand}_hand_to_{obj}_distance"
        )

assert len(CANONICAL_FEATURES) == 206

print("Canonical model features:", len(CANONICAL_FEATURES))


# ------------------------------------------------------------
# UNIVERSAL CONVERTER
# ------------------------------------------------------------

def convert_to_canonical(source):

    df = source.copy()

    data = {}

    # --------------------------------------------------------
    # PRESENCE
    # Supports:
    #   *_presence
    #   *_present
    # --------------------------------------------------------

    for obj in OBJECTS:

        canonical = f"{obj}_presence"
        old_present = f"{obj}_present"

        if canonical in df.columns:
            data[canonical] = df[canonical].to_numpy()

        elif old_present in df.columns:
            data[canonical] = df[old_present].to_numpy()

        else:
            raise KeyError(
                f"Missing presence column for {obj}"
            )


    # --------------------------------------------------------
    # CONFIDENCE
    # --------------------------------------------------------

    for obj in OBJECTS:

        col = f"{obj}_confidence"

        if col not in df.columns:
            raise KeyError(
                f"Missing confidence column: {col}"
            )

        data[col] = df[col].to_numpy()


    # --------------------------------------------------------
    # GEOMETRY
    #
    # Supports either:
    #
    # A) x1,y1,x2,y2
    #
    # OR
    #
    # B) center_x,center_y,width,height
    # --------------------------------------------------------

    for obj in OBJECTS:

        x1 = f"{obj}_x1"
        y1 = f"{obj}_y1"
        x2 = f"{obj}_x2"
        y2 = f"{obj}_y2"

        cx = f"{obj}_center_x"
        cy = f"{obj}_center_y"
        w  = f"{obj}_width"
        h  = f"{obj}_height"

        if all(
            c in df.columns
            for c in [x1, y1, x2, y2]
        ):

            data[x1] = df[x1].to_numpy()
            data[y1] = df[y1].to_numpy()
            data[x2] = df[x2].to_numpy()
            data[y2] = df[y2].to_numpy()

        elif all(
            c in df.columns
            for c in [cx, cy, w, h]
        ):

            data[x1] = (
                df[cx] - df[w] / 2
            ).to_numpy()

            data[y1] = (
                df[cy] - df[h] / 2
            ).to_numpy()

            data[x2] = (
                df[cx] + df[w] / 2
            ).to_numpy()

            data[y2] = (
                df[cy] + df[h] / 2
            ).to_numpy()

        else:
            raise KeyError(
                f"Cannot reconstruct raw geometry for {obj}"
            )

        # center_x
        if cx in df.columns:
            data[cx] = df[cx].to_numpy()
        else:
            data[cx] = (
                (data[x1] + data[x2]) / 2
            )


        # ----------------------------------------------------
        # NORMALIZED GEOMETRY
        # ----------------------------------------------------

        x1n = f"{obj}_x1_norm"
        y1n = f"{obj}_y1_norm"
        x2n = f"{obj}_x2_norm"
        y2n = f"{obj}_y2_norm"

        cxn = f"{obj}_center_x_norm"
        cyn = f"{obj}_center_y_norm"
        wn  = f"{obj}_width_norm"
        hn  = f"{obj}_height_norm"

        if all(
            c in df.columns
            for c in [x1n, y1n, x2n, y2n]
        ):

            data[x1n] = df[x1n].to_numpy()
            data[y1n] = df[y1n].to_numpy()
            data[x2n] = df[x2n].to_numpy()
            data[y2n] = df[y2n].to_numpy()

        elif all(
            c in df.columns
            for c in [cxn, cyn, wn, hn]
        ):

            data[x1n] = (
                df[cxn] - df[wn] / 2
            ).to_numpy()

            data[y1n] = (
                df[cyn] - df[hn] / 2
            ).to_numpy()

            data[x2n] = (
                df[cxn] + df[wn] / 2
            ).to_numpy()

            data[y2n] = (
                df[cyn] + df[hn] / 2
            ).to_numpy()

        else:
            raise KeyError(
                f"Cannot reconstruct normalized geometry "
                f"for {obj}"
            )

        # normalized center_x
        if cxn in df.columns:
            data[cxn] = df[cxn].to_numpy()
        else:
            data[cxn] = (
                (data[x1n] + data[x2n]) / 2
            )


    # --------------------------------------------------------
    # OBJECT-OBJECT DISTANCES
    # --------------------------------------------------------

    for col in DISTANCE_COLS:

        if col not in df.columns:
            raise KeyError(
                f"Missing object-object distance: {col}"
            )

        data[col] = df[col].to_numpy()


    # --------------------------------------------------------
    # HAND LANDMARKS
    #
    # Supports:
    #   left_hand_0_x
    #
    # OR:
    #   left_lm0_x
    # --------------------------------------------------------

    for hand in ["left", "right"]:

        for landmark in range(21):

            for coord in ["x", "y", "z"]:

                canonical = (
                    f"{hand}_hand_{landmark}_{coord}"
                )

                old_style = (
                    f"{hand}_lm{landmark}_{coord}"
                )

                if canonical in df.columns:
                    data[canonical] = (
                        df[canonical].to_numpy()
                    )

                elif old_style in df.columns:
                    data[canonical] = (
                        df[old_style].to_numpy()
                    )

                else:
                    raise KeyError(
                        f"Missing hand landmark: "
                        f"{canonical} / {old_style}"
                    )


    # --------------------------------------------------------
    # HAND-OBJECT DISTANCES
    # --------------------------------------------------------

    for hand in ["left", "right"]:

        for obj in OBJECTS:

            col = (
                f"{hand}_hand_to_{obj}_distance"
            )

            if col not in df.columns:
                raise KeyError(
                    f"Missing hand-object distance: {col}"
                )

            data[col] = df[col].to_numpy()


    # --------------------------------------------------------
    # BUILD DATAFRAME ONCE
    # --------------------------------------------------------

    out = pd.DataFrame(
        data,
        index=df.index
    )

    # Force exact canonical order
    out = out[CANONICAL_FEATURES]

    return out


# ------------------------------------------------------------
# FIND ALL 117 SOURCE FILES
# ------------------------------------------------------------

source_files = sorted(
    glob.glob(
        os.path.join(
            FEATURE_ROOT,
            "*",
            "*_final_features_with_hand_object_distance.csv"
        )
    )
)

print("Source files found:", len(source_files))

if len(source_files) != 117:
    print(
        "⚠️ WARNING: Expected 117 files, "
        f"but found {len(source_files)}"
    )


# ------------------------------------------------------------
# CONVERT ALL FILES
# ------------------------------------------------------------

success = []
failed = []

for i, source_path in enumerate(source_files, start=1):

    filename = os.path.basename(source_path)

    stem = filename.replace(
        "_final_features_with_hand_object_distance.csv",
        ""
    )

    output_dir = os.path.join(
        CANONICAL_ROOT,
        stem
    )

    os.makedirs(output_dir, exist_ok=True)

    output_path = os.path.join(
        output_dir,
        f"{stem}_canonical_features.csv"
    )

    print(
        f"\n[{i}/{len(source_files)}] {stem}"
    )

    try:

        source = pd.read_csv(source_path)

        canonical = convert_to_canonical(source)

        # ----------------------------------------------------
        # VALIDATE
        # ----------------------------------------------------

        assert canonical.shape[0] == source.shape[0]

        assert canonical.shape[1] == 206

        assert list(
            canonical.columns
        ) == CANONICAL_FEATURES

        nan_count = int(
            canonical.isna().sum().sum()
        )

        if nan_count != 0:
            raise ValueError(
                f"Canonical file contains "
                f"{nan_count} NaN values"
            )

        # ----------------------------------------------------
        # SAVE
        # ----------------------------------------------------

        canonical.to_csv(
            output_path,
            index=False
        )

        success.append(stem)

        print(
            f"  ✅ rows={len(canonical)}, "
            f"features=206, NaN=0"
        )

    except Exception as e:

        failed.append({
            "video": stem,
            "error": repr(e)
        })

        print(
            f"  ❌ FAILED: {repr(e)}"
        )


# ------------------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("CANONICAL CONVERSION COMPLETE")
print("=" * 90)

print("Source files :", len(source_files))
print("Successful   :", len(success))
print("Failed       :", len(failed))

if failed:

    print("\nFAILED FILES:")

    for item in failed:
        print(
            f"- {item['video']}: "
            f"{item['error']}"
        )

else:

    print("\n🎉 ALL FILES CONVERTED SUCCESSFULLY!")

print(
    "\nCanonical output directory:"
)
print(CANONICAL_ROOT)
```

```python
# ============================================================
# PERSON 2 — STEP 2
# VERIFY ALL 117 CANONICAL FEATURE FILES
# ============================================================

import os
import glob
import pandas as pd

CANONICAL_ROOT = "/content/person2_output/canonical_features"

OBJECTS = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# ------------------------------------------------------------
# BUILD EXPECTED CANONICAL FEATURE LIST
# ------------------------------------------------------------

CANONICAL_FEATURES = []

# Presence
for obj in OBJECTS:
    CANONICAL_FEATURES.append(
        f"{obj}_presence"
    )

# Confidence
for obj in OBJECTS:
    CANONICAL_FEATURES.append(
        f"{obj}_confidence"
    )

# Raw geometry
for obj in OBJECTS:
    CANONICAL_FEATURES += [
        f"{obj}_x1",
        f"{obj}_y1",
        f"{obj}_x2",
        f"{obj}_y2",
        f"{obj}_center_x"
    ]

# Normalized geometry
for obj in OBJECTS:
    CANONICAL_FEATURES += [
        f"{obj}_x1_norm",
        f"{obj}_y1_norm",
        f"{obj}_x2_norm",
        f"{obj}_y2_norm",
        f"{obj}_center_x_norm"
    ]

# Object-object distances
CANONICAL_FEATURES += [
    "cotton_to_dropper_distance",
    "cotton_to_seed_jar_distance",
    "cotton_to_target_plate_distance",
    "cotton_to_water_dish_distance",
    "dropper_to_seed_jar_distance",
    "dropper_to_target_plate_distance",
    "dropper_to_water_dish_distance",
    "seed_jar_to_target_plate_distance",
    "seed_jar_to_water_dish_distance",
    "target_plate_to_water_dish_distance"
]

# Hand landmarks
for hand in ["left", "right"]:
    for landmark in range(21):
        for coord in ["x", "y", "z"]:
            CANONICAL_FEATURES.append(
                f"{hand}_hand_{landmark}_{coord}"
            )

# Hand-object distances
for hand in ["left", "right"]:
    for obj in OBJECTS:
        CANONICAL_FEATURES.append(
            f"{hand}_hand_to_{obj}_distance"
        )

assert len(CANONICAL_FEATURES) == 206


# ------------------------------------------------------------
# FIND CANONICAL FILES
# ------------------------------------------------------------

canonical_files = sorted(
    glob.glob(
        os.path.join(
            CANONICAL_ROOT,
            "*",
            "*_canonical_features.csv"
        )
    )
)

print("=" * 90)
print("PERSON 2 — CANONICAL FEATURE VERIFICATION")
print("=" * 90)

print(
    "Canonical files found:",
    len(canonical_files)
)

if len(canonical_files) != 117:
    print(
        "⚠️ WARNING: Expected 117 files."
    )


# ------------------------------------------------------------
# VERIFY EACH FILE
# ------------------------------------------------------------

valid_files = []
invalid_files = []

for i, path in enumerate(
    canonical_files,
    start=1
):

    filename = os.path.basename(path)

    video_name = filename.replace(
        "_canonical_features.csv",
        ""
    )

    try:

        df = pd.read_csv(path)

        problems = []

        # ----------------------------------------------------
        # ROW CHECK
        # ----------------------------------------------------

        if len(df) == 0:
            problems.append(
                "0 rows"
            )

        # ----------------------------------------------------
        # COLUMN COUNT
        # ----------------------------------------------------

        if df.shape[1] != 206:
            problems.append(
                f"wrong column count: {df.shape[1]}"
            )

        # ----------------------------------------------------
        # EXACT COLUMN ORDER
        # ----------------------------------------------------

        if list(df.columns) != CANONICAL_FEATURES:
            problems.append(
                "column names/order mismatch"
            )

        # ----------------------------------------------------
        # NaN CHECK
        # ----------------------------------------------------

        nan_count = int(
            df.isna().sum().sum()
        )

        if nan_count != 0:
            problems.append(
                f"{nan_count} NaN values"
            )

        # ----------------------------------------------------
        # DUPLICATE COLUMN CHECK
        # ----------------------------------------------------

        duplicate_columns = (
            df.columns[
                df.columns.duplicated()
            ].tolist()
        )

        if duplicate_columns:
            problems.append(
                "duplicate columns"
            )

        # ----------------------------------------------------
        # FRAME CHECK
        # ----------------------------------------------------

        if "frame" not in df.columns:
            problems.append(
                "frame column missing"
            )

        else:

            if df["frame"].duplicated().any():
                problems.append(
                    "duplicate frame numbers"
                )

            frames = df["frame"].tolist()

            expected_frames = list(
                range(
                    int(min(frames)),
                    int(max(frames)) + 1
                )
            )

            if frames != expected_frames:
                problems.append(
                    "frame sequence is not continuous"
                )

        # ----------------------------------------------------
        # RESULT
        # ----------------------------------------------------

        if problems:

            invalid_files.append({
                "video": video_name,
                "problems": problems
            })

            print(
                f"[{i:03d}/117] ❌ {video_name}"
            )

            for problem in problems:
                print(
                    "       -",
                    problem
                )

        else:

            valid_files.append(
                video_name
            )

            print(
                f"[{i:03d}/117] ✅ {video_name} "
                f"({len(df)} frames)"
            )

    except Exception as e:

        invalid_files.append({
            "video": video_name,
            "problems": [
                repr(e)
            ]
        })

        print(
            f"[{i:03d}/117] ❌ {video_name}"
        )
        print(
            "       -",
            repr(e)
        )


# ------------------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("CANONICAL VERIFICATION COMPLETE")
print("=" * 90)

print(
    "Files found    :",
    len(canonical_files)
)

print(
    "VALID files    :",
    len(valid_files)
)

print(
    "INVALID files  :",
    len(invalid_files)
)

if invalid_files:

    print("\n❌ INVALID FILES:")

    for item in invalid_files:

        print(
            f"\n{item['video']}"
        )

        for problem in item["problems"]:
            print(
                "  -",
                problem
            )

else:

    print(
        "\n🎉 ALL CANONICAL FILES PASSED!"
    )

print(
    "\nExpected model features:",
    len(CANONICAL_FEATURES)
)

print(
    "Canonical directory:",
    CANONICAL_ROOT
)
```

```python
# ============================================================
# PERSON 2 — STEP 2 (CORRECTED)
# VERIFY ALL 117 CANONICAL FEATURE FILES
# ============================================================

import os
import glob
import pandas as pd

CANONICAL_ROOT = "/content/person2_output/canonical_features"

OBJECTS = [
    "cotton",
    "dropper",
    "seed_jar",
    "target_plate",
    "water_dish"
]

# ------------------------------------------------------------
# EXPECTED CANONICAL FEATURES
# ------------------------------------------------------------

CANONICAL_FEATURES = []

# 5 presence
for obj in OBJECTS:
    CANONICAL_FEATURES.append(
        f"{obj}_presence"
    )

# 5 confidence
for obj in OBJECTS:
    CANONICAL_FEATURES.append(
        f"{obj}_confidence"
    )

# 25 raw geometry
for obj in OBJECTS:
    CANONICAL_FEATURES += [
        f"{obj}_x1",
        f"{obj}_y1",
        f"{obj}_x2",
        f"{obj}_y2",
        f"{obj}_center_x"
    ]

# 25 normalized geometry
for obj in OBJECTS:
    CANONICAL_FEATURES += [
        f"{obj}_x1_norm",
        f"{obj}_y1_norm",
        f"{obj}_x2_norm",
        f"{obj}_y2_norm",
        f"{obj}_center_x_norm"
    ]

# 10 object-object distances
CANONICAL_FEATURES += [
    "cotton_to_dropper_distance",
    "cotton_to_seed_jar_distance",
    "cotton_to_target_plate_distance",
    "cotton_to_water_dish_distance",
    "dropper_to_seed_jar_distance",
    "dropper_to_target_plate_distance",
    "dropper_to_water_dish_distance",
    "seed_jar_to_target_plate_distance",
    "seed_jar_to_water_dish_distance",
    "target_plate_to_water_dish_distance"
]

# 126 hand landmarks
for hand in ["left", "right"]:
    for landmark in range(21):
        for coord in ["x", "y", "z"]:
            CANONICAL_FEATURES.append(
                f"{hand}_hand_{landmark}_{coord}"
            )

# 10 hand-object distances
for hand in ["left", "right"]:
    for obj in OBJECTS:
        CANONICAL_FEATURES.append(
            f"{hand}_hand_to_{obj}_distance"
        )

assert len(CANONICAL_FEATURES) == 206


# ------------------------------------------------------------
# FIND FILES
# ------------------------------------------------------------

canonical_files = sorted(
    glob.glob(
        os.path.join(
            CANONICAL_ROOT,
            "*",
            "*_canonical_features.csv"
        )
    )
)

print("=" * 90)
print("PERSON 2 — CANONICAL FEATURE VERIFICATION")
print("=" * 90)

print(
    "Canonical files found:",
    len(canonical_files)
)


# ------------------------------------------------------------
# VERIFY
# ------------------------------------------------------------

valid_files = []
invalid_files = []

for i, path in enumerate(
    canonical_files,
    start=1
):

    filename = os.path.basename(path)

    video_name = filename.replace(
        "_canonical_features.csv",
        ""
    )

    try:

        df = pd.read_csv(path)

        problems = []

        # ----------------------------------------------------
        # ROW CHECK
        # ----------------------------------------------------

        if len(df) == 0:
            problems.append(
                "0 rows"
            )

        # ----------------------------------------------------
        # COLUMN COUNT
        # ----------------------------------------------------

        if df.shape[1] != 206:
            problems.append(
                f"wrong column count: {df.shape[1]}"
            )

        # ----------------------------------------------------
        # EXACT COLUMN ORDER
        # ----------------------------------------------------

        if list(df.columns) != CANONICAL_FEATURES:
            problems.append(
                "column names/order mismatch"
            )

        # ----------------------------------------------------
        # NaN CHECK
        # ----------------------------------------------------

        nan_count = int(
            df.isna().sum().sum()
        )

        if nan_count != 0:
            problems.append(
                f"{nan_count} NaN values"
            )

        # ----------------------------------------------------
        # DUPLICATE COLUMN CHECK
        # ----------------------------------------------------

        duplicate_columns = (
            df.columns[
                df.columns.duplicated()
            ].tolist()
        )

        if duplicate_columns:
            problems.append(
                "duplicate columns"
            )

        # ----------------------------------------------------
        # NUMERIC CHECK
        # ----------------------------------------------------

        non_numeric = []

        for col in CANONICAL_FEATURES:

            if not pd.api.types.is_numeric_dtype(
                df[col]
            ):
                non_numeric.append(col)

        if non_numeric:

            problems.append(
                f"non-numeric columns: "
                f"{non_numeric[:5]}"
            )

        # ----------------------------------------------------
        # RESULT
        # ----------------------------------------------------

        if problems:

            invalid_files.append({
                "video": video_name,
                "problems": problems
            })

            print(
                f"[{i:03d}/{len(canonical_files)}] "
                f"❌ {video_name}"
            )

            for problem in problems:
                print(
                    "       -",
                    problem
                )

        else:

            valid_files.append(
                video_name
            )

            print(
                f"[{i:03d}/{len(canonical_files)}] "
                f"✅ {video_name} "
                f"({len(df)} frames)"
            )

    except Exception as e:

        invalid_files.append({
            "video": video_name,
            "problems": [
                repr(e)
            ]
        })

        print(
            f"[{i:03d}/{len(canonical_files)}] "
            f"❌ {video_name}"
        )

        print(
            "       -",
            repr(e)
        )


# ------------------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("CANONICAL VERIFICATION COMPLETE")
print("=" * 90)

print(
    "Files found    :",
    len(canonical_files)
)

print(
    "VALID files    :",
    len(valid_files)
)

print(
    "INVALID files  :",
    len(invalid_files)
)

print(
    "Expected features:",
    len(CANONICAL_FEATURES)
)

if invalid_files:

    print("\n❌ INVALID FILES:")

    for item in invalid_files:

        print(
            f"\n{item['video']}"
        )

        for problem in item["problems"]:
            print(
                "  -",
                problem
            )

else:

    print(
        "\n🎉 ALL 117 CANONICAL FILES PASSED!"
    )

print(
    "\nCanonical directory:"
)

print(
    CANONICAL_ROOT
)
```

```python
# ============================================================
# PERSON 2 — STEP 3
# REBUILD TRAIN / VAL / TEST ALIGNMENT
# USING CANONICAL FEATURES
# ============================================================

import os
import glob
import pandas as pd

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

CANONICAL_ROOT = "/content/person2_output/canonical_features"

OUTPUT_ROOT = "/content/person2_output/aligned_canonical"

os.makedirs(
    OUTPUT_ROOT,
    exist_ok=True
)

META_FILES = {
    "train": "/content/train (1).csv",
    "val": "/content/val.csv",
    "test": "/content/test.csv"
}


# ------------------------------------------------------------
# EXPECTED FEATURE COUNT
# ------------------------------------------------------------

EXPECTED_FEATURE_COUNT = 206


# ------------------------------------------------------------
# FIND CANONICAL FEATURE FILES
# ------------------------------------------------------------

canonical_files = glob.glob(
    os.path.join(
        CANONICAL_ROOT,
        "*",
        "*_canonical_features.csv"
    )
)

print(
    "Canonical feature files found:",
    len(canonical_files)
)


# ------------------------------------------------------------
# BUILD VIDEO -> FEATURE FILE MAP
# ------------------------------------------------------------

feature_map = {}

for path in canonical_files:

    filename = os.path.basename(path)

    stem = filename.replace(
        "_canonical_features.csv",
        ""
    )

    feature_map[
        stem.lower()
    ] = path


print(
    "Video names mapped:",
    len(feature_map)
)


# ------------------------------------------------------------
# PROCESS EACH SPLIT
# ------------------------------------------------------------

for split, meta_path in META_FILES.items():

    print("\n" + "=" * 100)
    print(f"PROCESSING {split.upper()}")
    print("=" * 100)

    # --------------------------------------------------------
    # LOAD METADATA
    # --------------------------------------------------------

    meta = pd.read_csv(
        meta_path
    )

    print(
        "Metadata rows:",
        len(meta)
    )

    aligned_parts = []
    excluded_rows = []

    # --------------------------------------------------------
    # PROCESS EACH TEMPORAL SEGMENT
    # --------------------------------------------------------

    for row_idx, row in meta.iterrows():

        video_file = str(
            row["video_file"]
        )

        video_stem = os.path.splitext(
            os.path.basename(video_file)
        )[0].lower()

        clip_id = str(
            row["clip_id"]
        )

        start_frame = int(
            row["start_frame"]
        )

        end_frame = int(
            row["end_frame"]
        )

        # Unique ID for THIS metadata row.
        #
        # Important:
        # clip_id can occur multiple times.
        #
        segment_uid = (
            f"{split}_{row_idx:04d}_{clip_id}"
        )


        # ----------------------------------------------------
        # INVALID RANGE
        # ----------------------------------------------------

        if start_frame > end_frame:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "reason": "Invalid temporal range"
            })

            continue


        # ----------------------------------------------------
        # FIND CANONICAL FEATURE FILE
        # ----------------------------------------------------

        feature_path = feature_map.get(
            video_stem
        )

        if feature_path is None:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "reason": "Canonical feature file not found"
            })

            continue


        # ----------------------------------------------------
        # LOAD CANONICAL FEATURES
        # ----------------------------------------------------

        features = pd.read_csv(
            feature_path
        )


        # ----------------------------------------------------
        # VALIDATE CANONICAL FILE
        # ----------------------------------------------------

        if features.shape[1] != EXPECTED_FEATURE_COUNT:

            raise ValueError(
                f"{video_stem}: expected "
                f"{EXPECTED_FEATURE_COUNT} features, "
                f"got {features.shape[1]}"
            )


        # ----------------------------------------------------
        # ADD FRAME COLUMN
        #
        # Canonical files contain 206 model features only.
        # Frame information is recovered from the original
        # source feature file.
        #
        # Since the canonical file has exactly the same row
        # order as its source, we recover frame numbers from
        # the corresponding original feature CSV.
        # ----------------------------------------------------

        original_path = os.path.join(
            "/content/person2_output/all_videos",
            video_stem,
            f"{video_stem}_final_features_with_hand_object_distance.csv"
        )


        if not os.path.exists(
            original_path
        ):

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "reason": "Original feature file not found"
            })

            continue


        original = pd.read_csv(
            original_path,
            usecols=["frame"]
        )


        if len(original) != len(features):

            raise ValueError(
                f"{video_stem}: canonical/source row count mismatch"
            )


        # Add frame to canonical features
        features = features.copy()

        features.insert(
            0,
            "frame",
            original["frame"].to_numpy()
        )


        # ----------------------------------------------------
        # SELECT TEMPORAL SEGMENT
        # ----------------------------------------------------

        segment = features[
            (features["frame"] >= start_frame) &
            (features["frame"] <= end_frame)
        ].copy()


        expected_frames = (
            end_frame -
            start_frame +
            1
        )

        actual_frames = len(
            segment
        )


        # ----------------------------------------------------
        # FRAME COUNT CHECK
        # ----------------------------------------------------

        if actual_frames != expected_frames:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "expected_frames": expected_frames,
                "actual_frames": actual_frames,
                "reason": "Frame count mismatch"
            })

            continue


        # ----------------------------------------------------
        # CONTINUITY CHECK
        # ----------------------------------------------------

        segment_frames = (
            segment["frame"]
            .astype(int)
            .tolist()
        )

        expected_frame_list = list(
            range(
                start_frame,
                end_frame + 1
            )
        )

        if segment_frames != expected_frame_list:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "expected_frames": expected_frames,
                "actual_frames": actual_frames,
                "reason": "Frame sequence not continuous"
            })

            continue


        # ----------------------------------------------------
        # CHECK FEATURE NaNs
        # ----------------------------------------------------

        feature_nan_count = int(
            segment.iloc[:, 1:].isna().sum().sum()
        )

        if feature_nan_count != 0:

            raise ValueError(
                f"{segment_uid}: "
                f"{feature_nan_count} feature NaNs detected"
            )


        # ----------------------------------------------------
        # ADD PERSON 1 METADATA
        # ----------------------------------------------------

        segment["segment_uid"] = segment_uid

        segment["clip_id"] = clip_id

        segment["video_id"] = row[
            "video_id"
        ]

        segment["activity"] = row[
            "activity"
        ]

        segment["label"] = row[
            "label"
        ]

        segment["step_id"] = row[
            "step_id"
        ]

        segment["behavior"] = row[
            "behavior"
        ]

        segment["variation"] = row[
            "variation"
        ]

        segment["metadata_row"] = row_idx


        # ----------------------------------------------------
        # SAVE IN MEMORY
        # ----------------------------------------------------

        aligned_parts.append(
            segment
        )


    # --------------------------------------------------------
    # COMBINE
    # --------------------------------------------------------

    if aligned_parts:

        aligned = pd.concat(
            aligned_parts,
            ignore_index=True
        )

    else:

        aligned = pd.DataFrame()


    # --------------------------------------------------------
    # SAVE ALIGNED FILE
    # --------------------------------------------------------

    aligned_path = os.path.join(
        OUTPUT_ROOT,
        f"{split}_aligned_canonical.csv"
    )

    aligned.to_csv(
        aligned_path,
        index=False
    )


    # --------------------------------------------------------
    # SAVE EXCLUDED METADATA
    # --------------------------------------------------------

    excluded_df = pd.DataFrame(
        excluded_rows
    )

    excluded_path = os.path.join(
        OUTPUT_ROOT,
        f"{split}_excluded.csv"
    )

    excluded_df.to_csv(
        excluded_path,
        index=False
    )


    # --------------------------------------------------------
    # SUMMARY
    # --------------------------------------------------------

    print(
        "\nOriginal metadata rows :",
        len(meta)
    )

    print(
        "Aligned segment rows   :",
        aligned["segment_uid"].nunique()
        if not aligned.empty
        else 0
    )

    print(
        "Aligned frame rows     :",
        len(aligned)
    )

    print(
        "Excluded metadata rows :",
        len(excluded_df)
    )

    if not aligned.empty:

        print(
            "Feature columns        :",
            aligned.shape[1] - 1 - 9
        )

        print(
            "Feature NaNs            :",
            int(
                aligned.iloc[:, 1:-9]
                .isna()
                .sum()
                .sum()
            )
        )

    print(
        "\nSaved:"
    )

    print(
        aligned_path
    )

    print(
        excluded_path
    )


print("\n" + "=" * 100)
print("STEP 3 — ALIGNMENT COMPLETE")
print("=" * 100)
```

```python
# ============================================================
# PERSON 2 — STEP 3.2
# CORRECTED ALIGNMENT USING ACTUAL SOURCE FILE PATHS
# ============================================================

import os
import glob
import pandas as pd

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

CANONICAL_ROOT = "/content/person2_output/canonical_features"
SOURCE_ROOT = "/content/person2_output/all_videos"
OUTPUT_ROOT = "/content/person2_output/aligned_canonical"

os.makedirs(
    OUTPUT_ROOT,
    exist_ok=True
)

META_FILES = {
    "train": "/content/train (1).csv",
    "val": "/content/val.csv",
    "test": "/content/test.csv"
}

EXPECTED_FEATURES = 206


# ------------------------------------------------------------
# FIND CANONICAL FILES
# ------------------------------------------------------------

canonical_files = glob.glob(
    os.path.join(
        CANONICAL_ROOT,
        "*",
        "*_canonical_features.csv"
    )
)

canonical_map = {}

for path in canonical_files:

    filename = os.path.basename(path)

    stem = filename.replace(
        "_canonical_features.csv",
        ""
    )

    canonical_map[stem.lower()] = path


# ------------------------------------------------------------
# FIND ORIGINAL SOURCE FEATURE FILES
#
# IMPORTANT:
# We scan the actual files instead of constructing paths.
# ------------------------------------------------------------

source_files = glob.glob(
    os.path.join(
        SOURCE_ROOT,
        "*",
        "*_final_features_with_hand_object_distance.csv"
    )
)

source_map = {}

for path in source_files:

    filename = os.path.basename(path)

    stem = filename.replace(
        "_final_features_with_hand_object_distance.csv",
        ""
    )

    source_map[stem.lower()] = path


# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

print("=" * 100)
print("PERSON 2 — CORRECTED ALIGNMENT")
print("=" * 100)

print(
    "Canonical files:",
    len(canonical_map)
)

print(
    "Source files   :",
    len(source_map)
)

print(
    "Expected       : 117"
)


# ------------------------------------------------------------
# CHECK MAP CONSISTENCY
# ------------------------------------------------------------

missing_source = sorted(
    set(canonical_map) -
    set(source_map)
)

missing_canonical = sorted(
    set(source_map) -
    set(canonical_map)
)

if missing_source:

    print("\n⚠️ Canonical files without source:")
    print(missing_source)

if missing_canonical:

    print("\n⚠️ Source files without canonical:")
    print(missing_canonical)

if not missing_source and not missing_canonical:

    print(
        "\n✅ Canonical/source maps contain "
        "the same 117 videos."
    )


# ------------------------------------------------------------
# PROCESS EACH SPLIT
# ------------------------------------------------------------

for split, meta_path in META_FILES.items():

    print("\n" + "=" * 100)
    print(f"PROCESSING {split.upper()}")
    print("=" * 100)

    meta = pd.read_csv(
        meta_path
    )

    print(
        "Metadata rows:",
        len(meta)
    )

    aligned_parts = []
    excluded_rows = []

    # --------------------------------------------------------
    # CACHE LOADED VIDEOS
    #
    # Prevent repeatedly reading the same source file.
    # --------------------------------------------------------

    source_cache = {}
    canonical_cache = {}

    # --------------------------------------------------------
    # PROCESS METADATA ROWS
    # --------------------------------------------------------

    for row_idx, row in meta.iterrows():

        video_file = str(
            row["video_file"]
        )

        video_stem = os.path.splitext(
            os.path.basename(video_file)
        )[0].lower()

        clip_id = str(
            row["clip_id"]
        )

        start_frame = int(
            row["start_frame"]
        )

        end_frame = int(
            row["end_frame"]
        )

        segment_uid = (
            f"{split}_{row_idx:04d}_{clip_id}"
        )


        # ----------------------------------------------------
        # INVALID TEMPORAL RANGE
        # ----------------------------------------------------

        if start_frame > end_frame:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "reason": "Invalid temporal range"
            })

            continue


        # ----------------------------------------------------
        # CHECK CANONICAL FILE
        # ----------------------------------------------------

        canonical_path = canonical_map.get(
            video_stem
        )

        if canonical_path is None:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "reason": "Canonical feature file not found"
            })

            continue


        # ----------------------------------------------------
        # CHECK SOURCE FILE
        # ----------------------------------------------------

        source_path = source_map.get(
            video_stem
        )

        if source_path is None:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "reason": "Source feature file not found"
            })

            continue


        # ----------------------------------------------------
        # LOAD CANONICAL FEATURES
        # ----------------------------------------------------

        if video_stem not in canonical_cache:

            canonical_cache[
                video_stem
            ] = pd.read_csv(
                canonical_path
            )

        canonical = canonical_cache[
            video_stem
        ]


        # ----------------------------------------------------
        # LOAD SOURCE FRAME COLUMN
        # ----------------------------------------------------

        if video_stem not in source_cache:

            source_cache[
                video_stem
            ] = pd.read_csv(
                source_path,
                usecols=["frame"]
            )

        source_frames = source_cache[
            video_stem
        ]


        # ----------------------------------------------------
        # VERIFY ROW COUNTS
        # ----------------------------------------------------

        if len(canonical) != len(source_frames):

            raise ValueError(
                f"{video_stem}: canonical rows "
                f"({len(canonical)}) != source rows "
                f"({len(source_frames)})"
            )


        # ----------------------------------------------------
        # VERIFY CANONICAL FEATURES
        # ----------------------------------------------------

        if canonical.shape[1] != EXPECTED_FEATURES:

            raise ValueError(
                f"{video_stem}: expected "
                f"{EXPECTED_FEATURES} features, got "
                f"{canonical.shape[1]}"
            )


        # ----------------------------------------------------
        # ADD FRAME
        # ----------------------------------------------------

        features = canonical.copy()

        features.insert(
            0,
            "frame",
            source_frames["frame"].to_numpy()
        )


        # ----------------------------------------------------
        # SELECT SEGMENT
        # ----------------------------------------------------

        segment = features[
            (features["frame"] >= start_frame) &
            (features["frame"] <= end_frame)
        ].copy()


        expected_frames = (
            end_frame -
            start_frame +
            1
        )

        actual_frames = len(
            segment
        )


        # ----------------------------------------------------
        # FRAME COUNT CHECK
        # ----------------------------------------------------

        if actual_frames != expected_frames:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "expected_frames": expected_frames,
                "actual_frames": actual_frames,
                "reason": "Frame count mismatch"
            })

            continue


        # ----------------------------------------------------
        # FRAME CONTINUITY
        # ----------------------------------------------------

        actual_frame_list = (
            segment["frame"]
            .astype(int)
            .tolist()
        )

        expected_frame_list = list(
            range(
                start_frame,
                end_frame + 1
            )
        )

        if actual_frame_list != expected_frame_list:

            excluded_rows.append({
                "split": split,
                "metadata_row": row_idx,
                "segment_uid": segment_uid,
                "clip_id": clip_id,
                "video_id": row["video_id"],
                "video_file": video_file,
                "start_frame": start_frame,
                "end_frame": end_frame,
                "expected_frames": expected_frames,
                "actual_frames": actual_frames,
                "reason": "Frame sequence not continuous"
            })

            continue


        # ----------------------------------------------------
        # FEATURE NaN CHECK
        # ----------------------------------------------------

        feature_nan_count = int(
            segment.iloc[:, 1:]
            .isna()
            .sum()
            .sum()
        )

        if feature_nan_count != 0:

            raise ValueError(
                f"{segment_uid}: "
                f"{feature_nan_count} feature NaNs"
            )


        # ----------------------------------------------------
        # ADD METADATA
        # ----------------------------------------------------

        segment["segment_uid"] = segment_uid

        segment["clip_id"] = clip_id

        segment["video_id"] = row[
            "video_id"
        ]

        segment["activity"] = row[
            "activity"
        ]

        segment["label"] = row[
            "label"
        ]

        segment["step_id"] = row[
            "step_id"
        ]

        segment["behavior"] = row[
            "behavior"
        ]

        segment["variation"] = row[
            "variation"
        ]

        segment["metadata_row"] = row_idx


        aligned_parts.append(
            segment
        )


    # --------------------------------------------------------
    # COMBINE
    # --------------------------------------------------------

    if aligned_parts:

        aligned = pd.concat(
            aligned_parts,
            ignore_index=True
        )

    else:

        aligned = pd.DataFrame()


    # --------------------------------------------------------
    # SAVE
    # --------------------------------------------------------

    aligned_path = os.path.join(
        OUTPUT_ROOT,
        f"{split}_aligned_canonical.csv"
    )

    aligned.to_csv(
        aligned_path,
        index=False
    )


    excluded_df = pd.DataFrame(
        excluded_rows
    )

    excluded_path = os.path.join(
        OUTPUT_ROOT,
        f"{split}_excluded.csv"
    )

    excluded_df.to_csv(
        excluded_path,
        index=False
    )


    # --------------------------------------------------------
    # SUMMARY
    # --------------------------------------------------------

    print(
        "\nOriginal metadata rows :",
        len(meta)
    )

    print(
        "Aligned segment rows   :",
        aligned["segment_uid"].nunique()
        if not aligned.empty
        else 0
    )

    print(
        "Aligned frame rows     :",
        len(aligned)
    )

    print(
        "Excluded metadata rows :",
        len(excluded_df)
    )

    if not aligned.empty:

        feature_columns = [
            c for c in aligned.columns
            if c not in [
                "frame",
                "segment_uid",
                "clip_id",
                "video_id",
                "activity",
                "label",
                "step_id",
                "behavior",
                "variation",
                "metadata_row"
            ]
        ]

        print(
            "Feature columns        :",
            len(feature_columns)
        )

        print(
            "Feature NaNs           :",
            int(
                aligned[feature_columns]
                .isna()
                .sum()
                .sum()
            )
        )

    print(
        "\nSaved:"
    )

    print(
        aligned_path
    )

    print(
        excluded_path
    )


print("\n" + "=" * 100)
print("STEP 3.2 — CORRECTED ALIGNMENT COMPLETE")
print("=" * 100)
```

```python
# ============================================================
# PERSON 2 — STEP 4
# COMPREHENSIVE ALIGNMENT VERIFICATION
# ============================================================

import os
import pandas as pd

ALIGN_ROOT = "/content/person2_output/aligned_canonical"

SPLIT_FILES = {
    "train": os.path.join(
        ALIGN_ROOT,
        "train_aligned_canonical.csv"
    ),
    "val": os.path.join(
        ALIGN_ROOT,
        "val_aligned_canonical.csv"
    ),
    "test": os.path.join(
        ALIGN_ROOT,
        "test_aligned_canonical.csv"
    )
}

EXPECTED_SEGMENTS = {
    "train": 1334,
    "val": 313,
    "test": 294
}

# Metadata columns added during alignment
META_COLUMNS = [
    "frame",
    "segment_uid",
    "clip_id",
    "video_id",
    "activity",
    "label",
    "step_id",
    "behavior",
    "variation",
    "metadata_row"
]

EXPECTED_FEATURE_COUNT = 206

EXPECTED_LABELS = {0, 1, 2, 3, 4, 5}


# ============================================================
# VERIFY EACH SPLIT
# ============================================================

results = {}

for split, path in SPLIT_FILES.items():

    print("\n" + "=" * 100)
    print(f"VERIFYING {split.upper()}")
    print("=" * 100)

    df = pd.read_csv(path)

    # --------------------------------------------------------
    # FEATURE COLUMNS
    # --------------------------------------------------------

    feature_columns = [
        c for c in df.columns
        if c not in META_COLUMNS
    ]

    print(
        "Total rows          :",
        len(df)
    )

    print(
        "Total columns       :",
        len(df.columns)
    )

    print(
        "Feature columns     :",
        len(feature_columns)
    )

    # --------------------------------------------------------
    # SEGMENTS
    # --------------------------------------------------------

    unique_segments = (
        df["segment_uid"]
        .nunique()
    )

    print(
        "Unique segments     :",
        unique_segments
    )

    print(
        "Expected segments   :",
        EXPECTED_SEGMENTS[split]
    )

    # --------------------------------------------------------
    # DUPLICATE SEGMENT UID
    # --------------------------------------------------------

    duplicate_segment_uids = (
        df["segment_uid"]
        .duplicated()
        .sum()
    )

    # This is expected because each segment has many frames.
    # What matters is whether the same segment UID appears
    # with multiple metadata definitions.

    segment_metadata_counts = (
        df.groupby("segment_uid")
        .agg({
            "clip_id": "nunique",
            "video_id": "nunique",
            "metadata_row": "nunique"
        })
    )

    bad_segment_metadata = (
        segment_metadata_counts[
            (segment_metadata_counts["clip_id"] > 1) |
            (segment_metadata_counts["video_id"] > 1) |
            (segment_metadata_counts["metadata_row"] > 1)
        ]
    )

    print(
        "Segments with conflicting metadata:",
        len(bad_segment_metadata)
    )

    # --------------------------------------------------------
    # DUPLICATE SEGMENT + FRAME
    # --------------------------------------------------------

    duplicate_segment_frames = (
        df.duplicated(
            subset=[
                "segment_uid",
                "frame"
            ]
        ).sum()
    )

    print(
        "Duplicate segment/frame:",
        duplicate_segment_frames
    )

    # --------------------------------------------------------
    # FEATURE NaNs
    # --------------------------------------------------------

    feature_nan_count = int(
        df[feature_columns]
        .isna()
        .sum()
        .sum()
    )

    print(
        "Feature NaNs        :",
        feature_nan_count
    )

    # --------------------------------------------------------
    # LABELS
    # --------------------------------------------------------

    labels = set(
        df["label"]
        .dropna()
        .astype(int)
        .unique()
        .tolist()
    )

    print(
        "Labels found        :",
        sorted(labels)
    )

    unexpected_labels = (
        labels - EXPECTED_LABELS
    )

    print(
        "Unexpected labels   :",
        sorted(unexpected_labels)
    )

    # --------------------------------------------------------
    # SEGMENT LENGTHS
    # --------------------------------------------------------

    segment_lengths = (
        df.groupby("segment_uid")["frame"]
        .nunique()
    )

    print(
        "Segment length min  :",
        int(segment_lengths.min())
    )

    print(
        "Segment length max  :",
        int(segment_lengths.max())
    )

    print(
        "Segment length mean :",
        round(float(segment_lengths.mean()), 4)
    )

    exactly_32 = int(
        (segment_lengths == 32).sum()
    )

    shorter_32 = int(
        (segment_lengths < 32).sum()
    )

    longer_32 = int(
        (segment_lengths > 32).sum()
    )

    print(
        "Exactly 32 frames   :",
        exactly_32
    )

    print(
        "Shorter than 32     :",
        shorter_32
    )

    print(
        "Longer than 32      :",
        longer_32
    )

    # --------------------------------------------------------
    # FRAME CONTINUITY
    # --------------------------------------------------------

    continuity_errors = []

    for segment_uid, group in df.groupby(
        "segment_uid",
        sort=False
    ):

        frames = (
            group["frame"]
            .astype(int)
            .tolist()
        )

        expected = list(
            range(
                min(frames),
                max(frames) + 1
            )
        )

        if frames != expected:

            continuity_errors.append(
                segment_uid
            )

    print(
        "Frame continuity errors:",
        len(continuity_errors)
    )

    # --------------------------------------------------------
    # MISSING ROWS INSIDE SEGMENTS
    # --------------------------------------------------------

    missing_frame_segments = []

    for segment_uid, group in df.groupby(
        "segment_uid",
        sort=False
    ):

        start = int(
            group["frame"].min()
        )

        end = int(
            group["frame"].max()
        )

        expected_count = (
            end - start + 1
        )

        actual_count = len(group)

        if expected_count != actual_count:

            missing_frame_segments.append(
                segment_uid
            )

    print(
        "Segments with missing frames:",
        len(missing_frame_segments)
    )

    # --------------------------------------------------------
    # VIDEO IDS
    # --------------------------------------------------------

    video_count = (
        df["video_id"]
        .nunique()
    )

    print(
        "Unique video IDs     :",
        video_count
    )

    # --------------------------------------------------------
    # SPLIT VALIDATION
    # --------------------------------------------------------

    results[split] = {
        "rows": len(df),
        "columns": len(df.columns),
        "features": len(feature_columns),
        "segments": unique_segments,
        "expected_segments": EXPECTED_SEGMENTS[split],
        "feature_nan_count": feature_nan_count,
        "unexpected_labels": sorted(
            unexpected_labels
        ),
        "duplicate_segment_frames":
            int(duplicate_segment_frames),
        "metadata_conflicts":
            len(bad_segment_metadata),
        "continuity_errors":
            len(continuity_errors),
        "missing_frame_segments":
            len(missing_frame_segments),
        "min_length":
            int(segment_lengths.min()),
        "max_length":
            int(segment_lengths.max()),
        "longer_than_32":
            longer_32
    }


# ============================================================
# CHECK TRAIN / VAL / TEST SEGMENT OVERLAP
# ============================================================

print("\n" + "=" * 100)
print("TRAIN / VAL / TEST OVERLAP CHECK")
print("=" * 100)

segment_sets = {}

for split, path in SPLIT_FILES.items():

    df = pd.read_csv(
        path,
        usecols=["segment_uid"]
    )

    segment_sets[split] = set(
        df["segment_uid"]
    )

train_val_overlap = (
    segment_sets["train"] &
    segment_sets["val"]
)

train_test_overlap = (
    segment_sets["train"] &
    segment_sets["test"]
)

val_test_overlap = (
    segment_sets["val"] &
    segment_sets["test"]
)

print(
    "Train ∩ Val  :",
    len(train_val_overlap)
)

print(
    "Train ∩ Test :",
    len(train_test_overlap)
)

print(
    "Val ∩ Test   :",
    len(val_test_overlap)
)


# ============================================================
# FINAL PASS / FAIL
# ============================================================

all_pass = True

for split, r in results.items():

    checks = [
        r["features"] == EXPECTED_FEATURE_COUNT,
        r["segments"] == r["expected_segments"],
        r["feature_nan_count"] == 0,
        len(r["unexpected_labels"]) == 0,
        r["duplicate_segment_frames"] == 0,
        r["metadata_conflicts"] == 0,
        r["continuity_errors"] == 0,
        r["missing_frame_segments"] == 0,
        r["max_length"] <= 32
    ]

    if all(checks):

        print(
            f"✅ {split.upper()} PASSED"
        )

    else:

        print(
            f"❌ {split.upper()} FAILED"
        )

        all_pass = False


if (
    len(train_val_overlap) == 0
    and len(train_test_overlap) == 0
    and len(val_test_overlap) == 0
):

    print(
        "✅ Train/Val/Test segment separation PASSED"
    )

else:

    print(
        "❌ Train/Val/Test segment separation FAILED"
    )

    all_pass = False


print("\n" + "=" * 100)

if all_pass:

    print(
        "🎉 STEP 4 PASSED — ALIGNMENT IS CLEAN"
    )

    print(
        "Ready for 32-frame sequence generation."
    )

else:

    print(
        "❌ STEP 4 FAILED — DO NOT GENERATE SEQUENCES YET"
    )

print("=" * 100)
```

```python
# ============================================================
# PERSON 2 — STEP 5
# GENERATE FIXED-LENGTH 32-FRAME SEQUENCES
# ============================================================

import os
import numpy as np
import pandas as pd

ALIGN_ROOT = "/content/person2_output/aligned_canonical"
SEQUENCE_ROOT = "/content/person2_output/sequences_32"

os.makedirs(SEQUENCE_ROOT, exist_ok=True)

SPLIT_FILES = {
    "train": os.path.join(
        ALIGN_ROOT,
        "train_aligned_canonical.csv"
    ),
    "val": os.path.join(
        ALIGN_ROOT,
        "val_aligned_canonical.csv"
    ),
    "test": os.path.join(
        ALIGN_ROOT,
        "test_aligned_canonical.csv"
    )
}

SEQUENCE_LENGTH = 32

# ============================================================
# IMPORTANT:
# These are the exact 206 model features verified in Step 4.
# ============================================================

META_COLUMNS = [
    "frame",
    "segment_uid",
    "clip_id",
    "video_id",
    "activity",
    "label",
    "step_id",
    "behavior",
    "variation",
    "metadata_row"
]

# Load once to establish exact feature order
sample_df = pd.read_csv(
    SPLIT_FILES["train"],
    nrows=1
)

FEATURE_COLUMNS = [
    c for c in sample_df.columns
    if c not in META_COLUMNS
]

assert len(FEATURE_COLUMNS) == 206, (
    f"Expected 206 features, found {len(FEATURE_COLUMNS)}"
)

print("=" * 100)
print("PERSON 2 — STEP 5: 32-FRAME SEQUENCE GENERATION")
print("=" * 100)

print(f"Feature count     : {len(FEATURE_COLUMNS)}")
print(f"Sequence length   : {SEQUENCE_LENGTH}")
print(f"Sequence shape    : ({SEQUENCE_LENGTH}, {len(FEATURE_COLUMNS)})")


# ============================================================
# PROCESS EACH SPLIT
# ============================================================

sequence_summary = {}

for split, path in SPLIT_FILES.items():

    print("\n" + "=" * 100)
    print(f"PROCESSING {split.upper()}")
    print("=" * 100)

    df = pd.read_csv(path)

    # --------------------------------------------------------
    # Safety checks
    # --------------------------------------------------------

    assert len(
        [c for c in df.columns if c not in META_COLUMNS]
    ) == 206

    assert df[FEATURE_COLUMNS].isna().sum().sum() == 0

    # --------------------------------------------------------
    # Storage
    # --------------------------------------------------------

    X_list = []
    y_list = []

    segment_ids = []
    clip_ids = []
    video_ids = []
    activities = []
    step_ids = []
    behaviors = []
    variations = []

    original_lengths = []

    # --------------------------------------------------------
    # GROUP BY SEGMENT UID
    # --------------------------------------------------------

    grouped = df.groupby(
        "segment_uid",
        sort=False
    )

    for segment_uid, group in grouped:

        # Preserve temporal order
        group = group.sort_values(
            "frame"
        ).reset_index(drop=True)

        # ----------------------------------------------------
        # Extract features
        # ----------------------------------------------------

        features = group[
            FEATURE_COLUMNS
        ].to_numpy(
            dtype=np.float32
        )

        original_length = len(features)

        # Safety: no segment should exceed 32
        if original_length > SEQUENCE_LENGTH:

            raise ValueError(
                f"{split}: segment {segment_uid} "
                f"has {original_length} frames (>32)"
            )

        # ----------------------------------------------------
        # Zero padding
        # ----------------------------------------------------

        if original_length < SEQUENCE_LENGTH:

            padding = np.zeros(
                (
                    SEQUENCE_LENGTH - original_length,
                    len(FEATURE_COLUMNS)
                ),
                dtype=np.float32
            )

            features = np.vstack(
                [
                    features,
                    padding
                ]
            )

        # ----------------------------------------------------
        # Final shape check
        # ----------------------------------------------------

        assert features.shape == (
            SEQUENCE_LENGTH,
            len(FEATURE_COLUMNS)
        )

        # ----------------------------------------------------
        # Label / metadata
        # ----------------------------------------------------

        label_values = (
            group["label"]
            .dropna()
            .astype(int)
            .unique()
        )

        if len(label_values) != 1:

            raise ValueError(
                f"{split}: segment {segment_uid} "
                f"has inconsistent labels: {label_values}"
            )

        label = int(label_values[0])

        # ----------------------------------------------------
        # Store
        # ----------------------------------------------------

        X_list.append(features)
        y_list.append(label)

        segment_ids.append(segment_uid)
        clip_ids.append(
            group["clip_id"].iloc[0]
        )
        video_ids.append(
            group["video_id"].iloc[0]
        )
        activities.append(
            group["activity"].iloc[0]
        )
        step_ids.append(
            group["step_id"].iloc[0]
        )
        behaviors.append(
            group["behavior"].iloc[0]
        )
        variations.append(
            group["variation"].iloc[0]
        )

        original_lengths.append(
            original_length
        )

    # --------------------------------------------------------
    # Convert to NumPy arrays
    # --------------------------------------------------------

    X = np.stack(
        X_list
    ).astype(
        np.float32
    )

    y = np.asarray(
        y_list,
        dtype=np.int64
    )

    # --------------------------------------------------------
    # Final validation
    # --------------------------------------------------------

    expected_shape = (
        len(segment_ids),
        SEQUENCE_LENGTH,
        len(FEATURE_COLUMNS)
    )

    assert X.shape == expected_shape

    assert y.shape == (
        len(segment_ids),
    )

    assert len(
        set(segment_ids)
    ) == len(segment_ids)

    assert not np.isnan(X).any()

    assert not np.isinf(X).any()

    assert set(
        np.unique(y)
    ).issubset(
        {0, 1, 2, 3, 4, 5}
    )

    # --------------------------------------------------------
    # Metadata table
    # --------------------------------------------------------

    sequence_metadata = pd.DataFrame({
        "sequence_index": np.arange(
            len(segment_ids)
        ),
        "segment_uid": segment_ids,
        "clip_id": clip_ids,
        "video_id": video_ids,
        "activity": activities,
        "label": y,
        "step_id": step_ids,
        "behavior": behaviors,
        "variation": variations,
        "original_length": original_lengths,
        "padded_frames": [
            SEQUENCE_LENGTH - n
            for n in original_lengths
        ]
    })

    # --------------------------------------------------------
    # Save arrays
    # --------------------------------------------------------

    X_path = os.path.join(
        SEQUENCE_ROOT,
        f"X_{split}.npy"
    )

    y_path = os.path.join(
        SEQUENCE_ROOT,
        f"y_{split}.npy"
    )

    metadata_path = os.path.join(
        SEQUENCE_ROOT,
        f"{split}_sequence_metadata.csv"
    )

    feature_names_path = os.path.join(
        SEQUENCE_ROOT,
        "feature_names.txt"
    )

    np.save(
        X_path,
        X
    )

    np.save(
        y_path,
        y
    )

    sequence_metadata.to_csv(
        metadata_path,
        index=False
    )

    # Save feature order once
    if not os.path.exists(
        feature_names_path
    ):

        with open(
            feature_names_path,
            "w"
        ) as f:

            for feature in FEATURE_COLUMNS:
                f.write(
                    feature + "\n"
                )

    # --------------------------------------------------------
    # Summary
    # --------------------------------------------------------

    sequence_summary[split] = {
        "sequences": len(X),
        "sequence_length": X.shape[1],
        "features": X.shape[2],
        "min_original_length": min(original_lengths),
        "max_original_length": max(original_lengths),
        "mean_original_length": float(
            np.mean(original_lengths)
        ),
        "padded_sequences": int(
            np.sum(
                np.asarray(original_lengths)
                < SEQUENCE_LENGTH
            )
        ),
        "nan_values": int(
            np.isnan(X).sum()
        ),
        "inf_values": int(
            np.isinf(X).sum()
        )
    }

    print(
        f"Sequences         : {X.shape[0]}"
    )

    print(
        f"X shape            : {X.shape}"
    )

    print(
        f"y shape            : {y.shape}"
    )

    print(
        f"Original min length: {min(original_lengths)}"
    )

    print(
        f"Original max length: {max(original_lengths)}"
    )

    print(
        f"Original mean      : "
        f"{np.mean(original_lengths):.4f}"
    )

    print(
        f"Padded sequences   : "
        f"{np.sum(np.asarray(original_lengths) < 32)}"
    )

    print(
        f"NaN values         : {np.isnan(X).sum()}"
    )

    print(
        f"Inf values         : {np.isinf(X).sum()}"
    )

    print(
        f"Saved X            : {X_path}"
    )

    print(
        f"Saved y            : {y_path}"
    )

    print(
        f"Saved metadata     : {metadata_path}"
    )


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("STEP 5 — SEQUENCE GENERATION SUMMARY")
print("=" * 100)

for split, summary in sequence_summary.items():

    print(
        f"\n{split.upper()}"
    )

    print(
        f"  Sequences       : {summary['sequences']}"
    )

    print(
        f"  X shape         : "
        f"({summary['sequences']}, "
        f"{summary['sequence_length']}, "
        f"{summary['features']})"
    )

    print(
        f"  y shape         : "
        f"({summary['sequences']},)"
    )

    print(
        f"  Padded sequences: "
        f"{summary['padded_sequences']}"
    )

    print(
        f"  NaNs            : "
        f"{summary['nan_values']}"
    )

    print(
        f"  Infs            : "
        f"{summary['inf_values']}"
    )

print("\n" + "=" * 100)
print("🎉 STEP 5 COMPLETE — 32-FRAME SEQUENCES GENERATED")
print("=" * 100)
```

```python
# ============================================================
# PERSON 2 — STEP 6
# FEATURE SCALING
# ============================================================

import os
import numpy as np
import joblib
from sklearn.preprocessing import StandardScaler

SEQUENCE_ROOT = "/content/person2_output/sequences_32"
SCALER_PATH = os.path.join(
    SEQUENCE_ROOT,
    "feature_scaler.pkl"
)

# ============================================================
# LOAD SEQUENCES
# ============================================================

X_train = np.load(
    os.path.join(SEQUENCE_ROOT, "X_train.npy")
)

X_val = np.load(
    os.path.join(SEQUENCE_ROOT, "X_val.npy")
)

X_test = np.load(
    os.path.join(SEQUENCE_ROOT, "X_test.npy")
)

y_train = np.load(
    os.path.join(SEQUENCE_ROOT, "y_train.npy")
)

y_val = np.load(
    os.path.join(SEQUENCE_ROOT, "y_val.npy")
)

y_test = np.load(
    os.path.join(SEQUENCE_ROOT, "y_test.npy")
)

print("=" * 100)
print("PERSON 2 — STEP 6: FEATURE SCALING")
print("=" * 100)

print("Original shapes:")
print("X_train:", X_train.shape)
print("X_val  :", X_val.shape)
print("X_test :", X_test.shape)


# ============================================================
# CHECK INPUTS
# ============================================================

assert X_train.shape == (1334, 32, 206)
assert X_val.shape == (313, 32, 206)
assert X_test.shape == (294, 32, 206)

assert not np.isnan(X_train).any()
assert not np.isnan(X_val).any()
assert not np.isnan(X_test).any()

assert not np.isinf(X_train).any()
assert not np.isinf(X_val).any()
assert not np.isinf(X_test).any()


# ============================================================
# RESHAPE FOR STANDARD SCALER
#
# StandardScaler expects:
#   (samples, features)
#
# Our data is:
#   (samples, frames, features)
#
# Therefore flatten samples + frames temporarily.
# ============================================================

n_train, seq_len, n_features = X_train.shape
n_val = X_val.shape[0]
n_test = X_test.shape[0]

X_train_2d = X_train.reshape(
    -1,
    n_features
)

X_val_2d = X_val.reshape(
    -1,
    n_features
)

X_test_2d = X_test.reshape(
    -1,
    n_features
)

print("\n2D shapes used for scaling:")
print("X_train_2d:", X_train_2d.shape)
print("X_val_2d  :", X_val_2d.shape)
print("X_test_2d :", X_test_2d.shape)


# ============================================================
# FIT SCALER ONLY ON TRAIN
# ============================================================

scaler = StandardScaler()

scaler.fit(
    X_train_2d
)

print("\n✅ StandardScaler fitted using TRAIN ONLY")


# ============================================================
# TRANSFORM ALL SPLITS
# ============================================================

X_train_scaled_2d = scaler.transform(
    X_train_2d
)

X_val_scaled_2d = scaler.transform(
    X_val_2d
)

X_test_scaled_2d = scaler.transform(
    X_test_2d
)


# ============================================================
# RESTORE 3D SEQUENCE SHAPE
# ============================================================

X_train_scaled = X_train_scaled_2d.reshape(
    n_train,
    seq_len,
    n_features
).astype(
    np.float32
)

X_val_scaled = X_val_scaled_2d.reshape(
    n_val,
    seq_len,
    n_features
).astype(
    np.float32
)

X_test_scaled = X_test_scaled_2d.reshape(
    n_test,
    seq_len,
    n_features
).astype(
    np.float32
)


# ============================================================
# VALIDATION
# ============================================================

assert X_train_scaled.shape == (
    1334,
    32,
    206
)

assert X_val_scaled.shape == (
    313,
    32,
    206
)

assert X_test_scaled.shape == (
    294,
    32,
    206
)

assert not np.isnan(
    X_train_scaled
).any()

assert not np.isnan(
    X_val_scaled
).any()

assert not np.isnan(
    X_test_scaled
).any()

assert not np.isinf(
    X_train_scaled
).any()

assert not np.isinf(
    X_val_scaled
).any()

assert not np.isinf(
    X_test_scaled
).any()


# ============================================================
# SAVE SCALED ARRAYS
# ============================================================

np.save(
    os.path.join(
        SEQUENCE_ROOT,
        "X_train_scaled.npy"
    ),
    X_train_scaled
)

np.save(
    os.path.join(
        SEQUENCE_ROOT,
        "X_val_scaled.npy"
    ),
    X_val_scaled
)

np.save(
    os.path.join(
        SEQUENCE_ROOT,
        "X_test_scaled.npy"
    ),
    X_test_scaled
)


# ============================================================
# SAVE SCALER
# ============================================================

joblib.dump(
    scaler,
    SCALER_PATH
)


# ============================================================
# REPORT
# ============================================================

print("\n" + "=" * 100)
print("SCALING RESULTS")
print("=" * 100)

print(
    "X_train_scaled:",
    X_train_scaled.shape
)

print(
    "X_val_scaled  :",
    X_val_scaled.shape
)

print(
    "X_test_scaled :",
    X_test_scaled.shape
)

print(
    "Train mean (overall):",
    float(X_train_scaled.mean())
)

print(
    "Train std (overall) :",
    float(X_train_scaled.std())
)

print(
    "Train NaNs:",
    int(np.isnan(X_train_scaled).sum())
)

print(
    "Val NaNs:",
    int(np.isnan(X_val_scaled).sum())
)

print(
    "Test NaNs:",
    int(np.isnan(X_test_scaled).sum())
)

print(
    "\nScaler saved:",
    SCALER_PATH
)

print(
    "\nSaved:"
)

print(
    os.path.join(
        SEQUENCE_ROOT,
        "X_train_scaled.npy"
    )
)

print(
    os.path.join(
        SEQUENCE_ROOT,
        "X_val_scaled.npy"
    )
)

print(
    os.path.join(
        SEQUENCE_ROOT,
        "X_test_scaled.npy"
    )
)

print("\n" + "=" * 100)
print("🎉 STEP 6 COMPLETE — FEATURES SCALED")
print("=" * 100)
```

```python
# ============================================================
# PERSON 2 — STEP 7
# PADDING / SCALING DIAGNOSTIC
# ============================================================

import os
import numpy as np
import pandas as pd

SEQUENCE_ROOT = "/content/person2_output/sequences_32"

X_train = np.load(
    os.path.join(
        SEQUENCE_ROOT,
        "X_train.npy"
    )
)

X_train_scaled = np.load(
    os.path.join(
        SEQUENCE_ROOT,
        "X_train_scaled.npy"
    )
)

# ============================================================
# LOAD ORIGINAL LENGTH INFORMATION
# ============================================================

metadata = pd.read_csv(
    os.path.join(
        SEQUENCE_ROOT,
        "train_sequence_metadata.csv"
    )
)

assert len(metadata) == len(X_train)

# ============================================================
# IDENTIFY REAL VS PADDED FRAMES
# ============================================================

original_lengths = (
    metadata["original_length"]
    .astype(int)
    .to_numpy()
)

real_mask = np.zeros(
    X_train.shape[:2],
    dtype=bool
)

for i, length in enumerate(
    original_lengths
):
    real_mask[i, :length] = True

padded_mask = ~real_mask

# ============================================================
# BASIC COUNTS
# ============================================================

total_frames = X_train.shape[0] * X_train.shape[1]

real_frames = int(
    real_mask.sum()
)

padded_frames = int(
    padded_mask.sum()
)

print("=" * 100)
print("PERSON 2 — STEP 7: PADDING / SCALING DIAGNOSTIC")
print("=" * 100)

print(
    "Total sequence frames :",
    total_frames
)

print(
    "Real frames            :",
    real_frames
)

print(
    "Padded frames          :",
    padded_frames
)

print(
    "Padding percentage     :",
    round(
        100 * padded_frames / total_frames,
        4
    ),
    "%"
)

# ============================================================
# CHECK PADDED VALUES BEFORE SCALING
# ============================================================

padded_values = X_train[
    padded_mask
]

print("\n" + "-" * 100)
print("PADDED VALUES BEFORE SCALING")
print("-" * 100)

print(
    "Padded value min:",
    float(padded_values.min())
)

print(
    "Padded value max:",
    float(padded_values.max())
)

print(
    "Padded value mean:",
    float(padded_values.mean())
)

print(
    "Unique padded values:",
    np.unique(padded_values)[:10]
)

# ============================================================
# CHECK REAL VALUES BEFORE SCALING
# ============================================================

real_values = X_train[
    real_mask
]

print("\n" + "-" * 100)
print("REAL VALUES BEFORE SCALING")
print("-" * 100)

print(
    "Real value min:",
    float(real_values.min())
)

print(
    "Real value max:",
    float(real_values.max())
)

print(
    "Real value mean:",
    float(real_values.mean())
)

# ============================================================
# CHECK PADDED VALUES AFTER SCALING
# ============================================================

scaled_padded = X_train_scaled[
    padded_mask
]

scaled_real = X_train_scaled[
    real_mask
]

print("\n" + "-" * 100)
print("PADDED VALUES AFTER SCALING")
print("-" * 100)

print(
    "Scaled padded min:",
    float(scaled_padded.min())
)

print(
    "Scaled padded max:",
    float(scaled_padded.max())
)

print(
    "Scaled padded mean:",
    float(scaled_padded.mean())
)

# ============================================================
# CHECK REAL VALUES AFTER SCALING
# ============================================================

print("\n" + "-" * 100)
print("REAL VALUES AFTER SCALING")
print("-" * 100)

print(
    "Scaled real min:",
    float(scaled_real.min())
)

print(
    "Scaled real max:",
    float(scaled_real.max())
)

print(
    "Scaled real mean:",
    float(scaled_real.mean())
)

print(
    "Scaled real std:",
    float(scaled_real.std())
)

# ============================================================
# PER-FEATURE PADDING IMPACT
# ============================================================

real_mean = np.mean(
    np.where(
        real_mask[:, :, None],
        X_train,
        np.nan
    ),
    axis=(0, 1)
)

real_std = np.nanstd(
    np.where(
        real_mask[:, :, None],
        X_train,
        np.nan
    ),
    axis=(0, 1)
)

all_mean = X_train.reshape(
    -1,
    X_train.shape[-1]
).mean(axis=0)

all_std = X_train.reshape(
    -1,
    X_train.shape[-1]
).std(axis=0)

mean_difference = np.abs(
    all_mean - real_mean
)

std_difference = np.abs(
    all_std - real_std
)

print("\n" + "-" * 100)
print("PADDING IMPACT ON FEATURE STATISTICS")
print("-" * 100)

print(
    "Mean difference — average:",
    float(
        np.nanmean(mean_difference)
    )
)

print(
    "Mean difference — maximum:",
    float(
        np.nanmax(mean_difference)
    )
)

print(
    "Std difference — average:",
    float(
        np.nanmean(std_difference)
    )
)

print(
    "Std difference — maximum:",
    float(
        np.nanmax(std_difference)
    )
)

# ============================================================
# HOW MANY FEATURES ARE NOTICEABLY AFFECTED?
# ============================================================

mean_threshold = 0.01
std_threshold = 0.01

affected_mean_features = int(
    np.sum(
        mean_difference >
        mean_threshold
    )
)

affected_std_features = int(
    np.sum(
        std_difference >
        std_threshold
    )
)

print(
    "\nFeatures with mean difference > 0.01:",
    affected_mean_features,
    "/ 206"
)

print(
    "Features with std difference > 0.01:",
    affected_std_features,
    "/ 206"
)

# ============================================================
# FINAL
# ============================================================

print("\n" + "=" * 100)
print("🎯 STEP 7 DIAGNOSTIC COMPLETE")
print("=" * 100)
```

```python
# ============================================================
# PERSON 2 — STEP 8
# PADDING-AWARE FEATURE SCALING
# ============================================================

import os
import numpy as np
import pandas as pd
import joblib
from sklearn.preprocessing import StandardScaler

SEQUENCE_ROOT = "/content/person2_output/sequences_32"

SCALER_PATH = os.path.join(
    SEQUENCE_ROOT,
    "feature_scaler_padding_aware.pkl"
)

# ============================================================
# LOAD ORIGINAL (UNSCALED) SEQUENCES
# ============================================================

X_train = np.load(
    os.path.join(
        SEQUENCE_ROOT,
        "X_train.npy"
    )
)

X_val = np.load(
    os.path.join(
        SEQUENCE_ROOT,
        "X_val.npy"
    )
)

X_test = np.load(
    os.path.join(
        SEQUENCE_ROOT,
        "X_test.npy"
    )
)

y_train = np.load(
    os.path.join(
        SEQUENCE_ROOT,
        "y_train.npy"
    )
)

y_val = np.load(
    os.path.join(
        SEQUENCE_ROOT,
        "y_val.npy"
    )
)

y_test = np.load(
    os.path.join(
        SEQUENCE_ROOT,
        "y_test.npy"
    )
)

# ============================================================
# LOAD SEQUENCE METADATA
# ============================================================

train_meta = pd.read_csv(
    os.path.join(
        SEQUENCE_ROOT,
        "train_sequence_metadata.csv"
    )
)

val_meta = pd.read_csv(
    os.path.join(
        SEQUENCE_ROOT,
        "val_sequence_metadata.csv"
    )
)

test_meta = pd.read_csv(
    os.path.join(
        SEQUENCE_ROOT,
        "test_sequence_metadata.csv"
    )
)

# ============================================================
# BASIC CHECKS
# ============================================================

assert X_train.shape == (1334, 32, 206)
assert X_val.shape == (313, 32, 206)
assert X_test.shape == (294, 32, 206)

assert len(train_meta) == len(X_train)
assert len(val_meta) == len(X_val)
assert len(test_meta) == len(X_test)

assert not np.isnan(X_train).any()
assert not np.isnan(X_val).any()
assert not np.isnan(X_test).any()

assert not np.isinf(X_train).any()
assert not np.isinf(X_val).any()
assert not np.isinf(X_test).any()

print("=" * 100)
print("PERSON 2 — STEP 8: PADDING-AWARE FEATURE SCALING")
print("=" * 100)

print(
    "X_train:",
    X_train.shape
)

print(
    "X_val  :",
    X_val.shape
)

print(
    "X_test :",
    X_test.shape
)


# ============================================================
# BUILD REAL-FRAME MASKS
# ============================================================

def build_real_mask(metadata, sequence_length=32):

    lengths = (
        metadata["original_length"]
        .astype(int)
        .to_numpy()
    )

    mask = np.zeros(
        (
            len(metadata),
            sequence_length
        ),
        dtype=bool
    )

    for i, length in enumerate(lengths):

        if length < 0 or length > sequence_length:

            raise ValueError(
                f"Invalid sequence length: {length}"
            )

        mask[
            i,
            :length
        ] = True

    return mask


train_mask = build_real_mask(
    train_meta
)

val_mask = build_real_mask(
    val_meta
)

test_mask = build_real_mask(
    test_meta
)


# ============================================================
# COUNT REAL / PADDED FRAMES
# ============================================================

train_real = int(
    train_mask.sum()
)

val_real = int(
    val_mask.sum()
)

test_real = int(
    test_mask.sum()
)

train_total = X_train.shape[0] * X_train.shape[1]
val_total = X_val.shape[0] * X_val.shape[1]
test_total = X_test.shape[0] * X_test.shape[1]

print("\nFrame counts:")

print(
    "Train real   :",
    train_real
)

print(
    "Train padded :",
    train_total - train_real
)

print(
    "Val real     :",
    val_real
)

print(
    "Val padded   :",
    val_total - val_real
)

print(
    "Test real    :",
    test_real
)

print(
    "Test padded  :",
    test_total - test_real
)


# ============================================================
# EXTRACT ONLY REAL TRAIN FRAMES
# ============================================================

X_train_real = X_train[
    train_mask
]

print("\nReal training frame matrix:")
print(
    X_train_real.shape
)

assert X_train_real.shape == (
    train_real,
    206
)


# ============================================================
# FIT SCALER ON REAL TRAIN FRAMES ONLY
# ============================================================

scaler = StandardScaler()

scaler.fit(
    X_train_real
)

print(
    "\n✅ StandardScaler fitted using REAL TRAIN FRAMES ONLY"
)


# ============================================================
# FUNCTION TO SCALE WHILE PRESERVING PADDING
# ============================================================

def scale_with_padding(
    X,
    real_mask,
    scaler
):

    X_scaled = np.zeros(
        X.shape,
        dtype=np.float32
    )

    real_values = X[
        real_mask
    ]

    scaled_real = scaler.transform(
        real_values
    ).astype(
        np.float32
    )

    X_scaled[
        real_mask
    ] = scaled_real

    # Padded frames remain exactly zero.
    return X_scaled


# ============================================================
# SCALE ALL SPLITS
# ============================================================

X_train_scaled_pa = scale_with_padding(
    X_train,
    train_mask,
    scaler
)

X_val_scaled_pa = scale_with_padding(
    X_val,
    val_mask,
    scaler
)

X_test_scaled_pa = scale_with_padding(
    X_test,
    test_mask,
    scaler
)


# ============================================================
# VALIDATION
# ============================================================

assert X_train_scaled_pa.shape == (
    1334,
    32,
    206
)

assert X_val_scaled_pa.shape == (
    313,
    32,
    206
)

assert X_test_scaled_pa.shape == (
    294,
    32,
    206
)

assert not np.isnan(
    X_train_scaled_pa
).any()

assert not np.isnan(
    X_val_scaled_pa
).any()

assert not np.isnan(
    X_test_scaled_pa
).any()

assert not np.isinf(
    X_train_scaled_pa
).any()

assert not np.isinf(
    X_val_scaled_pa
).any()

assert not np.isinf(
    X_test_scaled_pa
).any()


# ============================================================
# VERIFY PADDING REMAINS EXACTLY ZERO
# ============================================================

train_padding = X_train_scaled_pa[
    ~train_mask
]

val_padding = X_val_scaled_pa[
    ~val_mask
]

test_padding = X_test_scaled_pa[
    ~test_mask
]

assert np.all(
    train_padding == 0
)

assert np.all(
    val_padding == 0
)

assert np.all(
    test_padding == 0
)

print(
    "\n✅ All padded frames remain exactly zero"
)


# ============================================================
# VERIFY REAL TRAIN STATISTICS
# ============================================================

train_scaled_real = X_train_scaled_pa[
    train_mask
]

print("\n" + "-" * 100)
print("REAL TRAIN FRAME STATISTICS AFTER SCALING")
print("-" * 100)

print(
    "Mean:",
    float(
        train_scaled_real.mean()
    )
)

print(
    "Std :",
    float(
        train_scaled_real.std()
    )
)

print(
    "Min :",
    float(
        train_scaled_real.min()
    )
)

print(
    "Max :",
    float(
        train_scaled_real.max()
    )
)


# ============================================================
# PER-FEATURE CHECK
# ============================================================

per_feature_mean = (
    train_scaled_real.mean(axis=0)
)

per_feature_std = (
    train_scaled_real.std(axis=0)
)

print(
    "\nMaximum absolute per-feature mean:",
    float(
        np.max(
            np.abs(
                per_feature_mean
            )
        )
    )
)

print(
    "Largest per-feature std deviation from 1:",
    float(
        np.max(
            np.abs(
                per_feature_std - 1.0
            )
        )
    )
)


# ============================================================
# SAVE NEW SCALED ARRAYS
# ============================================================

train_scaled_path = os.path.join(
    SEQUENCE_ROOT,
    "X_train_scaled_padding_aware.npy"
)

val_scaled_path = os.path.join(
    SEQUENCE_ROOT,
    "X_val_scaled_padding_aware.npy"
)

test_scaled_path = os.path.join(
    SEQUENCE_ROOT,
    "X_test_scaled_padding_aware.npy"
)

np.save(
    train_scaled_path,
    X_train_scaled_pa
)

np.save(
    val_scaled_path,
    X_val_scaled_pa
)

np.save(
    test_scaled_path,
    X_test_scaled_pa
)


# ============================================================
# SAVE SCALER
# ============================================================

joblib.dump(
    scaler,
    SCALER_PATH
)


# ============================================================
# FINAL REPORT
# ============================================================

print("\n" + "=" * 100)
print("STEP 8 — PADDING-AWARE SCALING COMPLETE")
print("=" * 100)

print(
    "X_train_scaled_padding_aware:",
    X_train_scaled_pa.shape
)

print(
    "X_val_scaled_padding_aware  :",
    X_val_scaled_pa.shape
)

print(
    "X_test_scaled_padding_aware :",
    X_test_scaled_pa.shape
)

print(
    "Train NaNs:",
    int(
        np.isnan(
            X_train_scaled_pa
        ).sum()
    )
)

print(
    "Val NaNs:",
    int(
        np.isnan(
            X_val_scaled_pa
        ).sum()
    )
)

print(
    "Test NaNs:",
    int(
        np.isnan(
            X_test_scaled_pa
        ).sum()
    )
)

print(
    "\nScaler:",
    SCALER_PATH
)

print(
    "\nSaved:"
)

print(
    train_scaled_path
)

print(
    val_scaled_path
)

print(
    test_scaled_path
)

print("\n" + "=" * 100)
print(
    "🎉 STEP 8 COMPLETE — PADDING-AWARE DATA READY"
)
print("=" * 100)
```

```python
# ============================================================
# PERSON 2 — STEP 9
# FINAL DATASET VERIFICATION
# ============================================================

import os
import numpy as np
import pandas as pd
import joblib

SEQUENCE_ROOT = "/content/person2_output/sequences_32"

EXPECTED = {
    "train": {
        "samples": 1334,
        "shape": (1334, 32, 206)
    },
    "val": {
        "samples": 313,
        "shape": (313, 32, 206)
    },
    "test": {
        "samples": 294,
        "shape": (294, 32, 206)
    }
}

SPLITS = ["train", "val", "test"]

# ============================================================
# FILE PATHS
# ============================================================

FILES = {
    "train": {
        "X": os.path.join(
            SEQUENCE_ROOT,
            "X_train_scaled_padding_aware.npy"
        ),
        "y": os.path.join(
            SEQUENCE_ROOT,
            "y_train.npy"
        ),
        "meta": os.path.join(
            SEQUENCE_ROOT,
            "train_sequence_metadata.csv"
        )
    },

    "val": {
        "X": os.path.join(
            SEQUENCE_ROOT,
            "X_val_scaled_padding_aware.npy"
        ),
        "y": os.path.join(
            SEQUENCE_ROOT,
            "y_val.npy"
        ),
        "meta": os.path.join(
            SEQUENCE_ROOT,
            "val_sequence_metadata.csv"
        )
    },

    "test": {
        "X": os.path.join(
            SEQUENCE_ROOT,
            "X_test_scaled_padding_aware.npy"
        ),
        "y": os.path.join(
            SEQUENCE_ROOT,
            "y_test.npy"
        ),
        "meta": os.path.join(
            SEQUENCE_ROOT,
            "test_sequence_metadata.csv"
        )
    }
}

SCALER_PATH = os.path.join(
    SEQUENCE_ROOT,
    "feature_scaler_padding_aware.pkl"
)

FEATURE_NAMES_PATH = os.path.join(
    SEQUENCE_ROOT,
    "feature_names.txt"
)

# ============================================================
# HEADER
# ============================================================

print("=" * 100)
print("PERSON 2 — STEP 9: FINAL DATASET VERIFICATION")
print("=" * 100)


# ============================================================
# CHECK REQUIRED FILES
# ============================================================

print("\n" + "-" * 100)
print("1. REQUIRED FILE CHECK")
print("-" * 100)

all_files_exist = True

for split in SPLITS:

    for key, path in FILES[split].items():

        exists = os.path.exists(path)

        print(
            f"{split:5s} {key:5s}:",
            "✓" if exists else "✗",
            path
        )

        if not exists:
            all_files_exist = False

for key, path in {
    "scaler": SCALER_PATH,
    "feature_names": FEATURE_NAMES_PATH
}.items():

    exists = os.path.exists(path)

    print(
        f"{key:15s}:",
        "✓" if exists else "✗",
        path
    )

    if not exists:
        all_files_exist = False

assert all_files_exist, (
    "One or more required files are missing."
)

print("\n✅ All required files exist.")


# ============================================================
# LOAD FEATURE NAMES
# ============================================================

print("\n" + "-" * 100)
print("2. FEATURE NAME VERIFICATION")
print("-" * 100)

with open(
    FEATURE_NAMES_PATH,
    "r"
) as f:

    feature_names = [
        line.strip()
        for line in f
        if line.strip()
    ]

print(
    "Feature names:",
    len(feature_names)
)

assert len(feature_names) == 206

assert len(set(feature_names)) == 206

print(
    "Duplicate feature names:",
    len(feature_names) -
    len(set(feature_names))
)

print(
    "First 5:",
    feature_names[:5]
)

print(
    "Last 5:",
    feature_names[-5:]
)

print("\n✅ Exactly 206 unique feature names.")


# ============================================================
# LOAD SCALER
# ============================================================

print("\n" + "-" * 100)
print("3. SCALER VERIFICATION")
print("-" * 100)

scaler = joblib.load(
    SCALER_PATH
)

print(
    "Scaler type:",
    type(scaler).__name__
)

print(
    "Scaler features:",
    scaler.n_features_in_
)

assert scaler.n_features_in_ == 206

assert len(scaler.mean_) == 206
assert len(scaler.scale_) == 206

assert np.isfinite(
    scaler.mean_
).all()

assert np.isfinite(
    scaler.scale_
).all()

print(
    "Scaler mean shape:",
    scaler.mean_.shape
)

print(
    "Scaler scale shape:",
    scaler.scale_.shape
)

print("\n✅ Scaler contains exactly 206 features.")


# ============================================================
# LOAD AND VERIFY EACH SPLIT
# ============================================================

verification = {}

for split in SPLITS:

    print("\n" + "=" * 100)
    print(f"4. VERIFYING {split.upper()}")
    print("=" * 100)

    X = np.load(
        FILES[split]["X"]
    )

    y = np.load(
        FILES[split]["y"]
    )

    meta = pd.read_csv(
        FILES[split]["meta"]
    )

    expected_shape = EXPECTED[
        split
    ]["shape"]

    expected_samples = EXPECTED[
        split
    ]["samples"]

    # --------------------------------------------------------
    # SHAPES
    # --------------------------------------------------------

    print(
        "X shape:",
        X.shape
    )

    print(
        "Expected:",
        expected_shape
    )

    print(
        "y shape:",
        y.shape
    )

    print(
        "Metadata rows:",
        len(meta)
    )

    assert X.shape == expected_shape
    assert y.shape == (
        expected_samples,
    )
    assert len(meta) == expected_samples

    # --------------------------------------------------------
    # DTYPE
    # --------------------------------------------------------

    print(
        "X dtype:",
        X.dtype
    )

    print(
        "y dtype:",
        y.dtype
    )

    assert X.dtype == np.float32
    assert y.dtype == np.int64

    # --------------------------------------------------------
    # NaN / INF
    # --------------------------------------------------------

    nan_count = int(
        np.isnan(X).sum()
    )

    inf_count = int(
        np.isinf(X).sum()
    )

    print(
        "NaNs:",
        nan_count
    )

    print(
        "Infs:",
        inf_count
    )

    assert nan_count == 0
    assert inf_count == 0

    # --------------------------------------------------------
    # LABELS
    # --------------------------------------------------------

    labels = sorted(
        np.unique(y).tolist()
    )

    print(
        "Labels:",
        labels
    )

    assert set(labels).issubset(
        {0, 1, 2, 3, 4, 5}
    )

    # --------------------------------------------------------
    # METADATA / SEQUENCE IDS
    # --------------------------------------------------------

    assert (
        meta["sequence_index"].tolist()
        ==
        list(range(expected_samples))
    )

    assert meta[
        "segment_uid"
    ].nunique() == expected_samples

    print(
        "Unique segment UIDs:",
        meta["segment_uid"].nunique()
    )

    # --------------------------------------------------------
    # ORIGINAL LENGTHS
    # --------------------------------------------------------

    lengths = (
        meta["original_length"]
        .astype(int)
        .to_numpy()
    )

    padded = (
        meta["padded_frames"]
        .astype(int)
        .to_numpy()
    )

    assert np.all(
        lengths >= 1
    )

    assert np.all(
        lengths <= 32
    )

    assert np.all(
        padded == 32 - lengths
    )

    print(
        "Original length min:",
        lengths.min()
    )

    print(
        "Original length max:",
        lengths.max()
    )

    print(
        "Padded sequences:",
        int((lengths < 32).sum())
    )

    # --------------------------------------------------------
    # VERIFY PADDING IS EXACTLY ZERO
    # --------------------------------------------------------

    padding_values = []

    for i, length in enumerate(lengths):

        if length < 32:

            padding_values.append(
                X[
                    i,
                    length:,
                    :
                ]
            )

    if padding_values:

        padding_values = np.concatenate(
            padding_values,
            axis=0
        )

        padding_max_abs = float(
            np.max(
                np.abs(
                    padding_values
                )
            )
        )

    else:

        padding_max_abs = 0.0

    print(
        "Maximum absolute padded value:",
        padding_max_abs
    )

    assert padding_max_abs == 0.0

    print(
        "✓ Padding remains exactly zero"
    )

    # --------------------------------------------------------
    # VERIFY REAL FRAMES ARE FINITE
    # --------------------------------------------------------

    real_values = []

    for i, length in enumerate(lengths):

        real_values.append(
            X[
                i,
                :length,
                :
            ]
        )

    real_values = np.concatenate(
        real_values,
        axis=0
    )

    assert np.isfinite(
        real_values
    ).all()

    # --------------------------------------------------------
    # REAL DATA STATISTICS
    # --------------------------------------------------------

    real_mean = float(
        real_values.mean()
    )

    real_std = float(
        real_values.std()
    )

    print(
        "Real-frame mean:",
        real_mean
    )

    print(
        "Real-frame std :",
        real_std
    )

    # --------------------------------------------------------
    # SAVE RESULTS
    # --------------------------------------------------------

    verification[split] = {
        "samples": len(X),
        "shape": X.shape,
        "features": X.shape[-1],
        "nan": nan_count,
        "inf": inf_count,
        "labels": labels,
        "unique_segments":
            meta["segment_uid"].nunique(),
        "min_length":
            int(lengths.min()),
        "max_length":
            int(lengths.max()),
        "padded_sequences":
            int((lengths < 32).sum()),
        "padding_max_abs":
            padding_max_abs,
        "real_mean":
            real_mean,
        "real_std":
            real_std
    }

    print(
        f"\n✅ {split.upper()} PASSED"
    )


# ============================================================
# TRAIN / VAL / TEST LABEL DISTRIBUTION
# ============================================================

print("\n" + "=" * 100)
print("5. LABEL DISTRIBUTION")
print("=" * 100)

for split in SPLITS:

    y = np.load(
        FILES[split]["y"]
    )

    counts = pd.Series(
        y
    ).value_counts(
        sort=False
    ).sort_index()

    print(
        f"\n{split.upper()}"
    )

    for label in range(6):

        print(
            f"  Label {label}:",
            int(
                counts.get(
                    label,
                    0
                )
            )
        )


# ============================================================
# FINAL CROSS-SPLIT CHECK
# ============================================================

print("\n" + "=" * 100)
print("6. FINAL CROSS-SPLIT CHECK")
print("=" * 100)

train_meta = pd.read_csv(
    FILES["train"]["meta"],
    usecols=["segment_uid"]
)

val_meta = pd.read_csv(
    FILES["val"]["meta"],
    usecols=["segment_uid"]
)

test_meta = pd.read_csv(
    FILES["test"]["meta"],
    usecols=["segment_uid"]
)

train_segments = set(
    train_meta["segment_uid"]
)

val_segments = set(
    val_meta["segment_uid"]
)

test_segments = set(
    test_meta["segment_uid"]
)

train_val = (
    train_segments &
    val_segments
)

train_test = (
    train_segments &
    test_segments
)

val_test = (
    val_segments &
    test_segments
)

print(
    "Train ∩ Val :",
    len(train_val)
)

print(
    "Train ∩ Test:",
    len(train_test)
)

print(
    "Val ∩ Test  :",
    len(val_test)
)

assert len(train_val) == 0
assert len(train_test) == 0
assert len(val_test) == 0

print(
    "\n✅ No segment leakage between splits."
)


# ============================================================
# FINAL PASS / FAIL
# ============================================================

print("\n" + "=" * 100)
print("PERSON 2 — FINAL VERIFICATION")
print("=" * 100)

all_pass = True

for split in SPLITS:

    r = verification[split]

    checks = [
        r["samples"] == EXPECTED[split]["samples"],
        r["shape"] == EXPECTED[split]["shape"],
        r["features"] == 206,
        r["nan"] == 0,
        r["inf"] == 0,
        r["unique_segments"] == EXPECTED[split]["samples"],
        r["min_length"] >= 1,
        r["max_length"] <= 32,
        r["padding_max_abs"] == 0.0,
        r["labels"] == sorted(
            list(
                set(r["labels"]) &
                {0, 1, 2, 3, 4, 5}
            )
        )
    ]

    if all(checks):

        print(
            f"✅ {split.upper()} — PASS"
        )

    else:

        print(
            f"❌ {split.upper()} — FAIL"
        )

        all_pass = False


if (
    len(train_val) == 0
    and len(train_test) == 0
    and len(val_test) == 0
):

    print(
        "✅ SPLIT LEAKAGE — PASS"
    )

else:

    print(
        "❌ SPLIT LEAKAGE — FAIL"
    )

    all_pass = False


print("\n" + "=" * 100)

if all_pass:

    print(
        "🎉🎉🎉 PERSON 2 PIPELINE COMPLETE 🎉🎉🎉"
    )

    print(
        "\nFinal model inputs:"
    )

    print(
        "X_train:",
        EXPECTED["train"]["shape"]
    )

    print(
        "X_val  :",
        EXPECTED["val"]["shape"]
    )

    print(
        "X_test :",
        EXPECTED["test"]["shape"]
    )

    print(
        "\n206 spatial/interaction features"
    )

    print(
        "32 frames per sequence"
    )

    print(
        "Padding-aware StandardScaler"
    )

    print(
        "No NaNs / Infs"
    )

    print(
        "No segment leakage"
    )

    print(
        "\n✅ READY FOR PERSON 3 — GRU/LSTM"
    )

else:

    print(
        "❌ FINAL VERIFICATION FAILED"
    )

    print(
        "DO NOT hand the dataset to Person 3 yet."
    )

print("=" * 100)
```
